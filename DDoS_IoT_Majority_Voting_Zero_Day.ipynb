```python
import zipfile
import pandas as pd
import numpy as np

zip_path = "/content/drive/MyDrive/DDoS_ML_Cybersecurity/dataset/CSV-01-12.zip"

files = [
    "DrDoS_DNS.csv",
    "DrDoS_NetBIOS.csv",
    "DrDoS_LDAP.csv",
    "DrDoS_UDP.csv",
    "DrDoS_SNMP.csv"
]

for filename in files:
    print(f"\nChecking {filename}...")

    nan_total = 0
    inf_total = 0
    rows = 0

    with zipfile.ZipFile(zip_path, "r") as z:
        with z.open(f"01-12/{filename}") as f:

            for chunk in pd.read_csv(
                f,
                encoding="latin1",
                chunksize=100_000,
                low_memory=False
            ):
                chunk.columns = chunk.columns.str.strip()

                # Exclude non-feature columns
                feature_cols = [
                    col for col in chunk.select_dtypes(include="number").columns
                    if col != "Unnamed: 0"
                ]

                nan_total += chunk[feature_cols].isna().sum().sum()
                inf_total += np.isinf(chunk[feature_cols]).sum().sum()
                rows += len(chunk)

    print("Rows:", rows)
    print("NaN values:", int(nan_total))
    print("Infinite values:", int(inf_total))
```

```python
import zipfile
import pandas as pd
import numpy as np

zip_path = "/content/drive/MyDrive/DDoS_ML_Cybersecurity/dataset/CSV-01-12.zip"

files = {
    "DNS": "DrDoS_DNS.csv",
    "NetBIOS": "DrDoS_NetBIOS.csv",
    "LDAP": "DrDoS_LDAP.csv",
    "UDP": "DrDoS_UDP.csv",
    "SNMP": "DrDoS_SNMP.csv"
}

ATTACK_SAMPLES = 10_000
BENIGN_SAMPLES = 10_000

sampled_data = []

with zipfile.ZipFile(zip_path, "r") as z:

    for attack_type, filename in files.items():

        print(f"Processing {attack_type}...")

        attack_parts = []
        benign_parts = []

        attack_count = 0
        benign_count = 0

        with z.open(f"01-12/{filename}") as f:

            for chunk in pd.read_csv(
                f,
                encoding="latin1",
                chunksize=100_000,
                low_memory=False
            ):

                chunk.columns = chunk.columns.str.strip()

                # Remove index-like column
                chunk = chunk.drop(columns=["Unnamed: 0"])

                # Keep only numeric columns + Label
                numeric_cols = chunk.select_dtypes(
                    include="number"
                ).columns.tolist()

                chunk = chunk[numeric_cols + ["Label"]]

                # Clean infinite values
                chunk[numeric_cols] = chunk[numeric_cols].replace(
                    [np.inf, -np.inf],
                    np.nan
                )

                # Median imputation
                for col in numeric_cols:
                    if chunk[col].isna().any():
                        chunk[col] = chunk[col].fillna(
                            chunk[col].median()
                        )

                # Separate classes
                attack_rows = chunk[
                    chunk["Label"] == f"DrDoS_{attack_type}"
                ]

                benign_rows = chunk[
                    chunk["Label"] == "BENIGN"
                ]

                # Collect attack samples
                if attack_count < ATTACK_SAMPLES:
                    needed = ATTACK_SAMPLES - attack_count
                    selected = attack_rows.head(needed)

                    attack_parts.append(selected)
                    attack_count += len(selected)

                # Collect benign samples
                if benign_count < BENIGN_SAMPLES:
                    needed = BENIGN_SAMPLES - benign_count
                    selected = benign_rows.head(needed)

                    benign_parts.append(selected)
                    benign_count += len(selected)

                # Stop once enough samples are collected
                if (
                    attack_count >= ATTACK_SAMPLES
                    and benign_count >= BENIGN_SAMPLES
                ):
                    break

        attack_df = pd.concat(attack_parts, ignore_index=True)
        benign_df = pd.concat(benign_parts, ignore_index=True)

        attack_df["Attack_Type"] = attack_type
        benign_df["Attack_Type"] = attack_type

        combined = pd.concat(
            [attack_df, benign_df],
            ignore_index=True
        )

        sampled_data.append(combined)

        print(
            f"  Attack: {len(attack_df)} | "
            f"BENIGN: {len(benign_df)}"
        )

# Combine all five datasets
working_df = pd.concat(
    sampled_data,
    ignore_index=True
)

# Shuffle
working_df = working_df.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)

print("\nFinal working dataset:", working_df.shape)
print("\nLabels:")
print(working_df["Label"].value_counts())

print("\nAttack types:")
print(working_df["Attack_Type"].value_counts())
```

```python
print("Final working dataset:", working_df.shape)

print("\nLabel distribution:")
print(working_df["Label"].value_counts())

print("\nAttack-type distribution:")
print(working_df["Attack_Type"].value_counts())

print("\nMissing values:", working_df.isna().sum().sum())
print("Infinite values:", np.isinf(
    working_df.select_dtypes(include="number")
).sum().sum())
```

```python
# Check missing values before fixing
print("Missing values before:", working_df.isna().sum().sum())

# Numeric feature columns
numeric_features = working_df.select_dtypes(
    include="number"
).columns.tolist()

# Fill remaining NaNs using global feature medians
for col in numeric_features:
    median_value = working_df[col].median()

    # If a column somehow has no valid median, use 0
    if pd.isna(median_value):
        median_value = 0

    working_df[col] = working_df[col].fillna(median_value)

print("Missing values after:", working_df.isna().sum().sum())

print("Infinite values after:", np.isinf(
    working_df[numeric_features]
).sum().sum())
```

```python
from sklearn.preprocessing import StandardScaler

# Separate features and labels
X = working_df.drop(
    columns=["Label", "Attack_Type"]
)

y = working_df["Label"]

# Initialize scaler
scaler = StandardScaler()

# Apply Z-score normalization
X_scaled = scaler.fit_transform(X)

# Convert back to DataFrame
X_scaled = pd.DataFrame(
    X_scaled,
    columns=X.columns,
    index=X.index
)

print("Scaled feature shape:", X_scaled.shape)

print("\nMean of first 5 features:")
print(X_scaled.iloc[:, :5].mean())

print("\nStandard deviation of first 5 features:")
print(X_scaled.iloc[:, :5].std())
```

```python
print(y.value_counts())
```

```python
from imblearn.over_sampling import SMOTE
from sklearn.utils import resample

# --------------------------------------------------
# 1. Undersample BENIGN to 10,000
# --------------------------------------------------

data_for_sampling = X_scaled.copy()
data_for_sampling["Label"] = y.values

benign = data_for_sampling[
    data_for_sampling["Label"] == "BENIGN"
]

attacks = data_for_sampling[
    data_for_sampling["Label"] != "BENIGN"
]

benign_under = resample(
    benign,
    replace=False,
    n_samples=10_000,
    random_state=42
)

balanced_before_smote = pd.concat(
    [attacks, benign_under],
    ignore_index=True
)

print("Before SMOTE:")
print(balanced_before_smote["Label"].value_counts())

# --------------------------------------------------
# 2. Apply SMOTE
# --------------------------------------------------

X_before_smote = balanced_before_smote.drop(
    columns=["Label"]
)

y_before_smote = balanced_before_smote["Label"]

smote = SMOTE(
    random_state=42,
    k_neighbors=5
)

X_resampled, y_resampled = smote.fit_resample(
    X_before_smote,
    y_before_smote
)

print("\nAfter SMOTE:")
print(pd.Series(y_resampled).value_counts())

print("\nFinal shape:")
print(X_resampled.shape)
```

```python
from sklearn.feature_selection import mutual_info_classif

# Calculate Information Gain / Mutual Information
mi_scores = mutual_info_classif(
    X_resampled,
    y_resampled,
    random_state=42
)

# Create ranked feature table
feature_scores = pd.DataFrame({
    "Feature": X_resampled.columns,
    "Information_Gain": mi_scores
})

feature_scores = feature_scores.sort_values(
    by="Information_Gain",
    ascending=False
).reset_index(drop=True)

print("Top 20 features by Information Gain:")
print(feature_scores.head(20).to_string(index=False))
```

```python
from sklearn.feature_selection import mutual_info_classif

attack_types = [
    "DrDoS_DNS",
    "DrDoS_NetBIOS",
    "DrDoS_LDAP",
    "DrDoS_UDP",
    "DrDoS_SNMP"
]

y_resampled_series = pd.Series(
    y_resampled,
    name="Label"
)

attack_feature_rankings = {}

for attack in attack_types:

    print(f"\n{'='*60}")
    print(f"Feature selection: {attack} vs BENIGN")
    print(f"{'='*60}")

    # Select this attack + BENIGN
    mask = (
        (y_resampled_series == attack) |
        (y_resampled_series == "BENIGN")
    )

    X_attack = X_resampled.loc[mask]
    y_attack = y_resampled_series.loc[mask]

    # Attack = 1, BENIGN = 0
    y_binary = (y_attack == attack).astype(int)

    # Information Gain / Mutual Information
    mi = mutual_info_classif(
        X_attack,
        y_binary,
        random_state=42
    )

    ranking = pd.DataFrame({
        "Feature": X_attack.columns,
        "Information_Gain": mi
    }).sort_values(
        "Information_Gain",
        ascending=False
    ).reset_index(drop=True)

    attack_feature_rankings[attack] = ranking

    print(ranking.head(15).to_string(index=False))
```

```python
import matplotlib.pyplot as plt

for attack, ranking in attack_feature_rankings.items():

    scores = ranking["Information_Gain"].values
    cumulative = scores.cumsum() / scores.sum()

    plt.figure(figsize=(8, 5))
    plt.plot(
        range(1, len(scores) + 1),
        cumulative,
        marker="o",
        markersize=3
    )

    plt.xlabel("Number of Features")
    plt.ylabel("Cumulative Information Gain")
    plt.title(f"Feature Selection — {attack}")

    plt.axhline(
        0.95,
        linestyle="--",
        label="95% cumulative Information Gain"
    )

    plt.legend()
    plt.grid(True)
    plt.show()
```

```python
selected_feature_counts = {}

for attack, ranking in attack_feature_rankings.items():

    scores = ranking["Information_Gain"].values

    cumulative_gain = scores.cumsum() / scores.sum()

    # First feature count that reaches 95%
    n_features = (
        cumulative_gain >= 0.95
    ).argmax() + 1

    selected_feature_counts[attack] = n_features

    print(
        f"{attack}: {n_features} features "
        f"needed for 95% cumulative Information Gain"
    )
```

```python
selected_features = {}

for attack, ranking in attack_feature_rankings.items():

    n = selected_feature_counts[attack]

    selected_features[attack] = ranking.head(n)["Feature"].tolist()

    print(f"\n{attack}: {len(selected_features[attack])} features")
    print(selected_features[attack])
```

```python
import zipfile
import pandas as pd
import numpy as np

zip_path = "/content/drive/MyDrive/DDoS_ML_Cybersecurity/dataset/CSV-01-12.zip"

files = {
    "DNS": "DrDoS_DNS.csv",
    "NetBIOS": "DrDoS_NetBIOS.csv",
    "LDAP": "DrDoS_LDAP.csv",
    "UDP": "DrDoS_UDP.csv",
    "SNMP": "DrDoS_SNMP.csv"
}

ATTACK_SAMPLES = 50_000
RANDOM_STATE = 42

rng = np.random.default_rng(RANDOM_STATE)

sampled_data = []

for attack_type, filename in files.items():

    print(f"\nProcessing {attack_type}...")

    with zipfile.ZipFile(zip_path, "r") as z:

        # --------------------------------------------------
        # Get FIXED column structure from the header/sample
        # --------------------------------------------------
        with z.open(f"01-12/{filename}") as f:
            sample = pd.read_csv(
                f,
                encoding="latin1",
                nrows=1000,
                low_memory=False
            )

        sample.columns = sample.columns.str.strip()

        # Remove index-like column
        sample = sample.drop(columns=["Unnamed: 0"])

        # Fixed numeric feature list
        feature_cols = sample.select_dtypes(
            include="number"
        ).columns.tolist()

        columns = feature_cols + ["Label"]

        # --------------------------------------------------
        # Reservoirs
        # --------------------------------------------------
        attack_reservoir = []
        benign_rows_all = []

        attack_seen = 0

        # --------------------------------------------------
        # Read complete file in chunks
        # --------------------------------------------------
        with z.open(f"01-12/{filename}") as f:

            for chunk in pd.read_csv(
                f,
                encoding="latin1",
                chunksize=100_000,
                low_memory=False
            ):

                chunk.columns = chunk.columns.str.strip()

                # Remove index-like column
                chunk = chunk.drop(columns=["Unnamed: 0"])

                # Keep exactly the same columns
                chunk = chunk[columns]

                # Replace infinite values
                chunk[feature_cols] = chunk[
                    feature_cols
                ].replace(
                    [np.inf, -np.inf],
                    np.nan
                )

                # ------------------------------
                # Attack rows
                # ------------------------------
                attack_rows = chunk[
                    chunk["Label"] == f"DrDoS_{attack_type}"
                ]

                if len(attack_rows) > 0:

                    attack_array = attack_rows.to_numpy()

                    for row in attack_array:

                        attack_seen += 1

                        if len(attack_reservoir) < ATTACK_SAMPLES:

                            attack_reservoir.append(row)

                        else:

                            j = rng.integers(
                                0,
                                attack_seen
                            )

                            if j < ATTACK_SAMPLES:
                                attack_reservoir[j] = row

                # ------------------------------
                # BENIGN rows
                # ------------------------------
                benign_rows = chunk[
                    chunk["Label"] == "BENIGN"
                ]

                if len(benign_rows) > 0:
                    benign_rows_all.append(benign_rows)

    # --------------------------------------------------
    # Convert attack reservoir
    # --------------------------------------------------

    attack_df = pd.DataFrame(
        attack_reservoir,
        columns=columns
    )

    # Combine all BENIGN rows
    benign_df = pd.concat(
        benign_rows_all,
        ignore_index=True
    )

    # Add attack type
    attack_df["Attack_Type"] = attack_type
    benign_df["Attack_Type"] = attack_type

    combined = pd.concat(
        [attack_df, benign_df],
        ignore_index=True
    )

    sampled_data.append(combined)

    print(
        f"  Random attack samples: {len(attack_df)}"
    )

    print(
        f"  BENIGN samples: {len(benign_df)}"
    )


# --------------------------------------------------
# Combine all five datasets
# --------------------------------------------------

working_df = pd.concat(
    sampled_data,
    ignore_index=True
)

# Shuffle
working_df = working_df.sample(
    frac=1,
    random_state=RANDOM_STATE
).reset_index(drop=True)

print("\n========================================")
print("FINAL WORKING DATASET")
print("========================================")

print("Shape:", working_df.shape)

print("\nLabels:")
print(working_df["Label"].value_counts())

print("\nAttack types:")
print(working_df["Attack_Type"].value_counts())
```

```python
import zipfile
import pandas as pd
import numpy as np
import gc

zip_path = "/content/drive/MyDrive/DDoS_ML_Cybersecurity/dataset/CSV-01-12.zip"

files = {
    "DNS": "DrDoS_DNS.csv",
    "NetBIOS": "DrDoS_NetBIOS.csv",
    "LDAP": "DrDoS_LDAP.csv",
    "UDP": "DrDoS_UDP.csv",
    "SNMP": "DrDoS_SNMP.csv"
}

ATTACK_SAMPLES = 10_000

sampled_data = []

with zipfile.ZipFile(zip_path, "r") as z:

    for attack_type, filename in files.items():

        print(f"\nProcessing {attack_type}...")

        attack_parts = []
        benign_parts = []

        attack_count = 0

        with z.open(f"01-12/{filename}") as f:

            for chunk in pd.read_csv(
                f,
                encoding="latin1",
                chunksize=100_000,
                low_memory=False
            ):

                chunk.columns = chunk.columns.str.strip()

                # Remove index-like column
                chunk = chunk.drop(columns=["Unnamed: 0"])

                # Keep numeric features + Label
                numeric_cols = chunk.select_dtypes(
                    include="number"
                ).columns.tolist()

                chunk = chunk[numeric_cols + ["Label"]]

                # Replace infinite values with NaN
                chunk[numeric_cols] = chunk[numeric_cols].replace(
                    [np.inf, -np.inf],
                    np.nan
                )

                # Attack samples
                attack_rows = chunk[
                    chunk["Label"] == f"DrDoS_{attack_type}"
                ]

                if attack_count < ATTACK_SAMPLES:

                    needed = ATTACK_SAMPLES - attack_count

                    selected = attack_rows.head(needed)

                    attack_parts.append(selected)

                    attack_count += len(selected)

                # All available benign samples
                benign_rows = chunk[
                    chunk["Label"] == "BENIGN"
                ]

                if len(benign_rows) > 0:
                    benign_parts.append(benign_rows)

                # Stop once attack sample target is reached
                # and all chunks have been processed for BENIGN
                # so we retain all BENIGN rows.

        attack_df = pd.concat(
            attack_parts,
            ignore_index=True
        )

        benign_df = pd.concat(
            benign_parts,
            ignore_index=True
        )

        attack_df["Attack_Type"] = attack_type
        benign_df["Attack_Type"] = attack_type

        combined = pd.concat(
            [attack_df, benign_df],
            ignore_index=True
        )

        sampled_data.append(combined)

        print(
            f"Attack: {len(attack_df)} | "
            f"BENIGN: {len(benign_df)}"
        )

        del attack_parts, benign_parts
        gc.collect()

# Combine all five
working_df = pd.concat(
    sampled_data,
    ignore_index=True
)

# Shuffle
working_df = working_df.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)

print("\n==============================")
print("WORKING DATASET")
print("==============================")

print("Shape:", working_df.shape)

print("\nLabels:")
print(working_df["Label"].value_counts())

print("\nAttack types:")
print(working_df["Attack_Type"].value_counts())
```

```python
from sklearn.model_selection import train_test_split

attack_types = [
    "DrDoS_DNS",
    "DrDoS_NetBIOS",
    "DrDoS_LDAP",
    "DrDoS_UDP",
    "DrDoS_SNMP"
]

train_test_data = {}

for attack in attack_types:

    print(f"\nPreparing {attack} vs BENIGN...")

    # Select attack + BENIGN
    mask = (
        (working_df["Label"] == attack) |
        (working_df["Label"] == "BENIGN")
    )

    attack_df = working_df.loc[mask].copy()

    # Features
    X_attack = attack_df.drop(
        columns=["Label", "Attack_Type"]
    )

    # Binary target
    # 1 = attack
    # 0 = BENIGN
    y_attack = (
        attack_df["Label"] == attack
    ).astype(int)

    # 80/20 stratified split
    X_train, X_test, y_train, y_test = train_test_split(
        X_attack,
        y_attack,
        test_size=0.20,
        random_state=42,
        stratify=y_attack
    )

    train_test_data[attack] = {
        "X_train": X_train,
        "X_test": X_test,
        "y_train": y_train,
        "y_test": y_test
    }

    print("Train:", X_train.shape)
    print("Test :", X_test.shape)
    print(
        "Train labels:",
        y_train.value_counts().to_dict()
    )
    print(
        "Test labels:",
        y_test.value_counts().to_dict()
    )
```

```python
from sklearn.preprocessing import StandardScaler

scalers = {}

for attack in attack_types:

    X_train = train_test_data[attack]["X_train"]
    X_test = train_test_data[attack]["X_test"]

    scaler = StandardScaler()

    # Fit ONLY on training data
    X_train_scaled = scaler.fit_transform(X_train)

    # Transform test data using the training scaler
    X_test_scaled = scaler.transform(X_test)

    X_train_scaled = pd.DataFrame(
        X_train_scaled,
        columns=X_train.columns,
        index=X_train.index
    )

    X_test_scaled = pd.DataFrame(
        X_test_scaled,
        columns=X_test.columns,
        index=X_test.index
    )

    scalers[attack] = scaler

    train_test_data[attack]["X_train_scaled"] = X_train_scaled
    train_test_data[attack]["X_test_scaled"] = X_test_scaled

    print(
        f"{attack}: "
        f"Train {X_train_scaled.shape} | "
        f"Test {X_test_scaled.shape}"
    )
```

```python
from sklearn.utils import resample
from imblearn.over_sampling import SMOTE

for attack in attack_types:

    X_train = train_test_data[attack]["X_train_scaled"]
    y_train = train_test_data[attack]["y_train"]

    # Combine for undersampling
    train_df = X_train.copy()
    train_df["target"] = y_train.values

    # Separate classes
    benign = train_df[train_df["target"] == 0]
    attack_rows = train_df[train_df["target"] == 1]

    # Undersample BENIGN to match attack count
    benign_under = resample(
        benign,
        replace=False,
        n_samples=len(attack_rows),
        random_state=42
    )

    balanced_train = pd.concat(
        [benign_under, attack_rows],
        ignore_index=True
    )

    X_balanced = balanced_train.drop(columns=["target"])
    y_balanced = balanced_train["target"]

    # Apply SMOTE
    smote = SMOTE(
        random_state=42,
        k_neighbors=5
    )

    X_resampled, y_resampled = smote.fit_resample(
        X_balanced,
        y_balanced
    )

    # Store
    train_test_data[attack]["X_train_resampled"] = X_resampled
    train_test_data[attack]["y_train_resampled"] = y_resampled

    print(f"\n{attack}")
    print("Before sampling:", y_train.value_counts().to_dict())
    print(
        "After sampling:",
        pd.Series(y_resampled).value_counts().to_dict()
    )
    print("Final training shape:", X_resampled.shape)
```

```python
# Fix missing values using TRAINING medians only

for attack in attack_types:

    X_train = train_test_data[attack]["X_train"].copy()
    X_test = train_test_data[attack]["X_test"].copy()

    # Calculate medians ONLY from training data
    train_medians = X_train.median()

    # Fill training NaNs
    X_train = X_train.fillna(train_medians)

    # Fill test NaNs using the SAME training medians
    X_test = X_test.fillna(train_medians)

    # Store cleaned data
    train_test_data[attack]["X_train"] = X_train
    train_test_data[attack]["X_test"] = X_test

    # Verify
    print(
        f"{attack}: "
        f"Train NaN = {X_train.isna().sum().sum()}, "
        f"Test NaN = {X_test.isna().sum().sum()}"
    )
```

```python
# ============================================================
# 10 CANDIDATE MODELS
# Based on the base paper's 10 algorithms
# ============================================================

from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.base import BaseEstimator, ClassifierMixin
from sklearn.metrics import accuracy_score
import numpy as np


# ------------------------------------------------------------
# 1. GLM - Generalized Linear Model
# ------------------------------------------------------------
class GLMClassifier(BaseEstimator, ClassifierMixin):

    def __init__(self):
        self.model = None

    def fit(self, X, y):
        import statsmodels.api as sm

        X_array = np.asarray(X)
        X_array = sm.add_constant(X_array, has_constant="add")

        self.model = sm.GLM(
            y,
            X_array,
            family=sm.families.Binomial()
        ).fit()

        return self

    def predict_proba(self, X):
        import statsmodels.api as sm

        X_array = np.asarray(X)
        X_array = sm.add_constant(X_array, has_constant="add")

        probability = self.model.predict(X_array)

        return np.column_stack([
            1 - probability,
            probability
        ])

    def predict(self, X):
        probabilities = self.predict_proba(X)[:, 1]
        return (probabilities >= 0.5).astype(int)


# ------------------------------------------------------------
# 2. Linear Regression converted to binary classification
# ------------------------------------------------------------
class LinearRegressionClassifier(BaseEstimator, ClassifierMixin):

    def __init__(self):
        self.model = LinearRegression()

    def fit(self, X, y):
        self.model.fit(X, y)
        return self

    def predict_proba(self, X):
        probability = np.clip(
            self.model.predict(X),
            0,
            1
        )

        return np.column_stack([
            1 - probability,
            probability
        ])

    def predict(self, X):
        return (self.model.predict(X) >= 0.5).astype(int)


# ------------------------------------------------------------
# 3. Rule Induction
# Simple one-rule classifier:
# selects the feature + threshold giving the best training
# classification accuracy.
# ------------------------------------------------------------
class RuleInductionClassifier(BaseEstimator, ClassifierMixin):

    def __init__(self):
        self.feature_index = None
        self.threshold = None
        self.left_class = None
        self.right_class = None

    def fit(self, X, y):

        X = np.asarray(X)
        y = np.asarray(y)

        best_accuracy = -1

        for feature in range(X.shape[1]):

            values = X[:, feature]

            # Use quantiles to avoid testing thousands of thresholds
            thresholds = np.unique(
                np.quantile(
                    values,
                    np.linspace(0.05, 0.95, 19)
                )
            )

            for threshold in thresholds:

                left = y[values <= threshold]
                right = y[values > threshold]

                if len(left) == 0 or len(right) == 0:
                    continue

                left_class = int(np.round(left.mean()))
                right_class = int(np.round(right.mean()))

                predictions = np.where(
                    values <= threshold,
                    left_class,
                    right_class
                )

                accuracy = np.mean(predictions == y)

                if accuracy > best_accuracy:
                    best_accuracy = accuracy
                    self.feature_index = feature
                    self.threshold = threshold
                    self.left_class = left_class
                    self.right_class = right_class

        return self

    def predict(self, X):

        X = np.asarray(X)

        return np.where(
            X[:, self.feature_index] <= self.threshold,
            self.left_class,
            self.right_class
        )

    def predict_proba(self, X):

        predictions = self.predict(X)

        return np.column_stack([
            1 - predictions,
            predictions
        ])


# ============================================================
# CREATE ALL 10 CANDIDATE MODELS
# ============================================================

def create_all_models():

    models = {

        # 1. Support Vector Machine
        "SVM": SVC(
            kernel="rbf",
            probability=True,
            random_state=42
        ),

        # 2. Random Forest
        "RF": RandomForestClassifier(
            n_estimators=100,
            random_state=42,
            n_jobs=-1
        ),

        # 3. Decision Tree
        "DT": DecisionTreeClassifier(
            random_state=42
        ),

        # 4. Neural Network
        "NN": MLPClassifier(
            hidden_layer_sizes=(64, 32),
            max_iter=300,
            early_stopping=True,
            random_state=42
        ),

        # 5. Deep Learning
        "DL": MLPClassifier(
            hidden_layer_sizes=(128, 64, 32),
            max_iter=300,
            early_stopping=True,
            random_state=42
        ),

        # 6. Logistic Regression
        "LGR": LogisticRegression(
            max_iter=1000,
            random_state=42
        ),

        # 7. Generalized Linear Model
        "GLM": GLMClassifier(),

        # 8. Naive Bayes
        "NB": GaussianNB(),

        # 9. Linear Regression
        "LR": LinearRegressionClassifier(),

        # 10. Rule Induction
        "RI": RuleInductionClassifier()
    }

    return models


models = create_all_models()

print("10 candidate algorithms successfully created:\n")

for i, name in enumerate(models.keys(), 1):
    print(f"{i}. {name}")
```

```python
# ============================================================
# TRAIN + EVALUATE 10 MODELS ON DNS
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    mean_squared_error
)
import time
import pandas as pd
import numpy as np

attack = "DrDoS_DNS"

X_train = train_test_data[attack]["X_train_selected"]
y_train = train_test_data[attack]["y_train_resampled"]

X_test = train_test_data[attack]["X_test_selected"]
y_test = train_test_data[attack]["y_test"]

dns_models = create_all_models()

dns_results = []
dns_trained_models = {}

for name, model in dns_models.items():

    print(f"Training {name}...")

    start_time = time.time()

    try:
        model.fit(X_train, y_train)

        y_pred = model.predict(X_test)

        runtime = time.time() - start_time

        accuracy = accuracy_score(y_test, y_pred)
        precision = precision_score(
            y_test, y_pred, zero_division=0
        )
        recall = recall_score(
            y_test, y_pred, zero_division=0
        )
        f1 = f1_score(
            y_test, y_pred, zero_division=0
        )

        # RMSE using binary predictions
        rmse = np.sqrt(
            mean_squared_error(y_test, y_pred)
        )

        dns_results.append({
            "Attack": attack,
            "Model": name,
            "Accuracy": accuracy,
            "Precision": precision,
            "Recall": recall,
            "F1": f1,
            "RMSE": rmse,
            "Runtime_sec": runtime
        })

        dns_trained_models[name] = model

        print(
            f"  Accuracy: {accuracy:.4f} | "
            f"F1: {f1:.4f} | "
            f"Runtime: {runtime:.2f}s"
        )

    except Exception as e:

        print(f"  ERROR: {e}")

        dns_results.append({
            "Attack": attack,
            "Model": name,
            "Accuracy": np.nan,
            "Precision": np.nan,
            "Recall": np.nan,
            "F1": np.nan,
            "RMSE": np.nan,
            "Runtime_sec": np.nan
        })

# Convert results to DataFrame
dns_results_df = pd.DataFrame(dns_results)

print("\n" + "=" * 80)
print("DNS — 10 MODEL RESULTS")
print("=" * 80)

display(
    dns_results_df.sort_values(
        by="Accuracy",
        ascending=False
    ).reset_index(drop=True)
)
```

```python
from sklearn.metrics import confusion_matrix

# Add FPR and Classification Error to DNS results

fpr_values = []
classification_errors = []

for _, row in dns_results_df.iterrows():

    model_name = row["Model"]
    model = dns_trained_models[model_name]

    y_pred = model.predict(X_test)

    tn, fp, fn, tp = confusion_matrix(
        y_test,
        y_pred,
        labels=[0, 1]
    ).ravel()

    fpr = fp / (fp + tn) if (fp + tn) > 0 else 0
    classification_error = 1 - row["Accuracy"]

    fpr_values.append(fpr)
    classification_errors.append(classification_error)

dns_results_df["FPR"] = fpr_values
dns_results_df["Classification_Error"] = classification_errors

dns_results_df = dns_results_df.sort_values(
    by="Accuracy",
    ascending=False
).reset_index(drop=True)

display(
    dns_results_df[
        [
            "Model",
            "Accuracy",
            "Precision",
            "Recall",
            "F1",
            "FPR",
            "RMSE",
            "Runtime_sec",
            "Classification_Error"
        ]
    ]
)
```

```python
# ============================================================
# FUNCTION: TRAIN + EVALUATE ALL 10 MODELS
# ============================================================

def evaluate_all_models(attack):

    X_train = train_test_data[attack]["X_train_selected"]
    y_train = train_test_data[attack]["y_train_resampled"]

    X_test = train_test_data[attack]["X_test_selected"]
    y_test = train_test_data[attack]["y_test"]

    attack_models = create_all_models()

    results = []
    trained_models = {}

    print("=" * 70)
    print(f"TRAINING 10 MODELS — {attack}")
    print("=" * 70)

    for name, model in attack_models.items():

        print(f"\nTraining {name}...")

        start_time = time.time()

        try:
            model.fit(X_train, y_train)

            y_pred = model.predict(X_test)

            runtime = time.time() - start_time

            accuracy = accuracy_score(y_test, y_pred)
            precision = precision_score(
                y_test, y_pred, zero_division=0
            )
            recall = recall_score(
                y_test, y_pred, zero_division=0
            )
            f1 = f1_score(
                y_test, y_pred, zero_division=0
            )

            rmse = np.sqrt(
                mean_squared_error(y_test, y_pred)
            )

            tn, fp, fn, tp = confusion_matrix(
                y_test,
                y_pred,
                labels=[0, 1]
            ).ravel()

            fpr = fp / (fp + tn) if (fp + tn) > 0 else 0

            results.append({
                "Attack": attack,
                "Model": name,
                "Accuracy": accuracy,
                "Precision": precision,
                "Recall": recall,
                "F1": f1,
                "FPR": fpr,
                "RMSE": rmse,
                "Runtime_sec": runtime,
                "Classification_Error": 1 - accuracy
            })

            trained_models[name] = model

            print(
                f"  Accuracy: {accuracy:.4f} | "
                f"F1: {f1:.4f} | "
                f"FPR: {fpr:.6f} | "
                f"Runtime: {runtime:.2f}s"
            )

        except Exception as e:

            print(f"  ERROR: {e}")

            results.append({
                "Attack": attack,
                "Model": name,
                "Accuracy": np.nan,
                "Precision": np.nan,
                "Recall": np.nan,
                "F1": np.nan,
                "FPR": np.nan,
                "RMSE": np.nan,
                "Runtime_sec": np.nan,
                "Classification_Error": np.nan
            })

    return pd.DataFrame(results), trained_models
```

```python
dns_results_df, dns_trained_models = evaluate_all_models("DrDoS_DNS")
```

```python
# Combine results from all five attack types

all_model_results = pd.concat(
    [
        dns_results_df,
        netbios_results_df,
        ldap_results_df,
        udp_results_df,
        snmp_results_df
    ],
    ignore_index=True
)

all_model_results = all_model_results.sort_values(
    ["Attack", "Accuracy"],
    ascending=[True, False]
).reset_index(drop=True)

display(all_model_results)
```

```python
# ============================================================
# SELECT TOP 5 MODELS FOR EACH ATTACK
# ============================================================

selection_columns = [
    "Attack",
    "Model",
    "Accuracy",
    "F1",
    "FPR",
    "RMSE",
    "Runtime_sec"
]

ranking_df = all_model_results[selection_columns].copy()

# Rank models within each attack
ranking_df = ranking_df.sort_values(
    by=[
        "Attack",
        "Accuracy",
        "F1",
        "FPR",
        "RMSE",
        "Runtime_sec"
    ],
    ascending=[
        True,
        False,
        False,
        True,
        True,
        True
    ]
)

# Select top 5 for each attack
top5_models = (
    ranking_df
    .groupby("Attack", group_keys=False)
    .head(5)
    .reset_index(drop=True)
)

print("=" * 70)
print("SELECTED TOP 5 MODELS FOR EACH ATTACK")
print("=" * 70)

for attack in top5_models["Attack"].unique():

    print(f"\n{attack}")

    attack_top5 = top5_models[
        top5_models["Attack"] == attack
    ]

    for i, (_, row) in enumerate(
        attack_top5.iterrows(), start=1
    ):
        print(
            f"{i}. {row['Model']} | "
            f"Accuracy={row['Accuracy']:.4f} | "
            f"F1={row['F1']:.4f} | "
            f"FPR={row['FPR']:.6f} | "
            f"Runtime={row['Runtime_sec']:.2f}s"
        )

print("\n\nComplete Top-5 table:")
display(top5_models)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    mean_squared_error,
    roc_auc_score
)
import numpy as np
import pandas as pd
import time

# Map attack names to their trained model dictionaries
trained_model_map = {
    "DrDoS_DNS": dns_trained_models,
    "DrDoS_LDAP": ldap_trained_models,
    "DrDoS_NetBIOS": netbios_trained_models,
    "DrDoS_SNMP": snmp_trained_models,
    "DrDoS_UDP": udp_trained_models
}

ensemble_results = []
ensemble_predictions = {}

for attack in top5_models["Attack"].unique():

    print(f"\nRunning Majority Voting for {attack}...")

    # Get selected Top 5 models
    selected = top5_models[
        top5_models["Attack"] == attack
    ]["Model"].tolist()

    # Test data
    X_test = train_test_data[attack]["X_test_selected"]
    y_test = train_test_data[attack]["y_test"]

    # Trained models
    trained_models = trained_model_map[attack]

    start_time = time.time()

    # Collect predictions from the 5 classifiers
    model_predictions = []

    for model_name in selected:
        model = trained_models[model_name]
        pred = model.predict(X_test)
        model_predictions.append(pred)

    # Convert to array
    model_predictions = np.array(model_predictions)

    # Hard majority voting
    # 5 classifiers -> majority requires at least 3 votes
    vote_sum = np.sum(model_predictions, axis=0)

    y_ensemble = (vote_sum >= 3).astype(int)

    runtime = time.time() - start_time

    # Evaluation metrics
    accuracy = accuracy_score(y_test, y_ensemble)
    precision = precision_score(y_test, y_ensemble, zero_division=0)
    recall = recall_score(y_test, y_ensemble, zero_division=0)
    f1 = f1_score(y_test, y_ensemble, zero_division=0)

    rmse = np.sqrt(
        mean_squared_error(y_test, y_ensemble)
    )

    tn, fp, fn, tp = confusion_matrix(
        y_test,
        y_ensemble,
        labels=[0, 1]
    ).ravel()

    fpr = fp / (fp + tn) if (fp + tn) > 0 else 0

    classification_error = 1 - accuracy

    # Fraction of classifiers agreeing with final majority decision
    agreement = np.mean(
        model_predictions == y_ensemble,
        axis=0
    ).mean()

    # AUC using hard-voting output
    try:
        auc = roc_auc_score(y_test, y_ensemble)
    except:
        auc = np.nan

    ensemble_results.append({
        "Attack": attack,
        "Models": ", ".join(selected),
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "FPR": fpr,
        "RMSE": rmse,
        "Runtime_sec": runtime,
        "Classification_Error": classification_error,
        "Ensemble_Agreement": agreement,
        "AUC": auc
    })

    ensemble_predictions[attack] = {
        "y_true": y_test,
        "y_pred": y_ensemble,
        "model_predictions": model_predictions
    }

    print(f"Selected models: {selected}")
    print(f"Accuracy : {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall   : {recall:.4f}")
    print(f"F1 Score : {f1:.4f}")
    print(f"FPR      : {fpr:.6f}")
    print(f"RMSE     : {rmse:.4f}")
    print(f"AUC      : {auc:.4f}")
    print(f"Runtime  : {runtime:.4f}s")
    print(f"Agreement: {agreement:.4f}")


# Final ensemble results table
ensemble_results_df = pd.DataFrame(ensemble_results)

print("\n" + "=" * 80)
print("HARD MAJORITY VOTING RESULTS")
print("=" * 80)

display(ensemble_results_df)
```

```python
# Save the current hard majority voting results

hard_voting_results = ensemble_results_df.copy()

print("=" * 70)
print("HARD MAJORITY VOTING RESULTS SAVED")
print("=" * 70)

display(hard_voting_results)

print("\nVariables saved:")
print("✓ hard_voting_results")
print("✓ ensemble_predictions")
print("✓ top5_models")
```

```python
# Calculate accuracy-based weights for the selected Top-5 models

weighted_model_info = {}

for attack in top5_models["Attack"].unique():

    attack_top5 = top5_models[
        top5_models["Attack"] == attack
    ].copy()

    # Calculate normalized weights
    total_accuracy = attack_top5["Accuracy"].sum()

    attack_top5["Weight"] = (
        attack_top5["Accuracy"] / total_accuracy
    )

    weighted_model_info[attack] = attack_top5

    print("\n" + "=" * 70)
    print(f"WEIGHTS FOR {attack}")
    print("=" * 70)

    display(
        attack_top5[
            ["Model", "Accuracy", "Weight"]
        ].reset_index(drop=True)
    )

    print(
        f"Weight sum = "
        f"{attack_top5['Weight'].sum():.6f}"
    )
```

```python
weighted_ensemble_results = []
weighted_predictions = {}

for attack in top5_models["Attack"].unique():

    print(f"\nRunning Weighted Majority Voting for {attack}...")

    attack_info = weighted_model_info[attack]

    selected_models = attack_info["Model"].tolist()
    weights = attack_info["Weight"].values

    X_test = train_test_data[attack]["X_test_selected"]
    y_test = train_test_data[attack]["y_test"]

    trained_models = trained_model_map[attack]

    start_time = time.time()

    # Get predictions from each selected classifier
    model_predictions = []

    for model_name in selected_models:
        model = trained_models[model_name]
        pred = model.predict(X_test)
        model_predictions.append(pred)

    model_predictions = np.array(model_predictions)

    # Weighted vote
    weighted_votes = np.sum(
        model_predictions * weights[:, np.newaxis],
        axis=0
    )

    # Weighted majority threshold
    y_weighted = (weighted_votes >= 0.5).astype(int)

    runtime = time.time() - start_time

    # Evaluation
    accuracy = accuracy_score(y_test, y_weighted)
    precision = precision_score(
        y_test, y_weighted, zero_division=0
    )
    recall = recall_score(
        y_test, y_weighted, zero_division=0
    )
    f1 = f1_score(
        y_test, y_weighted, zero_division=0
    )

    rmse = np.sqrt(
        mean_squared_error(y_test, y_weighted)
    )

    tn, fp, fn, tp = confusion_matrix(
        y_test,
        y_weighted,
        labels=[0, 1]
    ).ravel()

    fpr = fp / (fp + tn) if (fp + tn) > 0 else 0

    classification_error = 1 - accuracy

    # Weighted ensemble agreement
    agreement = np.mean(
        model_predictions == y_weighted,
        axis=0
    ).mean()

    # AUC from weighted-voting output
    try:
        auc = roc_auc_score(
            y_test,
            y_weighted
        )
    except:
        auc = np.nan

    weighted_ensemble_results.append({
        "Attack": attack,
        "Models": ", ".join(selected_models),
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "FPR": fpr,
        "RMSE": rmse,
        "Runtime_sec": runtime,
        "Classification_Error": classification_error,
        "Ensemble_Agreement": agreement,
        "AUC": auc
    })

    weighted_predictions[attack] = {
        "y_true": y_test,
        "y_pred": y_weighted,
        "model_predictions": model_predictions,
        "weighted_votes": weighted_votes
    }

    print(f"Selected models: {selected_models}")
    print(f"Accuracy : {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall   : {recall:.4f}")
    print(f"F1 Score : {f1:.4f}")
    print(f"FPR      : {fpr:.6f}")
    print(f"RMSE     : {rmse:.4f}")
    print(f"AUC      : {auc:.4f}")
    print(f"Runtime  : {runtime:.4f}s")
    print(f"Agreement: {agreement:.4f}")


weighted_voting_results = pd.DataFrame(
    weighted_ensemble_results
)

print("\n" + "=" * 80)
print("WEIGHTED MAJORITY VOTING RESULTS")
print("=" * 80)

display(weighted_voting_results)
```

```python
# Compare Hard Majority Voting vs Weighted Majority Voting

comparison_df = hard_voting_results[
    [
        "Attack",
        "Accuracy",
        "Precision",
        "Recall",
        "F1",
        "FPR",
        "RMSE",
        "AUC"
    ]
].copy()

comparison_df = comparison_df.rename(columns={
    "Accuracy": "Hard_MV_Accuracy",
    "Precision": "Hard_MV_Precision",
    "Recall": "Hard_MV_Recall",
    "F1": "Hard_MV_F1",
    "FPR": "Hard_MV_FPR",
    "RMSE": "Hard_MV_RMSE",
    "AUC": "Hard_MV_AUC"
})

weighted_compare = weighted_voting_results[
    [
        "Attack",
        "Accuracy",
        "Precision",
        "Recall",
        "F1",
        "FPR",
        "RMSE",
        "AUC"
    ]
].copy()

weighted_compare = weighted_compare.rename(columns={
    "Accuracy": "Weighted_MV_Accuracy",
    "Precision": "Weighted_MV_Precision",
    "Recall": "Weighted_MV_Recall",
    "F1": "Weighted_MV_F1",
    "FPR": "Weighted_MV_FPR",
    "RMSE": "Weighted_MV_RMSE",
    "AUC": "Weighted_MV_AUC"
})

comparison_df = comparison_df.merge(
    weighted_compare,
    on="Attack"
)

print("=" * 100)
print("HARD VS WEIGHTED MAJORITY VOTING")
print("=" * 100)

display(comparison_df)
```

```python
from sklearn.metrics import accuracy_score
import numpy as np
import pandas as pd

prediction_comparison = []

for attack in top5_models["Attack"].unique():

    # Best individual model identified earlier
    best_model_name = model_vs_ensemble[
        model_vs_ensemble["Attack"] == attack
    ]["Best_Individual_Model"].iloc[0]

    best_model = trained_model_map[attack][best_model_name]

    X_test = train_test_data[attack]["X_test_selected"]
    y_test = train_test_data[attack]["y_test"]

    # Individual model prediction
    individual_pred = best_model.predict(X_test)

    # Ensemble prediction
    ensemble_pred = ensemble_predictions[attack]["y_pred"]

    # How often they agree
    agreement = np.mean(
        individual_pred == ensemble_pred
    )

    # Cases where they disagree
    disagreement_count = np.sum(
        individual_pred != ensemble_pred
    )

    # Cases where ensemble is correct but individual is wrong
    ensemble_correct_individual_wrong = np.sum(
        (ensemble_pred == y_test) &
        (individual_pred != y_test)
    )

    # Cases where individual is correct but ensemble is wrong
    individual_correct_ensemble_wrong = np.sum(
        (individual_pred == y_test) &
        (ensemble_pred != y_test)
    )

    prediction_comparison.append({
        "Attack": attack,
        "Best_Model": best_model_name,
        "Model_Accuracy": accuracy_score(
            y_test, individual_pred
        ),
        "Ensemble_Accuracy": accuracy_score(
            y_test, ensemble_pred
        ),
        "Model_Ensemble_Agreement": agreement,
        "Disagreement_Count": disagreement_count,
        "Ensemble_Correct_Model_Wrong":
            ensemble_correct_individual_wrong,
        "Model_Correct_Ensemble_Wrong":
            individual_correct_ensemble_wrong
    })

prediction_comparison_df = pd.DataFrame(
    prediction_comparison
)

print("=" * 110)
print("INDIVIDUAL MODEL VS ENSEMBLE PREDICTION COMPARISON")
print("=" * 110)

display(prediction_comparison_df)
```

```python
from statsmodels.stats.contingency_tables import mcnemar

mcnemar_results = []

for attack in top5_models["Attack"].unique():

    best_model_name = model_vs_ensemble[
        model_vs_ensemble["Attack"] == attack
    ]["Best_Individual_Model"].iloc[0]

    best_model = trained_model_map[attack][best_model_name]

    X_test = train_test_data[attack]["X_test_selected"]
    y_test = train_test_data[attack]["y_test"]

    individual_pred = best_model.predict(X_test)
    ensemble_pred = ensemble_predictions[attack]["y_pred"]

    # Correctness of each prediction
    individual_correct = individual_pred == y_test
    ensemble_correct = ensemble_pred == y_test

    # McNemar contingency table:
    #
    #                 Ensemble
    #                Correct  Wrong
    # Model Correct    a       b
    # Model Wrong      c       d

    a = np.sum(
        individual_correct & ensemble_correct
    )

    b = np.sum(
        individual_correct & ~ensemble_correct
    )

    c = np.sum(
        ~individual_correct & ensemble_correct
    )

    d = np.sum(
        ~individual_correct & ~ensemble_correct
    )

    table = np.array([
        [a, b],
        [c, d]
    ])

    # Exact test is preferable when the number
    # of discordant pairs is small.
    result = mcnemar(
        table,
        exact=True
    )

    mcnemar_results.append({
        "Attack": attack,
        "Both_Correct": a,
        "Model_Correct_Ensemble_Wrong": b,
        "Model_Wrong_Ensemble_Correct": c,
        "Both_Wrong": d,
        "Statistic": result.statistic,
        "p_value": result.pvalue
    })

mcnemar_df = pd.DataFrame(mcnemar_results)

print("=" * 110)
print("McNEMAR TEST: BEST INDIVIDUAL MODEL VS HARD MAJORITY VOTING")
print("=" * 110)

display(mcnemar_df)
```

```python
from statsmodels.stats.proportion import proportion_confint
import pandas as pd

confidence_results = []

for _, row in hard_voting_results.iterrows():

    attack = row["Attack"]

    y_true = ensemble_predictions[attack]["y_true"]
    y_pred = ensemble_predictions[attack]["y_pred"]

    correct = int(np.sum(y_true == y_pred))
    total = len(y_true)

    # 95% Wilson confidence interval
    lower, upper = proportion_confint(
        count=correct,
        nobs=total,
        alpha=0.05,
        method="wilson"
    )

    confidence_results.append({
        "Attack": attack,
        "Correct": correct,
        "Total": total,
        "Accuracy": row["Accuracy"],
        "95% CI Lower": lower,
        "95% CI Upper": upper,
        "CI Width": upper - lower
    })

confidence_intervals = pd.DataFrame(
    confidence_results
)

print("=" * 100)
print("95% CONFIDENCE INTERVALS FOR HARD MAJORITY VOTING ACCURACY")
print("=" * 100)

display(confidence_intervals)
```

```python
from sklearn.ensemble import AdaBoostClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    mean_squared_error,
    roc_auc_score
)
import time
import numpy as np
import pandas as pd

adaboost_results = []
adaboost_predictions = {}

for attack in top5_models["Attack"].unique():

    print(f"\nRunning AdaBoost for {attack}...")

    X_train = train_test_data[attack]["X_train_selected"]
    y_train = train_test_data[attack]["y_train_resampled"]

    X_test = train_test_data[attack]["X_test_selected"]
    y_test = train_test_data[attack]["y_test"]

    model = AdaBoostClassifier(
        n_estimators=100,
        random_state=42
    )

    start_time = time.time()

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    runtime = time.time() - start_time

    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(
        y_test, y_pred, zero_division=0
    )
    recall = recall_score(
        y_test, y_pred, zero_division=0
    )
    f1 = f1_score(
        y_test, y_pred, zero_division=0
    )

    rmse = np.sqrt(
        mean_squared_error(y_test, y_pred)
    )

    tn, fp, fn, tp = confusion_matrix(
        y_test,
        y_pred,
        labels=[0, 1]
    ).ravel()

    fpr = fp / (fp + tn) if (fp + tn) > 0 else 0

    try:
        y_prob = model.predict_proba(X_test)[:, 1]
        auc = roc_auc_score(y_test, y_prob)
    except:
        auc = np.nan

    adaboost_results.append({
        "Attack": attack,
        "Model": "AdaBoost",
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "FPR": fpr,
        "RMSE": rmse,
        "Runtime_sec": runtime,
        "Classification_Error": 1 - accuracy,
        "AUC": auc
    })

    adaboost_predictions[attack] = {
        "y_true": y_test,
        "y_pred": y_pred,
        "model": model
    }

    print(f"Accuracy : {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall   : {recall:.4f}")
    print(f"F1 Score : {f1:.4f}")
    print(f"FPR      : {fpr:.6f}")
    print(f"RMSE     : {rmse:.4f}")
    print(f"AUC      : {auc:.4f}")
    print(f"Runtime  : {runtime:.4f}s")


adaboost_results_df = pd.DataFrame(adaboost_results)

print("\n" + "=" * 90)
print("ADABOOST BENCHMARK RESULTS")
print("=" * 90)

display(adaboost_results_df)
```

```python
# LSTM benchmark
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

lstm_results = []
lstm_predictions = {}

for attack in top5_models["Attack"].unique():

    print(f"\nRunning LSTM for {attack}...")

    X_train = train_test_data[attack]["X_train_selected"]
    y_train = train_test_data[attack]["y_train_resampled"]

    X_test = train_test_data[attack]["X_test_selected"]
    y_test = train_test_data[attack]["y_test"]

    # Convert tabular data to sequence format:
    # samples × 20 features × 1 channel
    X_train_lstm = np.asarray(X_train).reshape(
        X_train.shape[0],
        X_train.shape[1],
        1
    )

    X_test_lstm = np.asarray(X_test).reshape(
        X_test.shape[0],
        X_test.shape[1],
        1
    )

    # Build LSTM
    model = Sequential([
        LSTM(
            64,
            input_shape=(X_train.shape[1], 1),
            return_sequences=False
        ),
        Dropout(0.2),
        Dense(32, activation="relu"),
        Dropout(0.2),
        Dense(1, activation="sigmoid")
    ])

    model.compile(
        optimizer="adam",
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    early_stopping = EarlyStopping(
        monitor="val_loss",
        patience=5,
        restore_best_weights=True
    )

    start_time = time.time()

    model.fit(
        X_train_lstm,
        y_train,
        epochs=30,
        batch_size=128,
        validation_split=0.1,
        callbacks=[early_stopping],
        verbose=0
    )

    # Probability and class prediction
    y_prob = model.predict(
        X_test_lstm,
        verbose=0
    ).ravel()

    y_pred = (y_prob >= 0.5).astype(int)

    runtime = time.time() - start_time

    # Metrics
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(
        y_test, y_pred, zero_division=0
    )
    recall = recall_score(
        y_test, y_pred, zero_division=0
    )
    f1 = f1_score(
        y_test, y_pred, zero_division=0
    )

    rmse = np.sqrt(
        mean_squared_error(y_test, y_pred)
    )

    tn, fp, fn, tp = confusion_matrix(
        y_test,
        y_pred,
        labels=[0, 1]
    ).ravel()

    fpr = fp / (fp + tn) if (fp + tn) > 0 else 0

    auc = roc_auc_score(
        y_test,
        y_prob
    )

    lstm_results.append({
        "Attack": attack,
        "Model": "LSTM",
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "FPR": fpr,
        "RMSE": rmse,
        "Runtime_sec": runtime,
        "Classification_Error": 1 - accuracy,
        "AUC": auc
    })

    lstm_predictions[attack] = {
        "y_true": y_test,
        "y_pred": y_pred,
        "y_prob": y_prob,
        "model": model
    }

    print(f"Accuracy : {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall   : {recall:.4f}")
    print(f"F1 Score : {f1:.4f}")
    print(f"FPR      : {fpr:.6f}")
    print(f"RMSE     : {rmse:.4f}")
    print(f"AUC      : {auc:.4f}")
    print(f"Runtime  : {runtime:.4f}s")


lstm_results_df = pd.DataFrame(lstm_results)

print("\n" + "=" * 90)
print("LSTM BENCHMARK RESULTS")
print("=" * 90)

display(lstm_results_df)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Input,
    Conv1D,
    MaxPooling1D,
    Flatten,
    Dense,
    Dropout
)
from tensorflow.keras.callbacks import EarlyStopping

cnn_results = []
cnn_predictions = {}

for attack in top5_models["Attack"].unique():

    print(f"\nRunning CNN for {attack}...")

    X_train = train_test_data[attack]["X_train_selected"]
    y_train = train_test_data[attack]["y_train_resampled"]

    X_test = train_test_data[attack]["X_test_selected"]
    y_test = train_test_data[attack]["y_test"]

    # Convert tabular data to 1D CNN input
    X_train_cnn = np.asarray(X_train).reshape(
        X_train.shape[0],
        X_train.shape[1],
        1
    )

    X_test_cnn = np.asarray(X_test).reshape(
        X_test.shape[0],
        X_test.shape[1],
        1
    )

    model = Sequential([
        Input(shape=(X_train.shape[1], 1)),

        Conv1D(
            filters=32,
            kernel_size=3,
            activation="relu"
        ),

        MaxPooling1D(
            pool_size=2
        ),

        Conv1D(
            filters=64,
            kernel_size=3,
            activation="relu"
        ),

        Flatten(),

        Dense(
            64,
            activation="relu"
        ),

        Dropout(0.2),

        Dense(
            1,
            activation="sigmoid"
        )
    ])

    model.compile(
        optimizer="adam",
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    early_stopping = EarlyStopping(
        monitor="val_loss",
        patience=5,
        restore_best_weights=True
    )

    start_time = time.time()

    model.fit(
        X_train_cnn,
        y_train,
        epochs=30,
        batch_size=128,
        validation_split=0.1,
        callbacks=[early_stopping],
        verbose=0
    )

    y_prob = model.predict(
        X_test_cnn,
        verbose=0
    ).ravel()

    y_pred = (y_prob >= 0.5).astype(int)

    runtime = time.time() - start_time

    # Metrics
    accuracy = accuracy_score(y_test, y_pred)

    precision = precision_score(
        y_test,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_test,
        y_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        y_pred,
        zero_division=0
    )

    rmse = np.sqrt(
        mean_squared_error(y_test, y_pred)
    )

    tn, fp, fn, tp = confusion_matrix(
        y_test,
        y_pred,
        labels=[0, 1]
    ).ravel()

    fpr = fp / (fp + tn) if (fp + tn) > 0 else 0

    auc = roc_auc_score(
        y_test,
        y_prob
    )

    cnn_results.append({
        "Attack": attack,
        "Model": "CNN",
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "FPR": fpr,
        "RMSE": rmse,
        "Runtime_sec": runtime,
        "Classification_Error": 1 - accuracy,
        "AUC": auc
    })

    cnn_predictions[attack] = {
        "y_true": y_test,
        "y_pred": y_pred,
        "y_prob": y_prob,
        "model": model
    }

    print(f"Accuracy : {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall   : {recall:.4f}")
    print(f"F1 Score : {f1:.4f}")
    print(f"FPR      : {fpr:.6f}")
    print(f"RMSE     : {rmse:.4f}")
    print(f"AUC      : {auc:.4f}")
    print(f"Runtime  : {runtime:.4f}s")


cnn_results_df = pd.DataFrame(cnn_results)

print("\n" + "=" * 90)
print("CNN BENCHMARK RESULTS")
print("=" * 90)

display(cnn_results_df)
```

```python
import zipfile
import pandas as pd
import numpy as np
from pathlib import Path

zip_path = Path.home() / "Downloads" / "CSV-01-12.zip"
chunk_size = 100_000

file_summaries = []

with zipfile.ZipFile(zip_path, "r") as z:
    csv_files = [
        name for name in z.namelist()
        if name.lower().endswith(".csv")
    ]

    for filename in csv_files:
        print(f"\nProcessing: {filename}")

        total_rows = 0
        label_counts = {}
        missing_counts = {}
        infinite_counts = {}

        with z.open(filename) as f:
            for chunk in pd.read_csv(
                f,
                chunksize=chunk_size,
                low_memory=False
            ):
                chunk.columns = chunk.columns.str.strip()
                total_rows += len(chunk)

                # Count labels
                counts = chunk["Label"].value_counts(dropna=False)
                for label, count in counts.items():
                    label_counts[label] = (
                        label_counts.get(label, 0) + int(count)
                    )

                # Count missing values
                missing = chunk.isna().sum()
                for col, count in missing.items():
                    missing_counts[col] = (
                        missing_counts.get(col, 0) + int(count)
                    )

                # Count positive/negative infinity in numeric columns
                numeric = chunk.select_dtypes(include=[np.number])
                inf_counts = np.isinf(numeric).sum()

                for col, count in inf_counts.items():
                    infinite_counts[col] = (
                        infinite_counts.get(col, 0) + int(count)
                    )

                del chunk

        file_summaries.append({
            "File": filename,
            "Rows": total_rows,
            "Labels": label_counts,
            "Missing cells": sum(missing_counts.values()),
            "Infinite cells": sum(infinite_counts.values())
        })

        print("Rows:", f"{total_rows:,}")
        print("Labels:", label_counts)
        print("Missing cells:", f"{sum(missing_counts.values()):,}")
        print("Infinite cells:", f"{sum(infinite_counts.values()):,}")

summary_df = pd.DataFrame(file_summaries)

print("\n=== FINAL DATASET SUMMARY ===")
display(summary_df[["File", "Rows", "Missing cells", "Infinite cells"]])

print("\nTotal rows:", f"{summary_df['Rows'].sum():,}")
```

```python
import zipfile
import pandas as pd
import numpy as np
import gc
from pathlib import Path

# --------------------------------------------------
# 1. Paths and settings
# --------------------------------------------------
zip_path = Path.home() / "Downloads" / "CSV-01-12.zip"

project_dir = Path.home() / "DDoS_IoT_Project"
processed_dir = project_dir / "processed"
processed_dir.mkdir(parents=True, exist_ok=True)

chunk_size = 25_000

exclude_columns = {
    "Unnamed: 0",
    "Flow ID",
    "Source IP",
    "Destination IP",
    "Timestamp",
    "SimillarHTTP",
    "Label"
}

# --------------------------------------------------
# 2. Find all CSV files in the ZIP
# --------------------------------------------------
with zipfile.ZipFile(zip_path, "r") as z:
    csv_files = [
        name for name in z.namelist()
        if name.lower().endswith(".csv")
    ]

print("CSV files:", len(csv_files))
print("Chunk size:", chunk_size)
print("Output directory:", processed_dir)

# --------------------------------------------------
# 3. Process each file in chunks
# --------------------------------------------------
total_rows = 0
total_chunks = 0

for file_index, filename in enumerate(csv_files, start=1):

    print(f"\n[{file_index}/{len(csv_files)}] Processing {filename}")

    file_rows = 0
    chunk_number = 0

    with zipfile.ZipFile(zip_path, "r") as z:
        with z.open(filename) as f:

            reader = pd.read_csv(
                f,
                chunksize=chunk_size,
                low_memory=False
            )

            for chunk in reader:

                # Standardize column names
                chunk.columns = chunk.columns.str.strip()

                # Preserve labels
                if "Label" not in chunk.columns:
                    raise ValueError(
                        f"Label column missing in {filename}"
                    )

                labels = chunk["Label"].astype("string").str.strip()

                # Identify numeric features
                numeric = chunk.drop(
                    columns=list(exclude_columns),
                    errors="ignore"
                ).select_dtypes(include=[np.number])

                # Convert infinite values to NaN
                numeric = numeric.replace(
                    [np.inf, -np.inf],
                    np.nan
                )

                # Use float32 to reduce storage and memory
                numeric = numeric.astype(np.float32)

                # Preserve labels and source category
                output = numeric.copy()
                output["Label"] = labels
                output["Source_File"] = Path(filename).stem

                # Write one Parquet file per chunk
                output_name = (
                    f"file_{file_index:02d}_"
                    f"chunk_{chunk_number:05d}.parquet"
                )

                output_path = processed_dir / output_name
                output.to_parquet(
                    output_path,
                    index=False,
                    compression="snappy"
                )

                rows = len(output)
                file_rows += rows
                total_rows += rows
                total_chunks += 1
                chunk_number += 1

                if chunk_number % 20 == 0:
                    print(
                        f"  Processed {file_rows:,} rows "
                        f"from this file"
                    )

                del chunk, numeric, output, labels
                gc.collect()

    print(f"  File complete: {file_rows:,} rows")

print("\n" + "=" * 60)
print("PREPROCESSING COMPLETE")
print("Total rows processed:", f"{total_rows:,}")
print("Total chunks saved:", f"{total_chunks:,}")
print("Output directory:", processed_dir)
print("=" * 60)
```

```python
from pathlib import Path
import pandas as pd
import numpy as np

processed_dir = Path.home() / "DDoS_IoT_Project" / "processed"

prefixes = {
    "DNS": "file_11_",
    "LDAP": "file_10_",
    "NetBIOS": "file_08_",
    "SNMP": "file_06_",
    "UDP": "file_04_",
}

RANDOM_SEED = 42

# Initial manageable sample sizes
ATTACK_SAMPLES_PER_KNOWN_TYPE = 50_000
SNMP_ZERO_DAY_SAMPLES = 50_000
BENIGN_TRAIN_SAMPLES = 3_000
BENIGN_TEST_SAMPLES = 1_000

rng = np.random.default_rng(RANDOM_SEED)

def sample_processed_chunks(prefix, label, n_samples, seed=42):
    """Reservoir-sample matching rows from processed Parquet chunks."""
    files = sorted(processed_dir.glob(f"{prefix}chunk_*.parquet"))
    reservoirs = []
    seen = 0
    rng_local = np.random.default_rng(seed)

    for file in files:
        df = pd.read_parquet(file)
        df["Label"] = df["Label"].astype(str).str.strip()

        matching = df[df["Label"] == label]
        if not matching.empty:
            for _, row in matching.iterrows():
                seen += 1
                if len(reservoirs) < n_samples:
                    reservoirs.append(row)
                else:
                    j = rng_local.integers(0, seen)
                    if j < n_samples:
                        reservoirs[j] = row

        del df, matching

    if not reservoirs:
        return pd.DataFrame()

    return pd.DataFrame(reservoirs).reset_index(drop=True)

# Collect known attack samples, excluding SNMP attacks
known_attack_frames = []

for i, attack in enumerate(["DNS", "LDAP", "NetBIOS", "UDP"]):
    label = {
        "DNS": "DrDoS_DNS",
        "LDAP": "DrDoS_LDAP",
        "NetBIOS": "DrDoS_NetBIOS",
        "UDP": "DrDoS_UDP",
    }[attack]

    print("Sampling known attack:", attack)

    sampled = sample_processed_chunks(
        prefixes[attack],
        label,
        ATTACK_SAMPLES_PER_KNOWN_TYPE,
        seed=RANDOM_SEED + i
    )

    sampled["Attack_Type"] = attack
    known_attack_frames.append(sampled)

known_attacks = pd.concat(known_attack_frames, ignore_index=True)

print("\nKnown attack samples:", len(known_attacks))
print(known_attacks["Attack_Type"].value_counts())

print("\nSampling held-out SNMP attacks...")
snmp_zero_day = sample_processed_chunks(
    prefixes["SNMP"],
    "DrDoS_SNMP",
    SNMP_ZERO_DAY_SAMPLES,
    seed=RANDOM_SEED + 100
)
print("Held-out SNMP attack samples:", len(snmp_zero_day))

print("\nSampling benign training traffic...")
benign_frames = []

for i, attack in enumerate(["DNS", "LDAP", "NetBIOS", "UDP"]):
    sampled = sample_processed_chunks(
        prefixes[attack],
        "BENIGN",
        BENIGN_TRAIN_SAMPLES // 4,
        seed=RANDOM_SEED + 200 + i
    )
    benign_frames.append(sampled)

benign_train = pd.concat(benign_frames, ignore_index=True)

print("Benign training samples:", len(benign_train))

print("\nSampling benign SNMP-file traffic for testing...")
benign_test = sample_processed_chunks(
    prefixes["SNMP"],
    "BENIGN",
    BENIGN_TEST_SAMPLES,
    seed=RANDOM_SEED + 300
)
print("Benign test samples:", len(benign_test))
```

```python

from pathlib import Path
import pandas as pd
import gc

# Set up directories
project_dir = Path.home() / "DDoS_IoT_Project"
sample_dir = project_dir / "processed" / "zero_day_samples"
sample_dir.mkdir(parents=True, exist_ok=True)

# Check expected sample sizes
assert len(known_attacks) == 200_000
assert len(snmp_zero_day) == 50_000
assert len(benign_train) == 3_000
assert len(benign_test) == 1_000

# Verify attack categories
assert set(known_attacks["Attack_Type"].unique()) == {
    "DNS", "LDAP", "NetBIOS", "UDP"
}
assert snmp_zero_day["Label"].eq("DrDoS_SNMP").all()
assert benign_train["Label"].eq("BENIGN").all()
assert benign_test["Label"].eq("BENIGN").all()

# Identify the 81 feature columns
excluded = {"Label", "Source_File", "Attack_Type"}
feature_columns = [
    c for c in known_attacks.columns if c not in excluded
]

assert len(feature_columns) == 81, (
    f"Expected 81 features, found {len(feature_columns)}"
)

for name, df in {
    "SNMP holdout": snmp_zero_day,
    "Benign training": benign_train,
    "Benign testing": benign_test,
}.items():
    missing = set(feature_columns) - set(df.columns)
    assert not missing, f"{name} is missing features: {missing}"

print("All sample and label checks passed.")
print("Feature count:", len(feature_columns))

# Save each sample separately
datasets = {
    "known_attacks": known_attacks,
    "snmp_zero_day": snmp_zero_day,
    "benign_train": benign_train,
    "benign_test": benign_test,
}

for name, df in datasets.items():
    output_path = sample_dir / f"{name}.parquet"
    df.to_parquet(output_path, index=False)
    print(f"Saved {name}: {output_path}")

print("\nAll four datasets saved successfully.")

```

```python

from pathlib import Path
import pandas as pd
import numpy as np
import gc

from sklearn.model_selection import train_test_split

# Load saved datasets
project_dir = Path.home() / "DDoS_IoT_Project"
sample_dir = project_dir / "processed" / "zero_day_samples"

known_attacks = pd.read_parquet(sample_dir / "known_attacks.parquet")
snmp_zero_day = pd.read_parquet(sample_dir / "snmp_zero_day.parquet")
benign_train = pd.read_parquet(sample_dir / "benign_train.parquet")
benign_test = pd.read_parquet(sample_dir / "benign_test.parquet")

# Define the common numerical features
excluded_columns = {"Label", "Source_File", "Attack_Type"}

feature_columns = [
    col for col in known_attacks.columns
    if col not in excluded_columns
]

print("Datasets loaded.")
print("Number of features:", len(feature_columns))

# Create binary labels for the supervised baseline
# 0 = benign, 1 = known attack
known_attacks = known_attacks.copy()
known_attacks["Target"] = 1

benign_supervised = benign_train.copy()
benign_supervised["Target"] = 0

# Combine known attacks and benign training traffic
supervised_data = pd.concat(
    [known_attacks, benign_supervised],
    ignore_index=True
)

# Split before fitting any imputer or model
train_data, val_data = train_test_split(
    supervised_data,
    test_size=0.20,
    random_state=42,
    stratify=supervised_data["Target"]
)

print("\nSupervised baseline split")
print("Training rows:", len(train_data))
print("Validation rows:", len(val_data))
print("\nTraining label counts:")
print(train_data["Target"].value_counts())
print("\nValidation label counts:")
print(val_data["Target"].value_counts())

# Prepare the benign-only data for the anomaly detector.
# These are kept separate from the supervised attack data.
X_benign_train = benign_train[feature_columns].copy()
X_benign_test = benign_test[feature_columns].copy()
X_snmp_test = snmp_zero_day[feature_columns].copy()

print("\nZero-day experiment")
print("Benign-only training rows:", len(X_benign_train))
print("Benign test rows:", len(X_benign_test))
print("Held-out SNMP attack rows:", len(X_snmp_test))

print("\nPreparation complete.")
```

```python

import numpy as np
import pandas as pd
import gc

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# 1. Extract supervised features and labels
X_train = train_data[feature_columns].copy()
y_train = train_data["Target"].to_numpy()

X_val = val_data[feature_columns].copy()
y_val = val_data["Target"].to_numpy()

# 2. Replace any remaining infinite values with NaN
X_train = X_train.replace([np.inf, -np.inf], np.nan)
X_val = X_val.replace([np.inf, -np.inf], np.nan)

# 3. Fit the supervised imputer on training data only
supervised_imputer = SimpleImputer(strategy="median")

X_train_imputed = supervised_imputer.fit_transform(X_train)
X_val_imputed = supervised_imputer.transform(X_val)

# 4. Prepare benign-only data for Isolation Forest
X_benign_train = X_benign_train.replace(
    [np.inf, -np.inf], np.nan
)
X_benign_test = X_benign_test.replace(
    [np.inf, -np.inf], np.nan
)
X_snmp_test = X_snmp_test.replace(
    [np.inf, -np.inf], np.nan
)

# Fit a separate imputer using benign training traffic only
anomaly_imputer = SimpleImputer(strategy="median")

X_benign_train_imputed = anomaly_imputer.fit_transform(
    X_benign_train
)
X_benign_test_imputed = anomaly_imputer.transform(
    X_benign_test
)
X_snmp_test_imputed = anomaly_imputer.transform(
    X_snmp_test
)

# 5. Verify dimensions and check for remaining missing values
print("Supervised preprocessing")
print("Training shape:", X_train_imputed.shape)
print("Validation shape:", X_val_imputed.shape)
print("Training missing values:", int(np.isnan(X_train_imputed).sum()))
print("Validation missing values:", int(np.isnan(X_val_imputed).sum()))

print("\nAnomaly-detector preprocessing")
print("Benign training shape:", X_benign_train_imputed.shape)
print("Benign test shape:", X_benign_test_imputed.shape)
print("SNMP test shape:", X_snmp_test_imputed.shape)

assert not np.isnan(X_train_imputed).any()
assert not np.isnan(X_val_imputed).any()
assert not np.isnan(X_benign_train_imputed).any()
assert not np.isnan(X_benign_test_imputed).any()
assert not np.isnan(X_snmp_test_imputed).any()

print("\nPreprocessing completed successfully.")
```

```python

import time
import joblib
from pathlib import Path

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score
)

# Output directory
project_dir = Path.home() / "DDoS_IoT_Project"
models_dir = project_dir / "models"
results_dir = project_dir / "results"

models_dir.mkdir(parents=True, exist_ok=True)
results_dir.mkdir(parents=True, exist_ok=True)

# Train Random Forest
rf_model = RandomForestClassifier(
    n_estimators=100,
    max_depth=20,
    min_samples_leaf=2,
    class_weight="balanced",
    random_state=42,
    n_jobs=1
)

print("Training Random Forest...", flush=True)
start_time = time.time()

rf_model.fit(X_train_imputed, y_train)

elapsed = time.time() - start_time
print(f"Training completed in {elapsed:.1f} seconds.")

# Predict on validation data
y_val_pred = rf_model.predict(X_val_imputed)
y_val_prob = rf_model.predict_proba(X_val_imputed)[:, 1]

# Calculate metrics
rf_metrics = {
    "Model": "Random Forest",
    "Accuracy": accuracy_score(y_val, y_val_pred),
    "Precision": precision_score(y_val, y_val_pred, zero_division=0),
    "Recall": recall_score(y_val, y_val_pred, zero_division=0),
    "F1_Score": f1_score(y_val, y_val_pred, zero_division=0),
    "ROC_AUC": roc_auc_score(y_val, y_val_prob),
}

print("\n--- Validation Results ---")
for metric, value in rf_metrics.items():
    if isinstance(value, (float, int)):
        print(f"{metric}: {value:.4f}")
    else:
        print(f"{metric}: {value}")

print("\nConfusion matrix:")
print(confusion_matrix(y_val, y_val_pred))

print("\nClassification report:")
print(classification_report(
    y_val,
    y_val_pred,
    target_names=["Benign", "Known Attack"],
    zero_division=0
))

# Save model and validation metrics
joblib.dump(
    rf_model,
    models_dir / "zero_day_supervised_random_forest.joblib"
)

pd.DataFrame([rf_metrics]).to_csv(
    results_dir / "zero_day_rf_validation_metrics.csv",
    index=False
)

print("\nModel saved.")
print("Metrics saved.")

```

```python

import time
import joblib
import numpy as np
import pandas as pd

from pathlib import Path
from sklearn.ensemble import IsolationForest
from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score
)

project_dir = Path.home() / "DDoS_IoT_Project"
models_dir = project_dir / "models"
results_dir = project_dir / "results"

models_dir.mkdir(parents=True, exist_ok=True)
results_dir.mkdir(parents=True, exist_ok=True)

# Train exclusively on benign training traffic
isolation_forest = IsolationForest(
    n_estimators=100,
    max_samples=512,
    contamination="auto",
    random_state=42,
    n_jobs=1
)

print("Training Isolation Forest on benign traffic only...", flush=True)
start_time = time.time()

isolation_forest.fit(X_benign_train_imputed)

print(
    f"Training completed in {time.time() - start_time:.1f} seconds."
)

# Combine the held-out benign and SNMP attack samples for evaluation.
# These samples are not used to fit the detector.
X_zero_day_test = np.vstack([
    X_benign_test_imputed,
    X_snmp_test_imputed
])

# Ground truth: 0 = benign, 1 = held-out SNMP attack
y_zero_day_true = np.concatenate([
    np.zeros(len(X_benign_test_imputed), dtype=int),
    np.ones(len(X_snmp_test_imputed), dtype=int)
])

# Isolation Forest predicts 1 for inliers and -1 for anomalies.
predictions = isolation_forest.predict(X_zero_day_test)

# Convert to our binary convention:
# 0 = predicted benign, 1 = predicted anomaly
y_zero_day_pred = (predictions == -1).astype(int)

# Lower decision_function values indicate more anomalous samples.
# Negating gives a score where higher means more anomalous.
anomaly_scores = -isolation_forest.decision_function(
    X_zero_day_test
)

# Evaluation metrics
metrics = {
    "Model": "Isolation Forest",
    "Accuracy": accuracy_score(y_zero_day_true, y_zero_day_pred),
    "Precision": precision_score(
        y_zero_day_true, y_zero_day_pred, zero_division=0
    ),
    "Recall": recall_score(
        y_zero_day_true, y_zero_day_pred, zero_division=0
    ),
    "F1_Score": f1_score(
        y_zero_day_true, y_zero_day_pred, zero_division=0
    ),
    "ROC_AUC": roc_auc_score(y_zero_day_true, anomaly_scores),
    "PR_AUC": average_precision_score(
        y_zero_day_true, anomaly_scores
    ),
}

print("\n--- Zero-Day Evaluation Results ---")
for name, value in metrics.items():
    if isinstance(value, (float, int)):
        print(f"{name}: {value:.4f}")
    else:
        print(f"{name}: {value}")

print("\nConfusion matrix:")
print(confusion_matrix(y_zero_day_true, y_zero_day_pred))

print("\nClassification report:")
print(classification_report(
    y_zero_day_true,
    y_zero_day_pred,
    target_names=["Benign", "SNMP Zero-Day"],
    zero_division=0
))

# Report false-positive rate separately
tn, fp, fn, tp = confusion_matrix(
    y_zero_day_true, y_zero_day_pred
).ravel()

fpr = fp / (fp + tn) if (fp + tn) else 0.0

print(f"False-positive rate: {fpr:.4f}")
print(f"SNMP attack detection rate: {tp / (tp + fn):.4f}")

# Save the detector and results
joblib.dump(
    isolation_forest,
    models_dir / "zero_day_isolation_forest.joblib"
)

pd.DataFrame([metrics | {"False_Positive_Rate": fpr}]).to_csv(
    results_dir / "zero_day_isolation_forest_metrics.csv",
    index=False
)

print("\nIsolation Forest model saved.")
print("Zero-day evaluation metrics saved.")
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Calculate anomaly scores.
# Higher values mean more anomalous.
benign_train_scores = -isolation_forest.decision_function(
    X_benign_train_imputed
)

benign_test_scores = -isolation_forest.decision_function(
    X_benign_test_imputed
)

snmp_test_scores = -isolation_forest.decision_function(
    X_snmp_test_imputed
)

# Summarize score distributions
score_summary = pd.DataFrame({
    "Benign_Training": pd.Series(benign_train_scores).describe(),
    "Benign_Test": pd.Series(benign_test_scores).describe(),
    "SNMP_Attacks": pd.Series(snmp_test_scores).describe()
})

print("Anomaly score summary:")
print(score_summary.round(5))

# Threshold corresponding to the default decision boundary
default_threshold = -0.0

print("\nDefault decision boundary for negated anomaly scores:",
      default_threshold)

print(
    "Benign training flagged as anomalous:",
    int((benign_train_scores > default_threshold).sum()),
    "/",
    len(benign_train_scores)
)

print(
    "Benign test flagged as anomalous:",
    int((benign_test_scores > default_threshold).sum()),
    "/",
    len(benign_test_scores)
)

print(
    "SNMP attacks flagged as anomalous:",
    int((snmp_test_scores > default_threshold).sum()),
    "/",
    len(snmp_test_scores)
)

# Plot the distributions
plt.figure(figsize=(10, 6))

plt.hist(
    benign_test_scores,
    bins=50,
    alpha=0.7,
    density=True,
    label="Benign test"
)

plt.hist(
    snmp_test_scores,
    bins=50,
    alpha=0.7,
    density=True,
    label="SNMP attacks"
)

plt.axvline(
    default_threshold,
    linestyle="--",
    label="Default decision boundary"
)

plt.xlabel("Anomaly score (higher = more anomalous)")
plt.ylabel("Density")
plt.title("Isolation Forest: Benign vs Held-Out SNMP Scores")
plt.legend()
plt.tight_layout()
plt.show()
```

```python

import numpy as np
import pandas as pd
import joblib

from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.ensemble import IsolationForest

# Split raw benign samples before fitting preprocessing or models
benign_fit_raw, benign_calibration_raw = train_test_split(
    benign_train,
    test_size=0.20,
    random_state=42
)

# Use benign fitting data only to fit the imputer
feature_columns = [
    c for c in benign_train.columns
    if c not in {"Label", "Source_File", "Attack_Type", "Target"}
]

calibration_imputer = SimpleImputer(strategy="median")

X_benign_fit = calibration_imputer.fit_transform(
    benign_fit_raw[feature_columns].replace([np.inf, -np.inf], np.nan)
)

X_benign_calibration = calibration_imputer.transform(
    benign_calibration_raw[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    )
)

# Fit Isolation Forest only on the benign fitting subset
calibrated_iforest = IsolationForest(
    n_estimators=100,
    max_samples=512,
    contamination="auto",
    random_state=42,
    n_jobs=1
)

print("Fitting Isolation Forest on 2,400 benign samples...")
calibrated_iforest.fit(X_benign_fit)

# Select a threshold from the independent benign calibration subset
calibration_scores = -calibrated_iforest.decision_function(
    X_benign_calibration
)

candidate_threshold = np.percentile(calibration_scores, 95)

calibration_fpr = np.mean(
    calibration_scores > candidate_threshold
)

print("\nIndependent calibration completed.")
print("Benign fitting samples:", len(benign_fit_raw))
print("Benign calibration samples:", len(benign_calibration_raw))
print("Candidate threshold:", round(candidate_threshold, 6))
print("Calibration false-positive rate:", f"{calibration_fpr:.2%}")

# Save the calibrated detector and preprocessing
project_dir = Path.home() / "DDoS_IoT_Project"
models_dir = project_dir / "models"
results_dir = project_dir / "results"

models_dir.mkdir(parents=True, exist_ok=True)
results_dir.mkdir(parents=True, exist_ok=True)

joblib.dump(
    calibrated_iforest,
    models_dir / "calibrated_isolation_forest.joblib"
)

joblib.dump(
    calibration_imputer,
    models_dir / "calibrated_isolation_forest_imputer.joblib"
)

pd.DataFrame([{
    "Threshold": candidate_threshold,
    "Calibration_FPR": calibration_fpr,
    "Fit_Samples": len(benign_fit_raw),
    "Calibration_Samples": len(benign_calibration_raw)
}]).to_csv(
    results_dir / "zero_day_threshold_calibration.csv",
    index=False
)

print("\nCalibrated model, imputer and threshold metadata saved.")
```

```python

import numpy as np
import pandas as pd
import joblib

from pathlib import Path
from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score
)

project_dir = Path.home() / "DDoS_IoT_Project"
models_dir = project_dir / "models"
results_dir = project_dir / "results"

# Load the saved calibrated model and imputer
calibrated_iforest = joblib.load(
    models_dir / "calibrated_isolation_forest.joblib"
)

calibration_imputer = joblib.load(
    models_dir / "calibrated_isolation_forest_imputer.joblib"
)

# Load the frozen threshold from the calibration result
threshold_df = pd.read_csv(
    results_dir / "zero_day_threshold_calibration.csv"
)
threshold = float(threshold_df.loc[0, "Threshold"])

print("Frozen threshold:", threshold)

# Prepare test features using the calibration-fitted imputer
X_benign_final = calibration_imputer.transform(
    benign_test[feature_columns].replace([np.inf, -np.inf], np.nan)
)

X_snmp_final = calibration_imputer.transform(
    snmp_zero_day[feature_columns].replace([np.inf, -np.inf], np.nan)
)

# Compute anomaly scores: higher means more anomalous
benign_scores = -calibrated_iforest.decision_function(
    X_benign_final
)
snmp_scores = -calibrated_iforest.decision_function(
    X_snmp_final
)

# Combine final evaluation data
y_true = np.concatenate([
    np.zeros(len(benign_scores), dtype=int),
    np.ones(len(snmp_scores), dtype=int)
])

scores = np.concatenate([benign_scores, snmp_scores])

# Apply the frozen threshold without further tuning
y_pred = (scores > threshold).astype(int)

# Calculate metrics
tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()

fpr = fp / (fp + tn) if (fp + tn) else 0.0
attack_detection_rate = tp / (tp + fn) if (tp + fn) else 0.0

metrics = {
    "Model": "Calibrated Isolation Forest",
    "Threshold": threshold,
    "Accuracy": accuracy_score(y_true, y_pred),
    "Precision": precision_score(y_true, y_pred, zero_division=0),
    "Recall": recall_score(y_true, y_pred, zero_division=0),
    "F1_Score": f1_score(y_true, y_pred, zero_division=0),
    "ROC_AUC": roc_auc_score(y_true, scores),
    "PR_AUC": average_precision_score(y_true, scores),
    "False_Positive_Rate": fpr,
    "SNMP_Detection_Rate": attack_detection_rate,
}

print("\n--- Final Zero-Day Evaluation ---")
for key, value in metrics.items():
    if isinstance(value, (float, int, np.floating, np.integer)):
        print(f"{key}: {value:.6f}")
    else:
        print(f"{key}: {value}")

print("\nConfusion matrix:")
print(confusion_matrix(y_true, y_pred))

print("\nClassification report:")
print(classification_report(
    y_true,
    y_pred,
    labels=[0, 1],
    target_names=["Benign", "SNMP Zero-Day"],
    zero_division=0
))

# Save final evaluation metrics
pd.DataFrame([metrics]).to_csv(
    results_dir / "zero_day_final_evaluation.csv",
    index=False
)

print("\nFinal evaluation saved to:")
print(results_dir / "zero_day_final_evaluation.csv")
```

```python

import pandas as pd
import numpy as np

# Use the original feature DataFrames, before imputation.
# Missing values are handled consistently by pandas hashing.
X_train_check = train_data[feature_columns]
X_val_check = val_data[feature_columns]

y_train_check = train_data["Target"].to_numpy()
y_val_check = val_data["Target"].to_numpy()

print("Checking exact duplicate feature rows...", flush=True)

# Hash each feature row to make comparison memory-efficient.
train_hashes = pd.util.hash_pandas_object(
    X_train_check, index=False
).to_numpy()

val_hashes = pd.util.hash_pandas_object(
    X_val_check, index=False
).to_numpy()

# Count rows whose hashes appear in both splits.
train_unique_hashes = np.unique(train_hashes)
val_unique_hashes = np.unique(val_hashes)

shared_hashes = np.intersect1d(
    train_unique_hashes,
    val_unique_hashes
)

train_shared_mask = np.isin(train_hashes, shared_hashes)
val_shared_mask = np.isin(val_hashes, shared_hashes)

print("\n--- Cross-split duplicate check ---")
print("Training rows:", len(train_hashes))
print("Validation rows:", len(val_hashes))
print("Unique training feature hashes:", len(train_unique_hashes))
print("Unique validation feature hashes:", len(val_unique_hashes))
print("Distinct feature hashes shared across splits:", len(shared_hashes))
print("Training rows matching a shared hash:", int(train_shared_mask.sum()))
print("Validation rows matching a shared hash:", int(val_shared_mask.sum()))

# Compare labels for rows with matching feature hashes.
train_labels_by_hash = pd.DataFrame({
    "hash": train_hashes,
    "label": y_train_check
})

val_labels_by_hash = pd.DataFrame({
    "hash": val_hashes,
    "label": y_val_check
})

train_label_sets = train_labels_by_hash.groupby("hash")["label"].agg(
    lambda values: frozenset(values)
)

val_label_sets = val_labels_by_hash.groupby("hash")["label"].agg(
    lambda values: frozenset(values)
)

shared_label_hashes = set(shared_hashes)

conflicting_hashes = [
    h for h in shared_label_hashes
    if train_label_sets[h] != val_label_sets[h]
]

print(
    "Shared hashes with different label sets:",
    len(conflicting_hashes)
)

# Check duplicates within each split
train_duplicate_rows = int(
    X_train_check.duplicated(keep="first").sum()
)

val_duplicate_rows = int(
    X_val_check.duplicated(keep="first").sum()
)

print("\n--- Within-split duplicates ---")
print("Duplicate rows within training:", train_duplicate_rows)
print("Duplicate rows within validation:", val_duplicate_rows)

print("\nDuplicate check complete.")
```

```python
import numpy as np
import pandas as pd

# RF: 0 = benign, 1 = attack
rf_pred = rf.predict(
    snmp.reindex(columns=feature_names)
)

# IF: 0 = anomalous, 1 = normal in sklearn's predict() convention
X_snmp_if = iso_imputer.transform(
    snmp.reindex(columns=feature_names)
)
if_pred = iso.predict(X_snmp_if)

# Convert IF output to our project convention:
# 0 = benign, 1 = attack
if_attack_pred = (if_pred == -1).astype(int)

disagreement = (rf_pred != if_attack_pred)

print("SNMP samples:", len(snmp))
print("RF predicts attack:", int((rf_pred == 1).sum()))
print("IF predicts attack:", int(if_attack_pred.sum()))
print("Models disagree:", int(disagreement.sum()))
print("Disagreement rate (%):", round(disagreement.mean() * 100, 2))

print("\nDisagreement patterns:")
print("RF attack, IF benign:", int(((rf_pred == 1) & (if_attack_pred == 0)).sum()))
print("RF benign, IF attack:", int(((rf_pred == 0) & (if_attack_pred == 1)).sum()))
```

```python
import numpy as np

# Convert both model predictions to:
# 0 = benign, 1 = attack

rf_pred = rf.predict(snmp.reindex(columns=feature_names))

X_snmp_if = iso_imputer.transform(
    snmp.reindex(columns=feature_names)
)
if_pred = iso.predict(X_snmp_if)
if_attack_pred = (if_pred == -1).astype(int)

# OR rule: flag attack if either model predicts attack
or_ensemble_pred = (
    (rf_pred == 1) | (if_attack_pred == 1)
).astype(int)

# Ground truth: every row in snmp is an attack sample
y_snmp = np.ones(len(snmp), dtype=int)

print("OR-rule ensemble on SNMP attacks")
print("Detection rate (%):", round(or_ensemble_pred.mean() * 100, 2))
print("Missed attacks:", int((or_ensemble_pred == 0).sum()))

# Evaluate on held-out benign samples too
X_benign_if = iso_imputer.transform(
    benign_test.reindex(columns=feature_names)
)
rf_benign_pred = rf.predict(
    benign_test.reindex(columns=feature_names)
)
if_benign_pred = (iso.predict(X_benign_if) == -1).astype(int)

or_benign_pred = (
    (rf_benign_pred == 1) | (if_benign_pred == 1)
).astype(int)

print("\nOR-rule ensemble on benign traffic")
print("False positives:", int(or_benign_pred.sum()))
print("False-positive rate (%):", round(or_benign_pred.mean() * 100, 2))
```

```python
import numpy as np

# Convert IF predictions: -1 = attack, 1 = benign
def get_if_attack_predictions(df):
    X = iso_imputer.transform(df.reindex(columns=feature_names))
    return (iso.predict(X) == -1).astype(int)

# SNMP attack samples
X_snmp = snmp.reindex(columns=feature_names)
rf_snmp = rf.predict(X_snmp)
if_snmp = get_if_attack_predictions(snmp)

# On disagreement, preserve the RF prediction.
# On agreement, both models already give the same prediction.
snmp_combined = np.where(rf_snmp != if_snmp, rf_snmp, if_snmp)

print("SNMP detection rate (%):", round(snmp_combined.mean() * 100, 2))
print("Missed SNMP attacks:", int((snmp_combined == 0).sum()))

# Benign samples
rf_benign = rf.predict(benign_test.reindex(columns=feature_names))
if_benign = get_if_attack_predictions(benign_test)

benign_combined = np.where(
    rf_benign != if_benign,
    rf_benign,
    if_benign
)

print("\nBenign false positives:", int(benign_combined.sum()))
print("Benign false-positive rate (%):", round(benign_combined.mean() * 100, 2))
```

```python
import numpy as np

# Random Forest attack probabilities
X_snmp = snmp.reindex(columns=feature_names)
X_benign = benign_test.reindex(columns=feature_names)

snmp_prob = rf.predict_proba(X_snmp)[:, 1]
benign_prob = rf.predict_proba(X_benign)[:, 1]

for name, probs in [
    ("SNMP attacks", snmp_prob),
    ("Benign traffic", benign_prob),
]:
    print(f"\n{name}")
    print("Minimum attack probability:", round(float(probs.min()), 4))
    print("Median attack probability:", round(float(np.median(probs)), 4))
    print("Maximum attack probability:", round(float(probs.max()), 4))

    for threshold in [0.50, 0.70, 0.90, 0.95]:
        uncertain = (probs >= 0.5) & (probs < threshold)
        print(
            f"Probability in [0.50, {threshold:.2f}):",
            int(uncertain.sum())
        )
```

```python
import numpy as np

def confidence_triggered_predict(df, low=0.30, high=0.70):
    X = df.reindex(columns=feature_names)

    # Supervised prediction and confidence
    probabilities = rf.predict_proba(X)[:, 1]
    rf_predictions = (probabilities >= 0.5).astype(int)

    # Only invoke the anomaly detector for uncertain samples
    uncertain = (probabilities >= low) & (probabilities <= high)
    final_predictions = rf_predictions.copy()

    if uncertain.any():
        X_uncertain = iso_imputer.transform(X.loc[uncertain])
        if_predictions = (iso.predict(X_uncertain) == -1).astype(int)

        # Conservative fallback: either model can flag an attack
        final_predictions[uncertain] = (
            (rf_predictions[uncertain] == 1) | (if_predictions == 1)
        ).astype(int)

    return final_predictions, uncertain

# Evaluate on SNMP attacks
snmp_predictions, snmp_routed = confidence_triggered_predict(snmp)
print("SNMP detection rate (%):", round(snmp_predictions.mean() * 100, 2))
print("Missed SNMP attacks:", int((snmp_predictions == 0).sum()))
print("SNMP samples routed to IF:", int(snmp_routed.sum()))

# Evaluate on benign traffic
benign_predictions, benign_routed = confidence_triggered_predict(benign_test)
print("\nBenign false positives:", int(benign_predictions.sum()))
print("Benign false-positive rate (%):", round(benign_predictions.mean() * 100, 2))
print("Benign samples routed to IF:", int(benign_routed.sum()))
```

```python
from pathlib import Path

zip_path = Path.home() / "Downloads" / "Bot-IoT_Dataset.zip"

signature = b"PK\x05\x06"
chunk_size = 16 * 1024 * 1024
overlap = len(signature) - 1

found_at = []
bytes_scanned = 0
carry = b""

with open(zip_path, "rb") as f:
    while True:
        chunk = f.read(chunk_size)
        if not chunk:
            break

        data = carry + chunk
        start = 0

        while True:
            pos = data.find(signature, start)
            if pos == -1:
                break
            found_at.append(bytes_scanned - len(carry) + pos)
            start = pos + 1

        bytes_scanned += len(chunk)
        carry = data[-overlap:]

print("Bytes scanned:", bytes_scanned)
print("End-record signature occurrences:", len(found_at))
print("First occurrence offsets:", found_at[:10])
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, confusion_matrix, roc_auc_score,
    average_precision_score
)

# Prepare the same test set for all methods
X_benign = benign_test.reindex(columns=feature_names)
X_snmp = snmp.reindex(columns=feature_names)

X_test = pd.concat([X_benign, X_snmp], ignore_index=True)
y_test = np.concatenate([
    np.zeros(len(X_benign), dtype=int),
    np.ones(len(X_snmp), dtype=int)
])

# Random Forest predictions and probabilities
rf_prob = rf.predict_proba(X_test)[:, 1]
rf_pred = (rf_prob >= 0.5).astype(int)

# Calibrated Isolation Forest scores and predictions
X_test_imputed = iso_imputer.transform(X_test)
if_scores = -iso.decision_function(X_test_imputed)
threshold = 0.011813
if_pred = (if_scores > threshold).astype(int)

# OR-rule: flag attack if either model flags attack
or_pred = np.maximum(rf_pred, if_pred)

# Compare on identical samples
results = []

for name, pred in [
    ("Random Forest", rf_pred),
    ("Calibrated Isolation Forest", if_pred),
    ("OR-rule Ensemble", or_pred)
]:
    tn, fp, fn, tp = confusion_matrix(
        y_test, pred, labels=[0, 1]
    ).ravel()

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, pred),
        "Precision": precision_score(
            y_test, pred, zero_division=0
        ),
        "Recall": recall_score(
            y_test, pred, zero_division=0
        ),
        "F1": f1_score(y_test, pred, zero_division=0),
        "False_Positive_Rate": fp / (fp + tn),
        "SNMP_Detection_Rate": tp / (tp + fn),
        "False_Positive_Count": fp,
        "Missed_Attacks": fn
    })

comparison = pd.DataFrame(results)
display(comparison.round(5))
```

```python
import zipfile
import pandas as pd
from pathlib import Path

ZIP_PATH = Path(
    r"C:\Users\CSE\Downloads\Processed_Network_dataset.zip"
)
OUTPUT_DIR = PROJECT / "processed" / "ton_iot"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

with zipfile.ZipFile(ZIP_PATH, "r") as z:
    with z.open("Network_dataset_13.csv") as f:
        ton = pd.read_csv(f, low_memory=False)

# Separate classes
normal = ton[ton["label"] == 0]
ddos = ton[ton["label"] == 1]

# Reproducible balanced sample
ddos_sample = ddos.sample(
    n=len(normal), random_state=42
)
ton_balanced = pd.concat(
    [normal, ddos_sample], ignore_index=True
).sample(frac=1, random_state=42).reset_index(drop=True)

# Save locally
output_path = OUTPUT_DIR / "ton_iot_ddos_balanced.csv"
ton_balanced.to_csv(output_path, index=False)

print("Saved to:", output_path)
print("Shape:", ton_balanced.shape)
print("\nLabel counts:")
print(ton_balanced["label"].value_counts())
print("\nType counts:")
print(ton_balanced["type"].value_counts())
```

```python
import pandas as pd
import numpy as np

TON_SAMPLE_PATH = (
    PROJECT / "processed" / "ton_iot" /
    "ton_iot_ddos_balanced.csv"
)

ton_balanced = pd.read_csv(TON_SAMPLE_PATH)

# Exclude labels and potentially identifying or time-dependent fields
exclude_cols = [
    "label", "type", "src_ip", "dst_ip", "ts"
]

candidate_features = [
    c for c in ton_balanced.select_dtypes(include=np.number).columns
    if c not in exclude_cols
]

print("Total rows:", len(ton_balanced))
print("Candidate numeric features:", len(candidate_features))
print("\nCandidate features:")
print(candidate_features)

print("\nMissing values in candidate features:")
print(
    ton_balanced[candidate_features]
    .isna().sum()
    .sort_values(ascending=False)
    .head(15)
)

constant_cols = [
    c for c in candidate_features
    if ton_balanced[c].nunique(dropna=False) <= 1
]
print("\nConstant features to remove:", constant_cols)

print("\nTarget distribution:")
print(ton_balanced["label"].value_counts())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    classification_report, confusion_matrix,
    roc_auc_score, average_precision_score
)

# Remove labels, identifiers, and constant features
features = [
    c for c in candidate_features
    if c not in constant_cols
]

X = ton_balanced[features].copy()
y = ton_balanced["label"].astype(int)

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

baseline = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=2000, random_state=42)
)

baseline.fit(X_train, y_train)

y_pred = baseline.predict(X_test)
y_prob = baseline.predict_proba(X_test)[:, 1]

print("Features:", len(features))
print("Training rows:", len(X_train))
print("Test rows:", len(X_test))

print("\nClassification report:")
print(classification_report(
    y_test, y_pred,
    target_names=["Normal", "DDoS"],
    digits=4
))

print("Confusion matrix:")
print(confusion_matrix(y_test, y_pred))

print(f"ROC-AUC: {roc_auc_score(y_test, y_prob):.4f}")
print(f"PR-AUC:  {average_precision_score(y_test, y_prob):.4f}")
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    classification_report, confusion_matrix,
    roc_auc_score, average_precision_score
)

ton_rf = RandomForestClassifier(
    n_estimators=300,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

ton_rf.fit(X_train, y_train)

rf_pred = ton_rf.predict(X_test)
rf_prob = ton_rf.predict_proba(X_test)[:, 1]

print("ToN-IoT Random Forest")
print("\nClassification report:")
print(classification_report(
    y_test, rf_pred,
    target_names=["Normal", "DDoS"],
    digits=4
))

print("Confusion matrix:")
print(confusion_matrix(y_test, rf_pred))

print(f"ROC-AUC: {roc_auc_score(y_test, rf_prob):.4f}")
print(f"PR-AUC:  {average_precision_score(y_test, rf_prob):.4f}")
```

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import StratifiedGroupKFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score
)

general_features = [
    "duration", "src_bytes", "dst_bytes", "missed_bytes",
    "src_pkts", "src_ip_bytes", "dst_pkts", "dst_ip_bytes"
]

X = ton_balanced[general_features].copy()
y = ton_balanced["label"].astype(int)

groups = pd.util.hash_pandas_object(X, index=False)

cv = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

fold_results = []

for fold, (train_idx, test_idx) in enumerate(
    cv.split(X, y, groups=groups), start=1
):
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

    model = RandomForestClassifier(
        n_estimators=300,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    )

    model.fit(X_train, y_train)

    pred = model.predict(X_test)
    prob = model.predict_proba(X_test)[:, 1]

    train_groups = set(groups.iloc[train_idx])
    test_groups = set(groups.iloc[test_idx])
    overlap = len(train_groups & test_groups)

    result = {
        "Fold": fold,
        "Train_rows": len(train_idx),
        "Test_rows": len(test_idx),
        "Test_normal": int((y_test == 0).sum()),
        "Test_DDoS": int((y_test == 1).sum()),
        "Pattern_overlap": overlap,
        "Accuracy": accuracy_score(y_test, pred),
        "DDoS_precision": precision_score(
            y_test, pred, zero_division=0
        ),
        "DDoS_recall": recall_score(
            y_test, pred, zero_division=0
        ),
        "DDoS_F1": f1_score(
            y_test, pred, zero_division=0
        ),
        "ROC_AUC": roc_auc_score(y_test, prob),
        "PR_AUC": average_precision_score(y_test, prob)
    }

    fold_results.append(result)

results_df = pd.DataFrame(fold_results)

print("Five-fold grouped evaluation:")
print(results_df.round(4).to_string(index=False))

print("\nMean metrics:")
print(
    results_df[
        ["Accuracy", "DDoS_precision", "DDoS_recall",
         "DDoS_F1", "ROC_AUC", "PR_AUC"]
    ].mean().round(4)
)

print("\nStandard deviation:")
print(
    results_df[
        ["Accuracy", "DDoS_precision", "DDoS_recall",
         "DDoS_F1", "ROC_AUC", "PR_AUC"]
    ].std().round(4)
)
```

```python
from sklearn.model_selection import StratifiedGroupKFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix
import pandas as pd

general_features = [
    "duration", "src_bytes", "dst_bytes", "missed_bytes",
    "src_pkts", "src_ip_bytes", "dst_pkts", "dst_ip_bytes"
]

X = ton_balanced[general_features].copy()
y = ton_balanced["label"].astype(int)
groups = pd.util.hash_pandas_object(X, index=False)

cv = StratifiedGroupKFold(
    n_splits=5, shuffle=True, random_state=42
)

splits = list(cv.split(X, y, groups=groups))
train_idx, test_idx = splits[0]

model = RandomForestClassifier(
    n_estimators=300,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)
model.fit(X.iloc[train_idx], y.iloc[train_idx])

pred = model.predict(X.iloc[test_idx])
prob = model.predict_proba(X.iloc[test_idx])[:, 1]

errors = X.iloc[test_idx].copy()
errors["actual"] = y.iloc[test_idx].to_numpy()
errors["predicted"] = pred
errors["attack_probability"] = prob
errors["error_type"] = "Correct"

errors.loc[
    (errors["actual"] == 1) & (errors["predicted"] == 0),
    "error_type"
] = "Missed DDoS"

errors.loc[
    (errors["actual"] == 0) & (errors["predicted"] == 1),
    "error_type"
] = "False alarm"

print("Fold 1 confusion matrix:")
print(confusion_matrix(y.iloc[test_idx], pred))

print("\nError counts:")
print(errors["error_type"].value_counts())

print("\nMissed DDoS samples:")
print(
    errors[errors["error_type"] == "Missed DDoS"]
    .sort_values("attack_probability")
    .head(10)
    .to_string(index=False)
)

print("\nFalse alarms:")
print(
    errors[errors["error_type"] == "False alarm"]
    .to_string(index=False)
)
```

```python
import numpy as np
from sklearn.ensemble import IsolationForest
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report

# Split benign training traffic only
X_benign_fit, X_benign_cal = train_test_split(
    X_benign_train,
    test_size=0.25,
    random_state=42
)

# Fit anomaly detector using benign traffic only
ton_iso = IsolationForest(
    n_estimators=300,
    contamination="auto",
    random_state=42,
    n_jobs=-1
)

ton_iso.fit(X_benign_fit)

# Higher scores indicate more anomalous traffic
cal_scores = -ton_iso.decision_function(X_benign_cal)

# Calibrate threshold using benign-only calibration data
ton_threshold = np.quantile(cal_scores, 0.95)

print("Benign detector-fitting rows:", len(X_benign_fit))
print("Benign calibration rows:", len(X_benign_cal))
print(f"Calibrated anomaly threshold: {ton_threshold:.6f}")

# Verify calibration false-positive rate
cal_pred = (cal_scores > ton_threshold).astype(int)
cal_fpr = cal_pred.mean()

print(f"Calibration false-positive rate: {cal_fpr:.4f}")
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

# Random Forest predictions from the existing Fold 1 model
rf_pred = pred.copy()
rf_prob = prob.copy()

# Isolation Forest anomaly scores on the same held-out test set
test_anomaly_scores = -ton_iso.decision_function(X_test)
iso_pred = (test_anomaly_scores > ton_threshold).astype(int)

# OR-rule ensemble: flag as attack if either model flags it
combined_pred = ((rf_pred == 1) | (iso_pred == 1)).astype(int)

def evaluate_detector(name, y_true, y_pred):
    tn, fp, fn, tp = confusion_matrix(
        y_true, y_pred, labels=[0, 1]
    ).ravel()

    return {
        "Detector": name,
        "Accuracy": accuracy_score(y_true, y_pred),
        "DDoS_Precision": precision_score(
            y_true, y_pred, zero_division=0
        ),
        "DDoS_Recall": recall_score(
            y_true, y_pred, zero_division=0
        ),
        "DDoS_F1": f1_score(
            y_true, y_pred, zero_division=0
        ),
        "False_Alarms": int(fp),
        "Missed_DDoS": int(fn),
        "Benign_FPR": fp / (tn + fp)
    }

comparison = pd.DataFrame([
    evaluate_detector("Random Forest", y_test, rf_pred),
    evaluate_detector("Isolation Forest", y_test, iso_pred),
    evaluate_detector("RF OR Isolation Forest", y_test, combined_pred)
])

print(comparison.round(4).to_string(index=False))

print("\nIsolation Forest detections among RF-missed DDoS rows:")
missed_mask = (y_test.to_numpy() == 1) & (rf_pred == 0)
print("RF-missed DDoS rows:", int(missed_mask.sum()))
print("Recovered by Isolation Forest:", int(iso_pred[missed_mask].sum()))

print("\nIsolation Forest detections among normal test rows:")
normal_mask = y_test.to_numpy() == 0
print("Normal rows flagged:", int(iso_pred[normal_mask].sum()))
print("Total normal rows:", int(normal_mask.sum()))
```

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import StratifiedGroupKFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix

# Training data only; the held-out Fold 1 test set is not used here
X_cal = X_train.reset_index(drop=True)
y_cal = y_train.reset_index(drop=True)
groups_cal = pd.util.hash_pandas_object(X_cal, index=False)

inner_cv = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=123
)

oof_prob = np.full(len(X_cal), np.nan)

for inner_train_idx, inner_valid_idx in inner_cv.split(
    X_cal, y_cal, groups=groups_cal
):
    inner_model = RandomForestClassifier(
        n_estimators=300,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    )

    inner_model.fit(
        X_cal.iloc[inner_train_idx],
        y_cal.iloc[inner_train_idx]
    )

    oof_prob[inner_valid_idx] = inner_model.predict_proba(
        X_cal.iloc[inner_valid_idx]
    )[:, 1]

valid_mask = ~np.isnan(oof_prob)
y_oof = y_cal.to_numpy()[valid_mask]
p_oof = oof_prob[valid_mask]

rows = []

for threshold in np.arange(0.05, 0.501, 0.01):
    pred_oof = (p_oof >= threshold).astype(int)
    tn, fp, fn, tp = confusion_matrix(
        y_oof, pred_oof, labels=[0, 1]
    ).ravel()

    fpr = fp / (tn + fp)
    recall = tp / (tp + fn)

    rows.append({
        "Threshold": round(float(threshold), 2),
        "DDoS_recall": recall,
        "Benign_FPR": fpr,
        "False_alarms": int(fp),
        "Missed_DDoS": int(fn)
    })

threshold_results = pd.DataFrame(rows)

eligible = threshold_results[
    threshold_results["Benign_FPR"] <= 0.05
].copy()

print("Out-of-fold calibration samples:", len(y_oof))
print("Out-of-fold class counts:")
print(pd.Series(y_oof).value_counts().sort_index())

print("\nBest threshold under the 5% FPR constraint:")
if len(eligible):
    best = eligible.sort_values(
        ["DDoS_recall", "Threshold"],
        ascending=[False, False]
    ).iloc[0]
    print(best.to_string())
else:
    print("No tested threshold met the FPR constraint.")

print("\nThreshold performance near the default 0.50:")
print(
    threshold_results[
        threshold_results["Threshold"].isin([0.10, 0.20, 0.30, 0.40, 0.50])
    ].round(4).to_string(index=False)
)
```

```python
import numpy as np
import pandas as pd

from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

# RF probabilities from the existing Fold 1 model
# pred/prob were generated on X_test using the Fold 1 training data.
thresholds = [0.50, 0.15]
evaluation_rows = []

for threshold in thresholds:
    threshold_pred = (prob >= threshold).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_test, threshold_pred, labels=[0, 1]
    ).ravel()

    evaluation_rows.append({
        "Threshold": threshold,
        "Accuracy": accuracy_score(y_test, threshold_pred),
        "DDoS_Precision": precision_score(
            y_test, threshold_pred, zero_division=0
        ),
        "DDoS_Recall": recall_score(
            y_test, threshold_pred, zero_division=0
        ),
        "DDoS_F1": f1_score(
            y_test, threshold_pred, zero_division=0
        ),
        "False_Alarms": int(fp),
        "Missed_DDoS": int(fn),
        "Benign_FPR": fp / (tn + fp)
    })

test_threshold_comparison = pd.DataFrame(evaluation_rows)
print(test_threshold_comparison.round(4).to_string(index=False))
```

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import StratifiedGroupKFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

general_features = [
    "duration", "src_bytes", "dst_bytes", "missed_bytes",
    "src_pkts", "src_ip_bytes", "dst_pkts", "dst_ip_bytes"
]

X_all = ton_balanced[general_features].reset_index(drop=True)
y_all = ton_balanced["label"].astype(int).reset_index(drop=True)
groups_all = pd.util.hash_pandas_object(X_all, index=False)

outer_cv = StratifiedGroupKFold(
    n_splits=5, shuffle=True, random_state=42
)

results = []

for fold, (outer_train, outer_test) in enumerate(
    outer_cv.split(X_all, y_all, groups=groups_all), start=1
):
    X_tr = X_all.iloc[outer_train].reset_index(drop=True)
    y_tr = y_all.iloc[outer_train].reset_index(drop=True)
    X_te = X_all.iloc[outer_test]
    y_te = y_all.iloc[outer_test]

    g_tr = pd.util.hash_pandas_object(X_tr, index=False)

    inner_cv = StratifiedGroupKFold(
        n_splits=5, shuffle=True, random_state=100 + fold
    )

    oof_prob = np.full(len(X_tr), np.nan)

    for inner_train, inner_valid in inner_cv.split(
        X_tr, y_tr, groups=g_tr
    ):
        inner_model = RandomForestClassifier(
            n_estimators=300,
            class_weight="balanced",
            random_state=42,
            n_jobs=-1
        )
        inner_model.fit(X_tr.iloc[inner_train], y_tr.iloc[inner_train])
        oof_prob[inner_valid] = inner_model.predict_proba(
            X_tr.iloc[inner_valid]
        )[:, 1]

    valid = ~np.isnan(oof_prob)
    y_oof = y_tr.to_numpy()[valid]
    p_oof = oof_prob[valid]

    threshold_rows = []

    for threshold in np.arange(0.05, 0.501, 0.01):
        oof_pred = (p_oof >= threshold).astype(int)
        tn, fp, fn, tp = confusion_matrix(
            y_oof, oof_pred, labels=[0, 1]
        ).ravel()

        threshold_rows.append({
            "threshold": float(threshold),
            "recall": tp / (tp + fn),
            "fpr": fp / (fp + tn)
        })

    eligible = [
        row for row in threshold_rows if row["fpr"] <= 0.05
    ]

    if eligible:
        chosen = max(
            eligible,
            key=lambda row: (row["recall"], row["threshold"])
        )
        threshold = chosen["threshold"]
    else:
        threshold = 0.50

    final_model = RandomForestClassifier(
        n_estimators=300,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    )
    final_model.fit(X_tr, y_tr)
    test_prob = final_model.predict_proba(X_te)[:, 1]

    fold_result = {}

    for name, cutoff in [
        ("Default", 0.50),
        ("Calibrated", threshold)
    ]:
        test_pred = (test_prob >= cutoff).astype(int)
        tn, fp, fn, tp = confusion_matrix(
            y_te, test_pred, labels=[0, 1]
        ).ravel()

        fold_result.update({
            f"{name}_threshold": cutoff,
            f"{name}_accuracy": accuracy_score(y_te, test_pred),
            f"{name}_DDoS_recall": recall_score(
                y_te, test_pred, zero_division=0
            ),
            f"{name}_DDoS_F1": f1_score(
                y_te, test_pred, zero_division=0
            ),
            f"{name}_benign_FPR": fp / (fp + tn),
            f"{name}_false_alarms": int(fp),
            f"{name}_missed_DDoS": int(fn)
        })

    fold_result["Fold"] = fold
    results.append(fold_result)

cv_results = pd.DataFrame(results)

display_cols = [
    "Fold",
    "Default_threshold", "Default_accuracy",
    "Default_DDoS_recall", "Default_benign_FPR",
    "Calibrated_threshold", "Calibrated_accuracy",
    "Calibrated_DDoS_recall", "Calibrated_benign_FPR"
]

print(cv_results[display_cols].round(4).to_string(index=False))

print("\nMean metrics:")
print(cv_results[
    [
        "Default_accuracy", "Default_DDoS_recall",
        "Default_benign_FPR", "Calibrated_accuracy",
        "Calibrated_DDoS_recall", "Calibrated_benign_FPR"
    ]
].mean().round(4))

print("\nCalibrated threshold per fold:")
print(cv_results[["Fold", "Calibrated_threshold"]].to_string(index=False))
```

```python
from pathlib import Path
import pandas as pd

PROJECT = Path(r"C:\Users\CSE\Downloads\DDoS_IoT_Project")

train_path = PROJECT / "UNSW_NB15_training-set.csv"
test_path = PROJECT / "UNSW_NB15_testing-set.csv"

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

print("Training shape:", train_df.shape)
print("Testing shape:", test_df.shape)

print("\nTraining labels (0 = normal, 1 = attack):")
print(train_df["label"].value_counts().sort_index())

print("\nTesting labels (0 = normal, 1 = attack):")
print(test_df["label"].value_counts().sort_index())

print("\nAttack categories in training data:")
print(train_df["attack_cat"].value_counts(dropna=False))
```

```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier

# Separate labels
y_train_unsw = train_df["label"].astype(int)
y_test_unsw = test_df["label"].astype(int)

# Exclude labels, category metadata, and record ID
excluded = ["label", "attack_cat", "id"]

X_train_unsw = train_df.drop(
    columns=[c for c in excluded if c in train_df.columns]
).copy()

X_test_unsw = test_df.drop(
    columns=[c for c in excluded if c in test_df.columns]
).copy()

# Identify categorical and numeric columns from training data
categorical_cols = X_train_unsw.select_dtypes(
    include=["object", "category", "string"]
).columns.tolist()

numeric_cols = [
    c for c in X_train_unsw.columns
    if c not in categorical_cols
]

# Build preprocessing using training data only
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median"))
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_cols),
    ("categorical", categorical_pipeline, categorical_cols)
])

print("Training feature columns:", X_train_unsw.shape[1])
print("Numeric features:", len(numeric_cols))
print("Categorical features:", categorical_cols)
print("Training rows:", len(X_train_unsw))
print("Testing rows:", len(X_test_unsw))
print("Training/testing feature columns match:",
      X_train_unsw.columns.tolist() == X_test_unsw.columns.tolist())
```

```python
import time
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline

unsw_rf_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=100,
        max_depth=20,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ))
])

start_time = time.time()

unsw_rf_pipeline.fit(X_train_unsw, y_train_unsw)

training_time = time.time() - start_time

print("UNSW-NB15 Random Forest trained successfully.")
print(f"Training time: {training_time:.2f} seconds")
print("Training rows:", len(X_train_unsw))
print("Features before encoding:", X_train_unsw.shape[1])
print("Classes:", unsw_rf_pipeline.named_steps["classifier"].classes_)
```

```python
import time
import numpy as np
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    classification_report
)

start_time = time.time()

y_pred_unsw = unsw_rf_pipeline.predict(X_test_unsw)
y_prob_unsw = unsw_rf_pipeline.predict_proba(X_test_unsw)[:, 1]

evaluation_time = time.time() - start_time

tn, fp, fn, tp = confusion_matrix(
    y_test_unsw, y_pred_unsw, labels=[0, 1]
).ravel()

unsw_metrics = {
    "Accuracy": accuracy_score(y_test_unsw, y_pred_unsw),
    "Precision": precision_score(y_test_unsw, y_pred_unsw, zero_division=0),
    "Recall": recall_score(y_test_unsw, y_pred_unsw, zero_division=0),
    "F1-score": f1_score(y_test_unsw, y_pred_unsw, zero_division=0),
    "ROC-AUC": roc_auc_score(y_test_unsw, y_prob_unsw),
    "PR-AUC": average_precision_score(y_test_unsw, y_prob_unsw),
    "False Positive Rate": fp / (fp + tn) if (fp + tn) else 0.0,
    "True Negatives": int(tn),
    "False Positives": int(fp),
    "False Negatives": int(fn),
    "True Positives": int(tp),
}

print("=== UNSW-NB15 OFFICIAL TEST SET RESULTS ===")
for metric, value in unsw_metrics.items():
    if isinstance(value, (float, np.floating)):
        print(f"{metric}: {value:.4f}")
    else:
        print(f"{metric}: {value}")

print(f"\nEvaluation time: {evaluation_time:.2f} seconds")

print("\nClassification report:")
print(classification_report(
    y_test_unsw,
    y_pred_unsw,
    labels=[0, 1],
    target_names=["Normal", "Attack"],
    zero_division=0
))
```

```python
import pandas as pd
from pathlib import Path

PROJECT = Path(r"C:\Users\CSE\Downloads\DDoS_IoT_Project")
data_dir = PROJECT / "processed" / "zero_day_samples"

benign_train = pd.read_parquet(data_dir / "benign_train.parquet")
benign_test = pd.read_parquet(data_dir / "benign_test.parquet")
known_attacks = pd.read_parquet(data_dir / "known_attacks.parquet")

metadata_cols = {"Label", "Source_File", "Attack_Type", "Target"}
feature_columns = [
    c for c in benign_train.columns if c not in metadata_cols
]

# Construct binary training labels: benign=0, attack=1
X_source_train = pd.concat(
    [
        benign_train[feature_columns],
        known_attacks[feature_columns],
    ],
    ignore_index=True
)

y_source_train = pd.Series(
    [0] * len(benign_train) + [1] * len(known_attacks),
    name="Label"
)

# Held-out source test: all benign test rows + a reproducible attack sample
attack_test = known_attacks.sample(n=1000, random_state=42)

X_source_test = pd.concat(
    [
        benign_test[feature_columns],
        attack_test[feature_columns],
    ],
    ignore_index=True
)

y_source_test = pd.Series(
    [0] * len(benign_test) + [1] * len(attack_test),
    name="Label"
)

print("Source training shape:", X_source_train.shape)
print("Source training labels:")
print(y_source_train.value_counts().sort_index().rename(index={0: "Benign", 1: "Attack"}))

print("\nSource test shape:", X_source_test.shape)
print("Source test labels:")
print(y_source_test.value_counts().sort_index().rename(index={0: "Benign", 1: "Attack"}))

print("\nMissing values in training features:", X_source_train.isna().sum().sum())
print("Missing values in test features:", X_source_test.isna().sum().sum())
print("Feature count:", len(feature_columns))
```

```python
from pathlib import Path
import numpy as np
import pandas as pd

from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, confusion_matrix, classification_report
)

PROJECT = Path(r"C:\Users\CSE\Downloads\DDoS_IoT_Project")
data_dir = PROJECT / "processed" / "zero_day_samples"
ton_path = PROJECT / "processed" / "ton_iot" / "ton_iot_ddos_balanced.csv"

# Load source-domain CIC data
benign_train = pd.read_parquet(data_dir / "benign_train.parquet")
known_attacks = pd.read_parquet(data_dir / "known_attacks.parquet")

# Load the held-out target-domain ToN-IoT sample
ton = pd.read_csv(ton_path)

# Five explicitly mapped features, in identical order
cic_features = [
    "Flow Duration",
    "Total Length of Fwd Packets",
    "Total Length of Bwd Packets",
    "Total Fwd Packets",
    "Total Backward Packets",
]

ton_features = [
    "duration",
    "src_bytes",
    "dst_bytes",
    "src_pkts",
    "dst_pkts",
]

# Build CIC source training data
X_benign = benign_train[cic_features].copy()
X_attack = known_attacks[cic_features].copy()

# CIC Flow Duration is conventionally measured in microseconds;
# convert it to seconds to match the ToN-IoT duration scale.
X_benign["Flow Duration"] = X_benign["Flow Duration"] / 1_000_000
X_attack["Flow Duration"] = X_attack["Flow Duration"] / 1_000_000

X_source = pd.concat([X_benign, X_attack], ignore_index=True)
y_source = np.concatenate([
    np.zeros(len(X_benign), dtype=int),
    np.ones(len(X_attack), dtype=int)
])

# Construct the target data using the corresponding ToN-IoT features
X_target = ton[ton_features].copy()
X_target.columns = cic_features
y_target = ton["label"].astype(int).to_numpy()

# Keep invalid numeric values from breaking preprocessing
X_source = X_source.replace([np.inf, -np.inf], np.nan)
X_target = X_target.replace([np.inf, -np.inf], np.nan)

# Fit imputation and model on CIC only; no target fitting or threshold tuning
transfer_model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("rf", RandomForestClassifier(
        n_estimators=200,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ))
])

transfer_model.fit(X_source, y_source)

# Evaluate directly on ToN-IoT at the default threshold
target_pred = transfer_model.predict(X_target)
target_prob = transfer_model.predict_proba(X_target)[:, 1]

tn, fp, fn, tp = confusion_matrix(
    y_target, target_pred, labels=[0, 1]
).ravel()

metrics = {
    "Experiment": "CIC-DDoS2019 -> ToN-IoT (exploratory)",
    "Features": len(cic_features),
    "Target_samples": len(y_target),
    "Accuracy": accuracy_score(y_target, target_pred),
    "Precision": precision_score(y_target, target_pred, zero_division=0),
    "Recall": recall_score(y_target, target_pred, zero_division=0),
    "F1": f1_score(y_target, target_pred, zero_division=0),
    "ROC_AUC": roc_auc_score(y_target, target_prob),
    "TN": tn, "FP": fp, "FN": fn, "TP": tp,
    "FPR": fp / (fp + tn) if (fp + tn) else np.nan,
}

print("Source training rows:", len(X_source))
print("Source class counts:", pd.Series(y_source).value_counts().sort_index().to_dict())
print("\nCross-dataset results:")
for key, value in metrics.items():
    print(f"{key}: {value}")

print("\nClassification report:")
print(classification_report(
    y_target, target_pred,
    target_names=["Normal", "DDoS"],
    zero_division=0
))
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import (
    precision_recall_curve,
    roc_curve,
    average_precision_score
)

score_df = pd.DataFrame({
    "Actual_Label": y_target,
    "Predicted_Attack_Probability": target_prob
})

print("Predicted attack probability by true class:")
print(
    score_df.groupby("Actual_Label")["Predicted_Attack_Probability"]
    .describe(percentiles=[0.1, 0.25, 0.5, 0.75, 0.9])
    .to_string()
)

print("\nAverage precision:", round(
    average_precision_score(y_target, target_prob), 4
))

precision, recall, thresholds = precision_recall_curve(
    y_target, target_prob
)

# Diagnostic only: best F1 threshold on target labels.
# This is NOT a valid final test threshold because it uses test labels.
f1_values = 2 * precision[:-1] * recall[:-1] / (
    precision[:-1] + recall[:-1] + 1e-12
)
best_idx = np.argmax(f1_values)

print("\nDiagnostic target-optimized threshold (not for final reporting):",
      round(float(thresholds[best_idx]), 6))
print("Diagnostic best F1:", round(float(f1_values[best_idx]), 4))
print("Diagnostic precision:", round(float(precision[best_idx]), 4))
print("Diagnostic recall:", round(float(recall[best_idx]), 4))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix
import numpy as np
import pandas as pd

# Use the existing CIC source data and feature mapping.
benign_test = pd.read_parquet(data_dir / "benign_test.parquet")

# Split known attacks so the threshold-validation attacks are not
# included in model training.
attack_train, attack_val = train_test_split(
    known_attacks,
    test_size=0.20,
    random_state=42,
    shuffle=True
)

X_train = pd.concat([
    benign_train[cic_features],
    attack_train[cic_features]
], ignore_index=True)

y_train = np.concatenate([
    np.zeros(len(benign_train), dtype=int),
    np.ones(len(attack_train), dtype=int)
])

X_val = pd.concat([
    benign_test[cic_features],
    attack_val[cic_features]
], ignore_index=True)

y_val = np.concatenate([
    np.zeros(len(benign_test), dtype=int),
    np.ones(len(attack_val), dtype=int)
])

# Apply the same duration conversion to all CIC data.
for frame in [X_train, X_val]:
    frame["Flow Duration"] = frame["Flow Duration"] / 1_000_000
    frame.replace([np.inf, -np.inf], np.nan, inplace=True)

source_model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("rf", RandomForestClassifier(
        n_estimators=200,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ))
])

source_model.fit(X_train, y_train)

# Select a threshold on CIC validation only:
# maximize recall among thresholds with validation FPR <= 5%.
val_prob = source_model.predict_proba(X_val)[:, 1]
candidate_thresholds = np.unique(
    np.concatenate(([0.0], val_prob, [1.0]))
)

eligible = []
for threshold in candidate_thresholds:
    pred = (val_prob >= threshold).astype(int)
    tn, fp, fn, tp = confusion_matrix(
        y_val, pred, labels=[0, 1]
    ).ravel()

    fpr = fp / (fp + tn) if (fp + tn) else 0.0
    recall = tp / (tp + fn) if (tp + fn) else 0.0

    if fpr <= 0.05:
        eligible.append((recall, threshold, fpr, tn, fp, fn, tp))

best = max(eligible, key=lambda row: (row[0], row[1]))
source_recall, fixed_threshold, source_fpr, *_ = best

print("CIC-only selected threshold:", fixed_threshold)
print("CIC validation recall:", source_recall)
print("CIC validation FPR:", source_fpr)

# Apply the fixed threshold to ToN-IoT without tuning on its labels.
target_pred_fixed = (
    source_model.predict_proba(X_target)[:, 1] >= fixed_threshold
).astype(int)

tn, fp, fn, tp = confusion_matrix(
    y_target, target_pred_fixed, labels=[0, 1]
).ravel()

print("\nCross-dataset results at CIC-selected threshold")
print("Threshold:", fixed_threshold)
print("Accuracy:", accuracy_score(y_target, target_pred_fixed))
print("Precision:", precision_score(y_target, target_pred_fixed, zero_division=0))
print("Recall:", recall_score(y_target, target_pred_fixed, zero_division=0))
print("F1:", f1_score(y_target, target_pred_fixed, zero_division=0))
print("TN, FP, FN, TP:", tn, fp, fn, tp)
print("ToN-IoT FPR:", fp / (fp + tn) if (fp + tn) else np.nan)
```

```python
import json
from pathlib import Path

notebook_path = Path(
    r"C:\Users\CSE\Downloads\DDoS_IoT_Project"
) / "DDoS_IoT_Majority_Voting_Zero_Day.ipynb"

with open(notebook_path, "r", encoding="utf-8") as f:
    nb = json.load(f)

keywords = [
    "majority", "voting", "fallback", "isolationforest",
    "isolation_forest", "zero_day_final", "final_pred",
    "combined_pred", "ensemble_pred", "confidence"
]

matches = []

for i, cell in enumerate(nb["cells"]):
    if cell.get("cell_type") != "code":
        continue

    source = "".join(cell.get("source", []))
    lines = source.splitlines()

    relevant = [
        (j + 1, line)
        for j, line in enumerate(lines)
        if any(k in line.lower() for k in keywords)
    ]

    if relevant:
        matches.append((i, relevant))

for cell_no, lines in matches:
    print(f"\n--- CELL {cell_no} ---")
    for line_no, line in lines:
        print(f"{line_no}: {line[:200]}")

    if sum(len(str(x)) for x in matches[:matches.index((cell_no, lines)) + 1]) > 10000:
        print("\nOutput limit reached.")
        break
```

```python

from pathlib import Path
from collections import Counter, defaultdict
import pandas as pd

processed_dir = (
    Path(r"C:\Users\CSE\Downloads\DDoS_IoT_Project")
    / "processed"
)

chunk_files = sorted(
    p for p in processed_dir.glob("*.parquet")
    if p.is_file()
)

print("Chunk files:", len(chunk_files))

# Inspect every file's row count and sample labels.
# Only a small sample from each file is read.
label_counts = Counter()
source_counts = Counter()
row_counts = []
errors = []

for i, path in enumerate(chunk_files, start=1):
    try:
        # Parquet metadata provides row count without loading the full file.
        import pyarrow.parquet as pq
        metadata = pq.ParquetFile(path).metadata
        row_counts.append(metadata.num_rows)

        sample = pd.read_parquet(
            path,
            columns=[
                c for c in ["Label", "Source_File"]
                if c in pq.ParquetFile(path).schema.names
            ]
        ).head(100)

        if "Label" in sample:
            label_counts.update(sample["Label"].astype(str))
        if "Source_File" in sample:
            source_counts.update(sample["Source_File"].astype(str))

    except Exception as e:
        errors.append((path.name, str(e)))

print("Total rows across chunks:", sum(row_counts))
print("Smallest/largest chunk:", min(row_counts), max(row_counts))
print("Sampled labels (not full-dataset counts):", label_counts.most_common(15))
print("Sampled source files:", source_counts.most_common(15))
print("Read errors:", len(errors))
if errors:
    print(errors[:5])
```

```python

from pathlib import Path
import pandas as pd

processed_dir = (
    Path(r"C:\Users\CSE\Downloads\DDoS_IoT_Project")
    / "processed"
)
zero_dir = processed_dir / "zero_day_samples"

chunk_files = sorted(processed_dir.glob("*.parquet"))

# Use the first 5 chunks for an initial diagnostic only.
sample_chunks = chunk_files[:5]

columns = [
    "Source Port",
    "Destination Port",
    "Protocol",
    "Flow Duration",
    "Total Fwd Packets",
    "Total Backward Packets",
    "Total Length of Fwd Packets",
    "Total Length of Bwd Packets",
    "Label",
    "Source_File",
]

chunk_parts = []
for path in sample_chunks:
    available = [
        c for c in columns
        if c in pd.read_parquet(path).columns
    ]
    chunk_parts.append(pd.read_parquet(path, columns=available))

chunk_sample = pd.concat(chunk_parts, ignore_index=True)
print("Rows inspected in first 5 chunks:", len(chunk_sample))

for filename in [
    "benign_train.parquet",
    "benign_test.parquet",
    "known_attacks.parquet",
    "snmp_zero_day.parquet",
]:
    path = zero_dir / filename
    df = pd.read_parquet(path)

    shared_cols = [
        c for c in columns
        if c in df.columns and c in chunk_sample.columns
    ]

    # Compare exact values across shared columns.
    chunk_keys = set(
        map(tuple, chunk_sample[shared_cols].astype(str).to_numpy())
    )
    zero_keys = set(
        map(tuple, df[shared_cols].astype(str).to_numpy())
    )

    overlap = len(chunk_keys & zero_keys)
    print(f"{filename}: {overlap} matching sampled-row keys")
```

```python
from pathlib import Path
from collections import defaultdict
import pyarrow.parquet as pq

PROJECT = Path(r"C:\Users\HP\DDoS_IoT_Project")
PROCESSED = PROJECT / "processed"

groups = defaultdict(list)

for path in PROCESSED.glob("file_*_chunk_*.parquet"):
    prefix = path.name.split("_chunk_")[0]
    groups[prefix].append(path)

print("=== CIC PROCESSED CHUNK INVENTORY ===")

for group in sorted(groups):
    chunks = sorted(groups[group])
    numbers = sorted(
        int(p.stem.split("_")[-1])
        for p in chunks
    )

    missing = sorted(set(range(numbers[-1] + 1)) - set(numbers))

    print(f"\n{group}")
    print("Chunk count:", len(chunks))
    print("First chunk:", numbers[0])
    print("Last chunk:", numbers[-1])
    print("Missing chunk numbers:", missing[:20] if missing else "None")

print("\n=== PARQUET READABILITY CHECK ===")

sample_files = [
    path
    for group in sorted(groups)
    for path in [sorted(groups[group])[0], sorted(groups[group])[-1]]
]

for path in sample_files:
    try:
        metadata = pq.read_metadata(path)
        print(
            f"OK | {path.name} | "
            f"{metadata.num_rows:,} rows | "
            f"{metadata.num_columns} columns"
        )
    except Exception as e:
        print(f"ERROR | {path.name} | {str(e)[:150]}")
```

```python
import json
from pathlib import Path

PROJECT = Path(r"C:\Users\HP\DDoS_IoT_Project")

NOTEBOOK = PROJECT / "DDoS_IoT_Majority_Voting_Zero_Day.ipynb"

with open(NOTEBOOK, "r", encoding="utf-8") as f:
    nb = json.load(f)

print("NOTEBOOK AUDIT")
print("=" * 60)

print("Total cells:", len(nb["cells"]))

cell_types = {}

for cell in nb["cells"]:
    kind = cell["cell_type"]
    cell_types[kind] = cell_types.get(kind, 0) + 1

print("Cell types:", cell_types)

print("\n--- EXECUTION SUMMARY ---")

executed = 0
unexecuted = 0
errors = 0

for cell in nb["cells"]:
    if cell["cell_type"] == "code":
        if cell.get("execution_count") is None:
            unexecuted += 1
        else:
            executed += 1

        if any(
            output.get("output_type") == "error"
            for output in cell.get("outputs", [])
        ):
            errors += 1

print("Executed code cells:", executed)
print("Unexecuted code cells:", unexecuted)
print("Cells containing saved errors:", errors)

print("\n--- NOTEBOOK SECTION HEADINGS ---")

for i, cell in enumerate(nb["cells"]):
    if cell["cell_type"] == "markdown":
        text_content = "".join(cell["source"]).strip()

        if text_content.startswith("#"):
            print(f"Cell {i}: {text_content[:150]}")
```

```python
from pathlib import Path
import re
import pandas as pd

PROJECT = Path(r"C:\Users\HP\DDoS_IoT_Project")
PROCESSED = PROJECT / "processed"

parquet_files = list(PROCESSED.glob("file_*.parquet"))

groups = {}

for path in parquet_files:
    match = re.match(
        r"(file_\d+)_chunk_(\d+)\.parquet",
        path.name
    )

    if match:
        prefix = match.group(1)
        chunk_number = int(match.group(2))

        groups.setdefault(prefix, []).append(chunk_number)

print("=" * 75)
print("CIC PARQUET CHUNK AUDIT")
print("=" * 75)

print("Total Parquet chunks:", len(parquet_files))

for prefix, chunks in sorted(groups.items()):
    print(
        f"{prefix}: {len(chunks)} chunks | "
        f"First: {min(chunks):05d} | "
        f"Last: {max(chunks):05d}"
    )

print("\n=== SAMPLE FILE SCHEMAS ===")

for prefix in sorted(groups):
    path = PROCESSED / f"{prefix}_chunk_00000.parquet"

    if path.exists():
        df = pd.read_parquet(path)

        print(f"\n{path.name}")
        print("Shape:", df.shape)
        print("Columns:", df.columns.tolist())
        print("Labels:")
        print(df["Label"].value_counts().to_dict())
        print("Source files:")
        print(df["Source_File"].value_counts().to_dict())
```

```python
from pathlib import Path
import re

PROJECT = Path(r"C:\Users\HP\DDoS_IoT_Project")
PROCESSED = PROJECT / "processed"

groups = {}

for path in PROCESSED.glob("file_*.parquet"):
    match = re.match(
        r"(file_\d+)_chunk_(\d+)\.parquet",
        path.name
    )

    if match:
        prefix = match.group(1)
        number = int(match.group(2))

        groups.setdefault(prefix, set()).add(number)

print("=== CHUNK COMPLETENESS AUDIT ===\n")

all_complete = True

for prefix, chunks in sorted(groups.items()):

    expected = set(range(max(chunks) + 1))
    missing = sorted(expected - chunks)

    status = "COMPLETE" if not missing else "MISSING CHUNKS"

    print(f"{prefix}: {status}")

    if missing:
        all_complete = False
        print("Missing:", missing[:30])

print("\nOverall status:",
      "ALL CHUNKS PRESENT" if all_complete
      else "SOME CHUNKS MISSING")
```

```python
from pathlib import Path
import pandas as pd
from collections import defaultdict
from itertools import combinations
import re
import time

PROJECT = Path(r"C:\Users\HP\DDoS_IoT_Project")
PROCESSED = PROJECT / "processed"

groups = defaultdict(set)
benign_counts = defaultdict(int)

start = time.time()

parquet_files = sorted(PROCESSED.glob("file_*.parquet"))

print("Scanning", len(parquet_files), "Parquet chunks...")
print("This may take a few minutes.\n")

feature_columns = None

for i, path in enumerate(parquet_files, start=1):

    match = re.match(r"(file_\d+)_chunk_", path.name)
    source_group = match.group(1)

    df = pd.read_parquet(path)

    if feature_columns is None:
        feature_columns = [
            c for c in df.columns
            if c not in ["Label", "Source_File"]
        ]

    labels = df["Label"].astype("string").str.strip().str.upper()
    benign = df.loc[labels == "BENIGN", feature_columns]

    benign_counts[source_group] += len(benign)

    if len(benign):
        hashes = pd.util.hash_pandas_object(
            benign,
            index=False
        )

        groups[source_group].update(hashes.tolist())

    del df, benign

    if i % 100 == 0:
        print(f"Processed {i}/{len(parquet_files)} chunks")

print("\n" + "=" * 75)
print("BENIGN DUPLICATE AUDIT")
print("=" * 75)

print("\nBenign rows and unique feature patterns by source:")

for source in sorted(benign_counts):
    print(
        f"{source}: "
        f"{benign_counts[source]:,} benign rows | "
        f"{len(groups[source]):,} unique patterns"
    )

print("\n=== CROSS-SOURCE OVERLAP ===")

for a, b in combinations(sorted(groups), 2):
    overlap = groups[a] & groups[b]

    if overlap:
        print(f"{a} vs {b}: {len(overlap):,} shared patterns")

total_unique = len(set().union(*groups.values()))

print("\nTotal benign rows:", sum(benign_counts.values()))
print("Unique benign feature patterns across sources:", total_unique)
print("Elapsed time:", round(time.time() - start, 2), "seconds")
```

```python
from collections import defaultdict

# Find how many source groups contain each benign pattern
pattern_sources = defaultdict(list)

for source, hashes in groups.items():
    for h in hashes:
        pattern_sources[h].append(source)

# Patterns appearing in multiple source groups
shared_patterns = {
    h: sources
    for h, sources in pattern_sources.items()
    if len(sources) > 1
}

# Sort by number of source groups sharing the pattern
top_shared = sorted(
    shared_patterns.items(),
    key=lambda x: len(x[1]),
    reverse=True
)

print("Total cross-source shared patterns:", len(shared_patterns))

print("\n=== MOST WIDELY SHARED PATTERNS ===")

for h, sources in top_shared[:15]:
    print(f"\nHash: {h}")
    print("Number of source groups:", len(sources))
    print("Source groups:", ", ".join(sorted(sources)))
```

```python
from pathlib import Path
import pandas as pd
import re
from collections import defaultdict

PROJECT = Path(r"C:\Users\HP\DDoS_IoT_Project")
PROCESSED = PROJECT / "processed"

# Hashes identified in the previous audit
target_hashes = set(shared_patterns.keys())

matched_records = defaultdict(list)

parquet_files = sorted(PROCESSED.glob("file_*.parquet"))

for i, path in enumerate(parquet_files, start=1):

    match = re.match(r"(file_\d+)_chunk_", path.name)
    source_group = match.group(1)

    df = pd.read_parquet(path)

    labels = df["Label"].astype("string").str.strip().str.upper()
    benign = df.loc[labels == "BENIGN"].copy()

    if not benign.empty:

        features = benign.drop(
            columns=["Label", "Source_File"],
            errors="ignore"
        )

        hashes = pd.util.hash_pandas_object(
            features,
            index=False
        )

        selected = benign.loc[hashes.isin(target_hashes)]

        for _, row in selected.iterrows():
            feature_values = row.drop(
                labels=["Label", "Source_File"]
            ).tolist()

            matched_records[
                int(pd.util.hash_pandas_object(
                    pd.DataFrame([feature_values]),
                    index=False
                ).iloc[0])
            ].append({
                "source": source_group,
                "label": row["Label"],
                "values": feature_values
            })

    del df, benign

    if i % 500 == 0:
        print(f"Scanned {i}/{len(parquet_files)} chunks")

print("\n=== SHARED RECORD INSPECTION ===")

for h, sources in sorted(
    shared_patterns.items(),
    key=lambda x: len(x[1]),
    reverse=True
)[:10]:

    records = matched_records.get(h, [])

    print("\nHash:", h)
    print("Expected source groups:", len(sources))
    print("Records retrieved:", len(records))

    if records:
        values = records[0]["values"]

        print("Zero-valued features:", sum(v == 0 for v in values))
        print("Total features:", len(values))
        print("First 15 feature values:", values[:15])
```

```python
from pathlib import Path
import pandas as pd
import re
from collections import defaultdict

PROJECT = Path(r"C:\Users\HP\DDoS_IoT_Project")
PROCESSED = PROJECT / "processed"

target_hashes = set(shared_patterns.keys())
matched_records = defaultdict(list)

parquet_files = sorted(PROCESSED.glob("file_*.parquet"))

for i, path in enumerate(parquet_files, start=1):

    match = re.match(r"(file_\d+)_chunk_", path.name)
    source_group = match.group(1)

    df = pd.read_parquet(path)

    labels = df["Label"].astype("string").str.strip().str.upper()
    benign = df.loc[labels == "BENIGN"].copy()

    if not benign.empty:

        features = benign.drop(
            columns=["Label", "Source_File"],
            errors="ignore"
        )

        hashes = pd.util.hash_pandas_object(
            features,
            index=False
        )

        matching_indices = hashes[hashes.isin(target_hashes)].index

        for idx in matching_indices:

            h = int(hashes.loc[idx])
            row = benign.loc[idx]

            matched_records[h].append({
                "source": source_group,
                "label": row["Label"],
                "values": row.drop(
                    labels=["Label", "Source_File"]
                ).tolist()
            })

    del df, benign

    if i % 500 == 0:
        print(f"Scanned {i}/{len(parquet_files)} chunks")

print("\n=== CORRECTED SHARED RECORD INSPECTION ===")

for h, sources in sorted(
    shared_patterns.items(),
    key=lambda x: len(x[1]),
    reverse=True
)[:10]:

    records = matched_records.get(h, [])

    print("\nHash:", h)
    print("Expected source groups:", len(sources))
    print("Records retrieved:", len(records))

    if records:
        values = records[0]["values"]

        print("Zero-valued features:", sum(v == 0 for v in values))
        print("Total features:", len(values))
        print("First 15 feature values:", values[:15])
```

```python
from pathlib import Path
import pandas as pd
import re
from collections import defaultdict, Counter

PROJECT = Path(r"C:\Users\HP\DDoS_IoT_Project")
PROCESSED = PROJECT / "processed"

target_hash = max(
    shared_patterns,
    key=lambda h: len(shared_patterns[h])
)

label_counts = Counter()
source_counts = defaultdict(Counter)

parquet_files = sorted(PROCESSED.glob("file_*.parquet"))

for i, path in enumerate(parquet_files, start=1):

    match = re.match(r"(file_\d+)_chunk_", path.name)
    source_group = match.group(1)

    df = pd.read_parquet(path)

    features = df.drop(
        columns=["Label", "Source_File"],
        errors="ignore"
    )

    hashes = pd.util.hash_pandas_object(
        features,
        index=False
    )

    selected = df.loc[hashes == target_hash]

    for _, row in selected.iterrows():
        label = str(row["Label"]).strip()

        label_counts[label] += 1
        source_counts[source_group][label] += 1

    del df, features

    if i % 500 == 0:
        print(f"Scanned {i}/{len(parquet_files)} chunks")

print("\n=== LABEL CONSISTENCY CHECK ===")

print("\nLabels associated with this feature pattern:")

for label, count in label_counts.items():
    print(f"{label}: {count}")

print("\nOccurrences by source group:")

for source, counts in sorted(source_counts.items()):
    print(source, dict(counts))
```

```python
from pathlib import Path
import pandas as pd
from collections import defaultdict, Counter

PROJECT = Path(r"C:\Users\HP\DDoS_IoT_Project")
PROCESSED = PROJECT / "processed"

target_hashes = set(shared_patterns.keys())

pattern_labels = defaultdict(Counter)

parquet_files = sorted(PROCESSED.glob("file_*.parquet"))

for i, path in enumerate(parquet_files, start=1):

    df = pd.read_parquet(path)

    features = df.drop(
        columns=["Label", "Source_File"],
        errors="ignore"
    )

    hashes = pd.util.hash_pandas_object(
        features,
        index=False
    )

    selected = df.loc[hashes.isin(target_hashes)]

    selected_hashes = hashes.loc[selected.index]

    for h, label in zip(
        selected_hashes,
        selected["Label"]
    ):
        pattern_labels[int(h)][str(label).strip()] += 1

    del df, features

    if i % 500 == 0:
        print(f"Scanned {i}/{len(parquet_files)} chunks")

print("\n=== ALL SHARED PATTERN LABEL AUDIT ===")

conflicts = []

for h, counts in pattern_labels.items():

    if len(counts) > 1:
        conflicts.append((h, counts))

print("Shared patterns examined:", len(pattern_labels))
print("Patterns with conflicting labels:", len(conflicts))

for h, counts in conflicts:
    print("\nHash:", h)
    print("Labels:", dict(counts))

print("\nPatterns with BENIGN-only labels:",
      sum(1 for c in pattern_labels.values()
          if set(c.keys()) == {"BENIGN"}))
```

```python
from pathlib import Path
import pandas as pd
import numpy as np

PROJECT = Path(r"C:\Users\HP\DDoS_IoT_Project")
PROCESSED = PROJECT / "processed"

OUTPUT = PROCESSED / "final_cic_known_attacks.parquet"

TARGET_CLASSES = [
    "BENIGN",
    "DrDoS_DNS",
    "DrDoS_LDAP",
    "DrDoS_NetBIOS",
    "DrDoS_UDP"
]

SAMPLES_PER_CLASS = 20000
SAMPLE_PER_CHUNK = 500
RANDOM_STATE = 42

rng = np.random.default_rng(RANDOM_STATE)

parquet_files = sorted(PROCESSED.glob("file_*.parquet"))

collected = {label: [] for label in TARGET_CLASSES}

print("Building balanced CIC dataset...")

for i, path in enumerate(parquet_files, 1):

    df = pd.read_parquet(path)

    for label in TARGET_CLASSES:

        subset = df[df["Label"].astype(str) == label]

        if len(subset) > SAMPLE_PER_CHUNK:
            subset = subset.sample(
                n=SAMPLE_PER_CHUNK,
                random_state=RANDOM_STATE + i
            )

        if not subset.empty:
            collected[label].append(subset)

    if i % 200 == 0:
        print(f"Processed {i}/{len(parquet_files)} chunks")

final_parts = []

for label in TARGET_CLASSES:

    if not collected[label]:
        raise ValueError(f"No records found for {label}")

    class_df = pd.concat(
        collected[label],
        ignore_index=True
    )

    if len(class_df) < SAMPLES_PER_CLASS:
        raise ValueError(
            f"Insufficient samples for {label}: {len(class_df)}"
        )

    class_df = class_df.sample(
        n=SAMPLES_PER_CLASS,
        random_state=RANDOM_STATE
    )

    final_parts.append(class_df)

final_df = pd.concat(final_parts, ignore_index=True)

final_df = final_df.sample(
    frac=1,
    random_state=RANDOM_STATE
).reset_index(drop=True)

final_df.to_parquet(OUTPUT, index=False)

print("\n========== FINAL DATASET ==========")
print("Shape:", final_df.shape)
print("\nClass distribution:")
print(final_df["Label"].value_counts())
print("\nSaved to:", OUTPUT)
```

```python
from pathlib import Path
import pandas as pd
import numpy as np
import joblib

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import (
    SelectKBest,
    mutual_info_classif
)

# ==========================================================
# 1. PATHS AND CONFIGURATION
# ==========================================================

PROJECT = Path(r"C:\Users\HP\DDoS_IoT_Project")

PROCESSED = PROJECT / "processed"
MODELS = PROJECT / "models"

MODELS.mkdir(parents=True, exist_ok=True)

ATTACKS = [
    "DrDoS_DNS",
    "DrDoS_LDAP",
    "DrDoS_NetBIOS",
    "DrDoS_UDP"
]

FEATURE_COUNT = 20
RANDOM_STATE = 42

# ==========================================================
# 2. LOAD DATA
# ==========================================================

train_df = pd.read_parquet(
    PROCESSED / "cic_train.parquet"
)

val_df = pd.read_parquet(
    PROCESSED / "cic_validation.parquet"
)

test_df = pd.read_parquet(
    PROCESSED / "cic_test.parquet"
)

print("Datasets loaded successfully.")

print("Training:", train_df.shape)
print("Validation:", val_df.shape)
print("Testing:", test_df.shape)

# ==========================================================
# 3. IDENTIFY FEATURES
# ==========================================================

feature_columns = [
    col for col in train_df.columns
    if col not in ["Label", "Source_File"]
]

print("\nTotal original features:", len(feature_columns))

def prepare_features(data):

    X = data[feature_columns].copy()

    X = X.apply(
        pd.to_numeric,
        errors="coerce"
    )

    X = X.replace(
        [np.inf, -np.inf],
        np.nan
    )

    return X

# ==========================================================
# 4. PICKLABLE MUTUAL INFORMATION FUNCTION
# ==========================================================

def mutual_info_score(X, y):

    return mutual_info_classif(
        X,
        y,
        random_state=RANDOM_STATE
    )

# ==========================================================
# 5. PREPROCESS EACH ATTACK SEPARATELY
# ==========================================================

processed_experiments = {}

for attack in ATTACKS:

    print("\n" + "=" * 65)
    print(f"PROCESSING: BENIGN vs {attack}")
    print("=" * 65)

    # ------------------------------------------------------
    # A. Filter binary classification datasets
    # ------------------------------------------------------

    train_part = train_df[
        train_df["Label"].isin(["BENIGN", attack])
    ].copy()

    val_part = val_df[
        val_df["Label"].isin(["BENIGN", attack])
    ].copy()

    test_part = test_df[
        test_df["Label"].isin(["BENIGN", attack])
    ].copy()

    # Binary labels: BENIGN = 0, ATTACK = 1

    y_train = (
        train_part["Label"] == attack
    ).astype(int)

    y_val = (
        val_part["Label"] == attack
    ).astype(int)

    y_test = (
        test_part["Label"] == attack
    ).astype(int)

    # ------------------------------------------------------
    # B. Extract numerical features
    # ------------------------------------------------------

    X_train = prepare_features(train_part)
    X_val = prepare_features(val_part)
    X_test = prepare_features(test_part)

    # ------------------------------------------------------
    # C. Missing-value imputation
    # Fit ONLY on training data
    # ------------------------------------------------------

    imputer = SimpleImputer(
        strategy="median"
    )

    X_train_imp = imputer.fit_transform(
        X_train
    )

    X_val_imp = imputer.transform(
        X_val
    )

    X_test_imp = imputer.transform(
        X_test
    )

    # ------------------------------------------------------
    # D. Feature selection
    # Fit ONLY on training data
    # ------------------------------------------------------

    selector = SelectKBest(
        score_func=mutual_info_score,
        k=FEATURE_COUNT
    )

    X_train_selected = selector.fit_transform(
        X_train_imp,
        y_train
    )

    X_val_selected = selector.transform(
        X_val_imp
    )

    X_test_selected = selector.transform(
        X_test_imp
    )

    selected_features = np.array(
        feature_columns
    )[selector.get_support()].tolist()

    # ------------------------------------------------------
    # E. Feature scaling
    # Fit ONLY on training data
    # ------------------------------------------------------

    scaler = StandardScaler()

    X_train_scaled = scaler.fit_transform(
        X_train_selected
    )

    X_val_scaled = scaler.transform(
        X_val_selected
    )

    X_test_scaled = scaler.transform(
        X_test_selected
    )

    # ------------------------------------------------------
    # F. Store processed experiment
    # ------------------------------------------------------

    processed_experiments[attack] = {

        "X_train": X_train_scaled,
        "X_val": X_val_scaled,
        "X_test": X_test_scaled,

        "y_train": y_train.to_numpy(),
        "y_val": y_val.to_numpy(),
        "y_test": y_test.to_numpy(),

        "selected_features": selected_features
    }

    # ------------------------------------------------------
    # G. Save individual preprocessing objects
    # ------------------------------------------------------

    preprocessing_objects = {

        "feature_columns": feature_columns,

        "imputer": imputer,

        "selector": selector,

        "scaler": scaler,

        "selected_features": selected_features
    }

    joblib.dump(
        preprocessing_objects,
        MODELS / f"{attack}_preprocessing.joblib"
    )

    # ------------------------------------------------------
    # H. Display experiment summary
    # ------------------------------------------------------

    print("\nTraining shape:", X_train_scaled.shape)
    print("Validation shape:", X_val_scaled.shape)
    print("Testing shape:", X_test_scaled.shape)

    print("\nTraining labels:", np.bincount(y_train))
    print("Validation labels:", np.bincount(y_val))
    print("Testing labels:", np.bincount(y_test))

    print("\nSelected features:")

    for feature in selected_features:
        print(" -", feature)

    print(f"\n{attack} preprocessing completed.")

# ==========================================================
# 6. SAVE ALL PROCESSED EXPERIMENTS
# ==========================================================

joblib.dump(
    processed_experiments,
    PROCESSED / "cic_binary_experiments.joblib",
    compress=3
)

print("\n" + "=" * 65)
print("ALL FOUR EXPERIMENTS PREPROCESSED SUCCESSFULLY!")
print("=" * 65)

print("\nExperiments created:")

for attack in ATTACKS:
    print(f"✓ BENIGN vs {attack}")

print("\nPreprocessing objects saved in:", MODELS)
print("Processed experiments saved in:", PROCESSED)
```

```python
import time
import joblib
import pandas as pd
import numpy as np

from pathlib import Path

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier,
    GradientBoostingClassifier,
    HistGradientBoostingClassifier,
    AdaBoostClassifier
)
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.neural_network import MLPClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

# =====================================================
# 1. CONFIGURATION
# =====================================================

PROJECT = Path(r"C:\Users\HP\DDoS_IoT_Project")
PROCESSED = PROJECT / "processed"
MODELS = PROJECT / "models"
RESULTS = PROJECT / "results"

RESULTS.mkdir(exist_ok=True)

processed_experiments = joblib.load(
    PROCESSED / "cic_binary_experiments.joblib"
)

# =====================================================
# 2. DEFINE CLASSIFIERS
# =====================================================

def get_models():

    return {
        "Logistic Regression": LogisticRegression(
            max_iter=1500,
            random_state=42
        ),

        "Decision Tree": DecisionTreeClassifier(
            max_depth=20,
            random_state=42
        ),

        "Random Forest": RandomForestClassifier(
            n_estimators=150,
            max_depth=25,
            n_jobs=-1,
            random_state=42
        ),

        "Extra Trees": ExtraTreesClassifier(
            n_estimators=150,
            max_depth=25,
            n_jobs=-1,
            random_state=42
        ),

        "Gradient Boosting": GradientBoostingClassifier(
            n_estimators=100,
            learning_rate=0.1,
            max_depth=3,
            random_state=42
        ),

        "HistGradientBoosting": HistGradientBoostingClassifier(
            max_iter=100,
            learning_rate=0.1,
            max_leaf_nodes=31,
            random_state=42
        ),

        "AdaBoost": AdaBoostClassifier(
            n_estimators=100,
            learning_rate=0.5,
            random_state=42
        ),

        "KNN": KNeighborsClassifier(
            n_neighbors=7,
            n_jobs=-1
        ),

        "Gaussian NB": GaussianNB(),

        "MLP": MLPClassifier(
            hidden_layer_sizes=(100, 50),
            max_iter=150,
            early_stopping=True,
            random_state=42
        )
    }


# =====================================================
# 3. EVALUATION FUNCTION
# =====================================================

def evaluate_model(model, X, y):

    predictions = model.predict(X)

    if hasattr(model, "predict_proba"):
        probabilities = model.predict_proba(X)[:, 1]

    elif hasattr(model, "decision_function"):
        probabilities = model.decision_function(X)

    else:
        probabilities = predictions

    tn, fp, fn, tp = confusion_matrix(
        y, predictions, labels=[0, 1]
    ).ravel()

    return {
        "Accuracy": accuracy_score(y, predictions),
        "Precision": precision_score(
            y, predictions, zero_division=0
        ),
        "Recall": recall_score(
            y, predictions, zero_division=0
        ),
        "F1": f1_score(
            y, predictions, zero_division=0
        ),
        "ROC_AUC": roc_auc_score(y, probabilities),
        "FPR": fp / (fp + tn) if (fp + tn) else 0
    }


# =====================================================
# 4. TRAIN ALL MODELS
# =====================================================

all_results = []
trained_models = {}

for attack, data in processed_experiments.items():

    print("\n" + "=" * 70)
    print(f"TRAINING MODELS: BENIGN vs {attack}")
    print("=" * 70)

    X_train = data["X_train"]
    y_train = data["y_train"]

    X_val = data["X_val"]
    y_val = data["y_val"]

    trained_models[attack] = {}

    models = get_models()

    for model_name, model in models.items():

        print(f"\nTraining {model_name}...")

        start_time = time.time()

        model.fit(X_train, y_train)

        training_time = time.time() - start_time

        metrics = evaluate_model(
            model,
            X_val,
            y_val
        )

        metrics["Attack"] = attack
        metrics["Model"] = model_name
        metrics["Training_Time"] = training_time

        all_results.append(metrics)

        trained_models[attack][model_name] = model

        print(
            f"F1: {metrics['F1']:.4f} | "
            f"AUC: {metrics['ROC_AUC']:.4f} | "
            f"FPR: {metrics['FPR']:.4f} | "
            f"Time: {training_time:.2f}s"
        )

    print(f"\nCompleted {attack}.")


# =====================================================
# 5. SAVE MODELS AND VALIDATION RESULTS
# =====================================================

results_df = pd.DataFrame(all_results)

results_df = results_df.sort_values(
    ["Attack", "F1"],
    ascending=[True, False]
)

results_df.to_csv(
    RESULTS / "cic_individual_model_validation.csv",
    index=False
)

joblib.dump(
    trained_models,
    MODELS / "cic_individual_models.joblib",
    compress=3
)


# =====================================================
# 6. DISPLAY RESULTS
# =====================================================

print("\n" + "=" * 70)
print("INDIVIDUAL CLASSIFIER VALIDATION RESULTS")
print("=" * 70)

for attack in processed_experiments:

    print(f"\n{attack}")

    display(
        results_df[
            results_df["Attack"] == attack
        ][
            [
                "Model",
                "Accuracy",
                "Precision",
                "Recall",
                "F1",
                "ROC_AUC",
                "FPR",
                "Training_Time"
            ]
        ].reset_index(drop=True)
    )

print("\nModels saved successfully.")
print("Results saved successfully.")
```

```python
import numpy as np
import pandas as pd
import joblib

from pathlib import Path

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

PROJECT = Path(r"C:\Users\HP\DDoS_IoT_Project")

PROCESSED = PROJECT / "processed"
MODELS = PROJECT / "models"
RESULTS = PROJECT / "results"

# Load saved models and validation results
trained_models = joblib.load(
    MODELS / "cic_individual_models.joblib"
)

processed_experiments = joblib.load(
    PROCESSED / "cic_binary_experiments.joblib"
)

results_df = pd.read_csv(
    RESULTS / "cic_individual_model_validation.csv"
)

ensemble_models = {}
ensemble_summary = []

# --------------------------------------------------
# Evaluation function
# --------------------------------------------------

def ensemble_metrics(y_true, predictions, probabilities):

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        predictions,
        labels=[0, 1]
    ).ravel()

    return {
        "Accuracy": accuracy_score(y_true, predictions),
        "Precision": precision_score(
            y_true, predictions, zero_division=0
        ),
        "Recall": recall_score(
            y_true, predictions, zero_division=0
        ),
        "F1": f1_score(
            y_true, predictions, zero_division=0
        ),
        "ROC_AUC": roc_auc_score(
            y_true, probabilities
        ),
        "FPR": fp / (fp + tn) if (fp + tn) else 0
    }


# --------------------------------------------------
# Build ensembles using validation results
# --------------------------------------------------

for attack, data in processed_experiments.items():

    print("\n" + "=" * 65)
    print("ENSEMBLE CONSTRUCTION:", attack)
    print("=" * 65)

    # Select top five using validation F1
    attack_results = results_df[
        results_df["Attack"] == attack
    ].copy()

    selected = attack_results.sort_values(
        ["F1", "ROC_AUC", "FPR"],
        ascending=[False, False, True]
    ).head(5)

    selected_names = selected["Model"].tolist()

    models = {
        name: trained_models[attack][name]
        for name in selected_names
    }

    print("Selected classifiers:", selected_names)

    X_val = data["X_val"]
    y_val = data["y_val"]

    # Validation probabilities
    val_probabilities = np.column_stack([
        model.predict_proba(X_val)[:, 1]
        for model in models.values()
    ])

    val_predictions = (val_probabilities >= 0.5).astype(int)

    # ----------------------------------------------
    # A. Majority voting
    # ----------------------------------------------

    majority_votes = (
        val_predictions.sum(axis=1) >= 3
    ).astype(int)

    majority_probability = (
        val_probabilities.mean(axis=1)
    )

    majority_result = ensemble_metrics(
        y_val,
        majority_votes,
        majority_probability
    )

    # ----------------------------------------------
    # B. F1-based weighted voting
    # ----------------------------------------------

    weights = selected["F1"].to_numpy()

    weights = weights / weights.sum()

    weighted_probability = np.average(
        val_probabilities,
        axis=1,
        weights=weights
    )

    weighted_predictions = (
        weighted_probability >= 0.5
    ).astype(int)

    weighted_result = ensemble_metrics(
        y_val,
        weighted_predictions,
        weighted_probability
    )

    # ----------------------------------------------
    # C. Adaptive validation-based weighting
    # ----------------------------------------------

    f1_scores = selected["F1"].to_numpy()

    fpr_scores = selected["FPR"].to_numpy()

    adaptive_weights = (
        (f1_scores + 1e-6) /
        (fpr_scores + 0.01)
    )

    adaptive_weights = (
        adaptive_weights / adaptive_weights.sum()
    )

    adaptive_probability = np.average(
        val_probabilities,
        axis=1,
        weights=adaptive_weights
    )

    adaptive_predictions = (
        adaptive_probability >= 0.5
    ).astype(int)

    adaptive_result = ensemble_metrics(
        y_val,
        adaptive_predictions,
        adaptive_probability
    )

    # ----------------------------------------------
    # Save ensemble configuration
    # ----------------------------------------------

    ensemble_models[attack] = {
        "selected_models": models,
        "selected_names": selected_names,
        "majority_weights": np.ones(len(models)) / len(models),
        "weighted_weights": weights,
        "adaptive_weights": adaptive_weights,
        "threshold": 0.5
    }

    for name, metrics in [
        ("Majority Voting", majority_result),
        ("Weighted Voting", weighted_result),
        ("Adaptive Weighted Voting", adaptive_result)
    ]:

        metrics["Attack"] = attack
        metrics["Ensemble"] = name

        ensemble_summary.append(metrics)

        print(f"\n{name}:")
        for metric, value in metrics.items():

            if metric not in ["Attack", "Ensemble"]:
                print(f"{metric}: {value:.5f}")


# --------------------------------------------------
# Save results
# --------------------------------------------------

ensemble_results = pd.DataFrame(ensemble_summary)

ensemble_results.to_csv(
    RESULTS / "cic_ensemble_validation.csv",
    index=False
)

joblib.dump(
    ensemble_models,
    MODELS / "cic_ensemble_configurations.joblib",
    compress=3
)

print("\n" + "=" * 65)
print("ENSEMBLE VALIDATION COMPLETED")
print("=" * 65)

print(ensemble_results.to_string(index=False))
```

```python
import numpy as np
import pandas as pd
import joblib

from pathlib import Path
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

PROJECT = Path(r"C:\Users\HP\DDoS_IoT_Project")

PROCESSED = PROJECT / "processed"
MODELS = PROJECT / "models"
RESULTS = PROJECT / "results"

RESULTS.mkdir(exist_ok=True)

# Load frozen models and ensemble configurations
trained_models = joblib.load(
    MODELS / "cic_individual_models.joblib"
)

ensemble_models = joblib.load(
    MODELS / "cic_ensemble_configurations.joblib"
)

experiments = joblib.load(
    PROCESSED / "cic_binary_experiments.joblib"
)

final_results = []
confusion_matrices = {}

def calculate_metrics(y_true, predictions, probabilities):

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        predictions,
        labels=[0, 1]
    ).ravel()

    return {
        "Accuracy": accuracy_score(y_true, predictions),
        "Precision": precision_score(
            y_true, predictions, zero_division=0
        ),
        "Recall": recall_score(
            y_true, predictions, zero_division=0
        ),
        "F1": f1_score(
            y_true, predictions, zero_division=0
        ),
        "ROC_AUC": roc_auc_score(
            y_true, probabilities
        ),
        "FPR": fp / (fp + tn) if (fp + tn) else 0,
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp
    }


for attack, data in experiments.items():

    print("\n" + "=" * 70)
    print("FINAL TEST EVALUATION:", attack)
    print("=" * 70)

    X_test = data["X_test"]
    y_test = data["y_test"]

    attack_confusions = {}

    # --------------------------------------------
    # A. Individual classifiers
    # --------------------------------------------

    for model_name, model in trained_models[attack].items():

        probabilities = model.predict_proba(X_test)[:, 1]

        predictions = (probabilities >= 0.5).astype(int)

        metrics = calculate_metrics(
            y_test,
            predictions,
            probabilities
        )

        metrics["Attack"] = attack
        metrics["Method"] = model_name

        final_results.append(metrics)

        attack_confusions[model_name] = confusion_matrix(
            y_test, predictions
        )

    # --------------------------------------------
    # B. Ensemble classifiers
    # --------------------------------------------

    config = ensemble_models[attack]

    names = config["selected_names"]
    models = config["selected_models"]

    probabilities_matrix = np.column_stack([
        models[name].predict_proba(X_test)[:, 1]
        for name in names
    ])

    binary_predictions = (
        probabilities_matrix >= 0.5
    ).astype(int)

    # Majority voting
    majority_pred = (
        binary_predictions.sum(axis=1) >= 3
    ).astype(int)

    majority_prob = probabilities_matrix.mean(axis=1)

    # Weighted voting
    weighted_prob = np.average(
        probabilities_matrix,
        axis=1,
        weights=config["weighted_weights"]
    )

    weighted_pred = (
        weighted_prob >= 0.5
    ).astype(int)

    # Adaptive weighted voting
    adaptive_prob = np.average(
        probabilities_matrix,
        axis=1,
        weights=config["adaptive_weights"]
    )

    adaptive_pred = (
        adaptive_prob >= 0.5
    ).astype(int)

    ensemble_predictions = {
        "Majority Voting": (majority_pred, majority_prob),
        "Weighted Voting": (weighted_pred, weighted_prob),
        "Adaptive Weighted Voting": (
            adaptive_pred, adaptive_prob
        )
    }

    for method, (pred, prob) in ensemble_predictions.items():

        metrics = calculate_metrics(
            y_test,
            pred,
            prob
        )

        metrics["Attack"] = attack
        metrics["Method"] = method

        final_results.append(metrics)

        attack_confusions[method] = confusion_matrix(
            y_test, pred
        )

    confusion_matrices[attack] = attack_confusions


# --------------------------------------------
# C. Save final results
# --------------------------------------------

final_df = pd.DataFrame(final_results)

final_df.to_csv(
    RESULTS / "cic_final_test_results.csv",
    index=False
)

joblib.dump(
    confusion_matrices,
    RESULTS / "cic_test_confusion_matrices.joblib"
)

print("\n" + "=" * 70)
print("FINAL TEST RESULTS")
print("=" * 70)

for attack in experiments:

    print(f"\n{attack}")

    display(
        final_df[
            final_df["Attack"] == attack
        ][
            [
                "Method",
                "Accuracy",
                "Precision",
                "Recall",
                "F1",
                "ROC_AUC",
                "FPR"
            ]
        ].sort_values("F1", ascending=False)
    )

print("\nFinal test results saved successfully.")
```

```python
import pandas as pd
import numpy as np
from pathlib import Path

BASE_DIR = Path(r"C:\Users\HP\DDoS_IoT_Project")
PROCESSED_DIR = BASE_DIR / "processed"

# Load the three original splits
train_df = pd.read_parquet(PROCESSED_DIR / "cic_train.parquet")
val_df = pd.read_parquet(PROCESSED_DIR / "cic_validation.parquet")
test_df = pd.read_parquet(PROCESSED_DIR / "cic_test.parquet")

splits = {
    "Train": train_df,
    "Validation": val_df,
    "Test": test_df
}

# Exclude labels and source metadata from feature comparison
feature_cols = [
    col for col in train_df.columns
    if col not in ["Label", "Source_File"]
]

print("=" * 65)
print("1. DATASET DIMENSIONS")
print("=" * 65)

for name, df in splits.items():
    print(f"{name}: {df.shape}")

print("\nNumber of features:", len(feature_cols))

# Generate hashes using feature values only
hashes = {}

for name, df in splits.items():
    hashes[name] = pd.util.hash_pandas_object(
        df[feature_cols],
        index=False
    )

# Check exact feature-row overlaps between splits
print("\n" + "=" * 65)
print("2. CROSS-SPLIT DUPLICATE FEATURE RECORDS")
print("=" * 65)

pairs = [
    ("Train", "Validation"),
    ("Train", "Test"),
    ("Validation", "Test")
]

for first, second in pairs:
    overlap = len(set(hashes[first]) & set(hashes[second]))
    print(f"{first} vs {second}: {overlap} matching feature rows")

# Check duplicates within each split
print("\n" + "=" * 65)
print("3. DUPLICATES WITHIN EACH SPLIT")
print("=" * 65)

for name, df in splits.items():
    duplicate_count = df[feature_cols].duplicated().sum()
    print(f"{name}: {duplicate_count} duplicate feature rows")

# Inspect source-file distribution
print("\n" + "=" * 65)
print("4. SOURCE FILE DISTRIBUTION")
print("=" * 65)

for name, df in splits.items():
    print(f"\n{name}:")
    print(df["Source_File"].value_counts().head(15))

# Check which source files are shared between splits
print("\n" + "=" * 65)
print("5. SOURCE FILE OVERLAP")
print("=" * 65)

source_sets = {
    name: set(df["Source_File"].dropna().unique())
    for name, df in splits.items()
}

for first, second in pairs:
    shared = source_sets[first] & source_sets[second]
    print(f"{first} vs {second}: {len(shared)} shared source files")
    if shared:
        print("Shared files:", list(shared)[:10])

print("\nAUDIT CELL COMPLETED.")
```

```python
import pandas as pd
from pathlib import Path

BASE_DIR = Path(r"C:\Users\HP\DDoS_IoT_Project")
PROCESSED_DIR = BASE_DIR / "processed"

train_df = pd.read_parquet(PROCESSED_DIR / "cic_train.parquet")
val_df = pd.read_parquet(PROCESSED_DIR / "cic_validation.parquet")
test_df = pd.read_parquet(PROCESSED_DIR / "cic_test.parquet")

feature_cols = [
    col for col in train_df.columns
    if col not in ["Label", "Source_File"]
]

combined = pd.concat(
    [train_df, val_df, test_df],
    ignore_index=True
)

# Group identical feature vectors and inspect their labels
duplicate_groups = (
    combined.groupby(feature_cols, dropna=False)["Label"]
    .agg(["count", "nunique", list])
    .reset_index()
)

duplicate_groups = duplicate_groups[
    duplicate_groups["count"] > 1
]

conflicting_groups = duplicate_groups[
    duplicate_groups["nunique"] > 1
]

print("=" * 60)
print("DUPLICATE LABEL INVESTIGATION")
print("=" * 60)

print("Total repeated feature groups:", len(duplicate_groups))
print("Groups with conflicting labels:", len(conflicting_groups))

print("\nRepeated feature groups with consistent labels:",
      len(duplicate_groups) - len(conflicting_groups))

if len(conflicting_groups) > 0:
    print("\nExamples of conflicting groups:")
    print(conflicting_groups[["count", "nunique", "list"]].head(10))

print("\nLabel distribution of conflicting records:")

if len(conflicting_groups) > 0:
    print(
        combined.merge(
            conflicting_groups[feature_cols],
            on=feature_cols,
            how="inner"
        )["Label"].value_counts()
    )

print("\nINVESTIGATION COMPLETED.")
```

```python
from pathlib import Path
import pandas as pd

BASE_DIR = Path(r"C:\Users\HP\DDoS_IoT_Project")
PROCESSED_DIR = BASE_DIR / "processed"
SOURCE_DIR = BASE_DIR / "Processed_Network_dataset"

print("=" * 65)
print("PROCESSED GROUP TO ATTACK CATEGORY MAPPING")
print("=" * 65)

for i in range(1, 12):

    group = f"file_{i:02d}"

    files = sorted(PROCESSED_DIR.glob(f"{group}_chunk_*.parquet"))

    labels = set()
    sources = set()
    total_rows = 0

    for file in files:
        df = pd.read_parquet(
            file,
            columns=["Label", "Source_File"]
        )

        total_rows += len(df)
        labels.update(df["Label"].dropna().unique())
        sources.update(df["Source_File"].dropna().unique())

    print(f"\n{group}")
    print("Chunks:", len(files))
    print("Total rows:", total_rows)
    print("Labels:", sorted(labels))
    print("Source metadata:", sorted(sources))


print("\n" + "=" * 65)
print("ORIGINAL CSV LABEL INFORMATION")
print("=" * 65)

for i in range(1, 24):

    file = SOURCE_DIR / f"Network_dataset_{i}.csv"

    # Inspect a sample, not the entire large CSV
    sample = pd.read_csv(file, nrows=1000, low_memory=False)

    print(f"\n{file.name}")
    print("Columns:", sample.columns.tolist())
    print("Possible label columns:",
          [c for c in sample.columns
           if any(x in c.lower() for x in ["label", "attack", "class"])])
    
    possible_labels = [
        c for c in sample.columns
        if any(x in c.lower() for x in ["label", "attack", "class"])
    ]

    for col in possible_labels:
        print(sample[col].value_counts().head(10).to_dict())

print("\nMAPPING VERIFICATION COMPLETED.")
```

```python
import joblib
import pandas as pd
import numpy as np
from pathlib import Path

BASE_DIR = Path(r"C:\Users\HP\DDoS_IoT_Project")
MODELS_DIR = BASE_DIR / "models"
PROCESSED_DIR = BASE_DIR / "processed"

# Load artifacts
calibrated_if = joblib.load(
    MODELS_DIR / "calibrated_isolation_forest.joblib"
)

original_if = joblib.load(
    MODELS_DIR / "zero_day_isolation_forest.joblib"
)

imputer = joblib.load(
    MODELS_DIR / "calibrated_isolation_forest_imputer.joblib"
)

individual_models = joblib.load(
    MODELS_DIR / "cic_individual_models.joblib"
)

# Load preprocessing for one attack
dns_preprocessing = joblib.load(
    MODELS_DIR / "DrDoS_DNS_preprocessing.joblib"
)

# Load small samples
benign = pd.read_parquet(
    PROCESSED_DIR / "zero_day_samples" / "benign_test.parquet"
).sample(200, random_state=42)

snmp = pd.read_parquet(
    PROCESSED_DIR / "zero_day_samples" / "snmp_zero_day.parquet"
).sample(200, random_state=42)

feature_cols = list(imputer.feature_names_in_)

X_benign = benign[feature_cols].apply(
    pd.to_numeric, errors="coerce"
).replace([np.inf, -np.inf], np.nan)

X_snmp = snmp[feature_cols].apply(
    pd.to_numeric, errors="coerce"
).replace([np.inf, -np.inf], np.nan)

X_benign_imp = imputer.transform(X_benign)
X_snmp_imp = imputer.transform(X_snmp)

print("=" * 65)
print("1. PREPROCESSING PIPELINE STRUCTURE")
print("=" * 65)

print("DNS preprocessing keys:", dns_preprocessing.keys())

print("\nDNS selector feature count:",
      dns_preprocessing["selector"].get_support().sum())

print("DNS scaler feature count:",
      dns_preprocessing["scaler"].n_features_in_)

print("\n" + "=" * 65)
print("2. ISOLATION FOREST COMPARISON")
print("=" * 65)

for name, model in [
    ("Calibrated Isolation Forest", calibrated_if),
    ("Original Isolation Forest", original_if)
]:

    benign_scores = -model.decision_function(X_benign_imp)
    snmp_scores = -model.decision_function(X_snmp_imp)

    print(f"\n{name}")
    print("Benign anomaly score range:",
          benign_scores.min(), benign_scores.max())
    print("SNMP anomaly score range:",
          snmp_scores.min(), snmp_scores.max())

    print("Benign mean score:", benign_scores.mean())
    print("SNMP mean score:", snmp_scores.mean())

    print("Benign scores above 0.011813:",
          np.mean(benign_scores > 0.011813))

    print("SNMP scores above 0.011813:",
          np.mean(snmp_scores > 0.011813))

print("\n" + "=" * 65)
print("3. SUPERVISED RANDOM FOREST")
print("=" * 65)

rf = joblib.load(
    MODELS_DIR / "zero_day_supervised_random_forest.joblib"
)

print("Feature names available:",
      hasattr(rf, "feature_names_in_"))

if hasattr(rf, "feature_names_in_"):
    print("First 10 expected features:",
          list(rf.feature_names_in_[:10]))

print("\n" + "=" * 65)
print("4. COMPLETED")
print("=" * 65)
```

```python
import joblib
import pandas as pd
import numpy as np
from pathlib import Path

BASE_DIR = Path(r"C:\Users\HP\DDoS_IoT_Project")
MODELS_DIR = BASE_DIR / "models"
PROCESSED_DIR = BASE_DIR / "processed"

# Load models
rf = joblib.load(
    MODELS_DIR / "zero_day_supervised_random_forest.joblib"
)

imputer = joblib.load(
    MODELS_DIR / "calibrated_isolation_forest_imputer.joblib"
)

# Load evaluation samples
benign = pd.read_parquet(
    PROCESSED_DIR / "zero_day_samples" / "benign_test.parquet"
).sample(500, random_state=42)

snmp = pd.read_parquet(
    PROCESSED_DIR / "zero_day_samples" / "snmp_zero_day.parquet"
).sample(500, random_state=42)

feature_cols = list(imputer.feature_names_in_)

def prepare_data(df):
    X = df[feature_cols].apply(pd.to_numeric, errors="coerce")
    X = X.replace([np.inf, -np.inf], np.nan)
    return imputer.transform(X)

X_benign = prepare_data(benign)
X_snmp = prepare_data(snmp)

# Obtain supervised predictions
benign_probs = rf.predict_proba(X_benign)[:, 1]
snmp_probs = rf.predict_proba(X_snmp)[:, 1]

print("=" * 65)
print("SUPERVISED RANDOM FOREST RESPONSE")
print("=" * 65)

for name, probs in [
    ("Benign", benign_probs),
    ("Held-out SNMP", snmp_probs)
]:
    print(f"\n{name}")
    print("Mean attack probability:", round(probs.mean(), 6))
    print("Minimum:", round(probs.min(), 6))
    print("Maximum:", round(probs.max(), 6))
    print("Predicted attack rate:", round(np.mean(probs >= 0.5), 4))

    print("Probability distribution:")
    print(pd.Series(probs).describe())

print("\nSUPERVISED BRANCH TEST COMPLETED.")
```

```python
import joblib
import pandas as pd
import numpy as np
from pathlib import Path
from sklearn.metrics import confusion_matrix

BASE_DIR = Path(r"C:\Users\HP\DDoS_IoT_Project")
MODELS_DIR = BASE_DIR / "models"
PROCESSED_DIR = BASE_DIR / "processed"

rf = joblib.load(
    MODELS_DIR / "zero_day_supervised_random_forest.joblib"
)

if_model = joblib.load(
    MODELS_DIR / "calibrated_isolation_forest.joblib"
)

imputer = joblib.load(
    MODELS_DIR / "calibrated_isolation_forest_imputer.joblib"
)

benign = pd.read_parquet(
    PROCESSED_DIR / "zero_day_samples" / "benign_test.parquet"
).sample(500, random_state=42)

snmp = pd.read_parquet(
    PROCESSED_DIR / "zero_day_samples" / "snmp_zero_day.parquet"
).sample(500, random_state=42)

data = pd.concat([benign, snmp], ignore_index=True)

y_true = np.concatenate([
    np.zeros(len(benign), dtype=int),
    np.ones(len(snmp), dtype=int)
])

feature_cols = list(imputer.feature_names_in_)

X = data[feature_cols].apply(
    pd.to_numeric, errors="coerce"
).replace([np.inf, -np.inf], np.nan)

X = imputer.transform(X)

# Supervised predictions
rf_probs = rf.predict_proba(X)[:, 1]
rf_pred = (rf_probs >= 0.5).astype(int)

# Anomaly predictions
anomaly_scores = -if_model.decision_function(X)
if_pred = (anomaly_scores > 0.011813).astype(int)

# Agreement categories
agreement = pd.DataFrame({
    "Actual": y_true,
    "RF_Prediction": rf_pred,
    "IF_Prediction": if_pred
})

agreement["Agreement"] = np.select(
    [
        (rf_pred == 0) & (if_pred == 0),
        (rf_pred == 1) & (if_pred == 1),
        (rf_pred == 1) & (if_pred == 0),
        (rf_pred == 0) & (if_pred == 1)
    ],
    [
        "Both classify as benign",
        "Both classify as attack",
        "RF only detects attack",
        "IF only detects attack"
    ]
)

print("=" * 65)
print("MODEL AGREEMENT ANALYSIS")
print("=" * 65)

print("\nOverall agreement:")
print(agreement["Agreement"].value_counts())

print("\nAgreement by actual class:")
print(pd.crosstab(
    agreement["Actual"],
    agreement["Agreement"]
))

print("\n" + "=" * 65)
print("SUPERVISED RANDOM FOREST CONFUSION MATRIX")
print("=" * 65)

print(confusion_matrix(y_true, rf_pred))

print("\n" + "=" * 65)
print("ISOLATION FOREST CONFUSION MATRIX")
print("=" * 65)

print(confusion_matrix(y_true, if_pred))

print("\nAGREEMENT ANALYSIS COMPLETED.")
```

```python
# Agreement categories
agreement = pd.DataFrame({
    "Actual": y_true,
    "RF_Prediction": rf_pred,
    "IF_Prediction": if_pred
})

agreement["Agreement"] = np.select(
    [
        (rf_pred == 0) & (if_pred == 0),
        (rf_pred == 1) & (if_pred == 1),
        (rf_pred == 1) & (if_pred == 0),
        (rf_pred == 0) & (if_pred == 1)
    ],
    [
        "Both classify as benign",
        "Both classify as attack",
        "RF only detects attack",
        "IF only detects attack"
    ],
    default="Unknown"
)

print("=" * 65)
print("MODEL AGREEMENT ANALYSIS")
print("=" * 65)

print("\nOverall agreement:")
print(agreement["Agreement"].value_counts())

print("\nAgreement by actual class:")
print(pd.crosstab(
    agreement["Actual"],
    agreement["Agreement"]
))

print("\n" + "=" * 65)
print("SUPERVISED RANDOM FOREST CONFUSION MATRIX")
print("=" * 65)

print(confusion_matrix(y_true, rf_pred))

print("\n" + "=" * 65)
print("ISOLATION FOREST CONFUSION MATRIX")
print("=" * 65)

print(confusion_matrix(y_true, if_pred))

print("\nAGREEMENT ANALYSIS COMPLETED.")
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score
from sklearn.metrics import f1_score, confusion_matrix

# Three decision mechanisms
rf_only = rf_pred

or_hybrid = np.logical_or(
    rf_pred == 1,
    if_pred == 1
).astype(int)

and_hybrid = np.logical_and(
    rf_pred == 1,
    if_pred == 1
).astype(int)


def evaluate_model(name, predictions):
    tn, fp, fn, tp = confusion_matrix(
        y_true, predictions
    ).ravel()

    return {
        "Model": name,
        "Accuracy": accuracy_score(y_true, predictions),
        "Precision": precision_score(y_true, predictions),
        "Recall": recall_score(y_true, predictions),
        "F1 Score": f1_score(y_true, predictions),
        "False Positive Rate": fp / (fp + tn),
        "False Negatives": fn,
        "False Positives": fp
    }


comparison = pd.DataFrame([
    evaluate_model("Random Forest", rf_only),
    evaluate_model("OR Hybrid", or_hybrid),
    evaluate_model("AND Hybrid", and_hybrid)
])

print("=" * 75)
print("HYBRID MODEL COMPARISON")
print("=" * 75)

print(comparison.to_string(index=False))

print("\nConfusion matrices:")

for name, predictions in [
    ("Random Forest", rf_only),
    ("OR Hybrid", or_hybrid),
    ("AND Hybrid", and_hybrid)
]:
    print(f"\n{name}")
    print(confusion_matrix(y_true, predictions))
```

```python
import numpy as np
import pandas as pd
import joblib

from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.metrics import confusion_matrix, classification_report

PROJECT = Path(r"C:\Users\HP\DDoS_IoT_Project")
DATA_DIR = PROJECT / "processed" / "zero_day_samples"
MODEL_DIR = PROJECT / "models"

# Load original datasets
known_attacks = pd.read_parquet(
    DATA_DIR / "known_attacks.parquet"
)

benign_train = pd.read_parquet(
    DATA_DIR / "benign_train.parquet"
)

benign_test = pd.read_parquet(
    DATA_DIR / "benign_test.parquet"
)

snmp_test = pd.read_parquet(
    DATA_DIR / "snmp_zero_day.parquet"
)

# Recover the original feature columns
excluded = {"Label", "Source_File", "Attack_Type"}

feature_columns = [
    col for col in known_attacks.columns
    if col not in excluded
]

# Reconstruct supervised training dataset
known_attacks = known_attacks.copy()
known_attacks["Target"] = 1

benign_supervised = benign_train.copy()
benign_supervised["Target"] = 0

supervised_data = pd.concat(
    [known_attacks, benign_supervised],
    ignore_index=True
)

# Reproduce the original split
train_data, val_data = train_test_split(
    supervised_data,
    test_size=0.20,
    random_state=42,
    stratify=supervised_data["Target"]
)

# Fit the original supervised imputation process
X_train = train_data[feature_columns].copy()
X_train = X_train.replace([np.inf, -np.inf], np.nan)

supervised_imputer = SimpleImputer(strategy="median")
supervised_imputer.fit(X_train)

# Prepare held-out evaluation data
test_data = pd.concat(
    [benign_test, snmp_test],
    ignore_index=True
)

X_test = test_data[feature_columns].copy()
X_test = X_test.replace([np.inf, -np.inf], np.nan)

X_test_imputed = supervised_imputer.transform(X_test)

y_true_correct = np.concatenate([
    np.zeros(len(benign_test), dtype=int),
    np.ones(len(snmp_test), dtype=int)
])

# Load the existing trained Random Forest
rf = joblib.load(
    MODEL_DIR / "zero_day_supervised_random_forest.joblib"
)

# Evaluate using correct preprocessing
rf_probs_correct = rf.predict_proba(X_test_imputed)[:, 1]
rf_pred_correct = (rf_probs_correct >= 0.5).astype(int)

print("=" * 70)
print("CORRECTED RANDOM FOREST ZERO-DAY EVALUATION")
print("=" * 70)

print("Training samples:", len(train_data))
print("Training features:", len(feature_columns))
print("Benign test samples:", len(benign_test))
print("SNMP test samples:", len(snmp_test))

print("\nConfusion Matrix:")
print(confusion_matrix(y_true_correct, rf_pred_correct))

print("\nClassification Report:")
print(classification_report(
    y_true_correct,
    rf_pred_correct,
    target_names=["Benign", "SNMP Zero-Day"],
    zero_division=0
))

print("\nMean attack probability:")
print("Benign:", rf_probs_correct[:len(benign_test)].mean())
print("SNMP:", rf_probs_correct[len(benign_test):].mean())

print("\nCorrected evaluation completed.")
```

```python
import joblib
import numpy as np
import pandas as pd

from pathlib import Path
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

PROJECT = Path(r"C:\Users\HP\DDoS_IoT_Project")
MODEL_DIR = PROJECT / "models"
DATA_DIR = PROJECT / "processed" / "zero_day_samples"

# Load Isolation Forest and its original imputer
if_model = joblib.load(
    MODEL_DIR / "calibrated_isolation_forest.joblib"
)

if_imputer = joblib.load(
    MODEL_DIR / "calibrated_isolation_forest_imputer.joblib"
)

# Prepare data using Isolation Forest's own imputer
feature_cols = list(if_imputer.feature_names_in_)

X_if = test_data[feature_cols].copy()

X_if = X_if.apply(
    pd.to_numeric,
    errors="coerce"
).replace([np.inf, -np.inf], np.nan)

X_if_imputed = if_imputer.transform(X_if)

# Isolation Forest anomaly scores
if_scores_correct = -if_model.decision_function(X_if_imputed)

# Previously calibrated threshold
IF_THRESHOLD = 0.011813

if_pred_correct = (
    if_scores_correct > IF_THRESHOLD
).astype(int)


def calculate_metrics(name, y_true, predictions):

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        predictions,
        labels=[0, 1]
    ).ravel()

    return {
        "Model": name,
        "Accuracy": accuracy_score(y_true, predictions),
        "Precision": precision_score(
            y_true, predictions, zero_division=0
        ),
        "Recall": recall_score(
            y_true, predictions, zero_division=0
        ),
        "F1 Score": f1_score(
            y_true, predictions, zero_division=0
        ),
        "FPR": fp / (fp + tn),
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp
    }


comparison = pd.DataFrame([
    calculate_metrics(
        "Random Forest",
        y_true_correct,
        rf_pred_correct
    ),
    calculate_metrics(
        "Isolation Forest",
        y_true_correct,
        if_pred_correct
    )
])

print("=" * 75)
print("FULL ZERO-DAY TEST SET: MODEL COMPARISON")
print("=" * 75)

print(comparison.to_string(index=False))

print("\nIsolation Forest confusion matrix:")
print(confusion_matrix(y_true_correct, if_pred_correct))

print("\nEvaluation samples:", len(y_true_correct))
print("Random Forest preprocessing: Supervised imputer")
print("Isolation Forest preprocessing: Anomaly imputer")
print("Isolation Forest threshold:", IF_THRESHOLD)
```

```python
import numpy as np
import pandas as pd

# Construct agreement dataframe
agreement_full = pd.DataFrame({
    "Actual": y_true_correct,
    "RF_Prediction": rf_pred_correct,
    "IF_Prediction": if_pred_correct
})

agreement_full["Agreement"] = np.select(
    [
        (rf_pred_correct == 0) & (if_pred_correct == 0),
        (rf_pred_correct == 1) & (if_pred_correct == 1),
        (rf_pred_correct == 1) & (if_pred_correct == 0),
        (rf_pred_correct == 0) & (if_pred_correct == 1)
    ],
    [
        "Both predict benign",
        "Both predict attack",
        "RF only predicts attack",
        "IF only predicts attack"
    ],
    default="Unknown"
)

print("=" * 75)
print("FULL ZERO-DAY MODEL AGREEMENT ANALYSIS")
print("=" * 75)

print("\nOverall agreement:")
print(agreement_full["Agreement"].value_counts())

print("\nAgreement by actual class:")
print(pd.crosstab(
    agreement_full["Actual"],
    agreement_full["Agreement"]
))

# Specifically inspect attacks missed by RF
rf_missed = agreement_full[
    (agreement_full["Actual"] == 1) &
    (agreement_full["RF_Prediction"] == 0)
]

print("\n" + "=" * 75)
print("RANDOM FOREST MISSED ATTACKS")
print("=" * 75)

print("Total RF missed attacks:", len(rf_missed))
print(
    "Recovered by Isolation Forest:",
    (rf_missed["IF_Prediction"] == 1).sum()
)

# Specifically inspect false positives introduced by IF
if_false_alarms = agreement_full[
    (agreement_full["Actual"] == 0) &
    (agreement_full["RF_Prediction"] == 0) &
    (agreement_full["IF_Prediction"] == 1)
]

print("\nAdditional benign false alarms from IF:", len(if_false_alarms))

print("\nAnalysis completed.")
```

```python
import numpy as np
import pandas as pd

# Separate benign and SNMP probabilities
benign_probs = rf_probs_correct[:len(benign_test)]
snmp_probs = rf_probs_correct[len(benign_test):]

# Confidence = distance from the decision boundary
rf_confidence = np.abs(rf_probs_correct - 0.5)

# Identify uncertain predictions
uncertain_mask = rf_confidence < 0.10

print("=" * 75)
print("RANDOM FOREST CONFIDENCE ANALYSIS")
print("=" * 75)

print("\nBenign probability distribution:")
print(pd.Series(benign_probs).describe())

print("\nSNMP probability distribution:")
print(pd.Series(snmp_probs).describe())

print("\nConfidence distribution:")
print(pd.Series(rf_confidence).describe())

print("\nUncertain predictions (probability between 0.40 and 0.60):")
print("Total:", uncertain_mask.sum())

print("\nUncertain benign samples:")
print(
    np.sum(
        uncertain_mask[:len(benign_test)]
    )
)

print("\nUncertain SNMP samples:")
print(
    np.sum(
        uncertain_mask[len(benign_test):]
    )
)

print("\nActual RF errors:")
print("False positives:", np.sum(
    rf_pred_correct[:len(benign_test)] == 1
))

print("False negatives:", np.sum(
    rf_pred_correct[len(benign_test):] == 0
))
```

```python
import numpy as np
import pandas as pd

# Identify classification errors
error_mask = (rf_pred_correct != y_true_correct)

# Identify uncertain predictions
uncertain_mask = (
    (rf_probs_correct >= 0.40) &
    (rf_probs_correct <= 0.60)
)

# Compare the two groups
print("=" * 75)
print("UNCERTAINTY VS CLASSIFICATION ERRORS")
print("=" * 75)

print("\nTotal classification errors:", error_mask.sum())
print("Total uncertain predictions:", uncertain_mask.sum())

print(
    "Errors inside uncertain group:",
    np.sum(error_mask & uncertain_mask)
)

print(
    "Errors outside uncertain group:",
    np.sum(error_mask & ~uncertain_mask)
)

print(
    "Correct predictions inside uncertain group:",
    np.sum(~error_mask & uncertain_mask)
)

# Detailed error analysis
error_analysis = pd.DataFrame({
    "Actual": y_true_correct,
    "RF_Probability": rf_probs_correct,
    "RF_Prediction": rf_pred_correct,
    "IF_Anomaly_Score": if_scores_correct,
    "Is_Uncertain": uncertain_mask,
    "Is_Error": error_mask
})

print("\nClassification errors:")
print(
    error_analysis[error_analysis["Is_Error"]].to_string(index=False)
)

print("\nUncertain predictions:")
print(
    error_analysis[error_analysis["Is_Uncertain"]].to_string(index=False)
)
```

```python
import pandas as pd
import numpy as np

PROJECT = Path(r"C:\Users\HP\DDoS_IoT_Project")
DATA_DIR = PROJECT / "processed" / "zero_day_samples"

# Top five important features
top_features = rf_importance["Feature"].head(5).tolist()

# Load datasets
datasets = {
    "Benign": pd.read_parquet(DATA_DIR / "benign_train.parquet"),
    "DNS": pd.read_parquet(DATA_DIR / "known_attacks.parquet"),
    "SNMP": pd.read_parquet(DATA_DIR / "snmp_zero_day.parquet")
}

# Extract attack categories from known_attacks
known = datasets["DNS"]

print("=" * 75)
print("TOP FIVE FEATURE DISTRIBUTIONS")
print("=" * 75)

for feature in top_features:

    print(f"\n\nFEATURE: {feature}")
    print("-" * 75)

    summary = []

    # Benign
    values = pd.to_numeric(
        datasets["Benign"][feature],
        errors="coerce"
    ).replace([np.inf, -np.inf], np.nan).dropna()

    summary.append({
        "Category": "Benign",
        "Mean": values.mean(),
        "Median": values.median(),
        "Std": values.std()
    })

    # Known attack categories
    for category in ["DNS", "LDAP", "NetBIOS", "UDP"]:

        category_data = known[
            known["Source_File"].str.contains(
                category,
                case=False,
                na=False
            )
        ]

        values = pd.to_numeric(
            category_data[feature],
            errors="coerce"
        ).replace([np.inf, -np.inf], np.nan).dropna()

        summary.append({
            "Category": category,
            "Mean": values.mean(),
            "Median": values.median(),
            "Std": values.std()
        })

    # SNMP
    values = pd.to_numeric(
        datasets["SNMP"][feature],
        errors="coerce"
    ).replace([np.inf, -np.inf], np.nan).dropna()

    summary.append({
        "Category": "SNMP",
        "Mean": values.mean(),
        "Median": values.median(),
        "Std": values.std()
    })

    print(pd.DataFrame(summary).round(4).to_string(index=False))
```

```python
import pandas as pd
import numpy as np
import joblib

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

# Load original training datasets
benign_train = pd.read_parquet(
    DATA_DIR / "benign_train.parquet"
)

known_attacks = pd.read_parquet(
    DATA_DIR / "known_attacks.parquet"
)

benign_test = pd.read_parquet(
    DATA_DIR / "benign_test.parquet"
)

snmp_test = pd.read_parquet(
    DATA_DIR / "snmp_zero_day.parquet"
)

# Reconstruct supervised training dataset
benign_train["Target"] = 0
known_attacks["Target"] = 1

supervised_data = pd.concat(
    [known_attacks, benign_train],
    ignore_index=True
)

feature_columns = [
    col for col in supervised_data.columns
    if col not in ["Label", "Source_File", "Attack_Type", "Target"]
]

X = supervised_data[feature_columns].replace(
    [np.inf, -np.inf], np.nan
)

y = supervised_data["Target"]

X_train, X_val, y_train, y_val = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Remove five most important features
top_5_features = rf_importance["Feature"].head(5).tolist()

reduced_features = [
    col for col in feature_columns
    if col not in top_5_features
]

print("Original features:", len(feature_columns))
print("Removed features:", top_5_features)
print("Remaining features:", len(reduced_features))

# Fit imputer only on training data
ablation_imputer = SimpleImputer(strategy="median")

X_train_reduced = ablation_imputer.fit_transform(
    X_train[reduced_features]
)

# Train Random Forest with the same hyperparameters
ablation_rf = RandomForestClassifier(
    n_estimators=100,
    max_depth=20,
    min_samples_leaf=2,
    class_weight="balanced",
    random_state=42,
    n_jobs=1
)

ablation_rf.fit(X_train_reduced, y_train)

# Prepare held-out test data
test_data = pd.concat(
    [benign_test, snmp_test],
    ignore_index=True
)

y_ablation_true = np.concatenate([
    np.zeros(len(benign_test), dtype=int),
    np.ones(len(snmp_test), dtype=int)
])

X_ablation_test = test_data[reduced_features].replace(
    [np.inf, -np.inf], np.nan
)

X_ablation_test = ablation_imputer.transform(
    X_ablation_test
)

# Predict
ablation_predictions = ablation_rf.predict(
    X_ablation_test
)

# Evaluate
print("\n" + "=" * 75)
print("RANDOM FOREST WITHOUT TOP FIVE FEATURES")
print("=" * 75)

print("\nConfusion Matrix:")
print(confusion_matrix(
    y_ablation_true,
    ablation_predictions
))

print("\nPerformance:")

print("Accuracy:",
      accuracy_score(y_ablation_true, ablation_predictions))

print("Precision:",
      precision_score(y_ablation_true, ablation_predictions))

print("Recall:",
      recall_score(y_ablation_true, ablation_predictions))

print("F1-score:",
      f1_score(y_ablation_true, ablation_predictions))

print("\nFeature ablation experiment completed.")
```

```python
# Identify packet-size-related features
packet_features = [
    col for col in feature_columns
    if any(term in col.lower() for term in [
        "packet length",
        "packet size",
        "segment size",
        "total length of fwd packets",
        "total length of bwd packets",
        "subflow fwd bytes",
        "subflow bwd bytes"
    ])
]

remaining_features = [
    col for col in feature_columns
    if col not in packet_features
]

print("Original features:", len(feature_columns))
print("Packet-size features removed:", len(packet_features))
print("Remaining features:", len(remaining_features))

print("\nRemoved features:")
print(packet_features)

# Fit a separate imputer on supervised training data
flow_imputer = SimpleImputer(strategy="median")

X_train_flow = flow_imputer.fit_transform(
    X_train[remaining_features]
)

# Train a new Random Forest
flow_rf = RandomForestClassifier(
    n_estimators=100,
    max_depth=20,
    min_samples_leaf=2,
    class_weight="balanced",
    random_state=42,
    n_jobs=1
)

flow_rf.fit(X_train_flow, y_train)

# Prepare test data
X_test_flow = flow_imputer.transform(
    test_data[remaining_features]
)

flow_predictions = flow_rf.predict(X_test_flow)

# Evaluate
print("\n" + "=" * 75)
print("RANDOM FOREST WITHOUT PACKET-SIZE FEATURES")
print("=" * 75)

print("\nConfusion Matrix:")
print(confusion_matrix(y_ablation_true, flow_predictions))

print("\nAccuracy:",
      accuracy_score(y_ablation_true, flow_predictions))

print("Precision:",
      precision_score(y_ablation_true, flow_predictions))

print("Recall:",
      recall_score(y_ablation_true, flow_predictions))

print("F1-score:",
      f1_score(y_ablation_true, flow_predictions))

print("\nPacket-size ablation completed.")
```

```python
import pandas as pd

flow_importance = pd.DataFrame({
    "Feature": remaining_features,
    "Importance": flow_rf.feature_importances_
})

flow_importance = flow_importance.sort_values(
    by="Importance",
    ascending=False
).reset_index(drop=True)

print("=" * 75)
print("RANDOM FOREST: NON-PACKET FEATURE IMPORTANCE")
print("=" * 75)

print(flow_importance.head(20).to_string(index=False))

print("\nCumulative importance:")

flow_importance["Cumulative"] = (
    flow_importance["Importance"].cumsum()
)

for n in [5, 10, 15, 20, 30, 40]:
    value = flow_importance.head(n)["Importance"].sum()

    print(
        f"Top {n:2d} features: "
        f"{value:.4f} ({value * 100:.2f}%)"
    )

print("\nTotal features:", len(remaining_features))
print(
    "Non-zero importance features:",
    (flow_importance["Importance"] > 0).sum()
)
```

```python
# Remove protocol and port-related features
network_identifier_features = [
    col for col in remaining_features
    if col.lower() in [
        "protocol",
        "source port",
        "destination port"
    ]
]

identifier_free_features = [
    col for col in remaining_features
    if col not in network_identifier_features
]

print("Original non-packet features:", len(remaining_features))
print("Removed features:", network_identifier_features)
print("Remaining features:", len(identifier_free_features))

# Fit a fresh imputer
identifier_imputer = SimpleImputer(strategy="median")

X_train_identifier = identifier_imputer.fit_transform(
    X_train[identifier_free_features]
)

# Train Random Forest
identifier_rf = RandomForestClassifier(
    n_estimators=100,
    max_depth=20,
    min_samples_leaf=2,
    class_weight="balanced",
    random_state=42,
    n_jobs=1
)

identifier_rf.fit(X_train_identifier, y_train)

# Transform test data
X_test_identifier = identifier_imputer.transform(
    test_data[identifier_free_features]
)

# Predictions
identifier_predictions = identifier_rf.predict(
    X_test_identifier
)

# Evaluation
print("\n" + "=" * 75)
print("RANDOM FOREST WITHOUT PROTOCOL AND PORT FEATURES")
print("=" * 75)

print("\nConfusion Matrix:")
print(confusion_matrix(
    y_ablation_true,
    identifier_predictions
))

print("\nAccuracy:",
      accuracy_score(y_ablation_true, identifier_predictions))

print("Precision:",
      precision_score(y_ablation_true, identifier_predictions))

print("Recall:",
      recall_score(y_ablation_true, identifier_predictions))

print("F1-score:",
      f1_score(y_ablation_true, identifier_predictions))

print("\nIdentifier-feature ablation completed.")
```

```python
attack_categories = [
    "DrDoS_DNS",
    "DrDoS_LDAP",
    "DrDoS_NetBIOS",
    "DrDoS_UDP"
]

leave_one_out_folds = {}

for held_out in attack_categories:

    training_categories = [
        category for category in attack_categories
        if category != held_out
    ]

    # Training attacks: three known categories
    fold_attack_train = known_attacks[
        known_attacks["Source_File"].isin(training_categories)
    ].copy()

    # Training benign: three corresponding source categories
    fold_benign_train = benign_train[
        benign_train["Source_File"].isin(training_categories)
    ].copy()

    # Testing attacks: completely held-out category
    fold_attack_test = known_attacks[
        known_attacks["Source_File"] == held_out
    ].copy()

    # Testing benign: held-out source category
    fold_benign_test = benign_train[
        benign_train["Source_File"] == held_out
    ].copy()

    leave_one_out_folds[held_out] = {
        "training_categories": training_categories,
        "attack_train": fold_attack_train,
        "benign_train": fold_benign_train,
        "attack_test": fold_attack_test,
        "benign_test": fold_benign_test
    }

    print("\n" + "=" * 65)
    print("HELD-OUT CATEGORY:", held_out)
    print("=" * 65)

    print("Training attack samples:", len(fold_attack_train))
    print("Training benign samples:", len(fold_benign_train))
    print("Testing attack samples:", len(fold_attack_test))
    print("Testing benign samples:", len(fold_benign_test))

    # Verify no source-category overlap
    train_sources = set(
        fold_attack_train["Source_File"].unique()
    ) | set(
        fold_benign_train["Source_File"].unique()
    )

    test_sources = set(
        fold_attack_test["Source_File"].unique()
    ) | set(
        fold_benign_test["Source_File"].unique()
    )

    print("Training sources:", sorted(train_sources))
    print("Testing sources:", sorted(test_sources))
    print("Source overlap:", train_sources.intersection(test_sources))

print("\nAll four experimental folds constructed.")
```

```python
import numpy as np
import pandas as pd
import time

from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    balanced_accuracy_score,
    roc_auc_score
)

loco_results = []
loco_models = {}

feature_columns = [
    col for col in known_attacks.columns
    if col not in ["Label", "Source_File", "Attack_Type", "Target"]
]

for held_out, fold in leave_one_out_folds.items():

    print("\n" + "=" * 75)
    print("RUNNING LOCO EXPERIMENT:", held_out)
    print("=" * 75)

    start_time = time.time()

    # --------------------------------------------------
    # 1. Construct training dataset
    # --------------------------------------------------

    attack_train = fold["attack_train"].copy()
    benign_train_fold = fold["benign_train"].copy()

    attack_train["Target"] = 1
    benign_train_fold["Target"] = 0

    train_data = pd.concat(
        [attack_train, benign_train_fold],
        ignore_index=True
    )

    X_train_fold = train_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    )

    y_train_fold = train_data["Target"]

    # --------------------------------------------------
    # 2. Construct held-out test dataset
    # --------------------------------------------------

    attack_test = fold["attack_test"].copy()
    benign_test_fold = fold["benign_test"].copy()

    attack_test["Target"] = 1
    benign_test_fold["Target"] = 0

    test_fold = pd.concat(
        [benign_test_fold, attack_test],
        ignore_index=True
    )

    X_test_fold = test_fold[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    )

    y_test_fold = test_fold["Target"]

    # --------------------------------------------------
    # 3. Fit preprocessing on training data only
    # --------------------------------------------------

    fold_imputer = SimpleImputer(strategy="median")

    X_train_imputed = fold_imputer.fit_transform(
        X_train_fold
    )

    X_test_imputed = fold_imputer.transform(
        X_test_fold
    )

    # --------------------------------------------------
    # 4. Train Random Forest
    # --------------------------------------------------

    fold_rf = RandomForestClassifier(
        n_estimators=100,
        max_depth=20,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=1
    )

    fold_rf.fit(
        X_train_imputed,
        y_train_fold
    )

    # --------------------------------------------------
    # 5. Predictions
    # --------------------------------------------------

    predictions = fold_rf.predict(X_test_imputed)

    probabilities = fold_rf.predict_proba(
        X_test_imputed
    )[:, 1]

    tn, fp, fn, tp = confusion_matrix(
        y_test_fold,
        predictions,
        labels=[0, 1]
    ).ravel()

    # --------------------------------------------------
    # 6. Evaluation metrics
    # --------------------------------------------------

    metrics = {
        "Held_Out_Category": held_out,
        "Training_Samples": len(train_data),
        "Testing_Samples": len(test_fold),
        "Accuracy": accuracy_score(y_test_fold, predictions),
        "Precision": precision_score(
            y_test_fold, predictions, zero_division=0
        ),
        "Recall": recall_score(
            y_test_fold, predictions, zero_division=0
        ),
        "F1_Score": f1_score(
            y_test_fold, predictions, zero_division=0
        ),
        "Balanced_Accuracy": balanced_accuracy_score(
            y_test_fold, predictions
        ),
        "ROC_AUC": roc_auc_score(
            y_test_fold, probabilities
        ),
        "FPR": fp / (fp + tn),
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp
    }

    loco_results.append(metrics)

    loco_models[held_out] = {
        "model": fold_rf,
        "imputer": fold_imputer,
        "features": feature_columns
    }

    print("\nConfusion Matrix:")
    print(confusion_matrix(y_test_fold, predictions))

    print("\nPerformance:")
    for metric, value in metrics.items():
        if isinstance(value, (float, np.floating)):
            print(f"{metric}: {value:.6f}")

    print(f"\nExecution time: {time.time() - start_time:.2f} seconds")

# ------------------------------------------------------
# 7. Overall comparison
# ------------------------------------------------------

loco_results_df = pd.DataFrame(loco_results)

print("\n" + "=" * 85)
print("FINAL LEAVE-ONE-ATTACK-CATEGORY-OUT RESULTS")
print("=" * 85)

print(
    loco_results_df[
        [
            "Held_Out_Category",
            "Accuracy",
            "Precision",
            "Recall",
            "F1_Score",
            "Balanced_Accuracy",
            "ROC_AUC",
            "FPR",
            "FN",
            "FP"
        ]
    ].round(5).to_string(index=False)
)

print("\nMean performance across four folds:")

print(
    loco_results_df[
        [
            "Accuracy",
            "Precision",
            "Recall",
            "F1_Score",
            "Balanced_Accuracy",
            "ROC_AUC",
            "FPR"
        ]
    ].mean().round(5)
)

print("\nAll four LOCO experiments completed.")
```

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, roc_auc_score

# Select DNS fold
fold = leave_one_out_folds["DrDoS_DNS"]

# Construct training data
attack_train = fold["attack_train"].copy()
benign_train_fold = fold["benign_train"].copy()

attack_train["Target"] = 1
benign_train_fold["Target"] = 0

dns_train_data = pd.concat(
    [attack_train, benign_train_fold],
    ignore_index=True
)

X_dns = dns_train_data[feature_columns].replace(
    [np.inf, -np.inf], np.nan
)

y_dns = dns_train_data["Target"]

# Internal stratified split
X_dns_train, X_dns_val, y_dns_train, y_dns_val = train_test_split(
    X_dns,
    y_dns,
    test_size=0.20,
    random_state=42,
    stratify=y_dns
)

# Fit imputer on training subset only
dns_imputer = SimpleImputer(strategy="median")

X_dns_train_imp = dns_imputer.fit_transform(X_dns_train)
X_dns_val_imp = dns_imputer.transform(X_dns_val)

# Train validation model
dns_rf = RandomForestClassifier(
    n_estimators=100,
    max_depth=20,
    min_samples_leaf=2,
    class_weight="balanced",
    random_state=42,
    n_jobs=1
)

dns_rf.fit(X_dns_train_imp, y_dns_train)

# Validation probabilities
dns_val_probs = dns_rf.predict_proba(X_dns_val_imp)[:, 1]

# Examine probability distributions
print("=" * 75)
print("DNS FOLD: VALIDATION PROBABILITY ANALYSIS")
print("=" * 75)

print("\nBenign probability distribution:")
print(
    pd.Series(
        dns_val_probs[y_dns_val.to_numpy() == 0]
    ).describe()
)

print("\nKnown attack probability distribution:")
print(
    pd.Series(
        dns_val_probs[y_dns_val.to_numpy() == 1]
    ).describe()
)

print("\nValidation ROC-AUC:", roc_auc_score(
    y_dns_val, dns_val_probs
))

# Find threshold maximizing recall with FPR <= 1%
threshold_candidates = np.unique(
    np.concatenate([
        np.linspace(0.01, 0.99, 200),
        dns_val_probs
    ])
)

valid_thresholds = []

for threshold in threshold_candidates:

    predictions = (
        dns_val_probs >= threshold
    ).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_dns_val,
        predictions,
        labels=[0, 1]
    ).ravel()

    fpr = fp / (fp + tn)
    recall = tp / (tp + fn)

    if fpr <= 0.01:
        valid_thresholds.append({
            "Threshold": threshold,
            "Recall": recall,
            "FPR": fpr,
            "TP": tp,
            "FN": fn,
            "FP": fp,
            "TN": tn
        })

threshold_results = pd.DataFrame(valid_thresholds)

best_threshold_row = threshold_results.sort_values(
    ["Recall", "Threshold"],
    ascending=[False, False]
).iloc[0]

print("\n" + "=" * 75)
print("VALIDATION-SELECTED THRESHOLD")
print("=" * 75)

print(best_threshold_row.to_string())

dns_selected_threshold = best_threshold_row["Threshold"]

print("\nSelected threshold:", dns_selected_threshold)
print("Threshold selection completed using validation data only.")
```

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    roc_auc_score
)

calibrated_loco_results = []
calibrated_loco_models = {}

MAX_VALIDATION_FPR = 0.01

for held_out, fold in leave_one_out_folds.items():

    print("\n" + "=" * 75)
    print("CALIBRATED LOCO:", held_out)
    print("=" * 75)

    # ------------------------------------------------
    # 1. Construct training dataset
    # ------------------------------------------------

    attack_train = fold["attack_train"].copy()
    benign_train_fold = fold["benign_train"].copy()

    attack_train["Target"] = 1
    benign_train_fold["Target"] = 0

    train_data = pd.concat(
        [attack_train, benign_train_fold],
        ignore_index=True
    )

    X = train_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    )

    y = train_data["Target"]

    # ------------------------------------------------
    # 2. Internal training-validation split
    # ------------------------------------------------

    X_train, X_val, y_train, y_val = train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )

    # ------------------------------------------------
    # 3. Fit imputer on training subset only
    # ------------------------------------------------

    imputer = SimpleImputer(strategy="median")

    X_train_imp = imputer.fit_transform(X_train)
    X_val_imp = imputer.transform(X_val)

    # ------------------------------------------------
    # 4. Train Random Forest
    # ------------------------------------------------

    model = RandomForestClassifier(
        n_estimators=100,
        max_depth=20,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=1
    )

    model.fit(X_train_imp, y_train)

    # ------------------------------------------------
    # 5. Select threshold using validation data
    # ------------------------------------------------

    val_probs = model.predict_proba(X_val_imp)[:, 1]

    thresholds = np.unique(
        np.concatenate([
            np.linspace(0.01, 0.99, 200),
            val_probs
        ])
    )

    valid_thresholds = []

    for threshold in thresholds:

        val_pred = (val_probs >= threshold).astype(int)

        tn, fp, fn, tp = confusion_matrix(
            y_val,
            val_pred,
            labels=[0, 1]
        ).ravel()

        fpr = fp / (fp + tn)
        recall = tp / (tp + fn)

        if fpr <= MAX_VALIDATION_FPR:

            valid_thresholds.append({
                "Threshold": threshold,
                "Recall": recall,
                "FPR": fpr
            })

    threshold_table = pd.DataFrame(valid_thresholds)

    best = threshold_table.sort_values(
        ["Recall", "Threshold"],
        ascending=[False, False]
    ).iloc[0]

    selected_threshold = float(best["Threshold"])

    # ------------------------------------------------
    # 6. Prepare untouched category test set
    # ------------------------------------------------

    attack_test = fold["attack_test"].copy()
    benign_test_fold = fold["benign_test"].copy()

    attack_test["Target"] = 1
    benign_test_fold["Target"] = 0

    test_data_fold = pd.concat(
        [benign_test_fold, attack_test],
        ignore_index=True
    )

    X_test = test_data_fold[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    )

    y_test = test_data_fold["Target"]

    X_test_imp = imputer.transform(X_test)

    test_probs = model.predict_proba(X_test_imp)[:, 1]

    # ------------------------------------------------
    # 7. Evaluate default and calibrated thresholds
    # ------------------------------------------------

    for method, threshold in [
        ("Default", 0.5),
        ("Calibrated", selected_threshold)
    ]:

        predictions = (test_probs >= threshold).astype(int)

        tn, fp, fn, tp = confusion_matrix(
            y_test,
            predictions,
            labels=[0, 1]
        ).ravel()

        result = {
            "Category": held_out,
            "Method": method,
            "Threshold": threshold,
            "Accuracy": accuracy_score(y_test, predictions),
            "Precision": precision_score(
                y_test, predictions, zero_division=0
            ),
            "Recall": recall_score(
                y_test, predictions, zero_division=0
            ),
            "F1_Score": f1_score(
                y_test, predictions, zero_division=0
            ),
            "Balanced_Accuracy": balanced_accuracy_score(
                y_test, predictions
            ),
            "ROC_AUC": roc_auc_score(y_test, test_probs),
            "FPR": fp / (fp + tn),
            "TN": tn,
            "FP": fp,
            "FN": fn,
            "TP": tp,
            "Validation_Recall": best["Recall"],
            "Validation_FPR": best["FPR"]
        }

        calibrated_loco_results.append(result)

    calibrated_loco_models[held_out] = {
        "model": model,
        "imputer": imputer,
        "threshold": selected_threshold,
        "features": feature_columns
    }

    print("Selected threshold:", round(selected_threshold, 6))
    print("Validation recall:", round(best["Recall"], 6))
    print("Validation FPR:", round(best["FPR"], 6))

    print("\nTest results:")
    print(
        pd.DataFrame([
            r for r in calibrated_loco_results
            if r["Category"] == held_out
        ])[
            ["Method", "Accuracy", "Precision", "Recall",
             "F1_Score", "FPR", "FN", "FP"]
        ].round(6).to_string(index=False)
    )

# ----------------------------------------------------
# 8. Overall comparison
# ----------------------------------------------------

calibrated_loco_df = pd.DataFrame(calibrated_loco_results)

print("\n" + "=" * 85)
print("FINAL CALIBRATED LOCO COMPARISON")
print("=" * 85)

print(
    calibrated_loco_df[
        ["Category", "Method", "Threshold",
         "Accuracy", "Precision", "Recall",
         "F1_Score", "FPR", "FN", "FP"]
    ].round(6).to_string(index=False)
)

print("\nMean performance by method:")

print(
    calibrated_loco_df.groupby("Method")[
        ["Accuracy", "Precision", "Recall",
         "F1_Score", "Balanced_Accuracy", "FPR"]
    ].mean().round(6)
)

print("\nAll calibrated LOCO experiments completed.")
```

```python
import pandas as pd
import numpy as np

# Separate default and calibrated results
default_results = calibrated_loco_df[
    calibrated_loco_df["Method"] == "Default"
].copy()

calibrated_results = calibrated_loco_df[
    calibrated_loco_df["Method"] == "Calibrated"
].copy()

# Aggregate totals
comparison = pd.DataFrame({
    "Method": ["Default", "Calibrated"],
    "Total_Attacks": [
        default_results["TP"].sum() + default_results["FN"].sum(),
        calibrated_results["TP"].sum() + calibrated_results["FN"].sum()
    ],
    "Detected_Attacks": [
        default_results["TP"].sum(),
        calibrated_results["TP"].sum()
    ],
    "Missed_Attacks": [
        default_results["FN"].sum(),
        calibrated_results["FN"].sum()
    ],
    "Benign_Samples": [
        default_results["TN"].sum() + default_results["FP"].sum(),
        calibrated_results["TN"].sum() + calibrated_results["FP"].sum()
    ],
    "False_Alarms": [
        default_results["FP"].sum(),
        calibrated_results["FP"].sum()
    ]
})

comparison["Detection_Rate"] = (
    comparison["Detected_Attacks"] /
    comparison["Total_Attacks"]
)

comparison["Overall_FPR"] = (
    comparison["False_Alarms"] /
    comparison["Benign_Samples"]
)

comparison["Missed_Attack_Reduction"] = np.nan

default_missed = comparison.loc[
    comparison["Method"] == "Default",
    "Missed_Attacks"
].iloc[0]

calibrated_missed = comparison.loc[
    comparison["Method"] == "Calibrated",
    "Missed_Attacks"
].iloc[0]

reduction = (
    (default_missed - calibrated_missed) /
    default_missed * 100
)

comparison.loc[
    comparison["Method"] == "Calibrated",
    "Missed_Attack_Reduction"
] = reduction

print("=" * 85)
print("AGGREGATED ZERO-DAY DETECTION ANALYSIS")
print("=" * 85)

print(comparison.round(6).to_string(index=False))

print("\nMissed attacks reduced by:", reduction, "%")

print("\nPer-category threshold transfer:")
print(
    calibrated_results[
        [
            "Category",
            "Threshold",
            "Validation_Recall",
            "Validation_FPR",
            "Recall",
            "FPR",
            "FN",
            "FP"
        ]
    ].round(6).to_string(index=False)
)
```

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    roc_auc_score
)

# --------------------------------------------------
# 1. Reconstruct supervised training dataset
# --------------------------------------------------

known_train = known_attacks.copy()
benign_train_data = benign_train.copy()

known_train["Target"] = 1
benign_train_data["Target"] = 0

supervised_data = pd.concat(
    [known_train, benign_train_data],
    ignore_index=True
)

X = supervised_data[feature_columns].replace(
    [np.inf, -np.inf], np.nan
)

y = supervised_data["Target"]

# --------------------------------------------------
# 2. Training-validation split
# --------------------------------------------------

X_train, X_val, y_train, y_val = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# --------------------------------------------------
# 3. Fit imputer on training data only
# --------------------------------------------------

snmp_imputer = SimpleImputer(strategy="median")

X_train_imp = snmp_imputer.fit_transform(X_train)
X_val_imp = snmp_imputer.transform(X_val)

# --------------------------------------------------
# 4. Train Random Forest
# --------------------------------------------------

snmp_rf = RandomForestClassifier(
    n_estimators=100,
    max_depth=20,
    min_samples_leaf=2,
    class_weight="balanced",
    random_state=42,
    n_jobs=1
)

snmp_rf.fit(X_train_imp, y_train)

# --------------------------------------------------
# 5. Select threshold using validation data
# --------------------------------------------------

val_probs = snmp_rf.predict_proba(X_val_imp)[:, 1]

threshold_candidates = np.unique(
    np.concatenate([
        np.linspace(0.01, 0.99, 200),
        val_probs
    ])
)

valid_thresholds = []

for threshold in threshold_candidates:

    predictions = (val_probs >= threshold).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_val,
        predictions,
        labels=[0, 1]
    ).ravel()

    fpr = fp / (fp + tn)
    recall = tp / (tp + fn)

    if fpr <= 0.01:

        valid_thresholds.append({
            "Threshold": threshold,
            "Recall": recall,
            "FPR": fpr
        })

threshold_table = pd.DataFrame(valid_thresholds)

best_threshold = threshold_table.sort_values(
    ["Recall", "Threshold"],
    ascending=[False, False]
).iloc[0]

selected_snmp_threshold = float(
    best_threshold["Threshold"]
)

print("=" * 75)
print("SNMP VALIDATION THRESHOLD")
print("=" * 75)

print(best_threshold.to_string())

# --------------------------------------------------
# 6. Prepare untouched SNMP test dataset
# --------------------------------------------------

snmp_test_data = pd.concat(
    [benign_test, snmp_test],
    ignore_index=True
)

y_snmp_test = np.concatenate([
    np.zeros(len(benign_test), dtype=int),
    np.ones(len(snmp_test), dtype=int)
])

X_snmp_test = snmp_test_data[feature_columns].replace(
    [np.inf, -np.inf], np.nan
)

X_snmp_test_imp = snmp_imputer.transform(X_snmp_test)

test_probs = snmp_rf.predict_proba(
    X_snmp_test_imp
)[:, 1]

# --------------------------------------------------
# 7. Compare default and calibrated thresholds
# --------------------------------------------------

snmp_comparison = []

for method, threshold in [
    ("Default", 0.5),
    ("Calibrated", selected_snmp_threshold)
]:

    predictions = (test_probs >= threshold).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_snmp_test,
        predictions,
        labels=[0, 1]
    ).ravel()

    snmp_comparison.append({
        "Method": method,
        "Threshold": threshold,
        "Accuracy": accuracy_score(
            y_snmp_test, predictions
        ),
        "Precision": precision_score(
            y_snmp_test, predictions, zero_division=0
        ),
        "Recall": recall_score(
            y_snmp_test, predictions, zero_division=0
        ),
        "F1_Score": f1_score(
            y_snmp_test, predictions, zero_division=0
        ),
        "Balanced_Accuracy": balanced_accuracy_score(
            y_snmp_test, predictions
        ),
        "ROC_AUC": roc_auc_score(
            y_snmp_test, test_probs
        ),
        "FPR": fp / (fp + tn),
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp
    })

snmp_comparison_df = pd.DataFrame(snmp_comparison)

print("\n" + "=" * 85)
print("SNMP ZERO-DAY: DEFAULT VS CALIBRATED THRESHOLD")
print("=" * 85)

print(snmp_comparison_df.round(6).to_string(index=False))

print("\nSNMP threshold experiment completed.")
```

```python
import pandas as pd
import numpy as np

# Four LOCO experiments
default_loco = calibrated_loco_df[
    calibrated_loco_df["Method"] == "Default"
]

calibrated_loco = calibrated_loco_df[
    calibrated_loco_df["Method"] == "Calibrated"
]

# SNMP experiment
default_snmp = snmp_comparison_df[
    snmp_comparison_df["Method"] == "Default"
].iloc[0]

calibrated_snmp = snmp_comparison_df[
    snmp_comparison_df["Method"] == "Calibrated"
].iloc[0]


def aggregate_counts(loco_data, snmp_data):

    return {
        "TP": int(loco_data["TP"].sum() + snmp_data["TP"]),
        "FN": int(loco_data["FN"].sum() + snmp_data["FN"]),
        "TN": int(loco_data["TN"].sum() + snmp_data["TN"]),
        "FP": int(loco_data["FP"].sum() + snmp_data["FP"])
    }


default_counts = aggregate_counts(
    default_loco, default_snmp
)

calibrated_counts = aggregate_counts(
    calibrated_loco, calibrated_snmp
)


def calculate_aggregate_metrics(name, counts):

    tp = counts["TP"]
    fn = counts["FN"]
    tn = counts["TN"]
    fp = counts["FP"]

    total = tp + fn + tn + fp

    precision = tp / (tp + fp)
    recall = tp / (tp + fn)
    specificity = tn / (tn + fp)

    f1 = 2 * precision * recall / (precision + recall)

    return {
        "Method": name,
        "Accuracy": (tp + tn) / total,
        "Precision": precision,
        "Recall": recall,
        "Specificity": specificity,
        "F1_Score": f1,
        "Balanced_Accuracy": (recall + specificity) / 2,
        "FPR": fp / (fp + tn),
        "TP": tp,
        "FN": fn,
        "TN": tn,
        "FP": fp
    }


final_comparison = pd.DataFrame([
    calculate_aggregate_metrics(
        "Default",
        default_counts
    ),
    calculate_aggregate_metrics(
        "Calibrated",
        calibrated_counts
    )
])

print("=" * 85)
print("AGGREGATED EVALUATION: FIVE UNSEEN ATTACK CATEGORIES")
print("=" * 85)

print(final_comparison.round(6).to_string(index=False))

# Missed attack reduction
missed_default = default_counts["FN"]
missed_calibrated = calibrated_counts["FN"]

reduction = (
    (missed_default - missed_calibrated)
    / missed_default * 100
)

print("\nTotal attacks evaluated:",
      default_counts["TP"] + default_counts["FN"])

print("Total benign samples evaluated:",
      default_counts["TN"] + default_counts["FP"])

print("Missed attacks reduced:", missed_default - missed_calibrated)

print("Relative missed-attack reduction:",
      round(reduction, 4), "%")

print("\nAggregation completed.")
```

```python
import pandas as pd
import numpy as np

duplicate_label_results = []

for category, fold in leave_one_out_folds.items():

    # Reconstruct training data
    attack_train = fold["attack_train"].copy()
    benign_train_fold = fold["benign_train"].copy()

    attack_train["Target"] = 1
    benign_train_fold["Target"] = 0

    train_data = pd.concat(
        [attack_train, benign_train_fold],
        ignore_index=True
    )

    # Reconstruct testing data
    attack_test = fold["attack_test"].copy()
    benign_test_fold = fold["benign_test"].copy()

    attack_test["Target"] = 1
    benign_test_fold["Target"] = 0

    test_data_fold = pd.concat(
        [benign_test_fold, attack_test],
        ignore_index=True
    )

    # Generate hashes
    train_features = train_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    ).fillna(-999999)

    test_features = test_data_fold[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    ).fillna(-999999)

    train_hashes = pd.util.hash_pandas_object(
        train_features, index=False
    )

    test_hashes = pd.util.hash_pandas_object(
        test_features, index=False
    )

    # Map each training hash to its unique labels
    train_lookup = pd.DataFrame({
        "Hash": train_hashes.to_numpy(),
        "Train_Target": train_data["Target"].to_numpy()
    })

    train_label_map = (
        train_lookup.groupby("Hash")["Train_Target"]
        .apply(lambda x: tuple(sorted(x.unique())))
        .to_dict()
    )

    # Identify matching test rows
    matched_indices = [
        i for i, h in enumerate(test_hashes)
        if h in train_label_map
    ]

    counts = {
        "Same-class duplicate": 0,
        "Opposite-class duplicate": 0,
        "Conflicting training labels": 0
    }

    class_counts = {
        0: {key: 0 for key in counts},
        1: {key: 0 for key in counts}
    }

    for i in matched_indices:

        test_label = int(test_data_fold.iloc[i]["Target"])
        matching_labels = train_label_map[test_hashes.iloc[i]]

        if len(matching_labels) > 1:
            relationship = "Conflicting training labels"

        elif test_label in matching_labels:
            relationship = "Same-class duplicate"

        else:
            relationship = "Opposite-class duplicate"

        counts[relationship] += 1
        class_counts[test_label][relationship] += 1

    print("\n" + "=" * 70)
    print("DUPLICATE LABEL ANALYSIS:", category)
    print("=" * 70)

    print("Total duplicate rows:", len(matched_indices))

    print("\nDuplicate relationships:")
    for key, value in counts.items():
        print(f"{key}: {value}")

    print("\nBy actual test class:")
    for label in [0, 1]:
        print(f"\nTest class {label}:")
        for key, value in class_counts[label].items():
            print(f"  {key}: {value}")

    duplicate_label_results.append({
        "Category": category,
        "Total_Duplicates": len(matched_indices),
        "Same_Class": counts["Same-class duplicate"],
        "Opposite_Class": counts["Opposite-class duplicate"],
        "Conflicting_Training_Labels": counts[
            "Conflicting training labels"
        ]
    })

duplicate_label_df = pd.DataFrame(duplicate_label_results)

print("\n" + "=" * 70)
print("FINAL DUPLICATE-LABEL SUMMARY")
print("=" * 70)

print(duplicate_label_df.to_string(index=False))
```

```python
# ==============================================
# STEP 53: LOCO FEATURE REDUCTION VALIDATION
# ==============================================

import numpy as np
import pandas as pd

from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    confusion_matrix
)

loco_feature_results = []

for category, fold in leave_one_out_folds.items():

    print("\n" + "=" * 75)
    print("HELD-OUT CATEGORY:", category)
    print("=" * 75)

    # ------------------------------------------
    # 1. PREPARE TRAINING DATA
    # ------------------------------------------

    attack_train = fold["attack_train"].copy()
    benign_train_fold = fold["benign_train"].copy()

    attack_train["Target"] = 1
    benign_train_fold["Target"] = 0

    train_data = pd.concat(
        [attack_train, benign_train_fold],
        ignore_index=True
    )

    # ------------------------------------------
    # 2. PREPARE TEST DATA
    # ------------------------------------------

    attack_test = fold["attack_test"].copy()
    benign_test_fold = fold["benign_test"].copy()

    attack_test["Target"] = 1
    benign_test_fold["Target"] = 0

    test_data = pd.concat(
        [benign_test_fold, attack_test],
        ignore_index=True
    )

    X_train_fold = train_data[feature_columns].copy()
    y_train_fold = train_data["Target"]

    X_test_fold = test_data[feature_columns].copy()
    y_test_fold = test_data["Target"]

    # Replace infinite values
    X_train_fold = X_train_fold.replace(
        [np.inf, -np.inf], np.nan
    )

    X_test_fold = X_test_fold.replace(
        [np.inf, -np.inf], np.nan
    )

    # ==========================================
    # MODEL A: ORIGINAL 81 FEATURES
    # ==========================================

    original_imputer = SimpleImputer(strategy="median")

    X_train_original = original_imputer.fit_transform(
        X_train_fold
    )

    X_test_original = original_imputer.transform(
        X_test_fold
    )

    rf_original = RandomForestClassifier(
        n_estimators=100,
        max_depth=20,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=1
    )

    rf_original.fit(
        X_train_original,
        y_train_fold
    )

    pred_original = rf_original.predict(
        X_test_original
    )

    cm_original = confusion_matrix(
        y_test_fold,
        pred_original
    )

    tn_a, fp_a, fn_a, tp_a = cm_original.ravel()

    # ==========================================
    # MODEL B: CORRELATION-REDUCED FEATURES
    # ==========================================

    # Identify non-constant features using training data only
    non_constant_features = [
        col for col in feature_columns
        if X_train_fold[col].nunique(dropna=True) > 1
    ]

    X_corr_fold = X_train_fold[
        non_constant_features
    ]

    # Calculate training-only correlations
    corr_matrix = X_corr_fold.corr().abs()

    upper = corr_matrix.where(
        np.triu(
            np.ones(corr_matrix.shape),
            k=1
        ).astype(bool)
    )

    correlation_threshold = 0.95

    features_to_remove = [
        col for col in upper.columns
        if any(upper[col] >= correlation_threshold)
    ]

    fold_reduced_features = [
        col for col in non_constant_features
        if col not in features_to_remove
    ]

    # Transform using reduced feature set
    X_train_reduced_fold = X_train_fold[
        fold_reduced_features
    ]

    X_test_reduced_fold = X_test_fold[
        fold_reduced_features
    ]

    reduced_fold_imputer = SimpleImputer(
        strategy="median"
    )

    X_train_reduced_imp = reduced_fold_imputer.fit_transform(
        X_train_reduced_fold
    )

    X_test_reduced_imp = reduced_fold_imputer.transform(
        X_test_reduced_fold
    )

    # Train reduced Random Forest
    rf_reduced_fold = RandomForestClassifier(
        n_estimators=100,
        max_depth=20,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=1
    )

    rf_reduced_fold.fit(
        X_train_reduced_imp,
        y_train_fold
    )

    pred_reduced = rf_reduced_fold.predict(
        X_test_reduced_imp
    )

    cm_reduced = confusion_matrix(
        y_test_fold,
        pred_reduced
    )

    tn_b, fp_b, fn_b, tp_b = cm_reduced.ravel()

    # ==========================================
    # 3. STORE RESULTS
    # ==========================================

    results = {
        "Category": category,

        "Original_Features": len(feature_columns),

        "Reduced_Features": len(fold_reduced_features),

        "Original_Accuracy": accuracy_score(
            y_test_fold, pred_original
        ),

        "Reduced_Accuracy": accuracy_score(
            y_test_fold, pred_reduced
        ),

        "Original_Precision": precision_score(
            y_test_fold, pred_original, zero_division=0
        ),

        "Reduced_Precision": precision_score(
            y_test_fold, pred_reduced, zero_division=0
        ),

        "Original_Recall": recall_score(
            y_test_fold, pred_original, zero_division=0
        ),

        "Reduced_Recall": recall_score(
            y_test_fold, pred_reduced, zero_division=0
        ),

        "Original_F1": f1_score(
            y_test_fold, pred_original, zero_division=0
        ),

        "Reduced_F1": f1_score(
            y_test_fold, pred_reduced, zero_division=0
        ),

        "Original_Balanced_Accuracy": balanced_accuracy_score(
            y_test_fold, pred_original
        ),

        "Reduced_Balanced_Accuracy": balanced_accuracy_score(
            y_test_fold, pred_reduced
        ),

        "Original_FP": fp_a,
        "Reduced_FP": fp_b,

        "Original_FN": fn_a,
        "Reduced_FN": fn_b
    }

    loco_feature_results.append(results)

    # ------------------------------------------
    # 4. DISPLAY FOLD RESULTS
    # ------------------------------------------

    print("\nOriginal model:")
    print("Features:", len(feature_columns))
    print("Confusion matrix:\n", cm_original)
    print("Accuracy:", round(results["Original_Accuracy"], 6))
    print("Recall:", round(results["Original_Recall"], 6))
    print("F1:", round(results["Original_F1"], 6))

    print("\nReduced model:")
    print("Features:", len(fold_reduced_features))
    print("Confusion matrix:\n", cm_reduced)
    print("Accuracy:", round(results["Reduced_Accuracy"], 6))
    print("Recall:", round(results["Reduced_Recall"], 6))
    print("F1:", round(results["Reduced_F1"], 6))

# ==============================================
# 5. FINAL COMPARISON
# ==============================================

loco_feature_df = pd.DataFrame(loco_feature_results)

print("\n" + "=" * 90)
print("FINAL LOCO: ORIGINAL VS REDUCED FEATURE MODEL")
print("=" * 90)

display(
    loco_feature_df.round(6)
)

print("\nAverage performance:")

display(
    loco_feature_df[
        [
            "Original_Accuracy",
            "Reduced_Accuracy",
            "Original_Recall",
            "Reduced_Recall",
            "Original_F1",
            "Reduced_F1",
            "Original_Balanced_Accuracy",
            "Reduced_Balanced_Accuracy"
        ]
    ].mean().to_frame("Average").round(6)
)
```

```python
# ==============================================
# STEP 54: SNMP ZERO-DAY FEATURE REDUCTION
# ==============================================

import numpy as np
import pandas as pd

from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    confusion_matrix,
    roc_auc_score
)

# ==========================================
# 1. PREPARE SUPERVISED TRAINING DATA
# ==========================================

X_snmp_train = X_sup_train[feature_columns].copy()

y_snmp_train = y_sup_train.copy()

# Prepare SNMP zero-day test data
X_snmp_test = snmp_zero_day[feature_columns].copy()

y_snmp_test = np.ones(len(X_snmp_test), dtype=int)

# Add benign test samples
X_benign_test = benign_test[feature_columns].copy()

y_benign_test = np.zeros(len(X_benign_test), dtype=int)

X_zero_day_test = pd.concat(
    [X_benign_test, X_snmp_test],
    ignore_index=True
)

y_zero_day_test = np.concatenate(
    [y_benign_test, y_snmp_test]
)

# Handle infinite values
X_snmp_train = X_snmp_train.replace(
    [np.inf, -np.inf], np.nan
)

X_zero_day_test = X_zero_day_test.replace(
    [np.inf, -np.inf], np.nan
)

print("Training samples:", len(X_snmp_train))
print("Zero-day test samples:", len(X_zero_day_test))

# ==========================================
# 2. MODEL A: ORIGINAL 81 FEATURES
# ==========================================

original_snmp_imputer = SimpleImputer(
    strategy="median"
)

X_train_original = original_snmp_imputer.fit_transform(
    X_snmp_train
)

X_test_original = original_snmp_imputer.transform(
    X_zero_day_test
)

rf_snmp_original = RandomForestClassifier(
    n_estimators=100,
    max_depth=20,
    min_samples_leaf=2,
    class_weight="balanced",
    random_state=42,
    n_jobs=1
)

rf_snmp_original.fit(
    X_train_original,
    y_snmp_train
)

original_probs = rf_snmp_original.predict_proba(
    X_test_original
)[:, 1]

original_predictions = (
    original_probs >= 0.5
).astype(int)

# ==========================================
# 3. MODEL B: CORRELATION-REDUCED FEATURES
# ==========================================

# Feature selection using training data only
non_constant_snmp_features = [
    col for col in feature_columns
    if X_snmp_train[col].nunique(dropna=True) > 1
]

snmp_corr_matrix = X_snmp_train[
    non_constant_snmp_features
].corr().abs()

snmp_upper = snmp_corr_matrix.where(
    np.triu(
        np.ones(snmp_corr_matrix.shape),
        k=1
    ).astype(bool)
)

snmp_features_to_remove = [
    col for col in snmp_upper.columns
    if any(snmp_upper[col] >= 0.95)
]

snmp_reduced_features = [
    col for col in non_constant_snmp_features
    if col not in snmp_features_to_remove
]

print("\nOriginal features:", len(feature_columns))
print("Reduced features:", len(snmp_reduced_features))

# Transform reduced data
X_train_snmp_reduced = X_snmp_train[
    snmp_reduced_features
]

X_test_snmp_reduced = X_zero_day_test[
    snmp_reduced_features
]

snmp_reduced_imputer = SimpleImputer(
    strategy="median"
)

X_train_snmp_reduced_imp = (
    snmp_reduced_imputer.fit_transform(
        X_train_snmp_reduced
    )
)

X_test_snmp_reduced_imp = (
    snmp_reduced_imputer.transform(
        X_test_snmp_reduced
    )
)

# Train reduced Random Forest
rf_snmp_reduced = RandomForestClassifier(
    n_estimators=100,
    max_depth=20,
    min_samples_leaf=2,
    class_weight="balanced",
    random_state=42,
    n_jobs=1
)

rf_snmp_reduced.fit(
    X_train_snmp_reduced_imp,
    y_snmp_train
)

reduced_probs = rf_snmp_reduced.predict_proba(
    X_test_snmp_reduced_imp
)[:, 1]

reduced_predictions = (
    reduced_probs >= 0.5
).astype(int)

# ==========================================
# 4. EVALUATION FUNCTION
# ==========================================

def evaluate_snmp_model(name, y_true, predictions, probabilities):

    cm = confusion_matrix(
        y_true,
        predictions
    )

    tn, fp, fn, tp = cm.ravel()

    print("\n" + "=" * 65)
    print(name)
    print("=" * 65)

    print("\nConfusion Matrix:")
    print(cm)

    print("\nAccuracy:", accuracy_score(y_true, predictions))
    print("Precision:", precision_score(y_true, predictions))
    print("Recall:", recall_score(y_true, predictions))
    print("F1:", f1_score(y_true, predictions))

    print(
        "Balanced Accuracy:",
        balanced_accuracy_score(y_true, predictions)
    )

    print(
        "ROC AUC:",
        roc_auc_score(y_true, probabilities)
    )

    print("\nFalse Positives:", fp)
    print("False Negatives:", fn)

    return {
        "Model": name,
        "Accuracy": accuracy_score(y_true, predictions),
        "Precision": precision_score(y_true, predictions),
        "Recall": recall_score(y_true, predictions),
        "F1": f1_score(y_true, predictions),
        "Balanced_Accuracy": balanced_accuracy_score(
            y_true, predictions
        ),
        "ROC_AUC": roc_auc_score(y_true, probabilities),
        "FP": fp,
        "FN": fn
    }

# ==========================================
# 5. COMPARE BOTH MODELS
# ==========================================

original_result = evaluate_snmp_model(
    "ORIGINAL 81-FEATURE MODEL",
    y_zero_day_test,
    original_predictions,
    original_probs
)

reduced_result = evaluate_snmp_model(
    "CORRELATION-REDUCED MODEL",
    y_zero_day_test,
    reduced_predictions,
    reduced_probs
)

snmp_comparison = pd.DataFrame([
    original_result,
    reduced_result
])

print("\n" + "=" * 75)
print("FINAL SNMP ZERO-DAY COMPARISON")
print("=" * 75)

display(
    snmp_comparison.set_index("Model").round(6)
)
```

```python
# ==============================================
# STEP 54: SNMP ZERO-DAY FEATURE REDUCTION
# ==============================================

import numpy as np
import pandas as pd

from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    confusion_matrix,
    roc_auc_score
)

# ==========================================
# 1. PREPARE TRAINING AND TEST DATA
# ==========================================

X_snmp_train = X_sup_train[feature_columns].copy()
y_snmp_train = y_sup_train.copy()

X_zero_day_test = snmp_test_data[feature_columns].copy()

# Convert string labels to binary labels
y_zero_day_test = (
    snmp_test_data["Label"] == "DrDoS_SNMP"
).astype(int)

# Handle infinite values
X_snmp_train = X_snmp_train.replace(
    [np.inf, -np.inf], np.nan
)

X_zero_day_test = X_zero_day_test.replace(
    [np.inf, -np.inf], np.nan
)

print("Training samples:", len(X_snmp_train))
print("Testing samples:", len(X_zero_day_test))

print("\nTest label distribution:")
print(y_zero_day_test.value_counts())

# ==========================================
# 2. MODEL A: ORIGINAL 81 FEATURES
# ==========================================

original_snmp_imputer = SimpleImputer(
    strategy="median"
)

X_train_original = original_snmp_imputer.fit_transform(
    X_snmp_train
)

X_test_original = original_snmp_imputer.transform(
    X_zero_day_test
)

rf_snmp_original = RandomForestClassifier(
    n_estimators=100,
    max_depth=20,
    min_samples_leaf=2,
    class_weight="balanced",
    random_state=42,
    n_jobs=1
)

rf_snmp_original.fit(
    X_train_original,
    y_snmp_train
)

original_probs = rf_snmp_original.predict_proba(
    X_test_original
)[:, 1]

original_predictions = (
    original_probs >= 0.5
).astype(int)

# ==========================================
# 3. MODEL B: CORRELATION-REDUCED FEATURES
# ==========================================

# Identify non-constant features using training data only
non_constant_snmp_features = [
    col for col in feature_columns
    if X_snmp_train[col].nunique(dropna=True) > 1
]

# Calculate correlation using training data only
snmp_corr_matrix = X_snmp_train[
    non_constant_snmp_features
].corr().abs()

snmp_upper = snmp_corr_matrix.where(
    np.triu(
        np.ones(snmp_corr_matrix.shape),
        k=1
    ).astype(bool)
)

# Remove highly correlated features
snmp_features_to_remove = [
    col for col in snmp_upper.columns
    if any(snmp_upper[col] >= 0.95)
]

snmp_reduced_features = [
    col for col in non_constant_snmp_features
    if col not in snmp_features_to_remove
]

print("\nOriginal features:", len(feature_columns))
print("Reduced features:", len(snmp_reduced_features))

# Prepare reduced datasets
X_train_snmp_reduced = X_snmp_train[
    snmp_reduced_features
]

X_test_snmp_reduced = X_zero_day_test[
    snmp_reduced_features
]

# Fit imputer only on training data
snmp_reduced_imputer = SimpleImputer(
    strategy="median"
)

X_train_snmp_reduced_imp = (
    snmp_reduced_imputer.fit_transform(
        X_train_snmp_reduced
    )
)

X_test_snmp_reduced_imp = (
    snmp_reduced_imputer.transform(
        X_test_snmp_reduced
    )
)

# Train reduced Random Forest
rf_snmp_reduced = RandomForestClassifier(
    n_estimators=100,
    max_depth=20,
    min_samples_leaf=2,
    class_weight="balanced",
    random_state=42,
    n_jobs=1
)

rf_snmp_reduced.fit(
    X_train_snmp_reduced_imp,
    y_snmp_train
)

reduced_probs = rf_snmp_reduced.predict_proba(
    X_test_snmp_reduced_imp
)[:, 1]

reduced_predictions = (
    reduced_probs >= 0.5
).astype(int)

# ==========================================
# 4. EVALUATION FUNCTION
# ==========================================

def evaluate_snmp_model(
    name, y_true, predictions, probabilities
):

    cm = confusion_matrix(y_true, predictions)

    tn, fp, fn, tp = cm.ravel()

    result = {
        "Model": name,
        "Accuracy": accuracy_score(y_true, predictions),
        "Precision": precision_score(y_true, predictions),
        "Recall": recall_score(y_true, predictions),
        "F1": f1_score(y_true, predictions),
        "Balanced_Accuracy": balanced_accuracy_score(
            y_true, predictions
        ),
        "ROC_AUC": roc_auc_score(y_true, probabilities),
        "FP": fp,
        "FN": fn
    }

    print("\n" + "=" * 65)
    print(name)
    print("=" * 65)

    print("\nConfusion Matrix:")
    print(cm)

    for metric, value in result.items():
        if metric != "Model":
            print(f"{metric}: {value}")

    return result

# ==========================================
# 5. COMPARE BOTH MODELS
# ==========================================

original_result = evaluate_snmp_model(
    "ORIGINAL 81-FEATURE MODEL",
    y_zero_day_test,
    original_predictions,
    original_probs
)

reduced_result = evaluate_snmp_model(
    "CORRELATION-REDUCED MODEL",
    y_zero_day_test,
    reduced_predictions,
    reduced_probs
)

snmp_comparison = pd.DataFrame([
    original_result,
    reduced_result
])

print("\n" + "=" * 75)
print("FINAL SNMP ZERO-DAY COMPARISON")
print("=" * 75)

display(
    snmp_comparison.set_index("Model").round(6)
)
```

```python
import time
import os
import joblib
import numpy as np
import pandas as pd

# ==========================================
# STEP 55: COMPUTATIONAL EFFICIENCY
# ==========================================

# Prepare original 81-feature training data
X_train_81 = X_snmp_train[feature_columns].copy()
X_test_81 = X_zero_day_test[feature_columns].copy()

# Prepare reduced 44-feature data
X_train_44 = X_snmp_train[snmp_reduced_features].copy()
X_test_44 = X_zero_day_test[snmp_reduced_features].copy()

# Handle infinite values
X_train_81 = X_train_81.replace([np.inf, -np.inf], np.nan)
X_test_81 = X_test_81.replace([np.inf, -np.inf], np.nan)

X_train_44 = X_train_44.replace([np.inf, -np.inf], np.nan)
X_test_44 = X_test_44.replace([np.inf, -np.inf], np.nan)

# ==========================================
# 1. TRAINING TIME
# ==========================================

start = time.perf_counter()

rf_benchmark_81 = RandomForestClassifier(
    n_estimators=100,
    max_depth=20,
    min_samples_leaf=2,
    class_weight="balanced",
    random_state=42,
    n_jobs=1
)

rf_benchmark_81.fit(
    X_train_original,
    y_snmp_train
)

training_time_81 = time.perf_counter() - start

start = time.perf_counter()

rf_benchmark_44 = RandomForestClassifier(
    n_estimators=100,
    max_depth=20,
    min_samples_leaf=2,
    class_weight="balanced",
    random_state=42,
    n_jobs=1
)

rf_benchmark_44.fit(
    X_train_snmp_reduced_imp,
    y_snmp_train
)

training_time_44 = time.perf_counter() - start

# ==========================================
# 2. INFERENCE TIME
# ==========================================

# Warm-up
rf_benchmark_81.predict(X_test_original[:1000])
rf_benchmark_44.predict(X_test_snmp_reduced_imp[:1000])

start = time.perf_counter()

rf_benchmark_81.predict(X_test_original)

inference_time_81 = time.perf_counter() - start

start = time.perf_counter()

rf_benchmark_44.predict(X_test_snmp_reduced_imp)

inference_time_44 = time.perf_counter() - start

# ==========================================
# 3. MODEL SIZE
# ==========================================

joblib.dump(rf_benchmark_81, "rf_81_benchmark.joblib")
joblib.dump(rf_benchmark_44, "rf_44_benchmark.joblib")

size_81 = os.path.getsize("rf_81_benchmark.joblib") / (1024 ** 2)
size_44 = os.path.getsize("rf_44_benchmark.joblib") / (1024 ** 2)

# ==========================================
# 4. RESULTS
# ==========================================

efficiency_results = pd.DataFrame({
    "Metric": [
        "Feature Count",
        "Training Time (seconds)",
        "Inference Time (seconds)",
        "Inference Time per Sample (ms)",
        "Model Size (MB)"
    ],

    "Original_81": [
        81,
        training_time_81,
        inference_time_81,
        inference_time_81 / len(X_test_81) * 1000,
        size_81
    ],

    "Reduced_44": [
        44,
        training_time_44,
        inference_time_44,
        inference_time_44 / len(X_test_44) * 1000,
        size_44
    ]
})

print("=" * 75)
print("COMPUTATIONAL EFFICIENCY COMPARISON")
print("=" * 75)

display(efficiency_results.round(6))

print("\nFeature reduction:",
      round((1 - 44/81) * 100, 2), "%")
```

```python
# ==============================================
# STEP 57: ENSEMBLE-BASED DDoS DETECTION
# ==============================================

import time
import numpy as np
import pandas as pd

from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier,
    VotingClassifier
)

from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    confusion_matrix,
    roc_auc_score
)

from xgboost import XGBClassifier

# ==========================================
# 1. PREPARE 44-FEATURE DATA
# ==========================================

X_train_ensemble = X_sup_train[
    snmp_reduced_features
].copy()

X_test_ensemble = X_zero_day_test[
    snmp_reduced_features
].copy()

y_train_ensemble = y_sup_train.copy()
y_test_ensemble = y_zero_day_test.copy()

# Handle infinite values
X_train_ensemble = X_train_ensemble.replace(
    [np.inf, -np.inf], np.nan
)

X_test_ensemble = X_test_ensemble.replace(
    [np.inf, -np.inf], np.nan
)

# Imputation
ensemble_imputer = SimpleImputer(
    strategy="median"
)

X_train_ensemble_imp = ensemble_imputer.fit_transform(
    X_train_ensemble
)

X_test_ensemble_imp = ensemble_imputer.transform(
    X_test_ensemble
)

print("Training samples:", len(X_train_ensemble))
print("Testing samples:", len(X_test_ensemble))
print("Features:", len(snmp_reduced_features))

# ==========================================
# 2. INITIALIZE CLASSIFIERS
# ==========================================

rf_model = RandomForestClassifier(
    n_estimators=100,
    max_depth=20,
    min_samples_leaf=2,
    class_weight="balanced",
    random_state=42,
    n_jobs=1
)

et_model = ExtraTreesClassifier(
    n_estimators=100,
    max_depth=20,
    min_samples_leaf=2,
    class_weight="balanced",
    random_state=42,
    n_jobs=1
)

xgb_model = XGBClassifier(
    n_estimators=100,
    max_depth=6,
    learning_rate=0.1,
    subsample=0.8,
    colsample_bytree=0.8,
    eval_metric="logloss",
    random_state=42,
    n_jobs=1
)

# ==========================================
# 3. TRAIN INDIVIDUAL CLASSIFIERS
# ==========================================

models = {
    "Random Forest": rf_model,
    "Extra Trees": et_model,
    "XGBoost": xgb_model
}

trained_models = {}

training_times = {}

for name, model in models.items():

    print(f"\nTraining {name}...")

    start = time.perf_counter()

    model.fit(
        X_train_ensemble_imp,
        y_train_ensemble
    )

    training_times[name] = time.perf_counter() - start

    trained_models[name] = model

    print(
        f"{name} training completed in "
        f"{training_times[name]:.2f} seconds"
    )

# ==========================================
# 4. CREATE SOFT VOTING ENSEMBLE
# ==========================================

voting_model = VotingClassifier(
    estimators=[
        ("rf", rf_model),
        ("et", et_model),
        ("xgb", xgb_model)
    ],
    voting="soft",
    weights=[1, 1, 1],
    n_jobs=1
)

print("\nTraining Soft Voting Ensemble...")

start = time.perf_counter()

voting_model.fit(
    X_train_ensemble_imp,
    y_train_ensemble
)

voting_training_time = time.perf_counter() - start

trained_models["Soft Voting"] = voting_model

training_times["Soft Voting"] = voting_training_time

print(
    "Soft Voting training completed in "
    f"{voting_training_time:.2f} seconds"
)

# ==========================================
# 5. EVALUATE ALL MODELS
# ==========================================

ensemble_results = []

for name, model in trained_models.items():

    probabilities = model.predict_proba(
        X_test_ensemble_imp
    )[:, 1]

    predictions = (
        probabilities >= 0.5
    ).astype(int)

    cm = confusion_matrix(
        y_test_ensemble,
        predictions
    )

    tn, fp, fn, tp = cm.ravel()

    result = {
        "Model": name,
        "Accuracy": accuracy_score(
            y_test_ensemble, predictions
        ),
        "Precision": precision_score(
            y_test_ensemble, predictions
        ),
        "Recall": recall_score(
            y_test_ensemble, predictions
        ),
        "F1": f1_score(
            y_test_ensemble, predictions
        ),
        "Balanced_Accuracy": balanced_accuracy_score(
            y_test_ensemble, predictions
        ),
        "ROC_AUC": roc_auc_score(
            y_test_ensemble, probabilities
        ),
        "FP": fp,
        "FN": fn,
        "Training_Time": training_times[name]
    }

    ensemble_results.append(result)

    print("\n" + "=" * 65)
    print(name)
    print("=" * 65)

    print("\nConfusion Matrix:")
    print(cm)

    for metric, value in result.items():
        if metric != "Model":
            print(f"{metric}: {value:.6f}")

# ==========================================
# 6. FINAL COMPARISON
# ==========================================

ensemble_comparison = pd.DataFrame(
    ensemble_results
)

print("\n" + "=" * 85)
print("ENSEMBLE PERFORMANCE COMPARISON")
print("=" * 85)

display(
    ensemble_comparison.set_index("Model").round(6)
)
```

```python
# ==============================================
# STEP 58: LOCO ENSEMBLE VALIDATION
# ==============================================

import numpy as np
import pandas as pd
import time

from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.utils.class_weight import compute_sample_weight

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    confusion_matrix,
    roc_auc_score
)

from xgboost import XGBClassifier

loco_ensemble_results = []

for category, fold in leave_one_out_folds.items():

    print("\n" + "=" * 75)
    print("HELD-OUT CATEGORY:", category)
    print("=" * 75)

    # ==========================================
    # 1. PREPARE FOLD DATA
    # ==========================================

    attack_train = fold["attack_train"].copy()
    benign_train_fold = fold["benign_train"].copy()

    attack_train["Target"] = 1
    benign_train_fold["Target"] = 0

    train_data = pd.concat(
        [attack_train, benign_train_fold],
        ignore_index=True
    )

    attack_test = fold["attack_test"].copy()
    benign_test_fold = fold["benign_test"].copy()

    attack_test["Target"] = 1
    benign_test_fold["Target"] = 0

    test_data = pd.concat(
        [benign_test_fold, attack_test],
        ignore_index=True
    )

    X_train_fold = train_data[feature_columns].copy()
    y_train_fold = train_data["Target"].astype(int)

    X_test_fold = test_data[feature_columns].copy()
    y_test_fold = test_data["Target"].astype(int)

    X_train_fold = X_train_fold.replace(
        [np.inf, -np.inf], np.nan
    )

    X_test_fold = X_test_fold.replace(
        [np.inf, -np.inf], np.nan
    )

    # ==========================================
    # 2. FOLD-SPECIFIC FEATURE SELECTION
    # ==========================================

    non_constant_features = [
        col for col in feature_columns
        if X_train_fold[col].nunique(dropna=True) > 1
    ]

    corr_matrix = X_train_fold[
        non_constant_features
    ].corr().abs()

    upper = corr_matrix.where(
        np.triu(
            np.ones(corr_matrix.shape),
            k=1
        ).astype(bool)
    )

    features_to_remove = [
        col for col in upper.columns
        if any(upper[col] >= 0.95)
    ]

    selected_features = [
        col for col in non_constant_features
        if col not in features_to_remove
    ]

    print("Selected features:", len(selected_features))

    X_train_selected = X_train_fold[selected_features]
    X_test_selected = X_test_fold[selected_features]

    # ==========================================
    # 3. IMPUTATION
    # ==========================================

    fold_imputer = SimpleImputer(strategy="median")

    X_train_imp = fold_imputer.fit_transform(
        X_train_selected
    )

    X_test_imp = fold_imputer.transform(
        X_test_selected
    )

    # ==========================================
    # 4. RANDOM FOREST
    # ==========================================

    rf = RandomForestClassifier(
        n_estimators=100,
        max_depth=20,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=1
    )

    start = time.perf_counter()

    rf.fit(X_train_imp, y_train_fold)

    rf_time = time.perf_counter() - start

    rf_probs = rf.predict_proba(X_test_imp)[:, 1]

    # ==========================================
    # 5. BALANCED XGBOOST
    # ==========================================

    xgb = XGBClassifier(
        n_estimators=100,
        max_depth=6,
        learning_rate=0.1,
        subsample=0.8,
        colsample_bytree=0.8,
        eval_metric="logloss",
        random_state=42,
        n_jobs=1
    )

    # Balance benign and attack sample contributions
    sample_weights = compute_sample_weight(
        class_weight="balanced",
        y=y_train_fold
    )

    start = time.perf_counter()

    xgb.fit(
        X_train_imp,
        y_train_fold,
        sample_weight=sample_weights
    )

    xgb_time = time.perf_counter() - start

    xgb_probs = xgb.predict_proba(X_test_imp)[:, 1]

    # ==========================================
    # 6. SOFT VOTING
    # ==========================================

    # Equal-weight probability averaging
    voting_probs = (rf_probs + xgb_probs) / 2

    voting_time = rf_time + xgb_time

    # ==========================================
    # 7. EVALUATE ALL THREE APPROACHES
    # ==========================================

    predictions_dict = {
        "Random Forest": (rf_probs >= 0.5).astype(int),
        "XGBoost": (xgb_probs >= 0.5).astype(int),
        "Soft Voting": (voting_probs >= 0.5).astype(int)
    }

    probabilities_dict = {
        "Random Forest": rf_probs,
        "XGBoost": xgb_probs,
        "Soft Voting": voting_probs
    }

    times_dict = {
        "Random Forest": rf_time,
        "XGBoost": xgb_time,
        "Soft Voting": voting_time
    }

    for model_name, predictions in predictions_dict.items():

        probabilities = probabilities_dict[model_name]

        cm = confusion_matrix(
            y_test_fold,
            predictions
        )

        tn, fp, fn, tp = cm.ravel()

        result = {
            "Category": category,
            "Model": model_name,
            "Features": len(selected_features),
            "Accuracy": accuracy_score(
                y_test_fold, predictions
            ),
            "Precision": precision_score(
                y_test_fold, predictions, zero_division=0
            ),
            "Recall": recall_score(
                y_test_fold, predictions, zero_division=0
            ),
            "F1": f1_score(
                y_test_fold, predictions, zero_division=0
            ),
            "Balanced_Accuracy": balanced_accuracy_score(
                y_test_fold, predictions
            ),
            "ROC_AUC": roc_auc_score(
                y_test_fold, probabilities
            ),
            "FP": fp,
            "FN": fn,
            "Training_Time": times_dict[model_name]
        }

        loco_ensemble_results.append(result)

        print("\n", model_name)
        print("Confusion Matrix:\n", cm)
        print("Recall:", round(result["Recall"], 6))
        print("F1:", round(result["F1"], 6))
        print("False Positives:", fp)
        print("False Negatives:", fn)

# ==========================================
# 8. FINAL LOCO COMPARISON
# ==========================================

loco_ensemble_df = pd.DataFrame(
    loco_ensemble_results
)

print("\n" + "=" * 90)
print("FINAL LOCO ENSEMBLE COMPARISON")
print("=" * 90)

display(
    loco_ensemble_df.round(6)
)

print("\nAVERAGE PERFORMANCE BY MODEL")

display(
    loco_ensemble_df.groupby("Model")[
        [
            "Accuracy",
            "Precision",
            "Recall",
            "F1",
            "Balanced_Accuracy",
            "ROC_AUC",
            "Training_Time"
        ]
    ].mean().round(6)
)

print("\nTOTAL FALSE POSITIVES AND FALSE NEGATIVES")

display(
    loco_ensemble_df.groupby("Model")[
        ["FP", "FN"]
    ].sum()
)
```

```python
# ==============================================
# STEP 59: UDP ENSEMBLE COMPLEMENTARITY ANALYSIS
# ==============================================

import numpy as np
import pandas as pd

from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.utils.class_weight import compute_sample_weight
from xgboost import XGBClassifier

from sklearn.metrics import confusion_matrix

# ==========================================
# 1. RECONSTRUCT UDP FOLD
# ==========================================

udp_fold = leave_one_out_folds["DrDoS_UDP"]

attack_train = udp_fold["attack_train"].copy()
benign_train_fold = udp_fold["benign_train"].copy()

attack_train["Target"] = 1
benign_train_fold["Target"] = 0

train_data = pd.concat(
    [attack_train, benign_train_fold],
    ignore_index=True
)

attack_test = udp_fold["attack_test"].copy()
benign_test_fold = udp_fold["benign_test"].copy()

attack_test["Target"] = 1
benign_test_fold["Target"] = 0

test_data = pd.concat(
    [benign_test_fold, attack_test],
    ignore_index=True
)

X_train_udp = train_data[feature_columns].copy()
y_train_udp = train_data["Target"].astype(int)

X_test_udp = test_data[feature_columns].copy()
y_test_udp = test_data["Target"].astype(int)

X_train_udp = X_train_udp.replace(
    [np.inf, -np.inf], np.nan
)

X_test_udp = X_test_udp.replace(
    [np.inf, -np.inf], np.nan
)

# ==========================================
# 2. SELECT FEATURES USING TRAINING DATA
# ==========================================

non_constant_udp = [
    col for col in feature_columns
    if X_train_udp[col].nunique(dropna=True) > 1
]

udp_corr = X_train_udp[non_constant_udp].corr().abs()

udp_upper = udp_corr.where(
    np.triu(
        np.ones(udp_corr.shape),
        k=1
    ).astype(bool)
)

udp_remove = [
    col for col in udp_upper.columns
    if any(udp_upper[col] >= 0.95)
]

udp_features = [
    col for col in non_constant_udp
    if col not in udp_remove
]

X_train_udp = X_train_udp[udp_features]
X_test_udp = X_test_udp[udp_features]

# ==========================================
# 3. IMPUTATION
# ==========================================

udp_imputer = SimpleImputer(strategy="median")

X_train_udp_imp = udp_imputer.fit_transform(X_train_udp)
X_test_udp_imp = udp_imputer.transform(X_test_udp)

# ==========================================
# 4. TRAIN RANDOM FOREST
# ==========================================

udp_rf = RandomForestClassifier(
    n_estimators=100,
    max_depth=20,
    min_samples_leaf=2,
    class_weight="balanced",
    random_state=42,
    n_jobs=1
)

udp_rf.fit(X_train_udp_imp, y_train_udp)

rf_probs = udp_rf.predict_proba(
    X_test_udp_imp
)[:, 1]

# ==========================================
# 5. TRAIN BALANCED XGBOOST
# ==========================================

udp_xgb = XGBClassifier(
    n_estimators=100,
    max_depth=6,
    learning_rate=0.1,
    subsample=0.8,
    colsample_bytree=0.8,
    eval_metric="logloss",
    random_state=42,
    n_jobs=1
)

udp_weights = compute_sample_weight(
    class_weight="balanced",
    y=y_train_udp
)

udp_xgb.fit(
    X_train_udp_imp,
    y_train_udp,
    sample_weight=udp_weights
)

xgb_probs = udp_xgb.predict_proba(
    X_test_udp_imp
)[:, 1]

# ==========================================
# 6. GENERATE PREDICTIONS
# ==========================================

rf_pred = (rf_probs >= 0.5).astype(int)
xgb_pred = (xgb_probs >= 0.5).astype(int)

soft_probs = (rf_probs + xgb_probs) / 2

soft_pred = (soft_probs >= 0.5).astype(int)

# ==========================================
# 7. ATTACK PREDICTION AGREEMENT
# ==========================================

attack_mask = y_test_udp.to_numpy() == 1

rf_attack = rf_pred[attack_mask]
xgb_attack = xgb_pred[attack_mask]
soft_attack = soft_pred[attack_mask]

agreement = pd.crosstab(
    pd.Series(rf_attack, name="Random_Forest"),
    pd.Series(xgb_attack, name="XGBoost")
)

print("=" * 70)
print("UDP ATTACK PREDICTION AGREEMENT")
print("=" * 70)

print("\nRows = Random Forest")
print("Columns = XGBoost")
print(agreement)

# ==========================================
# 8. SOFT VOTING RECOVERY ANALYSIS
# ==========================================

both_detect = np.sum(
    (rf_attack == 1) & (xgb_attack == 1)
)

rf_only = np.sum(
    (rf_attack == 1) & (xgb_attack == 0)
)

xgb_only = np.sum(
    (rf_attack == 0) & (xgb_attack == 1)
)

both_miss = np.sum(
    (rf_attack == 0) & (xgb_attack == 0)
)

soft_recovered = np.sum(
    (soft_attack == 1) &
    (rf_attack == 1) &
    (xgb_attack == 0)
)

print("\n" + "=" * 70)
print("ATTACK DETECTION COMPLEMENTARITY")
print("=" * 70)

print("Both detect:", both_detect)
print("Only Random Forest detects:", rf_only)
print("Only XGBoost detects:", xgb_only)
print("Both miss:", both_miss)
print("Attacks recovered by soft voting:", soft_recovered)

# ==========================================
# 9. PROBABILITY ANALYSIS OF XGBOOST MISSES
# ==========================================

xgb_missed = attack_mask & (xgb_pred == 0)

missed_analysis = pd.DataFrame({
    "RF_Probability": rf_probs[xgb_missed],
    "XGB_Probability": xgb_probs[xgb_missed],
    "Soft_Probability": soft_probs[xgb_missed],
    "Soft_Prediction": soft_pred[xgb_missed]
})

print("\n" + "=" * 70)
print("ANALYSIS OF XGBOOST MISSED UDP ATTACKS")
print("=" * 70)

print("Total XGBoost missed attacks:", len(missed_analysis))

print("\nProbability summary:")
display(missed_analysis.describe().round(6))

print("\nSoft voting predictions for XGBoost missed attacks:")
print(
    missed_analysis["Soft_Prediction"].value_counts()
)
```

```python
# ==============================================
# STEP 60: FOUR-CATEGORY ENSEMBLE COMPLEMENTARITY
# ==============================================

import numpy as np
import pandas as pd

complementarity_results = []

for category, fold in leave_one_out_folds.items():

    print("\n" + "=" * 70)
    print("CATEGORY:", category)
    print("=" * 70)

    # ------------------------------------------
    # Reconstruct fold datasets
    # ------------------------------------------

    attack_train = fold["attack_train"].copy()
    benign_train_fold = fold["benign_train"].copy()

    attack_train["Target"] = 1
    benign_train_fold["Target"] = 0

    train_data = pd.concat(
        [attack_train, benign_train_fold],
        ignore_index=True
    )

    attack_test = fold["attack_test"].copy()
    benign_test_fold = fold["benign_test"].copy()

    attack_test["Target"] = 1
    benign_test_fold["Target"] = 0

    test_data = pd.concat(
        [benign_test_fold, attack_test],
        ignore_index=True
    )

    X_train_fold = train_data[feature_columns].copy()
    y_train_fold = train_data["Target"].astype(int)

    X_test_fold = test_data[feature_columns].copy()
    y_test_fold = test_data["Target"].astype(int)

    X_train_fold = X_train_fold.replace(
        [np.inf, -np.inf], np.nan
    )

    X_test_fold = X_test_fold.replace(
        [np.inf, -np.inf], np.nan
    )

    # ------------------------------------------
    # Fold-specific feature selection
    # ------------------------------------------

    non_constant = [
        col for col in feature_columns
        if X_train_fold[col].nunique(dropna=True) > 1
    ]

    corr_matrix = X_train_fold[non_constant].corr().abs()

    upper = corr_matrix.where(
        np.triu(
            np.ones(corr_matrix.shape),
            k=1
        ).astype(bool)
    )

    remove_features = [
        col for col in upper.columns
        if any(upper[col] >= 0.95)
    ]

    selected_features = [
        col for col in non_constant
        if col not in remove_features
    ]

    X_train_selected = X_train_fold[selected_features]
    X_test_selected = X_test_fold[selected_features]

    # ------------------------------------------
    # Imputation
    # ------------------------------------------

    imputer = SimpleImputer(strategy="median")

    X_train_imp = imputer.fit_transform(X_train_selected)
    X_test_imp = imputer.transform(X_test_selected)

    # ------------------------------------------
    # Random Forest
    # ------------------------------------------

    rf = RandomForestClassifier(
        n_estimators=100,
        max_depth=20,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=1
    )

    rf.fit(X_train_imp, y_train_fold)

    rf_probs = rf.predict_proba(X_test_imp)[:, 1]

    rf_predictions = (
        rf_probs >= 0.5
    ).astype(int)

    # ------------------------------------------
    # XGBoost
    # ------------------------------------------

    xgb = XGBClassifier(
        n_estimators=100,
        max_depth=6,
        learning_rate=0.1,
        subsample=0.8,
        colsample_bytree=0.8,
        eval_metric="logloss",
        random_state=42,
        n_jobs=1
    )

    weights = compute_sample_weight(
        class_weight="balanced",
        y=y_train_fold
    )

    xgb.fit(
        X_train_imp,
        y_train_fold,
        sample_weight=weights
    )

    xgb_probs = xgb.predict_proba(X_test_imp)[:, 1]

    xgb_predictions = (
        xgb_probs >= 0.5
    ).astype(int)

    # ------------------------------------------
    # Soft voting
    # ------------------------------------------

    soft_probs = (rf_probs + xgb_probs) / 2

    soft_predictions = (
        soft_probs >= 0.5
    ).astype(int)

    # ------------------------------------------
    # Attack-only complementarity
    # ------------------------------------------

    attack_mask = y_test_fold.to_numpy() == 1

    rf_attack = rf_predictions[attack_mask]
    xgb_attack = xgb_predictions[attack_mask]
    soft_attack = soft_predictions[attack_mask]

    both_detect = np.sum(
        (rf_attack == 1) & (xgb_attack == 1)
    )

    rf_only = np.sum(
        (rf_attack == 1) & (xgb_attack == 0)
    )

    xgb_only = np.sum(
        (rf_attack == 0) & (xgb_attack == 1)
    )

    both_miss = np.sum(
        (rf_attack == 0) & (xgb_attack == 0)
    )

    soft_recovered = np.sum(
        (soft_attack == 1) &
        (rf_attack == 1) &
        (xgb_attack == 0)
    )

    soft_total_missed = np.sum(soft_attack == 0)

    result = {
        "Category": category,
        "Features": len(selected_features),
        "Both_Detect": both_detect,
        "RF_Only": rf_only,
        "XGB_Only": xgb_only,
        "Both_Miss": both_miss,
        "RF_Only_Recovered_By_Soft": soft_recovered,
        "Soft_Voting_Missed": soft_total_missed
    }

    complementarity_results.append(result)

    print("Selected features:", len(selected_features))
    print("Both detect:", both_detect)
    print("Only RF detects:", rf_only)
    print("Only XGBoost detects:", xgb_only)
    print("Both miss:", both_miss)
    print("Recovered by soft voting:", soft_recovered)
    print("Soft voting missed:", soft_total_missed)

# ==========================================
# FINAL COMPLEMENTARITY SUMMARY
# ==========================================

complementarity_df = pd.DataFrame(
    complementarity_results
)

print("\n" + "=" * 85)
print("FOUR-CATEGORY COMPLEMENTARITY SUMMARY")
print("=" * 85)

display(
    complementarity_df
)
```

```python
# ==============================================
# STEP 61: HARD VOTING VS SOFT VOTING
# ==============================================

import numpy as np
import pandas as pd
import time

from sklearn.ensemble import RandomForestClassifier
from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    confusion_matrix
)

from xgboost import XGBClassifier
from sklearn.utils.class_weight import compute_sample_weight

voting_results = []

for category, fold in leave_one_out_folds.items():

    print("\nProcessing:", category)

    # ------------------------------------------
    # Prepare training and test data
    # ------------------------------------------

    attack_train = fold["attack_train"].copy()
    benign_train_fold = fold["benign_train"].copy()

    attack_train["Target"] = 1
    benign_train_fold["Target"] = 0

    train_data = pd.concat(
        [attack_train, benign_train_fold],
        ignore_index=True
    )

    attack_test = fold["attack_test"].copy()
    benign_test_fold = fold["benign_test"].copy()

    attack_test["Target"] = 1
    benign_test_fold["Target"] = 0

    test_data = pd.concat(
        [benign_test_fold, attack_test],
        ignore_index=True
    )

    X_train = train_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    )

    y_train = train_data["Target"].astype(int)

    X_test = test_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    )

    y_test = test_data["Target"].astype(int)

    # ------------------------------------------
    # Training-only feature selection
    # ------------------------------------------

    non_constant = [
        col for col in feature_columns
        if X_train[col].nunique(dropna=True) > 1
    ]

    corr_matrix = X_train[non_constant].corr().abs()

    upper = corr_matrix.where(
        np.triu(
            np.ones(corr_matrix.shape),
            k=1
        ).astype(bool)
    )

    remove_features = [
        col for col in upper.columns
        if any(upper[col] >= 0.95)
    ]

    selected_features = [
        col for col in non_constant
        if col not in remove_features
    ]

    X_train = X_train[selected_features]
    X_test = X_test[selected_features]

    # ------------------------------------------
    # Imputation
    # ------------------------------------------

    imputer = SimpleImputer(strategy="median")

    X_train_imp = imputer.fit_transform(X_train)
    X_test_imp = imputer.transform(X_test)

    # ------------------------------------------
    # Random Forest
    # ------------------------------------------

    rf = RandomForestClassifier(
        n_estimators=100,
        max_depth=20,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=1
    )

    rf.fit(X_train_imp, y_train)

    rf_probs = rf.predict_proba(X_test_imp)[:, 1]
    rf_pred = (rf_probs >= 0.5).astype(int)

    # ------------------------------------------
    # XGBoost
    # ------------------------------------------

    xgb = XGBClassifier(
        n_estimators=100,
        max_depth=6,
        learning_rate=0.1,
        subsample=0.8,
        colsample_bytree=0.8,
        eval_metric="logloss",
        random_state=42,
        n_jobs=1
    )

    weights = compute_sample_weight(
        class_weight="balanced",
        y=y_train
    )

    xgb.fit(
        X_train_imp,
        y_train,
        sample_weight=weights
    )

    xgb_probs = xgb.predict_proba(X_test_imp)[:, 1]
    xgb_pred = (xgb_probs >= 0.5).astype(int)

    # ------------------------------------------
    # HARD VOTING
    # Tie-breaking rule: Random Forest
    # ------------------------------------------

    hard_pred = np.where(
        rf_pred == xgb_pred,
        rf_pred,
        rf_pred
    )

    # ------------------------------------------
    # SOFT VOTING
    # ------------------------------------------

    soft_probs = (rf_probs + xgb_probs) / 2
    soft_pred = (soft_probs >= 0.5).astype(int)

    # ------------------------------------------
    # Evaluate all four approaches
    # ------------------------------------------

    models = {
        "Random Forest": rf_pred,
        "XGBoost": xgb_pred,
        "Hard Voting": hard_pred,
        "Soft Voting": soft_pred
    }

    for model_name, predictions in models.items():

        tn, fp, fn, tp = confusion_matrix(
            y_test,
            predictions,
            labels=[0, 1]
        ).ravel()

        voting_results.append({
            "Category": category,
            "Model": model_name,
            "Features": len(selected_features),
            "Accuracy": accuracy_score(y_test, predictions),
            "Precision": precision_score(
                y_test, predictions, zero_division=0
            ),
            "Recall": recall_score(
                y_test, predictions, zero_division=0
            ),
            "F1": f1_score(
                y_test, predictions, zero_division=0
            ),
            "Balanced_Accuracy": balanced_accuracy_score(
                y_test, predictions
            ),
            "FP": fp,
            "FN": fn,
            "TP": tp,
            "TN": tn
        })

    print("Selected features:", len(selected_features))
    print("RF false negatives:", np.sum((y_test == 1) & (rf_pred == 0)))
    print("XGB false negatives:", np.sum((y_test == 1) & (xgb_pred == 0)))
    print("Hard voting false negatives:", np.sum((y_test == 1) & (hard_pred == 0)))
    print("Soft voting false negatives:", np.sum((y_test == 1) & (soft_pred == 0)))


# ==============================================
# FINAL RESULTS
# ==============================================

voting_df = pd.DataFrame(voting_results)

print("\n" + "=" * 85)
print("HARD VOTING VS SOFT VOTING — CATEGORY-WISE RESULTS")
print("=" * 85)

display(
    voting_df[
        [
            "Category", "Model", "Accuracy",
            "Precision", "Recall", "F1",
            "Balanced_Accuracy", "FP", "FN"
        ]
    ].round(6)
)

# ==============================================
# POOLED RESULTS ACROSS ALL FOUR FOLDS
# ==============================================

pooled_results = []

for model_name in voting_df["Model"].unique():

    subset = voting_df[
        voting_df["Model"] == model_name
    ]

    total_tp = subset["TP"].sum()
    total_tn = subset["TN"].sum()
    total_fp = subset["FP"].sum()
    total_fn = subset["FN"].sum()

    pooled_results.append({
        "Model": model_name,
        "Accuracy": (total_tp + total_tn) /
                    (total_tp + total_tn + total_fp + total_fn),

        "Precision": total_tp /
                     (total_tp + total_fp),

        "Recall": total_tp /
                  (total_tp + total_fn),

        "F1": 2 * total_tp /
              (2 * total_tp + total_fp + total_fn),

        "Balanced_Accuracy": (
            total_tp / (total_tp + total_fn) +
            total_tn / (total_tn + total_fp)
        ) / 2,

        "Total_FP": total_fp,
        "Total_FN": total_fn
    })

pooled_df = pd.DataFrame(pooled_results)

print("\n" + "=" * 85)
print("POOLED RESULTS — ALL FOUR ATTACK CATEGORIES")
print("=" * 85)

display(pooled_df.round(6))
```

```python
# ==============================================
# STEP 62: THREE-CLASSIFIER VOTING COMPARISON
# ==============================================

import numpy as np
import pandas as pd

from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier
)

from sklearn.impute import SimpleImputer
from sklearn.metrics import confusion_matrix, accuracy_score
from xgboost import XGBClassifier
from sklearn.utils.class_weight import compute_sample_weight

three_model_results = []

for category, fold in leave_one_out_folds.items():

    print("\nProcessing:", category)

    # Prepare data
    attack_train = fold["attack_train"].copy()
    benign_train_fold = fold["benign_train"].copy()

    attack_train["Target"] = 1
    benign_train_fold["Target"] = 0

    train_data = pd.concat(
        [attack_train, benign_train_fold],
        ignore_index=True
    )

    attack_test = fold["attack_test"].copy()
    benign_test_fold = fold["benign_test"].copy()

    attack_test["Target"] = 1
    benign_test_fold["Target"] = 0

    test_data = pd.concat(
        [benign_test_fold, attack_test],
        ignore_index=True
    )

    X_train = train_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    )
    y_train = train_data["Target"].astype(int)

    X_test = test_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    )
    y_test = test_data["Target"].astype(int)

    # Training-only feature selection
    non_constant = [
        col for col in feature_columns
        if X_train[col].nunique(dropna=True) > 1
    ]

    corr_matrix = X_train[non_constant].corr().abs()

    upper = corr_matrix.where(
        np.triu(
            np.ones(corr_matrix.shape),
            k=1
        ).astype(bool)
    )

    remove_features = [
        col for col in upper.columns
        if any(upper[col] >= 0.95)
    ]

    selected_features = [
        col for col in non_constant
        if col not in remove_features
    ]

    X_train = X_train[selected_features]
    X_test = X_test[selected_features]

    # Imputation
    imputer = SimpleImputer(strategy="median")

    X_train_imp = imputer.fit_transform(X_train)
    X_test_imp = imputer.transform(X_test)

    # Random Forest
    rf = RandomForestClassifier(
        n_estimators=100,
        max_depth=20,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=1
    )

    rf.fit(X_train_imp, y_train)

    # Extra Trees
    et = ExtraTreesClassifier(
        n_estimators=100,
        max_depth=20,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=1
    )

    et.fit(X_train_imp, y_train)

    # XGBoost
    xgb = XGBClassifier(
        n_estimators=100,
        max_depth=6,
        learning_rate=0.1,
        subsample=0.8,
        colsample_bytree=0.8,
        eval_metric="logloss",
        random_state=42,
        n_jobs=1
    )

    weights = compute_sample_weight(
        class_weight="balanced",
        y=y_train
    )

    xgb.fit(
        X_train_imp,
        y_train,
        sample_weight=weights
    )

    # Individual probabilities
    rf_prob = rf.predict_proba(X_test_imp)[:, 1]
    et_prob = et.predict_proba(X_test_imp)[:, 1]
    xgb_prob = xgb.predict_proba(X_test_imp)[:, 1]

    # Individual predictions
    rf_pred = (rf_prob >= 0.5).astype(int)
    et_pred = (et_prob >= 0.5).astype(int)
    xgb_pred = (xgb_prob >= 0.5).astype(int)

    # Hard majority voting: at least 2 of 3 votes
    hard_pred = (
        rf_pred + et_pred + xgb_pred >= 2
    ).astype(int)

    # Soft voting: average probabilities
    soft_prob = (
        rf_prob + et_prob + xgb_prob
    ) / 3

    soft_pred = (soft_prob >= 0.5).astype(int)

    # Evaluate
    predictions = {
        "Random Forest": rf_pred,
        "Extra Trees": et_pred,
        "XGBoost": xgb_pred,
        "Hard Majority Voting": hard_pred,
        "Soft Voting": soft_pred
    }

    for model_name, pred in predictions.items():

        tn, fp, fn, tp = confusion_matrix(
            y_test,
            pred,
            labels=[0, 1]
        ).ravel()

        three_model_results.append({
            "Category": category,
            "Model": model_name,
            "Features": len(selected_features),
            "Accuracy": accuracy_score(y_test, pred),
            "TP": tp,
            "TN": tn,
            "FP": fp,
            "FN": fn
        })

    print("Features:", len(selected_features))
    print("RF FN:", np.sum((y_test == 1) & (rf_pred == 0)))
    print("ET FN:", np.sum((y_test == 1) & (et_pred == 0)))
    print("XGB FN:", np.sum((y_test == 1) & (xgb_pred == 0)))
    print("Hard voting FN:", np.sum((y_test == 1) & (hard_pred == 0)))
    print("Soft voting FN:", np.sum((y_test == 1) & (soft_pred == 0)))


# Results
three_model_df = pd.DataFrame(three_model_results)

print("\nCATEGORY-WISE RESULTS")

display(three_model_df.round(6))

# Pooled results
pooled = []

for model_name in three_model_df["Model"].unique():

    subset = three_model_df[
        three_model_df["Model"] == model_name
    ]

    tp = subset["TP"].sum()
    tn = subset["TN"].sum()
    fp = subset["FP"].sum()
    fn = subset["FN"].sum()

    pooled.append({
        "Model": model_name,
        "Accuracy": (tp + tn) / (tp + tn + fp + fn),
        "Precision": tp / (tp + fp),
        "Recall": tp / (tp + fn),
        "F1": 2 * tp / (2 * tp + fp + fn),
        "False Positives": fp,
        "False Negatives": fn
    })

pooled_three_model_df = pd.DataFrame(pooled)

print("\nPOOLED THREE-CLASSIFIER COMPARISON")

display(pooled_three_model_df.round(6))
```

```python
# ==============================================
# STEP 63: MODEL STABILITY ACROSS RANDOM SEEDS
# ==============================================

import numpy as np
import pandas as pd

from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier
)

from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    recall_score,
    f1_score,
    confusion_matrix
)

from xgboost import XGBClassifier
from sklearn.utils.class_weight import compute_sample_weight

seeds = [21, 42, 55, 72, 99]

stability_results = []

for category, fold in leave_one_out_folds.items():

    print("\nProcessing:", category)

    # Prepare training data
    attack_train = fold["attack_train"].copy()
    benign_train_fold = fold["benign_train"].copy()

    attack_train["Target"] = 1
    benign_train_fold["Target"] = 0

    train_data = pd.concat(
        [attack_train, benign_train_fold],
        ignore_index=True
    )

    # Prepare test data
    attack_test = fold["attack_test"].copy()
    benign_test_fold = fold["benign_test"].copy()

    attack_test["Target"] = 1
    benign_test_fold["Target"] = 0

    test_data = pd.concat(
        [benign_test_fold, attack_test],
        ignore_index=True
    )

    X_train = train_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    )

    y_train = train_data["Target"].astype(int)

    X_test = test_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    )

    y_test = test_data["Target"].astype(int)

    # Training-only feature selection
    non_constant = [
        col for col in feature_columns
        if X_train[col].nunique(dropna=True) > 1
    ]

    corr_matrix = X_train[non_constant].corr().abs()

    upper = corr_matrix.where(
        np.triu(
            np.ones(corr_matrix.shape),
            k=1
        ).astype(bool)
    )

    remove_features = [
        col for col in upper.columns
        if any(upper[col] >= 0.95)
    ]

    selected_features = [
        col for col in non_constant
        if col not in remove_features
    ]

    X_train = X_train[selected_features]
    X_test = X_test[selected_features]

    # Fit imputer only on training data
    imputer = SimpleImputer(strategy="median")

    X_train_imp = imputer.fit_transform(X_train)
    X_test_imp = imputer.transform(X_test)

    weights = compute_sample_weight(
        class_weight="balanced",
        y=y_train
    )

    # Train using different random seeds
    for seed in seeds:

        models = {
            "Random Forest": RandomForestClassifier(
                n_estimators=100,
                max_depth=20,
                min_samples_leaf=2,
                class_weight="balanced",
                random_state=seed,
                n_jobs=1
            ),

            "Extra Trees": ExtraTreesClassifier(
                n_estimators=100,
                max_depth=20,
                min_samples_leaf=2,
                class_weight="balanced",
                random_state=seed,
                n_jobs=1
            ),

            "XGBoost": XGBClassifier(
                n_estimators=100,
                max_depth=6,
                learning_rate=0.1,
                subsample=0.8,
                colsample_bytree=0.8,
                eval_metric="logloss",
                random_state=seed,
                n_jobs=1
            )
        }

        probabilities = {}

        for name, model in models.items():

            if name == "XGBoost":
                model.fit(
                    X_train_imp,
                    y_train,
                    sample_weight=weights
                )
            else:
                model.fit(X_train_imp, y_train)

            probabilities[name] = (
                model.predict_proba(X_test_imp)[:, 1]
            )

        # Individual predictions
        rf_pred = (
            probabilities["Random Forest"] >= 0.5
        ).astype(int)

        et_pred = (
            probabilities["Extra Trees"] >= 0.5
        ).astype(int)

        xgb_pred = (
            probabilities["XGBoost"] >= 0.5
        ).astype(int)

        # Three-classifier hard majority voting
        hard_pred = (
            rf_pred + et_pred + xgb_pred >= 2
        ).astype(int)

        # Three-classifier soft voting
        soft_prob = np.mean(
            list(probabilities.values()),
            axis=0
        )

        soft_pred = (
            soft_prob >= 0.5
        ).astype(int)

        predictions = {
            "Random Forest": rf_pred,
            "Extra Trees": et_pred,
            "XGBoost": xgb_pred,
            "Hard Majority Voting": hard_pred,
            "Soft Voting": soft_pred
        }

        for model_name, pred in predictions.items():

            tn, fp, fn, tp = confusion_matrix(
                y_test,
                pred,
                labels=[0, 1]
            ).ravel()

            stability_results.append({
                "Category": category,
                "Seed": seed,
                "Model": model_name,
                "Recall": recall_score(y_test, pred),
                "F1": f1_score(y_test, pred),
                "FP": fp,
                "FN": fn
            })

    print("Completed:", category)


# ==============================================
# STABILITY SUMMARY
# ==============================================

stability_df = pd.DataFrame(stability_results)

stability_summary = (
    stability_df
    .groupby("Model")
    .agg(
        Mean_Recall=("Recall", "mean"),
        Std_Recall=("Recall", "std"),
        Mean_F1=("F1", "mean"),
        Std_F1=("F1", "std"),
        Mean_FP=("FP", "mean"),
        Mean_FN=("FN", "mean"),
        Min_FN=("FN", "min"),
        Max_FN=("FN", "max")
    )
    .reset_index()
)

print("\n" + "=" * 85)
print("MODEL STABILITY ACROSS FIVE RANDOM SEEDS")
print("=" * 85)

display(stability_summary.round(6))


# Category-wise stability
print("\nCATEGORY-WISE MEAN FALSE NEGATIVES")

category_stability = (
    stability_df
    .groupby(["Category", "Model"])
    .agg(
        Mean_FN=("FN", "mean"),
        Std_FN=("FN", "std"),
        Mean_Recall=("Recall", "mean")
    )
    .reset_index()
)

display(category_stability.round(6))
```

```python
# ==============================================
# STEP 64: PAIRED STATISTICAL COMPARISON
# ==============================================

import numpy as np
import pandas as pd

from scipy.stats import wilcoxon

# ------------------------------------------
# Prepare paired results
# ------------------------------------------

models_to_compare = [
    "Hard Majority Voting",
    "Soft Voting"
]

statistical_results = []

for model_name in models_to_compare:

    comparison = stability_df[
        stability_df["Model"].isin(
            ["Extra Trees", model_name]
        )
    ]

    recall_pivot = comparison.pivot(
        index=["Category", "Seed"],
        columns="Model",
        values="Recall"
    )

    f1_pivot = comparison.pivot(
        index=["Category", "Seed"],
        columns="Model",
        values="F1"
    )

    fn_pivot = comparison.pivot(
        index=["Category", "Seed"],
        columns="Model",
        values="FN"
    )

    # Positive difference means Extra Trees is better
    recall_diff = (
        recall_pivot["Extra Trees"] -
        recall_pivot[model_name]
    )

    f1_diff = (
        f1_pivot["Extra Trees"] -
        f1_pivot[model_name]
    )

    fn_diff = (
        fn_pivot[model_name] -
        fn_pivot["Extra Trees"]
    )

    # Wilcoxon signed-rank tests
    recall_test = wilcoxon(
        recall_diff,
        alternative="two-sided"
    )

    f1_test = wilcoxon(
        f1_diff,
        alternative="two-sided"
    )

    fn_test = wilcoxon(
        fn_diff,
        alternative="two-sided"
    )

    statistical_results.append({
        "Comparison": f"Extra Trees vs {model_name}",

        "Mean Recall Difference": recall_diff.mean(),

        "Mean F1 Difference": f1_diff.mean(),

        "Mean FN Reduction": fn_diff.mean(),

        "Recall Wilcoxon p-value": recall_test.pvalue,

        "F1 Wilcoxon p-value": f1_test.pvalue,

        "FN Wilcoxon p-value": fn_test.pvalue
    })


statistical_df = pd.DataFrame(
    statistical_results
)

print("=" * 85)
print("PAIRED STATISTICAL COMPARISON")
print("=" * 85)

display(statistical_df.round(8))
```

```python
# ==============================================
# STEP 65: HOLM MULTIPLE-COMPARISON CORRECTION
# ==============================================

import numpy as np
import pandas as pd
from statsmodels.stats.multitest import multipletests

pvalue_columns = [
    "Recall Wilcoxon p-value",
    "F1 Wilcoxon p-value",
    "FN Wilcoxon p-value"
]

correction_results = []

for comparison in statistical_df["Comparison"]:

    row = statistical_df[
        statistical_df["Comparison"] == comparison
    ].iloc[0]

    for metric in pvalue_columns:

        correction_results.append({
            "Comparison": comparison,
            "Metric": metric.replace(
                " Wilcoxon p-value", ""
            ),
            "Original_p": row[metric]
        })

correction_df = pd.DataFrame(correction_results)

# Apply Holm correction across all six tests
reject, corrected_p, _, _ = multipletests(
    correction_df["Original_p"].values,
    alpha=0.05,
    method="holm"
)

correction_df["Holm_Adjusted_p"] = corrected_p
correction_df["Significant_after_correction"] = reject

print("=" * 85)
print("HOLM MULTIPLE-COMPARISON CORRECTION")
print("=" * 85)

display(correction_df.round(8))
```

```python
# ==============================================
# STEP 66: FINAL SNMP ZERO-DAY MODEL COMPARISON
# ==============================================

import numpy as np
import pandas as pd
import time

from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier
)

from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    roc_auc_score
)

from xgboost import XGBClassifier
from sklearn.utils.class_weight import compute_sample_weight

# ------------------------------------------
# 1. Prepare training and SNMP test data
# ------------------------------------------

X_train_final = X_sup_train.copy()
y_train_final = y_sup_train.copy()

snmp_evaluation = snmp_test_data.copy()

X_snmp = snmp_evaluation[snmp_reduced_features].copy()

y_snmp = (
    snmp_evaluation["Label"]
    .astype(str)
    .ne("BENIGN")
    .astype(int)
)

# Replace infinite values
X_train_final = X_train_final[
    snmp_reduced_features
].replace([np.inf, -np.inf], np.nan)

X_snmp = X_snmp.replace(
    [np.inf, -np.inf], np.nan
)

print("Training samples:", len(X_train_final))
print("SNMP test samples:", len(X_snmp))
print("Selected features:", len(snmp_reduced_features))
print("\nSNMP test distribution:")
print(y_snmp.value_counts())

# ------------------------------------------
# 2. Fit imputer only on training data
# ------------------------------------------

snmp_imputer = SimpleImputer(strategy="median")

X_train_final_imp = snmp_imputer.fit_transform(
    X_train_final
)

X_snmp_imp = snmp_imputer.transform(X_snmp)

# ------------------------------------------
# 3. Define classifiers
# ------------------------------------------

rf = RandomForestClassifier(
    n_estimators=100,
    max_depth=20,
    min_samples_leaf=2,
    class_weight="balanced",
    random_state=42,
    n_jobs=1
)

et = ExtraTreesClassifier(
    n_estimators=100,
    max_depth=20,
    min_samples_leaf=2,
    class_weight="balanced",
    random_state=42,
    n_jobs=1
)

xgb = XGBClassifier(
    n_estimators=100,
    max_depth=6,
    learning_rate=0.1,
    subsample=0.8,
    colsample_bytree=0.8,
    eval_metric="logloss",
    random_state=42,
    n_jobs=1
)

# ------------------------------------------
# 4. Train models
# ------------------------------------------

sample_weights = compute_sample_weight(
    class_weight="balanced",
    y=y_train_final
)

models = {
    "Random Forest": rf,
    "Extra Trees": et,
    "XGBoost": xgb
}

probabilities = {}
training_times = {}

for name, model in models.items():

    start = time.time()

    if name == "XGBoost":
        model.fit(
            X_train_final_imp,
            y_train_final,
            sample_weight=sample_weights
        )
    else:
        model.fit(
            X_train_final_imp,
            y_train_final
        )

    training_times[name] = time.time() - start

    probabilities[name] = (
        model.predict_proba(X_snmp_imp)[:, 1]
    )

    print(
        name,
        "training time:",
        round(training_times[name], 2),
        "seconds"
    )

# ------------------------------------------
# 5. Individual predictions
# ------------------------------------------

rf_pred = (
    probabilities["Random Forest"] >= 0.5
).astype(int)

et_pred = (
    probabilities["Extra Trees"] >= 0.5
).astype(int)

xgb_pred = (
    probabilities["XGBoost"] >= 0.5
).astype(int)

# ------------------------------------------
# 6. Ensemble predictions
# ------------------------------------------

# Three-classifier hard majority voting
hard_pred = (
    rf_pred + et_pred + xgb_pred >= 2
).astype(int)

# Three-classifier soft voting
soft_prob = np.mean(
    list(probabilities.values()),
    axis=0
)

soft_pred = (
    soft_prob >= 0.5
).astype(int)

# ------------------------------------------
# 7. Evaluate all models
# ------------------------------------------

predictions = {
    "Random Forest": (rf_pred, probabilities["Random Forest"]),
    "Extra Trees": (et_pred, probabilities["Extra Trees"]),
    "XGBoost": (xgb_pred, probabilities["XGBoost"]),
    "Hard Majority Voting": (
        hard_pred,
        (rf_pred + et_pred + xgb_pred) / 3
    ),
    "Soft Voting": (soft_pred, soft_prob)
}

snmp_results = []

for name, (pred, prob) in predictions.items():

    tn, fp, fn, tp = confusion_matrix(
        y_snmp,
        pred,
        labels=[0, 1]
    ).ravel()

    snmp_results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_snmp, pred),
        "Precision": precision_score(y_snmp, pred),
        "Recall": recall_score(y_snmp, pred),
        "F1": f1_score(y_snmp, pred),
        "Balanced_Accuracy": balanced_accuracy_score(
            y_snmp, pred
        ),
        "ROC_AUC": roc_auc_score(y_snmp, prob),
        "FP": fp,
        "FN": fn,
        "Training_Time": training_times.get(
            name, sum(training_times.values())
        )
    })

snmp_results_df = pd.DataFrame(snmp_results)

print("\n" + "=" * 90)
print("SNMP ZERO-DAY EVALUATION — FINAL MODEL COMPARISON")
print("=" * 90)

display(snmp_results_df.round(6))
```

```python
# ==============================================
# STEP 67: VALIDATION-BASED THRESHOLD OPTIMISATION
# ==============================================

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer

from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier
)

from sklearn.metrics import confusion_matrix

from xgboost import XGBClassifier
from sklearn.utils.class_weight import compute_sample_weight

# ------------------------------------------
# 1. Training-validation split
# ------------------------------------------

X = X_sup_train[snmp_reduced_features].copy()
y = y_sup_train.copy()

X = X.replace([np.inf, -np.inf], np.nan)

X_train, X_val, y_train, y_val = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Validation samples:", len(X_val))

# ------------------------------------------
# 2. Fit imputer on training subset only
# ------------------------------------------

threshold_imputer = SimpleImputer(strategy="median")

X_train_imp = threshold_imputer.fit_transform(X_train)
X_val_imp = threshold_imputer.transform(X_val)

X_snmp_threshold = snmp_test_data[
    snmp_reduced_features
].replace([np.inf, -np.inf], np.nan)

X_snmp_imp = threshold_imputer.transform(
    X_snmp_threshold
)

y_snmp_true = (
    snmp_test_data["Label"]
    .astype(str)
    .ne("BENIGN")
    .astype(int)
)

# ------------------------------------------
# 3. Define models
# ------------------------------------------

models = {
    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        max_depth=20,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=1
    ),

    "Extra Trees": ExtraTreesClassifier(
        n_estimators=100,
        max_depth=20,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=1
    ),

    "XGBoost": XGBClassifier(
        n_estimators=100,
        max_depth=6,
        learning_rate=0.1,
        subsample=0.8,
        colsample_bytree=0.8,
        eval_metric="logloss",
        random_state=42,
        n_jobs=1
    )
}

weights = compute_sample_weight(
    class_weight="balanced",
    y=y_train
)

threshold_results = []
snmp_threshold_results = []

# ------------------------------------------
# 4. Train and optimise thresholds
# ------------------------------------------

for name, model in models.items():

    print("\nTraining:", name)

    if name == "XGBoost":
        model.fit(
            X_train_imp,
            y_train,
            sample_weight=weights
        )
    else:
        model.fit(
            X_train_imp,
            y_train
        )

    val_probs = model.predict_proba(X_val_imp)[:, 1]

    # Candidate thresholds from validation predictions
    candidates = np.unique(
        np.concatenate([
            np.linspace(0.01, 0.99, 200),
            val_probs
        ])
    )

    best_threshold = 0.5
    best_recall = -1
    best_fpr = 1

    # Maximise recall under validation FPR <= 1%
    for threshold in candidates:

        val_pred = (
            val_probs >= threshold
        ).astype(int)

        tn, fp, fn, tp = confusion_matrix(
            y_val,
            val_pred,
            labels=[0, 1]
        ).ravel()

        fpr = fp / (fp + tn)
        recall = tp / (tp + fn)

        if fpr <= 0.01:

            if recall > best_recall:
                best_recall = recall
                best_fpr = fpr
                best_threshold = threshold

    # --------------------------------------
    # Evaluate on untouched SNMP test data
    # --------------------------------------

    snmp_probs = model.predict_proba(X_snmp_imp)[:, 1]

    for threshold_type, threshold in [
        ("Default", 0.5),
        ("Calibrated", best_threshold)
    ]:

        snmp_pred = (
            snmp_probs >= threshold
        ).astype(int)

        tn, fp, fn, tp = confusion_matrix(
            y_snmp_true,
            snmp_pred,
            labels=[0, 1]
        ).ravel()

        snmp_threshold_results.append({
            "Model": name,
            "Threshold_Type": threshold_type,
            "Threshold": threshold,
            "TP": tp,
            "TN": tn,
            "FP": fp,
            "FN": fn,
            "Recall": tp / (tp + fn),
            "FPR": fp / (fp + tn),
            "Accuracy": (tp + tn) / (tp + tn + fp + fn)
        })

    threshold_results.append({
        "Model": name,
        "Selected_Threshold": best_threshold,
        "Validation_Recall": best_recall,
        "Validation_FPR": best_fpr
    })

# ------------------------------------------
# 5. Display threshold selection
# ------------------------------------------

threshold_df = pd.DataFrame(threshold_results)

print("\n" + "=" * 85)
print("VALIDATION-SELECTED THRESHOLDS")
print("=" * 85)

display(threshold_df.round(6))

# ------------------------------------------
# 6. Display SNMP test results
# ------------------------------------------

snmp_threshold_df = pd.DataFrame(
    snmp_threshold_results
)

print("\n" + "=" * 85)
print("SNMP RESULTS: DEFAULT VS CALIBRATED THRESHOLDS")
print("=" * 85)

display(snmp_threshold_df.round(6))
```

```python
# ==============================================
# STEP 68: THRESHOLD TRANSFERABILITY ACROSS LOCO
# ==============================================

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer

from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier
)

from sklearn.metrics import confusion_matrix

from xgboost import XGBClassifier
from sklearn.utils.class_weight import compute_sample_weight

threshold_transfer_results = []

for category, fold in leave_one_out_folds.items():

    print("\nProcessing:", category)

    # ------------------------------------------
    # Prepare fold data
    # ------------------------------------------

    attack_train = fold["attack_train"].copy()
    benign_train = fold["benign_train"].copy()

    attack_train["Target"] = 1
    benign_train["Target"] = 0

    train_data = pd.concat(
        [attack_train, benign_train],
        ignore_index=True
    )

    attack_test = fold["attack_test"].copy()
    benign_test = fold["benign_test"].copy()

    attack_test["Target"] = 1
    benign_test["Target"] = 0

    test_data = pd.concat(
        [benign_test, attack_test],
        ignore_index=True
    )

    X = train_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    )

    y = train_data["Target"].astype(int)

    X_test = test_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    )

    y_test = test_data["Target"].astype(int)

    # ------------------------------------------
    # Training-only feature selection
    # ------------------------------------------

    non_constant = [
        col for col in feature_columns
        if X[col].nunique(dropna=True) > 1
    ]

    corr_matrix = X[non_constant].corr().abs()

    upper = corr_matrix.where(
        np.triu(
            np.ones(corr_matrix.shape),
            k=1
        ).astype(bool)
    )

    remove_features = [
        col for col in upper.columns
        if any(upper[col] >= 0.95)
    ]

    selected = [
        col for col in non_constant
        if col not in remove_features
    ]

    X = X[selected]
    X_test = X_test[selected]

    # ------------------------------------------
    # Internal training-validation split
    # ------------------------------------------

    X_train, X_val, y_train, y_val = train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )

    imputer = SimpleImputer(strategy="median")

    X_train_imp = imputer.fit_transform(X_train)
    X_val_imp = imputer.transform(X_val)
    X_test_imp = imputer.transform(X_test)

    # ------------------------------------------
    # Define classifiers
    # ------------------------------------------

    models = {
        "Random Forest": RandomForestClassifier(
            n_estimators=100,
            max_depth=20,
            min_samples_leaf=2,
            class_weight="balanced",
            random_state=42,
            n_jobs=1
        ),

        "Extra Trees": ExtraTreesClassifier(
            n_estimators=100,
            max_depth=20,
            min_samples_leaf=2,
            class_weight="balanced",
            random_state=42,
            n_jobs=1
        ),

        "XGBoost": XGBClassifier(
            n_estimators=100,
            max_depth=6,
            learning_rate=0.1,
            subsample=0.8,
            colsample_bytree=0.8,
            eval_metric="logloss",
            random_state=42,
            n_jobs=1
        )
    }

    weights = compute_sample_weight(
        class_weight="balanced",
        y=y_train
    )

    # ------------------------------------------
    # Train and evaluate
    # ------------------------------------------

    for name, model in models.items():

        if name == "XGBoost":
            model.fit(
                X_train_imp,
                y_train,
                sample_weight=weights
            )
        else:
            model.fit(X_train_imp, y_train)

        val_probs = model.predict_proba(X_val_imp)[:, 1]

        test_probs = model.predict_proba(X_test_imp)[:, 1]

        # Select threshold using validation only
        candidates = np.unique(
            np.concatenate([
                np.linspace(0.001, 0.99, 300),
                val_probs
            ])
        )

        best_threshold = 0.5
        best_recall = -1
        best_fpr = 1

        for threshold in candidates:

            val_pred = (
                val_probs >= threshold
            ).astype(int)

            tn, fp, fn, tp = confusion_matrix(
                y_val,
                val_pred,
                labels=[0, 1]
            ).ravel()

            val_fpr = fp / (fp + tn)
            val_recall = tp / (tp + fn)

            if val_fpr <= 0.01:

                if val_recall > best_recall:
                    best_recall = val_recall
                    best_fpr = val_fpr
                    best_threshold = threshold

        # Evaluate default and calibrated thresholds
        for threshold_type, threshold in [
            ("Default", 0.5),
            ("Calibrated", best_threshold)
        ]:

            test_pred = (
                test_probs >= threshold
            ).astype(int)

            tn, fp, fn, tp = confusion_matrix(
                y_test,
                test_pred,
                labels=[0, 1]
            ).ravel()

            threshold_transfer_results.append({
                "Category": category,
                "Model": name,
                "Threshold_Type": threshold_type,
                "Threshold": threshold,
                "Validation_Recall": best_recall,
                "Validation_FPR": best_fpr,
                "Test_Recall": tp / (tp + fn),
                "Test_FPR": fp / (fp + tn),
                "Test_FP": fp,
                "Test_FN": fn
            })

        print(
            name,
            "| Threshold:",
            round(best_threshold, 5),
            "| Validation recall:",
            round(best_recall, 5),
            "| Test FN:",
            int(
                np.sum(
                    (y_test.to_numpy() == 1) &
                    (test_probs < best_threshold)
                )
            )
        )


# ------------------------------------------
# Final results
# ------------------------------------------

transfer_df = pd.DataFrame(
    threshold_transfer_results
)

print("\n" + "=" * 90)
print("LOCO THRESHOLD TRANSFERABILITY RESULTS")
print("=" * 90)

display(transfer_df.round(6))

# ------------------------------------------
# Aggregate comparison
# ------------------------------------------

aggregate_transfer = (
    transfer_df
    .groupby(["Model", "Threshold_Type"])
    .agg(
        Mean_Test_Recall=("Test_Recall", "mean"),
        Mean_Test_FPR=("Test_FPR", "mean"),
        Total_FP=("Test_FP", "sum"),
        Total_FN=("Test_FN", "sum")
    )
    .reset_index()
)

print("\nAGGREGATED THRESHOLD COMPARISON")

display(aggregate_transfer.round(6))
```

```python
# ==============================================
# STEP 69: CALIBRATED EXTRA TREES VS SOFT VOTING
# ==============================================

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer

from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier
)

from sklearn.metrics import confusion_matrix

from xgboost import XGBClassifier
from sklearn.utils.class_weight import compute_sample_weight

final_comparison_results = []

for category, fold in leave_one_out_folds.items():

    print("\nProcessing:", category)

    # ------------------------------------------
    # Prepare fold data
    # ------------------------------------------

    attack_train = fold["attack_train"].copy()
    benign_train = fold["benign_train"].copy()

    attack_train["Target"] = 1
    benign_train["Target"] = 0

    train_data = pd.concat(
        [attack_train, benign_train],
        ignore_index=True
    )

    attack_test = fold["attack_test"].copy()
    benign_test = fold["benign_test"].copy()

    attack_test["Target"] = 1
    benign_test["Target"] = 0

    test_data = pd.concat(
        [benign_test, attack_test],
        ignore_index=True
    )

    X = train_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    )

    y = train_data["Target"].astype(int)

    X_test = test_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    )

    y_test = test_data["Target"].astype(int)

    # ------------------------------------------
    # Training-only feature selection
    # ------------------------------------------

    non_constant = [
        col for col in feature_columns
        if X[col].nunique(dropna=True) > 1
    ]

    corr_matrix = X[non_constant].corr().abs()

    upper = corr_matrix.where(
        np.triu(
            np.ones(corr_matrix.shape),
            k=1
        ).astype(bool)
    )

    remove_features = [
        col for col in upper.columns
        if any(upper[col] >= 0.95)
    ]

    selected = [
        col for col in non_constant
        if col not in remove_features
    ]

    X = X[selected]
    X_test = X_test[selected]

    # ------------------------------------------
    # Internal validation split
    # ------------------------------------------

    X_train, X_val, y_train, y_val = train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )

    imputer = SimpleImputer(strategy="median")

    X_train_imp = imputer.fit_transform(X_train)
    X_val_imp = imputer.transform(X_val)
    X_test_imp = imputer.transform(X_test)

    # ------------------------------------------
    # Train three classifiers
    # ------------------------------------------

    rf = RandomForestClassifier(
        n_estimators=100,
        max_depth=20,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=1
    )

    et = ExtraTreesClassifier(
        n_estimators=100,
        max_depth=20,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=1
    )

    xgb = XGBClassifier(
        n_estimators=100,
        max_depth=6,
        learning_rate=0.1,
        subsample=0.8,
        colsample_bytree=0.8,
        eval_metric="logloss",
        random_state=42,
        n_jobs=1
    )

    weights = compute_sample_weight(
        class_weight="balanced",
        y=y_train
    )

    rf.fit(X_train_imp, y_train)
    et.fit(X_train_imp, y_train)

    xgb.fit(
        X_train_imp,
        y_train,
        sample_weight=weights
    )

    # ------------------------------------------
    # Obtain validation probabilities
    # ------------------------------------------

    et_val_prob = et.predict_proba(X_val_imp)[:, 1]

    rf_val_prob = rf.predict_proba(X_val_imp)[:, 1]
    xgb_val_prob = xgb.predict_proba(X_val_imp)[:, 1]

    soft_val_prob = (
        et_val_prob + rf_val_prob + xgb_val_prob
    ) / 3

    # ------------------------------------------
    # Obtain test probabilities
    # ------------------------------------------

    et_test_prob = et.predict_proba(X_test_imp)[:, 1]

    rf_test_prob = rf.predict_proba(X_test_imp)[:, 1]
    xgb_test_prob = xgb.predict_proba(X_test_imp)[:, 1]

    soft_test_prob = (
        et_test_prob + rf_test_prob + xgb_test_prob
    ) / 3

    # ------------------------------------------
    # Select threshold using validation data
    # ------------------------------------------

    probability_sets = {
        "Extra Trees": (et_val_prob, et_test_prob),
        "Soft Voting": (soft_val_prob, soft_test_prob)
    }

    for model_name, (val_probs, test_probs) in probability_sets.items():

        candidates = np.unique(
            np.concatenate([
                np.linspace(0.001, 0.99, 300),
                val_probs
            ])
        )

        best_threshold = 0.5
        best_recall = -1
        best_fpr = 1

        for threshold in candidates:

            val_pred = (
                val_probs >= threshold
            ).astype(int)

            tn, fp, fn, tp = confusion_matrix(
                y_val,
                val_pred,
                labels=[0, 1]
            ).ravel()

            val_fpr = fp / (fp + tn)
            val_recall = tp / (tp + fn)

            if val_fpr <= 0.01:

                if val_recall > best_recall:
                    best_recall = val_recall
                    best_fpr = val_fpr
                    best_threshold = threshold

        # --------------------------------------
        # Evaluate on unseen attack category
        # --------------------------------------

        test_pred = (
            test_probs >= best_threshold
        ).astype(int)

        tn, fp, fn, tp = confusion_matrix(
            y_test,
            test_pred,
            labels=[0, 1]
        ).ravel()

        final_comparison_results.append({
            "Category": category,
            "Model": model_name,
            "Threshold": best_threshold,
            "Validation_Recall": best_recall,
            "Validation_FPR": best_fpr,
            "Test_Recall": tp / (tp + fn),
            "Test_FPR": fp / (fp + tn),
            "FP": fp,
            "FN": fn
        })

        print(
            model_name,
            "| Threshold:", round(best_threshold, 5),
            "| Test FP:", fp,
            "| Test FN:", fn
        )


# ------------------------------------------
# Category-wise results
# ------------------------------------------

final_comparison_df = pd.DataFrame(
    final_comparison_results
)

print("\n" + "=" * 90)
print("CALIBRATED EXTRA TREES VS CALIBRATED SOFT VOTING")
print("=" * 90)

display(final_comparison_df.round(6))


# ------------------------------------------
# Pooled results
# ------------------------------------------

pooled_final = (
    final_comparison_df
    .groupby("Model")
    .agg(
        Total_FP=("FP", "sum"),
        Total_FN=("FN", "sum"),
        Mean_Test_Recall=("Test_Recall", "mean"),
        Mean_Test_FPR=("Test_FPR", "mean")
    )
    .reset_index()
)

print("\nPOOLED FINAL COMPARISON")

display(pooled_final.round(6))
```

```python
# ==============================================
# STEP 70: LEAKAGE-SAFE CALIBRATED ENSEMBLE
# ==============================================

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer

from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier
)

from sklearn.metrics import confusion_matrix

from xgboost import XGBClassifier
from sklearn.utils.class_weight import compute_sample_weight

corrected_results = []

for category, fold in leave_one_out_folds.items():

    print("\nProcessing:", category)

    # ------------------------------------------
    # 1. Prepare fold data
    # ------------------------------------------

    attack_train = fold["attack_train"].copy()
    benign_train = fold["benign_train"].copy()

    attack_train["Target"] = 1
    benign_train["Target"] = 0

    train_data = pd.concat(
        [attack_train, benign_train],
        ignore_index=True
    )

    attack_test = fold["attack_test"].copy()
    benign_test = fold["benign_test"].copy()

    attack_test["Target"] = 1
    benign_test["Target"] = 0

    test_data = pd.concat(
        [benign_test, attack_test],
        ignore_index=True
    )

    X = train_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    )

    y = train_data["Target"].astype(int)

    X_test = test_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    )

    y_test = test_data["Target"].astype(int)

    # ------------------------------------------
    # 2. Split BEFORE feature selection
    # ------------------------------------------

    X_train, X_val, y_train, y_val = train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )

    # ------------------------------------------
    # 3. Feature selection using training only
    # ------------------------------------------

    non_constant = [
        col for col in X_train.columns
        if X_train[col].nunique(dropna=True) > 1
    ]

    corr_matrix = X_train[non_constant].corr().abs()

    upper = corr_matrix.where(
        np.triu(
            np.ones(corr_matrix.shape),
            k=1
        ).astype(bool)
    )

    remove_features = [
        col for col in upper.columns
        if any(upper[col] >= 0.95)
    ]

    selected = [
        col for col in non_constant
        if col not in remove_features
    ]

    X_train = X_train[selected]
    X_val = X_val[selected]
    X_test = X_test[selected]

    # ------------------------------------------
    # 4. Fit imputer on training only
    # ------------------------------------------

    imputer = SimpleImputer(strategy="median")

    X_train_imp = imputer.fit_transform(X_train)
    X_val_imp = imputer.transform(X_val)
    X_test_imp = imputer.transform(X_test)

    # ------------------------------------------
    # 5. Train classifiers
    # ------------------------------------------

    rf = RandomForestClassifier(
        n_estimators=100,
        max_depth=20,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=1
    )

    et = ExtraTreesClassifier(
        n_estimators=100,
        max_depth=20,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=1
    )

    xgb = XGBClassifier(
        n_estimators=100,
        max_depth=6,
        learning_rate=0.1,
        subsample=0.8,
        colsample_bytree=0.8,
        eval_metric="logloss",
        random_state=42,
        n_jobs=1
    )

    weights = compute_sample_weight(
        class_weight="balanced",
        y=y_train
    )

    rf.fit(X_train_imp, y_train)
    et.fit(X_train_imp, y_train)

    xgb.fit(
        X_train_imp,
        y_train,
        sample_weight=weights
    )

    # ------------------------------------------
    # 6. Obtain validation probabilities
    # ------------------------------------------

    rf_val = rf.predict_proba(X_val_imp)[:, 1]
    et_val = et.predict_proba(X_val_imp)[:, 1]
    xgb_val = xgb.predict_proba(X_val_imp)[:, 1]

    soft_val = (rf_val + et_val + xgb_val) / 3

    # ------------------------------------------
    # 7. Obtain test probabilities
    # ------------------------------------------

    rf_test = rf.predict_proba(X_test_imp)[:, 1]
    et_test = et.predict_proba(X_test_imp)[:, 1]
    xgb_test = xgb.predict_proba(X_test_imp)[:, 1]

    soft_test = (rf_test + et_test + xgb_test) / 3

    # ------------------------------------------
    # 8. Threshold selection
    # ------------------------------------------

    threshold_sets = {
        "Extra Trees": (
            et_val,
            et_test
        ),

        "Soft Voting": (
            soft_val,
            soft_test
        )
    }

    for model_name, (val_probs, test_probs) in threshold_sets.items():

        candidates = np.unique(
            np.concatenate([
                np.linspace(0.001, 0.99, 300),
                val_probs
            ])
        )

        best_threshold = 0.5
        best_recall = -1
        best_fpr = 1

        for threshold in candidates:

            val_pred = (
                val_probs >= threshold
            ).astype(int)

            tn, fp, fn, tp = confusion_matrix(
                y_val,
                val_pred,
                labels=[0, 1]
            ).ravel()

            val_fpr = fp / (fp + tn)
            val_recall = tp / (tp + fn)

            if val_fpr <= 0.01:

                if val_recall > best_recall:
                    best_recall = val_recall
                    best_fpr = val_fpr
                    best_threshold = threshold

        # --------------------------------------
        # 9. Evaluate on held-out category
        # --------------------------------------

        test_pred = (
            test_probs >= best_threshold
        ).astype(int)

        tn, fp, fn, tp = confusion_matrix(
            y_test,
            test_pred,
            labels=[0, 1]
        ).ravel()

        corrected_results.append({
            "Category": category,
            "Model": model_name,
            "Features": len(selected),
            "Threshold": best_threshold,
            "Validation_Recall": best_recall,
            "Validation_FPR": best_fpr,
            "Test_Recall": tp / (tp + fn),
            "Test_FPR": fp / (fp + tn),
            "FP": fp,
            "FN": fn
        })

        print(
            model_name,
            "| Features:", len(selected),
            "| Threshold:", round(best_threshold, 5),
            "| FP:", fp,
            "| FN:", fn
        )


# ------------------------------------------
# 10. Display category-wise results
# ------------------------------------------

corrected_df = pd.DataFrame(corrected_results)

print("\n" + "=" * 90)
print("CORRECTED LEAKAGE-SAFE EVALUATION")
print("=" * 90)

display(corrected_df.round(6))


# ------------------------------------------
# 11. Aggregate results
# ------------------------------------------

corrected_summary = (
    corrected_df
    .groupby("Model")
    .agg(
        Total_FP=("FP", "sum"),
        Total_FN=("FN", "sum"),
        Mean_Recall=("Test_Recall", "mean"),
        Mean_FPR=("Test_FPR", "mean")
    )
    .reset_index()
)

print("\nFINAL AGGREGATED RESULTS")

display(corrected_summary.round(6))
```

```python
import pandas as pd
import numpy as np
from pathlib import Path
from pandas.util import hash_pandas_object

PROJECT = Path(r"C:\Users\HP\DDoS_IoT_Project")
DATA_DIR = PROJECT / "processed" / "zero_day_samples"

files = {
    "Known attacks": "known_attacks.parquet",
    "SNMP attacks": "snmp_zero_day.parquet",
    "Benign training": "benign_train.parquet",
    "Benign testing": "benign_test.parquet"
}

datasets = {}

for name, filename in files.items():

    df = pd.read_parquet(DATA_DIR / filename)

    # Exclude labels and source metadata.
    feature_cols = [
        c for c in df.columns
        if c not in ["Label", "Source_File", "Attack_Type", "Target"]
    ]

    X = df[feature_cols].copy()

    X = X.replace([np.inf, -np.inf], np.nan)

    # Generate a hash for each feature row.
    row_hashes = hash_pandas_object(
        X,
        index=False
    )

    datasets[name] = set(row_hashes)

    print(
        f"{name}: {len(row_hashes)} rows, "
        f"{len(datasets[name])} unique feature rows"
    )

print("\n" + "=" * 70)
print("CROSS-DATASET EXACT DUPLICATE AUDIT")
print("=" * 70)

names = list(datasets.keys())

for i in range(len(names)):

    for j in range(i + 1, len(names)):

        a = names[i]
        b = names[j]

        overlap = datasets[a].intersection(datasets[b])

        print(f"\n{a} vs {b}")
        print("Exact duplicate feature rows:", len(overlap))
```

```python
import numpy as np
import pandas as pd
from scipy.stats import ks_2samp

# Extract only the 81 model features
X_train_benign = benign_train[feature_columns].copy()
X_test_benign = benign_test_clean[feature_columns].copy()

results = []

for feature in feature_columns:

    train_values = pd.to_numeric(
        X_train_benign[feature], errors="coerce"
    ).replace([np.inf, -np.inf], np.nan).dropna()

    test_values = pd.to_numeric(
        X_test_benign[feature], errors="coerce"
    ).replace([np.inf, -np.inf], np.nan).dropna()

    if len(train_values) == 0 or len(test_values) == 0:
        continue

    # Kolmogorov-Smirnov test
    ks_stat, ks_pvalue = ks_2samp(
        train_values,
        test_values
    )

    # Standardised Mean Difference
    pooled_std = np.sqrt(
        (train_values.var() + test_values.var()) / 2
    )

    if pooled_std > 0:
        smd = (
            train_values.mean() - test_values.mean()
        ) / pooled_std
    else:
        smd = 0.0

    results.append({
        "Feature": feature,
        "Train Mean": train_values.mean(),
        "Test Mean": test_values.mean(),
        "KS Statistic": ks_stat,
        "KS p-value": ks_pvalue,
        "Absolute SMD": abs(smd),
        "Train Missing %": (
            X_train_benign[feature].isna().mean() * 100
        ),
        "Test Missing %": (
            X_test_benign[feature].isna().mean() * 100
        )
    })

distribution_df = pd.DataFrame(results)

distribution_df = distribution_df.sort_values(
    "KS Statistic",
    ascending=False
)

print("=" * 85)
print("BENIGN TRAFFIC DISTRIBUTION COMPARISON")
print("=" * 85)

print("\nTraining benign rows:", len(X_train_benign))
print("Testing benign rows:", len(X_test_benign))
print("Features compared:", len(distribution_df))

print("\nTop 15 features with largest distribution differences:")

print(
    distribution_df.head(15).round(5).to_string(index=False)
)

print("\n" + "=" * 85)
print("OVERALL DISTRIBUTION SHIFT SUMMARY")
print("=" * 85)

print(
    "Features with KS statistic > 0.10:",
    (distribution_df["KS Statistic"] > 0.10).sum()
)

print(
    "Features with KS statistic > 0.20:",
    (distribution_df["KS Statistic"] > 0.20).sum()
)

print(
    "Features with absolute SMD > 0.20:",
    (distribution_df["Absolute SMD"] > 0.20).sum()
)

print(
    "Features with absolute SMD > 0.50:",
    (distribution_df["Absolute SMD"] > 0.50).sum()
)

print(
    "Features with KS p-value < 0.05:",
    (distribution_df["KS p-value"] < 0.05).sum()
)
```

```python
import numpy as np
import pandas as pd

from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    confusion_matrix
)

# Suspicious features
remove_features = [
    "Init_Win_bytes_forward",
    "Fwd Header Length",
    "Fwd Header Length.1"
]

# Remaining features
ablation_features = [
    f for f in feature_columns
    if f not in remove_features
]

print("Original features:", len(feature_columns))
print("Removed features:", len(remove_features))
print("Remaining features:", len(ablation_features))

# Reconstruct original supervised training data
X_train_ablation = X_train[ablation_features].copy()
X_test_ablation = X_clean[ablation_features].copy()

y_train_ablation = y_train.copy()

# Fit imputer exclusively on training data
ablation_test_imputer = SimpleImputer(strategy="median")

X_train_ablation_imp = ablation_test_imputer.fit_transform(
    X_train_ablation
)

X_test_ablation_imp = ablation_test_imputer.transform(
    X_test_ablation
)

# Train reduced-feature Random Forest
ablation_test_rf = RandomForestClassifier(
    n_estimators=100,
    max_depth=20,
    min_samples_leaf=2,
    class_weight="balanced",
    random_state=42,
    n_jobs=1
)

ablation_test_rf.fit(
    X_train_ablation_imp,
    y_train_ablation
)

# Predictions
ablation_probabilities = ablation_test_rf.predict_proba(
    X_test_ablation_imp
)[:, 1]

ablation_predictions = (
    ablation_probabilities >= 0.5
).astype(int)

# Evaluate
tn, fp, fn, tp = confusion_matrix(
    y_clean,
    ablation_predictions,
    labels=[0, 1]
).ravel()

print("\n" + "=" * 70)
print("SUSPICIOUS FEATURE ABLATION RESULTS")
print("=" * 70)

print("Accuracy:", accuracy_score(y_clean, ablation_predictions))
print("Precision:", precision_score(y_clean, ablation_predictions))
print("Recall:", recall_score(y_clean, ablation_predictions))
print("F1 Score:", f1_score(y_clean, ablation_predictions))
print("Balanced Accuracy:", balanced_accuracy_score(y_clean, ablation_predictions))
print("TN:", tn)
print("FP:", fp)
print("FN:", fn)
print("TP:", tp)
```

```python
from sklearn.model_selection import train_test_split

# Load original supervised training pool
known_data = pd.read_parquet(
    DATA_DIR / "known_attacks.parquet"
)

benign_data = pd.read_parquet(
    DATA_DIR / "benign_train.parquet"
)

# Combine attacks and benign records
supervised_data = pd.concat(
    [known_data, benign_data],
    ignore_index=True
)

# Extract features and binary labels
X_supervised = supervised_data[feature_columns].copy()

y_supervised = np.where(
    supervised_data["Label"] == "BENIGN",
    0,
    1
)

# Reproduce original stratified split
X_train_ablation, X_validation_ablation, \
y_train_ablation, y_validation_ablation = train_test_split(
    X_supervised,
    y_supervised,
    test_size=0.20,
    random_state=42,
    stratify=y_supervised
)

# Remove suspicious features
remove_features = [
    "Init_Win_bytes_forward",
    "Fwd Header Length",
    "Fwd Header Length.1"
]

ablation_features = [
    f for f in feature_columns
    if f not in remove_features
]

print("Original training pool:", len(supervised_data))
print("Training records:", len(X_train_ablation))
print("Validation records:", len(X_validation_ablation))
print("Original features:", len(feature_columns))
print("Remaining features:", len(ablation_features))

print("\nTraining class distribution:")
print(pd.Series(y_train_ablation).value_counts())
```

```python
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    confusion_matrix
)

# Select remaining features
X_train_reduced = X_train_ablation[ablation_features]
X_test_reduced = X_clean[ablation_features]

# Fit imputer only on training data
ablation_imputer = SimpleImputer(strategy="median")

X_train_reduced_imp = ablation_imputer.fit_transform(
    X_train_reduced
)

X_test_reduced_imp = ablation_imputer.transform(
    X_test_reduced
)

# Train reduced-feature RF
ablation_rf = RandomForestClassifier(
    n_estimators=100,
    max_depth=20,
    min_samples_leaf=2,
    class_weight="balanced",
    random_state=42,
    n_jobs=1
)

ablation_rf.fit(
    X_train_reduced_imp,
    y_train_ablation
)

# Predict on clean SNMP test data
ablation_probabilities = ablation_rf.predict_proba(
    X_test_reduced_imp
)[:, 1]

ablation_predictions = (
    ablation_probabilities >= 0.5
).astype(int)

# Calculate metrics
tn, fp, fn, tp = confusion_matrix(
    y_clean,
    ablation_predictions,
    labels=[0, 1]
).ravel()

ablation_results = {
    "Accuracy": accuracy_score(y_clean, ablation_predictions),
    "Precision": precision_score(y_clean, ablation_predictions),
    "Recall": recall_score(y_clean, ablation_predictions),
    "F1 Score": f1_score(y_clean, ablation_predictions),
    "Balanced Accuracy": balanced_accuracy_score(
        y_clean, ablation_predictions
    ),
    "FPR": fp / (fp + tn),
    "TN": tn,
    "FP": fp,
    "FN": fn,
    "TP": tp
}

print("=" * 70)
print("78-FEATURE ABLATION MODEL RESULTS")
print("=" * 70)

for metric, value in ablation_results.items():
    print(f"{metric:22s}: {value}")
    
```

```python
# Compare predictions from original and ablation models

original_predictions = (
    clean_probabilities >= 0.5
).astype(int)

reduced_predictions = (
    ablation_probabilities >= 0.5
).astype(int)

comparison = pd.DataFrame({
    "Actual": y_clean,
    "Original Prediction": original_predictions,
    "Ablation Prediction": reduced_predictions,
    "Original Probability": clean_probabilities,
    "Ablation Probability": ablation_probabilities
})

comparison["Prediction Changed"] = (
    comparison["Original Prediction"]
    != comparison["Ablation Prediction"]
)

comparison["Original Correct"] = (
    comparison["Original Prediction"]
    == comparison["Actual"]
)

comparison["Ablation Correct"] = (
    comparison["Ablation Prediction"]
    == comparison["Actual"]
)

print("=" * 70)
print("PREDICTION CHANGE ANALYSIS")
print("=" * 70)

print(
    "Total predictions changed:",
    comparison["Prediction Changed"].sum()
)

print(
    "\nOriginal correct, ablation incorrect:",
    (
        comparison["Original Correct"]
        & ~comparison["Ablation Correct"]
    ).sum()
)

print(
    "Original incorrect, ablation correct:",
    (
        ~comparison["Original Correct"]
        & comparison["Ablation Correct"]
    ).sum()
)

print("\nChanged predictions by actual class:")

print(
    comparison[comparison["Prediction Changed"]]
    ["Actual"]
    .value_counts()
    .rename(index={0: "Benign", 1: "SNMP Attack"})
)
```

```python
import networkx as nx
import pandas as pd
import numpy as np

# Calculate correlations using training data only
corr_matrix = X_train_ablation[
    feature_columns
].corr(method="spearman").abs()

# Create a graph of highly correlated features
G = nx.Graph()

G.add_nodes_from(feature_columns)

for i in range(len(feature_columns)):
    for j in range(i + 1, len(feature_columns)):

        corr = corr_matrix.iloc[i, j]

        if corr >= 0.95:
            G.add_edge(
                feature_columns[i],
                feature_columns[j],
                weight=corr
            )

# Identify connected feature groups
correlation_groups = [
    sorted(list(group))
    for group in nx.connected_components(G)
    if len(group) > 1
]

correlation_groups.sort(
    key=len,
    reverse=True
)

print("=" * 75)
print("HIGHLY CORRELATED FEATURE GROUPS")
print("=" * 75)

for index, group in enumerate(correlation_groups, 1):

    print(f"\nGroup {index} ({len(group)} features):")

    for feature in group:
        print(" -", feature)

print("\nTotal redundant feature groups:", len(correlation_groups))

print(
    "Features involved in correlated groups:",
    len(set(
        feature
        for group in correlation_groups
        for feature in group
    ))
)
```

```python
import pandas as pd
import numpy as np

# Use only the original training partition
X_duplicate_check = X_train_ablation[feature_columns].copy()

duplicate_pairs = []

for i in range(len(feature_columns)):

    for j in range(i + 1, len(feature_columns)):

        feature_a = feature_columns[i]
        feature_b = feature_columns[j]

        values_a = X_duplicate_check[feature_a]
        values_b = X_duplicate_check[feature_b]

        identical = values_a.equals(values_b)

        if identical:

            duplicate_pairs.append({
                "Feature 1": feature_a,
                "Feature 2": feature_b
            })

duplicate_columns_df = pd.DataFrame(duplicate_pairs)

print("=" * 75)
print("EXACT DUPLICATE FEATURE COLUMN AUDIT")
print("=" * 75)

print("Total features:", len(feature_columns))
print("Exact duplicate column pairs:", len(duplicate_pairs))

if len(duplicate_pairs) > 0:
    print("\nIdentical feature pairs:")
    print(duplicate_columns_df.to_string(index=False))
else:
    print("\nNo exact duplicate feature columns found.")
```

```python
import pandas as pd
import numpy as np

X_audit = X_train_ablation[feature_columns]

feature_audit = pd.DataFrame({
    "Feature": feature_columns,
    "Unique Values": [
        X_audit[col].nunique(dropna=False)
        for col in feature_columns
    ],
    "Missing Values": [
        X_audit[col].isna().sum()
        for col in feature_columns
    ],
    "Most Frequent Value": [
        X_audit[col].mode(dropna=False).iloc[0]
        for col in feature_columns
    ],
    "Most Frequent Count": [
        X_audit[col].value_counts(dropna=False).iloc[0]
        for col in feature_columns
    ]
})

feature_audit = feature_audit.sort_values(
    "Unique Values"
)

print("=" * 75)
print("FEATURE UNIQUENESS AUDIT")
print("=" * 75)

print("\nConstant features:")

print(
    feature_audit[
        feature_audit["Unique Values"] == 1
    ].to_string(index=False)
)

print("\nFeatures with two or fewer unique values:")

print(
    feature_audit[
        feature_audit["Unique Values"] <= 2
    ].to_string(index=False)
)

print("\nSUMMARY")
print("Total features:", len(feature_columns))
print(
    "Constant features:",
    (feature_audit["Unique Values"] == 1).sum()
)
print(
    "Non-constant features:",
    (feature_audit["Unique Values"] > 1).sum()
)
```

```python
constant_features = feature_audit.loc[
    feature_audit["Unique Values"] == 1,
    "Feature"
].tolist()

print("=" * 75)
print("CONSTANT FEATURE CROSS-DATASET VALIDATION")
print("=" * 75)

constant_check = []

for feature in constant_features:

    train_unique = X_train_ablation[feature].nunique(
        dropna=False
    )

    test_unique = X_clean[feature].nunique(
        dropna=False
    )

    train_value = X_train_ablation[feature].iloc[0]

    test_values = X_clean[feature].unique()

    constant_check.append({
        "Feature": feature,
        "Training Unique Values": train_unique,
        "Testing Unique Values": test_unique,
        "Training Value": train_value,
        "Testing Values": str(test_values[:5]),
        "Constant in Both": test_unique == 1
    })

constant_check_df = pd.DataFrame(constant_check)

print(constant_check_df.to_string(index=False))

print("\nConstant in both datasets:",
      constant_check_df["Constant in Both"].sum(),
      "out of", len(constant_features))
```

```python
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    confusion_matrix
)

# Identify non-constant features
nonconstant_features = [
    f for f in feature_columns
    if f not in constant_features
]

print("Original features:", len(feature_columns))
print("Removed constant features:", len(constant_features))
print("Remaining features:", len(nonconstant_features))

# Prepare training and testing data
X_train_69 = X_train_ablation[nonconstant_features]
X_test_69 = X_clean[nonconstant_features]

# Fit imputer only on training data
imputer_69 = SimpleImputer(strategy="median")

X_train_69_imp = imputer_69.fit_transform(X_train_69)
X_test_69_imp = imputer_69.transform(X_test_69)

# Train Random Forest
rf_69 = RandomForestClassifier(
    n_estimators=100,
    max_depth=20,
    min_samples_leaf=2,
    class_weight="balanced",
    random_state=42,
    n_jobs=1
)

rf_69.fit(X_train_69_imp, y_train_ablation)

# Predictions
probabilities_69 = rf_69.predict_proba(X_test_69_imp)[:, 1]
predictions_69 = (probabilities_69 >= 0.5).astype(int)

# Evaluation
tn, fp, fn, tp = confusion_matrix(
    y_clean,
    predictions_69,
    labels=[0, 1]
).ravel()

results_69 = {
    "Accuracy": accuracy_score(y_clean, predictions_69),
    "Precision": precision_score(y_clean, predictions_69),
    "Recall": recall_score(y_clean, predictions_69),
    "F1 Score": f1_score(y_clean, predictions_69),
    "Balanced Accuracy": balanced_accuracy_score(y_clean, predictions_69),
    "FPR": fp / (fp + tn),
    "TN": tn,
    "FP": fp,
    "FN": fn,
    "TP": tp
}

print("\n" + "=" * 70)
print("69-FEATURE RANDOM FOREST RESULTS")
print("=" * 70)

for metric, value in results_69.items():
    print(f"{metric:22s}: {value}")
```

```python
# Identify exact duplicate non-constant columns
X_unique_check = X_train_ablation[nonconstant_features]

duplicate_groups = []

for i in range(len(nonconstant_features)):

    feature_a = nonconstant_features[i]

    for j in range(i + 1, len(nonconstant_features)):

        feature_b = nonconstant_features[j]

        if X_unique_check[feature_a].equals(
            X_unique_check[feature_b]
        ):

            duplicate_groups.append(
                (feature_a, feature_b)
            )

print("=" * 75)
print("NON-CONSTANT EXACT DUPLICATE AUDIT")
print("=" * 75)

print("Non-constant features:", len(nonconstant_features))
print("Exact duplicate pairs:", len(duplicate_groups))

for feature_a, feature_b in duplicate_groups:
    print(f"{feature_a} <--> {feature_b}")
```

```python
print("=" * 75)
print("TRAINING vs TESTING DUPLICATE FEATURE VERIFICATION")
print("=" * 75)

duplicate_pairs_to_check = [
    ("Total Fwd Packets", "Subflow Fwd Packets"),
    ("Total Backward Packets", "Subflow Bwd Packets"),
    ("Total Length of Fwd Packets", "Subflow Fwd Bytes"),
    ("Total Length of Bwd Packets", "Subflow Bwd Bytes"),
    ("Fwd Packet Length Mean", "Avg Fwd Segment Size"),
    ("Bwd Packet Length Mean", "Avg Bwd Segment Size"),
    ("Fwd PSH Flags", "RST Flag Count"),
    ("Fwd Header Length", "Fwd Header Length.1")
]

verification_results = []

for feature_a, feature_b in duplicate_pairs_to_check:

    train_identical = X_train_ablation[feature_a].equals(
        X_train_ablation[feature_b]
    )

    test_identical = X_clean[feature_a].equals(
        X_clean[feature_b]
    )

    verification_results.append({
        "Feature Pair": f"{feature_a} / {feature_b}",
        "Training Identical": train_identical,
        "Testing Identical": test_identical
    })

verification_df = pd.DataFrame(verification_results)

print(verification_df.to_string(index=False))

print("\nTraining duplicates confirmed:",
      verification_df["Training Identical"].sum(), "/ 8")

print("Testing duplicates confirmed:",
      verification_df["Testing Identical"].sum(), "/ 8")
```

```python
# Feature importance stability across multiple random seeds

seeds = [42, 21, 7, 101, 202]

importance_runs = []

print("=" * 75)
print("FEATURE IMPORTANCE STABILITY ANALYSIS")
print("=" * 75)

for seed in seeds:

    # Create a new stratified training/validation split
    X_train_seed, X_val_seed, y_train_seed, y_val_seed = train_test_split(
        X_supervised[nonconstant_features],
        y_supervised,
        test_size=0.20,
        random_state=seed,
        stratify=y_supervised
    )

    # Fit imputer only on training data
    seed_imputer = SimpleImputer(strategy="median")

    X_train_seed_imp = seed_imputer.fit_transform(X_train_seed)

    # Train Random Forest
    seed_model = RandomForestClassifier(
        n_estimators=100,
        max_depth=20,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=seed,
        n_jobs=1
    )

    seed_model.fit(X_train_seed_imp, y_train_seed)

    # Store feature importance
    seed_importance = pd.DataFrame({
        "Feature": nonconstant_features,
        f"Seed_{seed}": seed_model.feature_importances_
    })

    importance_runs.append(seed_importance)

    print(f"Seed {seed}: Model trained successfully")

# Combine importance values
importance_stability = importance_runs[0]

for run in importance_runs[1:]:
    importance_stability = importance_stability.merge(
        run,
        on="Feature"
    )

seed_columns = [f"Seed_{seed}" for seed in seeds]

# Calculate stability statistics
importance_stability["Mean Importance"] = (
    importance_stability[seed_columns].mean(axis=1)
)

importance_stability["Std Importance"] = (
    importance_stability[seed_columns].std(axis=1)
)

importance_stability["Rank"] = (
    importance_stability["Mean Importance"]
    .rank(ascending=False, method="min")
    .astype(int)
)

importance_stability = importance_stability.sort_values(
    "Mean Importance",
    ascending=False
).reset_index(drop=True)

print("\n" + "=" * 75)
print("TOP 20 FEATURES — IMPORTANCE STABILITY")
print("=" * 75)

display(
    importance_stability[
        ["Feature", "Mean Importance", "Std Importance", "Rank"]
    ].head(20).style.format({
        "Mean Importance": "{:.6f}",
        "Std Importance": "{:.6f}"
    })
)
```

```python
# Calculate pairwise Spearman rank correlations

from scipy.stats import spearmanr
from itertools import combinations

rank_agreement = []

for seed_a, seed_b in combinations(seeds, 2):

    rank_a = importance_stability[f"Rank_{seed_a}"]
    rank_b = importance_stability[f"Rank_{seed_b}"]

    correlation, p_value = spearmanr(rank_a, rank_b)

    rank_agreement.append({
        "Seed Pair": f"{seed_a} vs {seed_b}",
        "Spearman Correlation": correlation,
        "P-value": p_value
    })

rank_agreement_df = pd.DataFrame(rank_agreement)

print("=" * 75)
print("PAIRWISE FEATURE RANKING AGREEMENT")
print("=" * 75)

display(
    rank_agreement_df.style.format({
        "Spearman Correlation": "{:.4f}",
        "P-value": "{:.6e}"
    })
)

print("\nMean Spearman correlation:",
      rank_agreement_df["Spearman Correlation"].mean())

print("Minimum Spearman correlation:",
      rank_agreement_df["Spearman Correlation"].min())

print("Maximum Spearman correlation:",
      rank_agreement_df["Spearman Correlation"].max())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create correlation matrix
rank_matrix = pd.DataFrame(
    np.eye(len(seeds)),
    index=seeds,
    columns=seeds
)

for _, row in rank_agreement_df.iterrows():

    seed_a, seed_b = map(
        int,
        row["Seed Pair"].split(" vs ")
    )

    correlation = row["Spearman Correlation"]

    rank_matrix.loc[seed_a, seed_b] = correlation
    rank_matrix.loc[seed_b, seed_a] = correlation

# Plot heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    rank_matrix,
    annot=True,
    fmt=".3f",
    cmap="coolwarm",
    vmin=0,
    vmax=1,
    square=True
)

plt.title("Feature Importance Ranking Agreement Across Random Seeds")
plt.xlabel("Random Seed")
plt.ylabel("Random Seed")
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# STEP 140: CORRECTED LOCO WITH COMPLETE PREDICTION STORAGE
# ============================================================

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer

from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
from sklearn.metrics import confusion_matrix

from xgboost import XGBClassifier
from sklearn.utils.class_weight import compute_sample_weight

loco_prediction_store = {}
loco_stored_results = []

for category, fold in leave_one_out_folds.items():

    print("\nProcessing:", category)

    # ------------------------------------------------
    # 1. Prepare training and testing data
    # ------------------------------------------------

    attack_train = fold["attack_train"].copy()
    benign_train = fold["benign_train"].copy()

    attack_train["Target"] = 1
    benign_train["Target"] = 0

    train_data = pd.concat(
        [attack_train, benign_train],
        ignore_index=True
    )

    attack_test = fold["attack_test"].copy()
    benign_test = fold["benign_test"].copy()

    attack_test["Target"] = 1
    benign_test["Target"] = 0

    test_data = pd.concat(
        [benign_test, attack_test],
        ignore_index=True
    )

    X = train_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    )

    y = train_data["Target"].astype(int)

    X_test = test_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    )

    y_test = test_data["Target"].astype(int)

    # ------------------------------------------------
    # 2. Split BEFORE feature selection
    # ------------------------------------------------

    X_train, X_val, y_train, y_val = train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )

    # ------------------------------------------------
    # 3. Training-only feature selection
    # ------------------------------------------------

    non_constant = [
        col for col in X_train.columns
        if X_train[col].nunique(dropna=True) > 1
    ]

    corr_matrix = X_train[non_constant].corr().abs()

    upper = corr_matrix.where(
        np.triu(
            np.ones(corr_matrix.shape),
            k=1
        ).astype(bool)
    )

    remove_features = [
        col for col in upper.columns
        if any(upper[col] >= 0.95)
    ]

    selected = [
        col for col in non_constant
        if col not in remove_features
    ]

    X_train = X_train[selected]
    X_val = X_val[selected]
    X_test = X_test[selected]

    # ------------------------------------------------
    # 4. Training-only imputation
    # ------------------------------------------------

    imputer = SimpleImputer(strategy="median")

    X_train_imp = imputer.fit_transform(X_train)
    X_val_imp = imputer.transform(X_val)
    X_test_imp = imputer.transform(X_test)

    # ------------------------------------------------
    # 5. Train three classifiers
    # ------------------------------------------------

    rf = RandomForestClassifier(
        n_estimators=100,
        max_depth=20,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=1
    )

    et = ExtraTreesClassifier(
        n_estimators=100,
        max_depth=20,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=1
    )

    xgb = XGBClassifier(
        n_estimators=100,
        max_depth=6,
        learning_rate=0.1,
        subsample=0.8,
        colsample_bytree=0.8,
        eval_metric="logloss",
        random_state=42,
        n_jobs=1
    )

    weights = compute_sample_weight(
        class_weight="balanced",
        y=y_train
    )

    rf.fit(X_train_imp, y_train)
    et.fit(X_train_imp, y_train)

    xgb.fit(
        X_train_imp,
        y_train,
        sample_weight=weights
    )

    # ------------------------------------------------
    # 6. Validation probabilities
    # ------------------------------------------------

    rf_val = rf.predict_proba(X_val_imp)[:, 1]
    et_val = et.predict_proba(X_val_imp)[:, 1]
    xgb_val = xgb.predict_proba(X_val_imp)[:, 1]

    soft_val = (rf_val + et_val + xgb_val) / 3

    # ------------------------------------------------
    # 7. Test probabilities
    # ------------------------------------------------

    rf_test = rf.predict_proba(X_test_imp)[:, 1]
    et_test = et.predict_proba(X_test_imp)[:, 1]
    xgb_test = xgb.predict_proba(X_test_imp)[:, 1]

    soft_test = (rf_test + et_test + xgb_test) / 3

    # ------------------------------------------------
    # 8. Threshold calibration
    # ------------------------------------------------

    model_probabilities = {
        "Extra Trees": (et_val, et_test),
        "Soft Voting": (soft_val, soft_test)
    }

    fold_storage = {
        "y_test": y_test.to_numpy(),
        "features": selected.copy(),
        "models": {}
    }

    for model_name, (val_probs, test_probs) in model_probabilities.items():

        candidates = np.unique(
            np.concatenate([
                np.linspace(0.001, 0.99, 300),
                val_probs
            ])
        )

        best_threshold = 0.5
        best_recall = -1
        best_fpr = 1

        for threshold in candidates:

            val_pred = (
                val_probs >= threshold
            ).astype(int)

            tn, fp, fn, tp = confusion_matrix(
                y_val,
                val_pred,
                labels=[0, 1]
            ).ravel()

            val_fpr = fp / (fp + tn)
            val_recall = tp / (tp + fn)

            if val_fpr <= 0.01:

                if val_recall > best_recall:
                    best_recall = val_recall
                    best_fpr = val_fpr
                    best_threshold = threshold

        # ------------------------------------------------
        # 9. Test predictions
        # ------------------------------------------------

        test_pred = (
            test_probs >= best_threshold
        ).astype(int)

        tn, fp, fn, tp = confusion_matrix(
            y_test,
            test_pred,
            labels=[0, 1]
        ).ravel()

        fold_storage["models"][model_name] = {
            "validation_probabilities": val_probs.copy(),
            "test_probabilities": test_probs.copy(),
            "threshold": best_threshold,
            "predictions": test_pred.copy(),
            "FP": fp,
            "FN": fn
        }

        loco_stored_results.append({
            "Category": category,
            "Model": model_name,
            "Features": len(selected),
            "Threshold": best_threshold,
            "Validation_Recall": best_recall,
            "Validation_FPR": best_fpr,
            "Test_Recall": tp / (tp + fn),
            "Test_FPR": fp / (fp + tn),
            "FP": fp,
            "FN": fn
        })

        print(
            model_name,
            "| Features:", len(selected),
            "| Threshold:", round(best_threshold, 6),
            "| FP:", fp,
            "| FN:", fn
        )

    loco_prediction_store[category] = fold_storage


# ============================================================
# 10. Aggregate stored results
# ============================================================

loco_stored_df = pd.DataFrame(loco_stored_results)

loco_stored_summary = (
    loco_stored_df.groupby("Model")
    .agg(
        Total_FP=("FP", "sum"),
        Total_FN=("FN", "sum"),
        Mean_Recall=("Test_Recall", "mean"),
        Mean_FPR=("Test_FPR", "mean")
    )
    .reset_index()
)

print("\n" + "=" * 75)
print("STORED LOCO RESULTS")
print("=" * 75)

display(loco_stored_df.round(6))

print("\nAGGREGATE RESULTS")
display(loco_stored_summary.round(6))

print("\nStored folds:", len(loco_prediction_store))
print("Total stored predictions:",
      sum(
          len(fold["y_test"])
          for fold in loco_prediction_store.values()
      ))
```

```python
# ============================================================
# STEP 141: EXTRA TREES vs SOFT VOTING COMPLEMENTARITY
# ============================================================

from sklearn.metrics import confusion_matrix

comparison_results = []

for category, fold_data in loco_prediction_store.items():

    y_true = fold_data["y_test"]

    et_pred = fold_data["models"]["Extra Trees"]["predictions"]
    soft_pred = fold_data["models"]["Soft Voting"]["predictions"]

    # Agreement
    agreement = np.mean(et_pred == soft_pred)

    # Disagreement
    disagreement = np.sum(et_pred != soft_pred)

    # Individual correct/incorrect predictions
    et_correct = et_pred == y_true
    soft_correct = soft_pred == y_true

    # Complementary detection
    et_only_correct = np.sum(et_correct & ~soft_correct)
    soft_only_correct = np.sum(soft_correct & ~et_correct)

    # Actual attacks missed
    attack_mask = y_true == 1

    et_missed = np.sum(attack_mask & (et_pred == 0))
    soft_missed = np.sum(attack_mask & (soft_pred == 0))

    # Actual benign incorrectly classified
    benign_mask = y_true == 0

    et_fp = np.sum(benign_mask & (et_pred == 1))
    soft_fp = np.sum(benign_mask & (soft_pred == 1))

    comparison_results.append({
        "Category": category,
        "Agreement_%": agreement * 100,
        "Disagreements": disagreement,
        "ET_only_correct": et_only_correct,
        "Soft_only_correct": soft_only_correct,
        "ET_missed_attacks": et_missed,
        "Soft_missed_attacks": soft_missed,
        "ET_false_positives": et_fp,
        "Soft_false_positives": soft_fp
    })


comparison_df = pd.DataFrame(comparison_results)

print("=" * 85)
print("EXTRA TREES vs SOFT VOTING — COMPLEMENTARITY ANALYSIS")
print("=" * 85)

display(comparison_df.round(4))

print("\nAGGREGATE COMPLEMENTARITY")

for column in [
    "Disagreements",
    "ET_only_correct",
    "Soft_only_correct",
    "ET_missed_attacks",
    "Soft_missed_attacks",
    "ET_false_positives",
    "Soft_false_positives"
]:
    print(f"{column}: {comparison_df[column].sum()}")

total_samples = sum(
    len(fold["y_test"])
    for fold in loco_prediction_store.values()
)

total_disagreements = comparison_df["Disagreements"].sum()

print("\nOverall agreement:",
      f"{(1 - total_disagreements / total_samples) * 100:.6f}%")
```

```python
# ============================================================
# STEP 142: DISAGREEMENT ANALYSIS BY ACTUAL CLASS
# ============================================================

disagreement_details = []

for category, fold_data in loco_prediction_store.items():

    y_true = fold_data["y_test"]

    et_pred = fold_data["models"]["Extra Trees"]["predictions"]
    soft_pred = fold_data["models"]["Soft Voting"]["predictions"]

    disagreement_mask = et_pred != soft_pred

    for actual_class in [0, 1]:

        class_mask = (
            disagreement_mask &
            (y_true == actual_class)
        )

        et_values = et_pred[class_mask]
        soft_values = soft_pred[class_mask]

        disagreement_details.append({
            "Category": category,
            "Actual Class": (
                "Benign" if actual_class == 0 else "Attack"
            ),
            "Disagreements": int(class_mask.sum()),
            "ET Predicted Attack": int((et_values == 1).sum()),
            "ET Predicted Benign": int((et_values == 0).sum()),
            "Soft Predicted Attack": int((soft_values == 1).sum()),
            "Soft Predicted Benign": int((soft_values == 0).sum())
        })

disagreement_df = pd.DataFrame(disagreement_details)

print("=" * 85)
print("DISAGREEMENT ANALYSIS BY ACTUAL CLASS")
print("=" * 85)

display(disagreement_df)

print("\nTOTAL DISAGREEMENTS BY CLASS:")

display(
    disagreement_df.groupby("Actual Class")["Disagreements"]
    .sum()
    .reset_index()
)
```

```python
# ============================================================
# STEP 144: THREE-MODEL HARD VOTING EVALUATION
# ============================================================

from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
from sklearn.metrics import confusion_matrix
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split

from xgboost import XGBClassifier
from sklearn.utils.class_weight import compute_sample_weight

hard_voting_store = {}
hard_voting_results = []

for category, fold in leave_one_out_folds.items():

    print("\nProcessing:", category)

    # 1. Prepare data
    attack_train = fold["attack_train"].copy()
    benign_train = fold["benign_train"].copy()

    attack_train["Target"] = 1
    benign_train["Target"] = 0

    train_data = pd.concat(
        [attack_train, benign_train],
        ignore_index=True
    )

    attack_test = fold["attack_test"].copy()
    benign_test = fold["benign_test"].copy()

    attack_test["Target"] = 1
    benign_test["Target"] = 0

    test_data = pd.concat(
        [benign_test, attack_test],
        ignore_index=True
    )

    X = train_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    )

    y = train_data["Target"].astype(int)

    X_test = test_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    )

    y_test = test_data["Target"].astype(int)

    # 2. Split before feature selection
    X_train, X_val, y_train, y_val = train_test_split(
        X, y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )

    # 3. Training-only feature selection
    non_constant = [
        col for col in X_train.columns
        if X_train[col].nunique(dropna=True) > 1
    ]

    corr_matrix = X_train[non_constant].corr().abs()

    upper = corr_matrix.where(
        np.triu(
            np.ones(corr_matrix.shape),
            k=1
        ).astype(bool)
    )

    remove_features = [
        col for col in upper.columns
        if any(upper[col] >= 0.95)
    ]

    selected = [
        col for col in non_constant
        if col not in remove_features
    ]

    X_train = X_train[selected]
    X_val = X_val[selected]
    X_test = X_test[selected]

    # 4. Training-only imputation
    imputer = SimpleImputer(strategy="median")

    X_train_imp = imputer.fit_transform(X_train)
    X_val_imp = imputer.transform(X_val)
    X_test_imp = imputer.transform(X_test)

    # 5. Train classifiers
    rf = RandomForestClassifier(
        n_estimators=100,
        max_depth=20,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=1
    )

    et = ExtraTreesClassifier(
        n_estimators=100,
        max_depth=20,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=1
    )

    xgb = XGBClassifier(
        n_estimators=100,
        max_depth=6,
        learning_rate=0.1,
        subsample=0.8,
        colsample_bytree=0.8,
        eval_metric="logloss",
        random_state=42,
        n_jobs=1
    )

    weights = compute_sample_weight(
        class_weight="balanced",
        y=y_train
    )

    rf.fit(X_train_imp, y_train)
    et.fit(X_train_imp, y_train)

    xgb.fit(
        X_train_imp,
        y_train,
        sample_weight=weights
    )

    # 6. Individual validation predictions
    rf_val = rf.predict(X_val_imp)
    et_val = et.predict(X_val_imp)
    xgb_val = xgb.predict(X_val_imp)

    # 7. Individual test predictions
    rf_test = rf.predict(X_test_imp)
    et_test = et.predict(X_test_imp)
    xgb_test = xgb.predict(X_test_imp)

    # 8. Majority voting
    hard_val_votes = rf_val + et_val + xgb_val
    hard_test_votes = rf_test + et_test + xgb_test

    hard_val = (hard_val_votes >= 2).astype(int)
    hard_test = (hard_test_votes >= 2).astype(int)

    # 9. Validation-based threshold selection
    # Hard-vote output is binary (0 or 1).
    # Evaluate the two possible decision thresholds.
    candidates = [0, 1]

    best_threshold = 1
    best_recall = -1
    best_fpr = 1

    for threshold in candidates:

        val_pred = (hard_val >= threshold).astype(int)

        tn, fp, fn, tp = confusion_matrix(
            y_val,
            val_pred,
            labels=[0, 1]
        ).ravel()

        val_fpr = fp / (fp + tn)
        val_recall = tp / (tp + fn)

        if val_fpr <= 0.01 and val_recall > best_recall:
            best_threshold = threshold
            best_recall = val_recall
            best_fpr = val_fpr

    # 10. Apply selected threshold to test predictions
    final_hard_test = (
        hard_test >= best_threshold
    ).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_test,
        final_hard_test,
        labels=[0, 1]
    ).ravel()

    hard_voting_store[category] = {
        "y_test": y_test.to_numpy(),
        "RF_predictions": rf_test.copy(),
        "ET_predictions": et_test.copy(),
        "XGB_predictions": xgb_test.copy(),
        "Hard_Voting_predictions": final_hard_test.copy(),
        "threshold": best_threshold,
        "features": selected.copy()
    }

    hard_voting_results.append({
        "Category": category,
        "Features": len(selected),
        "Threshold": best_threshold,
        "Validation_Recall": best_recall,
        "Validation_FPR": best_fpr,
        "Test_Recall": tp / (tp + fn),
        "Test_FPR": fp / (fp + tn),
        "FP": fp,
        "FN": fn
    })

    print(
        "Hard Voting",
        "| Features:", len(selected),
        "| Threshold:", best_threshold,
        "| FP:", fp,
        "| FN:", fn
    )

hard_voting_df = pd.DataFrame(hard_voting_results)

print("\n" + "=" * 75)
print("HARD VOTING RESULTS")
print("=" * 75)

display(hard_voting_df.round(6))

print("\nStored folds:", len(hard_voting_store))
```

```python
# ============================================================
# STEP 145: THREE-MODEL AGGREGATE PERFORMANCE COMPARISON
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score
)

model_predictions = {
    "Extra Trees": [],
    "Hard Voting": [],
    "Soft Voting": []
}

all_actual = []

for category in leave_one_out_folds:

    y_true = loco_prediction_store[category]["y_test"]

    all_actual.extend(y_true)

    model_predictions["Extra Trees"].extend(
        loco_prediction_store[category]["models"]
        ["Extra Trees"]["predictions"]
    )

    model_predictions["Soft Voting"].extend(
        loco_prediction_store[category]["models"]
        ["Soft Voting"]["predictions"]
    )

    model_predictions["Hard Voting"].extend(
        hard_voting_store[category]["Hard_Voting_predictions"]
    )

all_actual = np.array(all_actual)

three_model_results = []

for model, predictions in model_predictions.items():

    predictions = np.array(predictions)

    tn, fp, fn, tp = confusion_matrix(
        all_actual,
        predictions,
        labels=[0, 1]
    ).ravel()

    three_model_results.append({
        "Model": model,
        "Accuracy": accuracy_score(all_actual, predictions),
        "Precision": precision_score(all_actual, predictions),
        "Recall": recall_score(all_actual, predictions),
        "Specificity": tn / (tn + fp),
        "F1": f1_score(all_actual, predictions),
        "Balanced Accuracy": balanced_accuracy_score(
            all_actual, predictions
        ),
        "FPR": fp / (fp + tn),
        "FP": fp,
        "FN": fn
    })

three_model_summary = pd.DataFrame(three_model_results)

print("=" * 100)
print("THREE-MODEL AGGREGATE LOCO COMPARISON")
print("=" * 100)

display(
    three_model_summary.style.format({
        "Accuracy": "{:.6%}",
        "Precision": "{:.6%}",
        "Recall": "{:.6%}",
        "Specificity": "{:.6%}",
        "F1": "{:.6%}",
        "Balanced Accuracy": "{:.6%}",
        "FPR": "{:.6%}"
    })
)
```

```python
# ============================================================
# STEP 146: CATEGORY-WISE THREE-MODEL COMPARISON
# ============================================================

category_comparison = []

for category in leave_one_out_folds:

    y_true = loco_prediction_store[category]["y_test"]

    predictions = {
        "Extra Trees": loco_prediction_store[category]
            ["models"]["Extra Trees"]["predictions"],

        "Hard Voting": hard_voting_store[category]
            ["Hard_Voting_predictions"],

        "Soft Voting": loco_prediction_store[category]
            ["models"]["Soft Voting"]["predictions"]
    }

    for model, y_pred in predictions.items():

        tn, fp, fn, tp = confusion_matrix(
            y_true,
            y_pred,
            labels=[0, 1]
        ).ravel()

        category_comparison.append({
            "Category": category,
            "Model": model,
            "FP": fp,
            "FN": fn,
            "Recall": tp / (tp + fn),
            "FPR": fp / (fp + tn)
        })

category_comparison_df = pd.DataFrame(category_comparison)

print("=" * 85)
print("CATEGORY-WISE PERFORMANCE COMPARISON")
print("=" * 85)

display(
    category_comparison_df.style.format({
        "Recall": "{:.6%}",
        "FPR": "{:.6%}"
    })
)
```

```python
# ============================================================
# STEP 147: INVESTIGATE HARD VOTING FALSE NEGATIVES
# ============================================================

hard_fn_analysis = []

for category in leave_one_out_folds:

    store = hard_voting_store[category]

    y_true = store["y_test"]

    rf_pred = store["RF_predictions"]
    et_pred = store["ET_predictions"]
    xgb_pred = store["XGB_predictions"]

    hard_pred = store["Hard_Voting_predictions"]

    # Identify actual attacks missed by Hard Voting
    fn_indices = np.where(
        (y_true == 1) & (hard_pred == 0)
    )[0]

    for idx in fn_indices:

        votes = int(
            rf_pred[idx] +
            et_pred[idx] +
            xgb_pred[idx]
        )

        hard_fn_analysis.append({
            "Category": category,
            "RF": int(rf_pred[idx]),
            "Extra Trees": int(et_pred[idx]),
            "XGBoost": int(xgb_pred[idx]),
            "Attack Votes": votes
        })

hard_fn_df = pd.DataFrame(hard_fn_analysis)

print("=" * 75)
print("HARD VOTING FALSE-NEGATIVE INVESTIGATION")
print("=" * 75)

print("\nTotal missed attacks:", len(hard_fn_df))

print("\nIndividual classifier voting patterns:")

display(
    hard_fn_df.groupby(
        ["RF", "Extra Trees", "XGBoost", "Attack Votes"]
    ).size().reset_index(name="Count")
)

print("\nMissed attacks by category:")

display(
    hard_fn_df.groupby("Category").size().reset_index(
        name="Missed Attacks"
    )
)
```

```python
# ============================================================
# STEP 149: MISSED VS CORRECTLY DETECTED ATTACK COMPARISON
# ============================================================

comparison_features = [
    "Flow Duration",
    "Flow IAT Mean",
    "Flow IAT Max",
    "Fwd IAT Total",
    "Fwd IAT Mean",
    "Fwd Packets/s",
    "Flow Packets/s",
    "Packet Length Variance",
    "Idle Mean",
    "Idle Max",
    "Source Port"
]

missed_attack_data = []
detected_attack_data = []

for category in leave_one_out_folds:

    store = hard_voting_store[category]
    fold_data = leave_one_out_folds[category]

    y_true = store["y_test"]

    rf_pred = store["RF_predictions"]
    et_pred = store["ET_predictions"]
    xgb_pred = store["XGB_predictions"]

    X_test = pd.concat(
        [
            fold_data["benign_test"],
            fold_data["attack_test"]
        ],
        ignore_index=True
    )

    # Attacks missed by all three classifiers
    missed_mask = (
        (y_true == 1) &
        (rf_pred == 0) &
        (et_pred == 0) &
        (xgb_pred == 0)
    )

    # Attacks correctly detected by all three classifiers
    detected_mask = (
        (y_true == 1) &
        (rf_pred == 1) &
        (et_pred == 1) &
        (xgb_pred == 1)
    )

    missed_attack_data.append(
        X_test.loc[missed_mask, comparison_features]
    )

    detected_attack_data.append(
        X_test.loc[detected_mask, comparison_features]
    )

missed_comparison = pd.concat(
    missed_attack_data,
    ignore_index=True
)

detected_comparison = pd.concat(
    detected_attack_data,
    ignore_index=True
)

comparison_summary = pd.DataFrame({
    "Missed_Attacks_Median": missed_comparison.median(),
    "Detected_Attacks_Median": detected_comparison.median()
})

comparison_summary["Difference"] = (
    comparison_summary["Missed_Attacks_Median"] -
    comparison_summary["Detected_Attacks_Median"]
)

print("Missed attack samples:", len(missed_comparison))
print("Correctly detected by all three:", len(detected_comparison))

print("\nFEATURE MEDIAN COMPARISON")

display(comparison_summary)
```

```python
# ============================================================
# STEP 159: REPRODUCE CORRECTED EXTRA TREES MODELS
# AND EXTRACT FEATURE IMPORTANCE
# ============================================================

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.ensemble import ExtraTreesClassifier

reproduced_et_models = {}
importance_results = []
verification_results = []

for category, fold in leave_one_out_folds.items():

    print("\nProcessing:", category)

    # ------------------------------------------------
    # 1. Reconstruct original training data
    # ------------------------------------------------

    attack_train = fold["attack_train"].copy()
    benign_train = fold["benign_train"].copy()

    attack_train["Target"] = 1
    benign_train["Target"] = 0

    train_data = pd.concat(
        [attack_train, benign_train],
        ignore_index=True
    )

    attack_test = fold["attack_test"].copy()
    benign_test = fold["benign_test"].copy()

    attack_test["Target"] = 1
    benign_test["Target"] = 0

    test_data = pd.concat(
        [benign_test, attack_test],
        ignore_index=True
    )

    X = train_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    )

    y = train_data["Target"].astype(int)

    X_test = test_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    )

    # ------------------------------------------------
    # 2. Reproduce original train-validation split
    # ------------------------------------------------

    X_train, X_val, y_train, y_val = train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )

    # ------------------------------------------------
    # 3. Reproduce training-only feature selection
    # ------------------------------------------------

    non_constant = [
        col for col in X_train.columns
        if X_train[col].nunique(dropna=True) > 1
    ]

    corr_matrix = X_train[non_constant].corr().abs()

    upper = corr_matrix.where(
        np.triu(
            np.ones(corr_matrix.shape),
            k=1
        ).astype(bool)
    )

    remove_features = [
        col for col in upper.columns
        if any(upper[col] >= 0.95)
    ]

    selected = [
        col for col in non_constant
        if col not in remove_features
    ]

    # Verify selected features
    original_features = loco_prediction_store[category]["features"]

    assert selected == original_features, (
        f"Feature selection mismatch in {category}"
    )

    X_train = X_train[selected]
    X_test = X_test[selected]

    # ------------------------------------------------
    # 4. Reproduce original imputation
    # ------------------------------------------------

    imputer = SimpleImputer(strategy="median")

    X_train_imp = imputer.fit_transform(X_train)
    X_test_imp = imputer.transform(X_test)

    # ------------------------------------------------
    # 5. Reproduce original Extra Trees model
    # ------------------------------------------------

    et = ExtraTreesClassifier(
        n_estimators=100,
        max_depth=20,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=1
    )

    et.fit(X_train_imp, y_train)

    # ------------------------------------------------
    # 6. Verify against original stored probabilities
    # ------------------------------------------------

    reproduced_probabilities = et.predict_proba(
        X_test_imp
    )[:, 1]

    original_probabilities = (
        loco_prediction_store[category]
        ["models"]["Extra Trees"]["test_probabilities"]
    )

    max_difference = np.max(
        np.abs(
            reproduced_probabilities -
            original_probabilities
        )
    )

    probabilities_match = np.allclose(
        reproduced_probabilities,
        original_probabilities,
        rtol=1e-7,
        atol=1e-9
    )

    verification_results.append({
        "Category": category,
        "Features": len(selected),
        "Maximum Probability Difference": max_difference,
        "Exact Reproduction": probabilities_match
    })

    # ------------------------------------------------
    # 7. Store reproduced model
    # ------------------------------------------------

    reproduced_et_models[category] = {
        "model": et,
        "imputer": imputer,
        "features": selected
    }

    # ------------------------------------------------
    # 8. Extract feature importance
    # ------------------------------------------------

    for feature, importance in zip(
        selected,
        et.feature_importances_
    ):

        importance_results.append({
            "Category": category,
            "Feature": feature,
            "Importance": importance
        })

# ============================================================
# 9. Display reproduction verification
# ============================================================

verification_df = pd.DataFrame(
    verification_results
)

print("\n" + "=" * 85)
print("MODEL REPRODUCTION VERIFICATION")
print("=" * 85)

display(verification_df)

# ============================================================
# 10. Display Source Port importance
# ============================================================

importance_df = pd.DataFrame(importance_results)

source_port_df = importance_df[
    importance_df["Feature"] == "Source Port"
].copy()

print("\n" + "=" * 85)
print("SOURCE PORT IMPORTANCE ACROSS CORRECTED LOCO FOLDS")
print("=" * 85)

display(source_port_df)

# ============================================================
# 11. Display top 10 features per fold
# ============================================================

print("\n" + "=" * 85)
print("TOP 10 FEATURES PER FOLD")
print("=" * 85)

for category in leave_one_out_folds:

    print("\n", category)

    display(
        importance_df[
            importance_df["Category"] == category
        ]
        .nlargest(10, "Importance")
        .reset_index(drop=True)
    )
```

```python
# ============================================================
# STEP 161: INBOUND FEATURE ABLATION
# CORRECTED LOCO EXTRA TREES
# ============================================================

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.ensemble import ExtraTreesClassifier
from sklearn.metrics import confusion_matrix

inbound_ablation_results = []

for category, fold in leave_one_out_folds.items():

    print("\nProcessing:", category)

    # ------------------------------------------------
    # 1. Reconstruct original training data
    # ------------------------------------------------

    attack_train = fold["attack_train"].copy()
    benign_train = fold["benign_train"].copy()

    attack_train["Target"] = 1
    benign_train["Target"] = 0

    train_data = pd.concat(
        [attack_train, benign_train],
        ignore_index=True
    )

    attack_test = fold["attack_test"].copy()
    benign_test = fold["benign_test"].copy()

    attack_test["Target"] = 1
    benign_test["Target"] = 0

    test_data = pd.concat(
        [benign_test, attack_test],
        ignore_index=True
    )

    X = train_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    )

    y = train_data["Target"].astype(int)

    X_test = test_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    )

    y_test = test_data["Target"].astype(int)

    # ------------------------------------------------
    # 2. Original train-validation split
    # ------------------------------------------------

    X_train, X_val, y_train, y_val = train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )

    # ------------------------------------------------
    # 3. Reproduce original feature selection
    # ------------------------------------------------

    non_constant = [
        col for col in X_train.columns
        if X_train[col].nunique(dropna=True) > 1
    ]

    corr_matrix = X_train[non_constant].corr().abs()

    upper = corr_matrix.where(
        np.triu(
            np.ones(corr_matrix.shape),
            k=1
        ).astype(bool)
    )

    remove_features = [
        col for col in upper.columns
        if any(upper[col] >= 0.95)
    ]

    selected = [
        col for col in non_constant
        if col not in remove_features
    ]

    # Remove only Inbound
    ablation_features = [
        col for col in selected
        if col != "Inbound"
    ]

    X_train = X_train[ablation_features]
    X_val = X_val[ablation_features]
    X_test = X_test[ablation_features]

    # ------------------------------------------------
    # 4. Training-only imputation
    # ------------------------------------------------

    imputer = SimpleImputer(strategy="median")

    X_train_imp = imputer.fit_transform(X_train)
    X_val_imp = imputer.transform(X_val)
    X_test_imp = imputer.transform(X_test)

    # ------------------------------------------------
    # 5. Train Extra Trees
    # ------------------------------------------------

    et_ablation = ExtraTreesClassifier(
        n_estimators=100,
        max_depth=20,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=1
    )

    et_ablation.fit(X_train_imp, y_train)

    # ------------------------------------------------
    # 6. Validation probabilities and threshold
    # ------------------------------------------------

    val_probs = et_ablation.predict_proba(
        X_val_imp
    )[:, 1]

    test_probs = et_ablation.predict_proba(
        X_test_imp
    )[:, 1]

    candidates = np.unique(
        np.concatenate([
            np.linspace(0.001, 0.99, 300),
            val_probs
        ])
    )

    best_threshold = 0.5
    best_recall = -1
    best_fpr = 1

    for threshold in candidates:

        val_pred = (
            val_probs >= threshold
        ).astype(int)

        tn, fp, fn, tp = confusion_matrix(
            y_val,
            val_pred,
            labels=[0, 1]
        ).ravel()

        val_fpr = fp / (fp + tn)
        val_recall = tp / (tp + fn)

        if val_fpr <= 0.01 and val_recall > best_recall:

            best_threshold = threshold
            best_recall = val_recall
            best_fpr = val_fpr

    # ------------------------------------------------
    # 7. Evaluate held-out category
    # ------------------------------------------------

    test_pred = (
        test_probs >= best_threshold
    ).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_test,
        test_pred,
        labels=[0, 1]
    ).ravel()

    inbound_ablation_results.append({
        "Category": category,
        "Features": len(ablation_features),
        "Threshold": best_threshold,
        "Validation_Recall": best_recall,
        "Validation_FPR": best_fpr,
        "Test_Recall": tp / (tp + fn),
        "Test_FPR": fp / (fp + tn),
        "FP": fp,
        "FN": fn
    })

    print(
        "Features:", len(ablation_features),
        "| Threshold:", round(best_threshold, 6),
        "| FP:", fp,
        "| FN:", fn
    )

# ============================================================
# 8. Display ablation results
# ============================================================

inbound_ablation_df = pd.DataFrame(
    inbound_ablation_results
)

print("\n" + "=" * 85)
print("EXTRA TREES — WITHOUT INBOUND")
print("=" * 85)

display(
    inbound_ablation_df.style.format({
        "Threshold": "{:.6f}",
        "Validation_Recall": "{:.6%}",
        "Validation_FPR": "{:.6%}",
        "Test_Recall": "{:.6%}",
        "Test_FPR": "{:.6%}"
    })
)

print("\nAggregate FP:", inbound_ablation_df["FP"].sum())
print("Aggregate FN:", inbound_ablation_df["FN"].sum())
```

```python
# ============================================================
# STEP 163: STORE INBOUND ABLATION PREDICTIONS
# ============================================================

inbound_ablation_prediction_store = {}

for category, fold in leave_one_out_folds.items():

    # Reconstruct training data
    attack_train = fold["attack_train"].copy()
    benign_train = fold["benign_train"].copy()

    attack_train["Target"] = 1
    benign_train["Target"] = 0

    train_data = pd.concat(
        [attack_train, benign_train],
        ignore_index=True
    )

    # Reconstruct test data
    attack_test = fold["attack_test"].copy()
    benign_test = fold["benign_test"].copy()

    attack_test["Target"] = 1
    benign_test["Target"] = 0

    test_data = pd.concat(
        [benign_test, attack_test],
        ignore_index=True
    )

    X = train_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    )

    y = train_data["Target"].astype(int)

    X_test = test_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    )

    y_test = test_data["Target"].astype(int)

    # Original split
    X_train, X_val, y_train, y_val = train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )

    # Training-only feature selection
    non_constant = [
        col for col in X_train.columns
        if X_train[col].nunique(dropna=True) > 1
    ]

    corr_matrix = X_train[non_constant].corr().abs()

    upper = corr_matrix.where(
        np.triu(
            np.ones(corr_matrix.shape),
            k=1
        ).astype(bool)
    )

    remove_features = [
        col for col in upper.columns
        if any(upper[col] >= 0.95)
    ]

    selected = [
        col for col in non_constant
        if col not in remove_features
        and col != "Inbound"
    ]

    X_train = X_train[selected]
    X_val = X_val[selected]
    X_test = X_test[selected]

    # Imputation
    imputer = SimpleImputer(strategy="median")

    X_train_imp = imputer.fit_transform(X_train)
    X_val_imp = imputer.transform(X_val)
    X_test_imp = imputer.transform(X_test)

    # Train Extra Trees
    model = ExtraTreesClassifier(
        n_estimators=100,
        max_depth=20,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=1
    )

    model.fit(X_train_imp, y_train)

    # Validation probabilities
    val_probs = model.predict_proba(X_val_imp)[:, 1]

    # Test probabilities
    test_probs = model.predict_proba(X_test_imp)[:, 1]

    # Select threshold using validation FPR constraint
    candidates = np.unique(
        np.concatenate([
            np.linspace(0.001, 0.99, 300),
            val_probs
        ])
    )

    best_threshold = 0.5
    best_recall = -1

    for threshold in candidates:

        val_pred = (val_probs >= threshold).astype(int)

        tn, fp, fn, tp = confusion_matrix(
            y_val, val_pred, labels=[0, 1]
        ).ravel()

        fpr = fp / (fp + tn)
        recall = tp / (tp + fn)

        if fpr <= 0.01 and recall > best_recall:
            best_threshold = threshold
            best_recall = recall

    test_pred = (
        test_probs >= best_threshold
    ).astype(int)

    # Store predictions and actual labels
    inbound_ablation_prediction_store[category] = {
        "y_test": y_test.to_numpy(),
        "predictions": test_pred,
        "threshold": best_threshold,
        "features": selected
    }

print("Stored ablation predictions for all four folds.")

# Compare identities of missed attacks
comparison_results = []

for category in leave_one_out_folds:

    y_true = inbound_ablation_prediction_store[category]["y_test"]

    original_pred = (
        loco_prediction_store[category]
        ["models"]["Extra Trees"]["predictions"]
    )

    ablation_pred = (
        inbound_ablation_prediction_store[category]
        ["predictions"]
    )

    original_fn = (y_true == 1) & (original_pred == 0)
    ablation_fn = (y_true == 1) & (ablation_pred == 0)

    newly_missed = ablation_fn & ~original_fn

    comparison_results.append({
        "Category": category,
        "Original_FN": original_fn.sum(),
        "Ablation_FN": ablation_fn.sum(),
        "Newly_Missed": newly_missed.sum()
    })

    print(
        category,
        "| Original FN:", original_fn.sum(),
        "| Ablation FN:", ablation_fn.sum(),
        "| Newly missed:", newly_missed.sum()
    )

print("\nCATEGORY-WISE COMPARISON")

display(pd.DataFrame(comparison_results))
```

```python
# ============================================================
# STEP 164: CHECK OVERLAP BETWEEN THE TWO SETS OF MISSED ATTACKS
# ============================================================

print("Available columns in all_model_misses_df:")
print(all_model_misses_df.columns.tolist())

print("\nAvailable fold information:")
for category, fold in leave_one_out_folds.items():
    print(category, ":", list(fold.keys()))

# Collect positions of the 7 attacks missed without Inbound
ablation_misses = {}

for category, result in inbound_ablation_prediction_store.items():

    y_true = result["y_test"]
    y_pred = result["predictions"]

    missed_positions = np.where(
        (y_true == 1) & (y_pred == 0)
    )[0]

    ablation_misses[category] = missed_positions

    print(
        f"\n{category}:",
        missed_positions.tolist()
    )

# Display the original 18 all-model misses
print("\nOriginal 18 all-model misses:")
display(all_model_misses_df)

# Compare category-wise positions
print("\nOverlap comparison:")

overlap_results = []

for category, positions in ablation_misses.items():

    original_rows = all_model_misses_df[
        all_model_misses_df["Category"] == category
    ]

    original_positions = set(
        original_rows["Test_Index"].astype(int)
    )

    ablation_positions = set(positions.tolist())

    overlap = ablation_positions.intersection(
        original_positions
    )

    overlap_results.append({
        "Category": category,
        "Ablation_FN": len(ablation_positions),
        "Original_All_Model_FN": len(original_positions),
        "Overlap": len(overlap),
        "Ablation_Only": len(ablation_positions - original_positions),
        "Original_Only": len(original_positions - ablation_positions)
    })

display(pd.DataFrame(overlap_results))
```

```python
# ============================================================
# STEP 165: INSPECT THE SEVEN ATTACKS MISSED WITHOUT INBOUND
# ============================================================

important_features = [
    "Source Port",
    "Destination Port",
    "Protocol",
    "Flow Duration",
    "Total Fwd Packets",
    "Total Backward Packets",
    "Total Length of Fwd Packets",
    "Total Length of Bwd Packets",
    "Flow Bytes/s",
    "Flow Packets/s",
    "Flow IAT Mean",
    "Flow IAT Max",
    "Fwd Packets/s",
    "Min Packet Length",
    "Max Packet Length",
    "Packet Length Mean",
    "Packet Length Variance",
    "Inbound"
]

missed_sample_rows = []

for category, result in inbound_ablation_prediction_store.items():

    fold = leave_one_out_folds[category]

    # Maintain the exact test ordering used during evaluation:
    # benign samples first, followed by attack samples.
    test_data = pd.concat(
        [
            fold["benign_test"],
            fold["attack_test"]
        ],
        ignore_index=True
    )

    y_true = result["y_test"]
    y_pred = result["predictions"]

    missed_positions = np.where(
        (y_true == 1) & (y_pred == 0)
    )[0]

    original_rows = all_model_misses_df[
        all_model_misses_df["Category"] == category
    ]

    original_indices = set(
        original_rows["Test_Index"].astype(int)
    )

    for idx in missed_positions:

        row = test_data.iloc[idx]

        sample = {
            "Category": category,
            "Test_Index": idx,
            "Previously_Missed_By_All_3": (
                idx in original_indices
            )
        }

        for feature in important_features:
            sample[feature] = row[feature]

        missed_sample_rows.append(sample)

missed_samples_df = pd.DataFrame(missed_sample_rows)

print("Total missed attack samples:", len(missed_samples_df))

display(
    missed_samples_df.style.format(precision=4)
)
```

```python
# ============================================================
# STEP 166: COMPARE MISSED AND DETECTED ATTACK CHARACTERISTICS
# ============================================================

comparison_features = [
    "Flow Duration",
    "Total Fwd Packets",
    "Total Backward Packets",
    "Total Length of Fwd Packets",
    "Total Length of Bwd Packets",
    "Flow Bytes/s",
    "Flow Packets/s",
    "Flow IAT Mean",
    "Fwd Packets/s",
    "Packet Length Mean",
    "Packet Length Variance",
    "Inbound"
]

comparison_tables = []

for category, result in inbound_ablation_prediction_store.items():

    fold = leave_one_out_folds[category]

    test_data = pd.concat(
        [
            fold["benign_test"],
            fold["attack_test"]
        ],
        ignore_index=True
    )

    y_true = result["y_test"]
    y_pred = result["predictions"]

    attack_mask = (y_true == 1)

    missed_mask = attack_mask & (y_pred == 0)
    detected_mask = attack_mask & (y_pred == 1)

    missed_data = test_data.loc[missed_mask, comparison_features]
    detected_data = test_data.loc[detected_mask, comparison_features]

    print(f"\n{'='*70}")
    print(category)
    print("Missed attacks:", len(missed_data))
    print("Detected attacks:", len(detected_data))

    if len(missed_data) == 0:
        print("No missed attacks in this category.")
        continue

    summary = pd.DataFrame({
        "Missed_Median": missed_data.median(),
        "Detected_Median": detected_data.median(),
        "Missed_Mean": missed_data.mean(),
        "Detected_Mean": detected_data.mean()
    })

    comparison_tables.append({
        "Category": category,
        "Summary": summary
    })

    display(summary.round(4))
```

```python
# ============================================================
# STEP 168: COMPARE ORIGINAL AND ABLATION CONFIDENCE
# ============================================================

ablation_probability_store = {}
confidence_comparison = []

for category, fold in leave_one_out_folds.items():

    attack_train = fold["attack_train"].copy()
    benign_train = fold["benign_train"].copy()

    attack_train["Target"] = 1
    benign_train["Target"] = 0

    train_data = pd.concat(
        [attack_train, benign_train],
        ignore_index=True
    )

    attack_test = fold["attack_test"].copy()
    benign_test = fold["benign_test"].copy()

    attack_test["Target"] = 1
    benign_test["Target"] = 0

    test_data = pd.concat(
        [benign_test, attack_test],
        ignore_index=True
    )

    X = train_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    )

    y = train_data["Target"].astype(int)

    X_test = test_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    )

    y_test = test_data["Target"].astype(int)

    X_train, X_val, y_train, y_val = train_test_split(
        X, y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )

    non_constant = [
        col for col in X_train.columns
        if X_train[col].nunique(dropna=True) > 1
    ]

    corr_matrix = X_train[non_constant].corr().abs()

    upper = corr_matrix.where(
        np.triu(
            np.ones(corr_matrix.shape),
            k=1
        ).astype(bool)
    )

    remove_features = [
        col for col in upper.columns
        if any(upper[col] >= 0.95)
    ]

    selected = [
        col for col in non_constant
        if col not in remove_features
        and col != "Inbound"
    ]

    imputer = SimpleImputer(strategy="median")

    X_train_imp = imputer.fit_transform(
        X_train[selected]
    )

    X_val_imp = imputer.transform(
        X_val[selected]
    )

    X_test_imp = imputer.transform(
        X_test[selected]
    )

    model = ExtraTreesClassifier(
        n_estimators=100,
        max_depth=20,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=1
    )

    model.fit(X_train_imp, y_train)

    val_probs = model.predict_proba(X_val_imp)[:, 1]
    test_probs = model.predict_proba(X_test_imp)[:, 1]

    candidates = np.unique(
        np.concatenate([
            np.linspace(0.001, 0.99, 300),
            val_probs
        ])
    )

    best_threshold = 0.5
    best_recall = -1

    for threshold in candidates:

        val_pred = (val_probs >= threshold).astype(int)

        tn, fp, fn, tp = confusion_matrix(
            y_val, val_pred, labels=[0, 1]
        ).ravel()

        fpr = fp / (fp + tn)
        recall = tp / (tp + fn)

        if fpr <= 0.01 and recall > best_recall:
            best_threshold = threshold
            best_recall = recall

    ablation_probability_store[category] = {
        "test_probabilities": test_probs,
        "threshold": best_threshold
    }

    # Identify missed attacks
    missed_positions = np.where(
        (y_test.to_numpy() == 1) &
        (inbound_ablation_prediction_store[category]["predictions"] == 0)
    )[0]

    original = loco_prediction_store[category]["models"]["Extra Trees"]

    for idx in missed_positions:

        original_prob = original["test_probabilities"][idx]
        ablation_prob = test_probs[idx]

        confidence_comparison.append({
            "Category": category,
            "Test_Index": idx,
            "Original_Probability": original_prob,
            "Original_Threshold": original["threshold"],
            "Ablation_Probability": ablation_prob,
            "Ablation_Threshold": best_threshold,
            "Probability_Change": ablation_prob - original_prob
        })

confidence_df = pd.DataFrame(confidence_comparison)

print("Seven missed attacks — probability comparison:")

display(
    confidence_df.round(6)
)
```

```python
# ============================================================
# STEP 169: COMPARE CLASSIFICATION MARGINS
# ============================================================

margin_df = confidence_df.copy()

margin_df["Original_Margin"] = (
    margin_df["Original_Probability"]
    - margin_df["Original_Threshold"]
)

margin_df["Ablation_Margin"] = (
    margin_df["Ablation_Probability"]
    - margin_df["Ablation_Threshold"]
)

margin_df["Margin_Change"] = (
    margin_df["Ablation_Margin"]
    - margin_df["Original_Margin"]
)

display(
    margin_df[
        [
            "Category",
            "Test_Index",
            "Original_Margin",
            "Ablation_Margin",
            "Margin_Change"
        ]
    ].round(6)
)

print("\nSummary:")
print(
    "Original positive margins:",
    (margin_df["Original_Margin"] > 0).sum()
)

print(
    "Ablation negative margins:",
    (margin_df["Ablation_Margin"] < 0).sum()
)

print(
    "Mean original margin:",
    round(margin_df["Original_Margin"].mean(), 6)
)

print(
    "Mean ablation margin:",
    round(margin_df["Ablation_Margin"].mean(), 6)
)
```

```python
# ============================================================
# STEP 170: FINAL INBOUND ABLATION PERFORMANCE COMPARISON
# ============================================================

original_total = {
    "TN": 2975,
    "FP": 25,
    "FN": 0,
    "TP": 200000
}

ablation_total = {
    "TN": 2975,
    "FP": 25,
    "FN": 7,
    "TP": 199993
}

def calculate_metrics(cm):

    tn, fp = cm["TN"], cm["FP"]
    fn, tp = cm["FN"], cm["TP"]

    accuracy = (tn + tp) / (tn + fp + fn + tp)
    precision = tp / (tp + fp)
    recall = tp / (tp + fn)
    specificity = tn / (tn + fp)
    f1 = 2 * precision * recall / (precision + recall)
    balanced_accuracy = (recall + specificity) / 2
    fpr = fp / (fp + tn)

    return {
        "Accuracy (%)": accuracy * 100,
        "Precision (%)": precision * 100,
        "Recall (%)": recall * 100,
        "Specificity (%)": specificity * 100,
        "F1 (%)": f1 * 100,
        "Balanced Accuracy (%)": balanced_accuracy * 100,
        "FPR (%)": fpr * 100,
        "FP": fp,
        "FN": fn
    }

final_ablation_comparison = pd.DataFrame({
    "Original Extra Trees": calculate_metrics(original_total),
    "Without Inbound": calculate_metrics(ablation_total)
}).T

display(final_ablation_comparison.round(6))
```

```python
# ============================================================
# STEP 175: MASTER EXPERIMENTAL PERFORMANCE SUMMARY
# ============================================================

# 1. Recreate LOCO ensemble comparison

models_cm = {
    "Extra Trees": {"TN": 2975, "FP": 25, "FN": 0, "TP": 200000},
    "Soft Voting": {"TN": 2973, "FP": 27, "FN": 0, "TP": 200000},
    "Hard Voting": {"TN": 2992, "FP": 8, "FN": 37, "TP": 199963}
}

comparison_rows = []

for model_name, cm in models_cm.items():

    tn, fp = cm["TN"], cm["FP"]
    fn, tp = cm["FN"], cm["TP"]

    accuracy = (tp + tn) / (tp + tn + fp + fn)
    precision = tp / (tp + fp)
    recall = tp / (tp + fn)
    specificity = tn / (tn + fp)
    f1 = 2 * precision * recall / (precision + recall)
    balanced_accuracy = (recall + specificity) / 2
    fpr = fp / (fp + tn)

    comparison_rows.append({
        "Model": model_name,
        "Accuracy (%)": accuracy * 100,
        "Precision (%)": precision * 100,
        "Recall (%)": recall * 100,
        "F1 (%)": f1 * 100,
        "Balanced Accuracy (%)": balanced_accuracy * 100,
        "FPR (%)": fpr * 100,
        "FP": fp,
        "FN": fn
    })

ensemble_comparison_df = pd.DataFrame(comparison_rows)


# 2. Build master results

master_results = []

for _, row in ensemble_comparison_df.iterrows():

    master_results.append({
        "Experiment": "LOCO",
        **row.to_dict()
    })


# SNMP zero-day results

for _, row in snmp_results_df.iterrows():

    master_results.append({
        "Experiment": "SNMP Zero-Day",
        "Model": row["Model"],
        "Accuracy (%)": row["Accuracy"] * 100,
        "Precision (%)": row["Precision"] * 100,
        "Recall (%)": row["Recall"] * 100,
        "F1 (%)": row["F1"] * 100,
        "Balanced Accuracy (%)": row["Balanced_Accuracy"] * 100,
        "FPR (%)": np.nan,
        "FP": row["FP"],
        "FN": row["FN"]
    })


# Inbound ablation

for model_name, row in final_ablation_comparison.iterrows():

    master_results.append({
        "Experiment": "Inbound Ablation",
        "Model": model_name,
        "Accuracy (%)": row["Accuracy (%)"],
        "Precision (%)": row["Precision (%)"],
        "Recall (%)": row["Recall (%)"],
        "F1 (%)": row["F1 (%)"],
        "Balanced Accuracy (%)": row["Balanced Accuracy (%)"],
        "FPR (%)": row["FPR (%)"],
        "FP": row["FP"],
        "FN": row["FN"]
    })


master_results_df = pd.DataFrame(master_results)

print("MASTER EXPERIMENTAL RESULTS")
print("Total configurations:", len(master_results_df))

display(master_results_df.round(6))
```

```python
# ============================================================
# STEP 176: ADD EVALUATION DETAILS TO MASTER RESULTS
# ============================================================

master_results_df["Benign_Samples"] = np.where(
    master_results_df["Experiment"] == "SNMP Zero-Day",
    1000,
    3000
)

master_results_df["Attack_Samples"] = np.where(
    master_results_df["Experiment"] == "SNMP Zero-Day",
    50000,
    200000
)

master_results_df["Total_Test_Samples"] = (
    master_results_df["Benign_Samples"]
    + master_results_df["Attack_Samples"]
)

# Calculate FPR from false positives and benign sample count
master_results_df["FPR (%)"] = (
    master_results_df["FP"]
    / master_results_df["Benign_Samples"]
    * 100
)

# Add evaluation grouping
master_results_df["Evaluation_Type"] = (
    master_results_df["Experiment"].map({
        "LOCO": "Attack-category holdout",
        "SNMP Zero-Day": "Held-out SNMP category",
        "Inbound Ablation": "LOCO feature ablation"
    })
)

display(
    master_results_df.round(6)
)
```

```python
# ============================================================
# STEP 177: VALIDATE MASTER TABLE METRICS
# ============================================================

validation_rows = []

for _, row in master_results_df.iterrows():

    tn = row["Benign_Samples"] - row["FP"]
    tp = row["Attack_Samples"] - row["FN"]

    fp = row["FP"]
    fn = row["FN"]

    total = tn + fp + fn + tp

    accuracy = (tn + tp) / total
    precision = tp / (tp + fp)
    recall = tp / (tp + fn)
    specificity = tn / (tn + fp)

    f1 = 2 * precision * recall / (precision + recall)

    balanced_accuracy = (
        recall + specificity
    ) / 2

    fpr = fp / (fp + tn)

    checks = {
        "Accuracy": accuracy * 100,
        "Precision": precision * 100,
        "Recall": recall * 100,
        "F1": f1 * 100,
        "Balanced Accuracy": balanced_accuracy * 100,
        "FPR": fpr * 100
    }

    differences = {}

    for metric, calculated in checks.items():

        reported = row[
            "FPR (%)" if metric == "FPR"
            else metric + " (%)"
        ]

        differences[metric] = abs(calculated - reported)

    validation_rows.append({
        "Experiment": row["Experiment"],
        "Model": row["Model"],
        "Max_Metric_Difference": max(differences.values()),
        "Validation_Status": (
            "PASS" if max(differences.values()) < 0.00001
            else "CHECK"
        )
    })

validation_df = pd.DataFrame(validation_rows)

display(validation_df)
```

```python
import pandas as pd
import numpy as np

general_features = [
    "duration", "src_bytes", "dst_bytes", "missed_bytes",
    "src_pkts", "src_ip_bytes", "dst_pkts", "dst_ip_bytes"
]

X_audit = ton_balanced[general_features].copy()
y_audit = ton_balanced["label"].astype(int)

# Identify exact feature-vector duplicates
duplicate_mask = X_audit.duplicated(keep=False)

duplicate_data = X_audit.loc[duplicate_mask].copy()
duplicate_data["label"] = y_audit.loc[duplicate_mask].values

# Count distinct labels associated with each feature vector
group_audit = (
    duplicate_data
    .groupby(general_features, dropna=False)["label"]
    .agg(["size", "nunique"])
    .reset_index()
)

conflicting_groups = group_audit[group_audit["nunique"] > 1]

print("TON-IOT DUPLICATE GROUP AUDIT")
print("=" * 65)

print("Total rows:", len(X_audit))
print("Unique feature vectors:", len(X_audit.drop_duplicates()))
print("Rows belonging to duplicate groups:", int(duplicate_mask.sum()))
print("Duplicate feature groups:", len(group_audit))
print("Groups with conflicting labels:", len(conflicting_groups))

print("\nLabel distribution:")
print(y_audit.value_counts().sort_index())

print("\nFeature missing values:")
print(X_audit.isna().sum())

print("\nConflicting duplicate groups:")
display(conflicting_groups.head(10))
```

```python
# Inspect the two conflicting feature groups in detail

for group_number, row in conflicting_groups.iterrows():

    feature_values = row[general_features].to_dict()

    mask = pd.Series(True, index=ton_balanced.index)

    for feature, value in feature_values.items():
        mask &= ton_balanced[feature].eq(value)

    matching_rows = ton_balanced.loc[mask]

    print("\n" + "=" * 70)
    print(f"CONFLICTING GROUP: {group_number}")
    print("=" * 70)

    print("Number of samples:", len(matching_rows))

    print("\nLabel counts:")
    print(matching_rows["label"].value_counts().sort_index())

    print("\nOriginal attack types:")
    print(matching_rows["type"].value_counts(dropna=False))

    print("\nSample source information:")
    available_columns = [
        c for c in ["src_ip", "dst_ip", "src_port",
                    "dst_port", "proto", "service", "type", "label"]
        if c in matching_rows.columns
    ]

    display(matching_rows[available_columns].head(10))
```

```python
# Quantify feature-level label ambiguity

total_rows = len(ton_balanced)

ambiguous_rows = int(
    group_audit.loc[
        group_audit["nunique"] > 1, "size"
    ].sum()
)

ambiguous_groups = len(conflicting_groups)

# Minimum unavoidable classification errors if each
# identical feature group receives one common prediction
minimum_errors = int(
    group_audit.loc[
        group_audit["nunique"] > 1
    ].apply(
        lambda row: min(
            row["size"] - 1,
            1
        ),
        axis=1
    ).sum()
)

print("FEATURE-LEVEL LABEL AMBIGUITY")
print("=" * 60)

print("Total dataset rows:", total_rows)
print("Ambiguous feature groups:", ambiguous_groups)
print("Rows in ambiguous groups:", ambiguous_rows)

print(
    "Percentage of dataset affected:",
    round(ambiguous_rows / total_rows * 100, 4), "%"
)

print(
    "Minimum unavoidable errors across conflicting groups:",
    minimum_errors
)

print(
    "Minimum error percentage from these groups:",
    round(minimum_errors / total_rows * 100, 4), "%"
)
```

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import confusion_matrix

# Load CIC datasets
data_dir = project_dir / "processed" / "zero_day_samples"

benign_train = pd.read_parquet(data_dir / "benign_train.parquet")
benign_test = pd.read_parquet(data_dir / "benign_test.parquet")
known_attacks = pd.read_parquet(data_dir / "known_attacks.parquet")

cic_features = [
    "Flow Duration",
    "Total Length of Fwd Packets",
    "Total Length of Bwd Packets",
    "Total Fwd Packets",
    "Total Backward Packets"
]

# Reproduce the original attack split
attack_train, attack_val = train_test_split(
    known_attacks,
    test_size=0.20,
    random_state=42,
    shuffle=True
)

# Construct source training data
X_train = pd.concat([
    benign_train[cic_features],
    attack_train[cic_features]
], ignore_index=True)

y_train = np.concatenate([
    np.zeros(len(benign_train), dtype=int),
    np.ones(len(attack_train), dtype=int)
])

# Construct source validation data
X_val = pd.concat([
    benign_test[cic_features],
    attack_val[cic_features]
], ignore_index=True)

y_val = np.concatenate([
    np.zeros(len(benign_test), dtype=int),
    np.ones(len(attack_val), dtype=int)
])

# Apply original duration conversion
for frame in [X_train, X_val]:
    frame["Flow Duration"] /= 1_000_000
    frame.replace([np.inf, -np.inf], np.nan, inplace=True)

# Rebuild original model
source_model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("rf", RandomForestClassifier(
        n_estimators=200,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ))
])

source_model.fit(X_train, y_train)

# Reproduce threshold selection
val_prob = source_model.predict_proba(X_val)[:, 1]

candidate_thresholds = np.unique(
    np.concatenate(([0.0], val_prob, [1.0]))
)

eligible = []

for threshold in candidate_thresholds:

    pred = (val_prob >= threshold).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_val, pred, labels=[0, 1]
    ).ravel()

    fpr = fp / (fp + tn)
    recall = tp / (tp + fn)

    if fpr <= 0.05:
        eligible.append((recall, threshold, fpr, tn, fp, fn, tp))

best = max(eligible, key=lambda row: (row[0], row[1]))

source_recall, fixed_threshold, source_fpr, tn, fp, fn, tp = best

print("CIC CALIBRATION RECONSTRUCTION")
print("=" * 65)

print("Training samples:", len(X_train))
print("Validation samples:", len(X_val))
print("Selected threshold:", fixed_threshold)

print("\nValidation confusion matrix:")
print(np.array([[tn, fp], [fn, tp]]))

print("\nTN:", tn)
print("FP:", fp)
print("FN:", fn)
print("TP:", tp)

print("\nValidation recall:", source_recall)
print("Validation FPR:", source_fpr)
print("FPR constraint satisfied:", source_fpr <= 0.05)
```

```python
from scipy.stats import ks_2samp
import pandas as pd
import numpy as np

comparison_pairs = {
    "Benign": (cic_benign_audit, ton_benign_audit),
    "Attack": (cic_attack_audit, ton_attack_audit)
}

ks_rows = []

for traffic_type, (source, target) in comparison_pairs.items():

    for feature in cic_features:

        source_values = (
            source[feature]
            .replace([np.inf, -np.inf], np.nan)
            .dropna()
        )

        target_values = (
            target[feature]
            .replace([np.inf, -np.inf], np.nan)
            .dropna()
        )

        statistic, p_value = ks_2samp(
            source_values,
            target_values
        )

        ks_rows.append({
            "Traffic_Type": traffic_type,
            "Feature": feature,
            "CIC_Samples": len(source_values),
            "ToN_Samples": len(target_values),
            "KS_Statistic": statistic,
            "P_Value": p_value
        })

ks_results = pd.DataFrame(ks_rows)

print("CIC vs ToN-IoT: KOLMOGOROV-SMIRNOV TEST")
print("=" * 85)

display(
    ks_results.sort_values(
        ["Traffic_Type", "KS_Statistic"],
        ascending=[True, False]
    ).round(6)
)
```

```python
import numpy as np
import pandas as pd

duplicate_audit = []

for category, fold in leave_one_out_folds.items():

    attack_train = fold["attack_train"].copy()
    benign_train = fold["benign_train"].copy()

    attack_train["Target"] = 1
    benign_train["Target"] = 0

    train_data = pd.concat(
        [attack_train, benign_train],
        ignore_index=True
    )

    attack_test = fold["attack_test"].copy()
    benign_test = fold["benign_test"].copy()

    attack_test["Target"] = 1
    benign_test["Target"] = 0

    test_data = pd.concat(
        [benign_test, attack_test],
        ignore_index=True
    )

    X_train = train_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    ).fillna(-999999)

    X_test = test_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    ).fillna(-999999)

    train_hashes = pd.util.hash_pandas_object(
        X_train, index=False
    )

    test_hashes = pd.util.hash_pandas_object(
        X_test, index=False
    )

    train_lookup = pd.DataFrame({
        "Hash": train_hashes.to_numpy(),
        "Train_Label": train_data["Target"].to_numpy()
    })

    label_map = (
        train_lookup.groupby("Hash")["Train_Label"]
        .apply(lambda x: set(x))
        .to_dict()
    )

    matched = [
        i for i, h in enumerate(test_hashes)
        if h in label_map
    ]

    same_class = 0
    opposite_class = 0
    conflicting = 0

    for i in matched:

        test_label = int(test_data.iloc[i]["Target"])
        train_labels = label_map[test_hashes.iloc[i]]

        if len(train_labels) > 1:
            conflicting += 1
        elif test_label in train_labels:
            same_class += 1
        else:
            opposite_class += 1

    duplicate_audit.append({
        "Held_Out_Category": category,
        "Training_Rows": len(train_data),
        "Testing_Rows": len(test_data),
        "Duplicate_Test_Rows": len(matched),
        "Same_Class": same_class,
        "Opposite_Class": opposite_class,
        "Conflicting_Training_Labels": conflicting
    })

duplicate_audit_df = pd.DataFrame(duplicate_audit)

print("=" * 95)
print("LOCO TRAIN-TEST EXACT DUPLICATE AUDIT")
print("=" * 95)

display(duplicate_audit_df)

print("\nTOTALS")
display(duplicate_audit_df.select_dtypes("number").sum().to_frame("Total"))
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import accuracy_score, confusion_matrix

duplicate_prediction_results = []

for category, fold in leave_one_out_folds.items():

    # Reconstruct test data in the original experiment's order
    attack_test = fold["attack_test"].copy()
    benign_test = fold["benign_test"].copy()

    attack_test["Target"] = 1
    benign_test["Target"] = 0

    test_data = pd.concat(
        [benign_test, attack_test],
        ignore_index=True
    )

    X_test = test_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    ).fillna(-999999)

    y_test = test_data["Target"].to_numpy()

    # Identify exact matches with the corresponding training fold
    attack_train = fold["attack_train"].copy()
    benign_train = fold["benign_train"].copy()

    X_train = pd.concat(
        [attack_train, benign_train],
        ignore_index=True
    )[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    ).fillna(-999999)

    train_hashes = pd.util.hash_pandas_object(
        X_train, index=False
    )

    test_hashes = pd.util.hash_pandas_object(
        X_test, index=False
    )

    duplicate_mask = test_hashes.isin(
        set(train_hashes)
    ).to_numpy()

    # Retrieve the actual calibrated Extra Trees predictions
    stored = loco_prediction_store[category]["models"]["Extra Trees"]

    y_stored = loco_prediction_store[category]["y_test"]
    predictions = stored["predictions"]

    # Verify alignment before evaluation
    assert len(predictions) == len(y_test)
    assert np.array_equal(y_stored, y_test)

    for group_name, mask in [
        ("Duplicate", duplicate_mask),
        ("Non-Duplicate", ~duplicate_mask)
    ]:

        if mask.sum() == 0:
            continue

        actual = y_test[mask]
        predicted = predictions[mask]

        tn, fp, fn, tp = confusion_matrix(
            actual, predicted, labels=[0, 1]
        ).ravel()

        duplicate_prediction_results.append({
            "Category": category,
            "Group": group_name,
            "Samples": int(mask.sum()),
            "Correct": int((actual == predicted).sum()),
            "Incorrect": int((actual != predicted).sum()),
            "Accuracy": accuracy_score(actual, predicted),
            "FP": int(fp),
            "FN": int(fn)
        })

duplicate_prediction_df = pd.DataFrame(
    duplicate_prediction_results
)

print("=" * 95)
print("CALIBRATED EXTRA TREES: DUPLICATE VS NON-DUPLICATE PERFORMANCE")
print("=" * 95)

display(duplicate_prediction_df.round(6))

print("\nOVERALL GROUP TOTALS")

display(
    duplicate_prediction_df.groupby("Group")[
        ["Samples", "Correct", "Incorrect", "FP", "FN"]
    ].sum()
)
```

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split

validation_duplicate_results = []

for category, fold in leave_one_out_folds.items():

    attack_train = fold["attack_train"].copy()
    benign_train = fold["benign_train"].copy()

    attack_train["Target"] = 1
    benign_train["Target"] = 0

    train_data = pd.concat(
        [attack_train, benign_train],
        ignore_index=True
    )

    X = train_data[feature_columns].replace(
        [np.inf, -np.inf], np.nan
    )

    y = train_data["Target"].astype(int)

    # Reproduce the original internal split exactly
    X_train, X_val, y_train, y_val = train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )

    X_train_hash = X_train.fillna(-999999)
    X_val_hash = X_val.fillna(-999999)

    train_hashes = pd.util.hash_pandas_object(
        X_train_hash, index=False
    )

    val_hashes = pd.util.hash_pandas_object(
        X_val_hash, index=False
    )

    train_lookup = pd.DataFrame({
        "Hash": train_hashes.to_numpy(),
        "Label": y_train.to_numpy()
    })

    label_map = (
        train_lookup.groupby("Hash")["Label"]
        .apply(lambda x: set(x))
        .to_dict()
    )

    matched = [
        i for i, h in enumerate(val_hashes)
        if h in label_map
    ]

    same_class = 0
    opposite_class = 0
    conflicting = 0

    for i in matched:

        validation_label = int(y_val.iloc[i])
        training_labels = label_map[val_hashes.iloc[i]]

        if len(training_labels) > 1:
            conflicting += 1
        elif validation_label in training_labels:
            same_class += 1
        else:
            opposite_class += 1

    validation_duplicate_results.append({
        "Category": category,
        "Training_Rows": len(X_train),
        "Validation_Rows": len(X_val),
        "Duplicate_Validation_Rows": len(matched),
        "Same_Class": same_class,
        "Opposite_Class": opposite_class,
        "Conflicting_Training_Labels": conflicting
    })

validation_duplicate_df = pd.DataFrame(
    validation_duplicate_results
)

print("=" * 100)
print("LOCO INTERNAL TRAIN-VALIDATION DUPLICATE AUDIT")
print("=" * 100)

display(validation_duplicate_df)

print("\nTOTALS")

display(
    validation_duplicate_df.select_dtypes("number")
    .sum()
    .to_frame("Total")
)
```

```python
from collections import defaultdict
import pandas as pd
from pathlib import Path

print("=" * 90)
print("COMPLETE PARQUET SOURCE-FILE MAPPING")
print("=" * 90)

source_mapping = defaultdict(lambda: {
    "Chunks": 0,
    "Rows": 0,
    "Source_Categories": set(),
    "Labels": set()
})

for i, path in enumerate(parquet_files):

    df = pd.read_parquet(
        path,
        columns=["Source_File", "Label"]
    )

    group = path.stem.split("_chunk_")[0]

    source_mapping[group]["Chunks"] += 1
    source_mapping[group]["Rows"] += len(df)

    source_mapping[group]["Source_Categories"].update(
        df["Source_File"].dropna().astype(str).unique()
    )

    source_mapping[group]["Labels"].update(
        df["Label"].dropna().astype(str).unique()
    )

    if (i + 1) % 250 == 0:
        print(f"Processed {i + 1}/{len(parquet_files)} files")

mapping_rows = []

for group, values in sorted(source_mapping.items()):

    mapping_rows.append({
        "File_Group": group,
        "Chunks": values["Chunks"],
        "Rows": values["Rows"],
        "Source_Categories": ", ".join(
            sorted(values["Source_Categories"])
        ),
        "Number_of_Categories": len(
            values["Source_Categories"]
        )
    })

source_mapping_df = pd.DataFrame(mapping_rows)

display(source_mapping_df)

print("\nTotal file groups:", len(source_mapping_df))
print("Total chunks:", source_mapping_df["Chunks"].sum())
print("Total rows:", source_mapping_df["Rows"].sum())
```

```python
attack_categories = [
    "DrDoS_DNS",
    "DrDoS_LDAP",
    "DrDoS_NetBIOS",
    "DrDoS_UDP"
]

leave_one_out_folds = {}

for held_out in attack_categories:

    training_categories = [
        category for category in attack_categories
        if category != held_out
    ]

    fold_attack_train = known_attacks[
        known_attacks["Source_File"].isin(training_categories)
    ].copy()

    fold_benign_train = benign_train[
        benign_train["Source_File"].isin(training_categories)
    ].copy()

    fold_attack_test = known_attacks[
        known_attacks["Source_File"] == held_out
    ].copy()

    fold_benign_test = benign_train[
        benign_train["Source_File"] == held_out
    ].copy()

    leave_one_out_folds[held_out] = {
        "training_categories": training_categories,
        "attack_train": fold_attack_train,
        "benign_train": fold_benign_train,
        "attack_test": fold_attack_test,
        "benign_test": fold_benign_test
    }

    print(
        held_out,
        "| Attack train:", len(fold_attack_train),
        "| Benign train:", len(fold_benign_train),
        "| Attack test:", len(fold_attack_test),
        "| Benign test:", len(fold_benign_test)
    )
```

```python
feature_columns = [
    col for col in known_attacks.columns
    if col not in [
        "Label",
        "Source_File",
        "Attack_Type",
        "Target"
    ]
]

print("Number of initial features:", len(feature_columns))
print("\nFeatures:")
print(feature_columns)
```

```python
print("Checking feature compatibility...")

for category, fold in leave_one_out_folds.items():

    print(
        category,
        "| Attack features:",
        len(fold["attack_train"][feature_columns].columns),
        "| Benign features:",
        len(fold["benign_train"][feature_columns].columns)
    )
```
